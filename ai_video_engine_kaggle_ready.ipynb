````python
"""Generated Kaggle sync bundle. Run generate_bundle.py to regenerate."""
import os
from pathlib import Path

BASE_DIR = Path(os.environ.get('ENGINE_BASE_DIR', '/kaggle/working/ai-video-engine'))
BASE_DIR.mkdir(parents=True, exist_ok=True)

FILES = {'STATUS.md': '# KANG - Current status (refresh: 30 October 2026, 22:45 IST)\n'
              '\n'
              'This summary separates reported test evidence, implementation and final '
              'acceptance.\n'
              '\n'
              '## Current product state\n'
              '- Workflow G: narration + captions + scene markers + NEW approved local visuals '
              'track (prototype).\n'
              '  `kang.py visuals-template scenes.json` -> visuals.csv (brief/asset/licence per '
              'scene, manual).\n'
              '  `kang.py visuals media words scenes visuals.csv --assets DIR --preview '
              'S01,S02,S03 [--win-root]`\n'
              '  -> gate visual_selection -> integration preview mp4 + Premiere package (visual '
              'track at scene times).\n'
              '- First integration run KANG_018 (Dwarika S01-S03, 81.4 s): TEST PLACEHOLDER assets '
              'only (not real\n'
              '  visuals), auto(--yes) approval = NOT human; transcript = draft v4 (KANG_016, not '
              'accepted).\n'
              '  Sandbox QA PASS (xml structure, preview duration, glyphs/shaping, '
              'caption-vs-words sync).\n'
              "- Real visuals for Dwarika: NOT_STARTED (needs user's own images/clips + licence).\n"
              '- Premiere 24.6.1 actual import (with visual track): NOT_TESTED. Windows package '
              'check 2026-10-31 (Antigravity, file-level, no GUI): unzip to '
              'E:\\kang\\packages\\dwarika_visuals_v1 PASS; XML 25 fps/17309 frames, A1 '
              'dwarika.mp3, V1 9 clipitems 0-81.40 s, 22 markers, all 4 media paths exist on disk '
              "PASS; SRT UTF-8 PASS. GUI import 2026-10-31 by user: **FAIL** — Premiere 24 'File "
              'Import Failure ... sequence.prproj: The project could not be loaded, it may be '
              "damaged or contain outdated elements.' Diagnosis kit out/kang_xml_diag.zip (T1 "
              'audio only, T2 +markers, T3 +stills, T4 +mp4 loop, T5 full with valid GUID uuid; '
              'suspect: non-GUID <uuid>). Result: T1–T5 all FAIL; T0 empty sequence (no media, '
              'valid GUID) also FAIL → cause is not our XML content; Premiere-side XML importer '
              "suspected. Test B (Premiere's own FCP XML export re-import) also fails once the "
              "source sequence is deleted → Premiere FCP XML importer broken on user's install "
              "(environment). CAVEAT 10:27: user's Project panel was collapsed (list area hidden; "
              "Sequence 03 existed but was invisible) → T1–T5/T0/Test B 'fail' verdicts may be "
              'invisible successes; only the first sequence.xml import had a real error popup. '
              'Re-verify with maximized panel. 10:38: T0 XML opened via File > Open Project = PASS '
              "(project T0_empty_sequence, sequence 'T0 empty', 25 fps) → XML path works on this "
              'install through Open Project; Import-into-project route unreliable. 10:39: real '
              "sequence.xml via Open Project = FAIL (same 'project could not be loaded'). T0 PASS "
              'vs package FAIL → our XML content is the cause. Hardened bisection kit '
              'out/kang_xml_diag3.zip (H1 audio, H2 +ascii markers, H3 +hindi markers, H4 +stills, '
              'H5 +mp4 loop, H6 full; T0-style track enabled/locked, audio format, file '
              'timecode/width/height, masterclipid). Results pending. T6 EDL: importer accepted '
              'file (title parsed, PAL/sequence dialogs). Next: T6 EDL import test '
              '(out/kang_edl_test.zip); options: official Premiere repair/reinstall, EDL path, or '
              'ExtendScript builder.\n'
              '- Transcript acceptance: false (32 blank + 14 pending listening rows); changed-word '
              'timing unverified.\n'
              '- qa.check_sync compares ASS onsets with words.json, NOT with audio: serialization '
              'check, not speech sync.\n'
              '- Spelling policy: per-video opt-in --spelling-policy; global glossary neutral '
              '(frozen).\n'
              '\n'
              '## Pending acceptance list (does not block visual development)\n'
              '1. Transcript listening decisions (reviews/dwarika_names_alldiffer.csv).\n'
              '2. Changed-word timing verification.\n'
              '3. Human-reference subtitle sync against audio.\n'
              '4. Actual Premiere import/playback (scene-marker package and visual-track '
              'package).\n'
              '\n'
              '- Latest test run: see build_report.json (build_and_check), 30 Oct 2026 22:45 IST.\n'
              '\n'
              '## Reported evidence (not independently reproduced here)\n'
              '- 29 Oct 2026 Kaggle T4 / kernel v7: full Dwarika run, 831 s processing;\n'
              '  8 detected gaps, 245 recovered words, 6/6 suspect repairs;\n'
              '  unfilled 0, 2-3 s review gaps 0; 16 cuts auto-approved, NOT human-approved.\n'
              '- Current (30 Oct 2026, after readiness patch): see "Latest test run" line below.\n'
              '  Older 111/126-test and 32/34-file records are historical.\n'
              '- v7 run evidence and v6-derived package acceptance are different checks.\n'
              '\n'
              '## Code changes (readiness patch, applied 30 Oct 2026)\n'
              '- Package generation emits draft_package_ready or\n'
              '  package_generated_pending_verification (or qa_failed), never final accepted.\n'
              '- Text acceptance and changed-word timing review are separate metadata fields.\n'
              '- Unknown full-file audio sync is explicitly NOT_TESTED.\n'
              '- Text acceptance metadata records whether changed-word timing remains unverified.\n'
              '- These changes still require repository regression tests and a fresh long-form '
              'run.\n'
              '\n'
              '- Earlier 29 Oct review-5: cut approval re-verified by plan hash + source media '
              'sha256; review timing labels.\n'
              '- Historical: 143 pytest + 21 legacy PASS, 37 files, 30 Oct 2026 15:27 IST. Caption '
              'option --english-script asr|latin|devanagari (core/script_style.py, '
              'config/english_script.json; captions only, words.json unchanged).\n'
              '\n'
              '## Next acceptance work\n'
              '1. Listen to flagged regions, fill keep/fix/delete decisions and apply human '
              'review.\n'
              '2. Re-align or manually verify changed-word timing; do not mark it verified without '
              'evidence.\n'
              '3. Regenerate captions/scenes/package from the corrected revision.\n'
              '4. Test actual Premiere import, full playback and selected visuals.\n'
              '5. Complete the 100-word hand-check and separate broader long-form audit.\n'
              '\n'
              '## Historical notes (preserved verbatim below)\n'
              'The following text may contain superseded current-status labels; dates and context '
              'apply.\n'
              '\n'
              '---\n'
              '\n'
              '# Kang AI Creative Automation — Status (Master Roadmap v2.0, 29 Oct 2026; earlier '
              'notes said "v4.0")\n'
              '\n'
              'Last verified: 2026-10-29 IST (111 pytest + 21 legacy, 32 bundle files in sync). '
              '`python build_and_check.py` → **PASS** (see build_report.json).\n'
              '\n'
              '## Quick start (VS Code, local)\n'
              '```bash\n'
              'pip install -r requirements.txt          # ffmpeg 6+ with libass/HarfBuzz must be '
              'on PATH\n'
              'python scripts/fetch_fonts.py            # Mukta ExtraBold + Noto Sans Devanagari + '
              'Montserrat (OFL)\n'
              'python build_and_check.py                # Phase 0 gate: must say RESULT: PASS\n'
              'python kang.py transcribe voice.mp3 --device cpu\n'
              'python kang.py fix-words runs/<id>/words.json     # config/glossary.json\n'
              'python kang.py promo voice.mp3 runs/<id>/words.json --start 0 --dur 20\n'
              '```\n'
              'Har command ek naya `runs/<date>_KANG_NNN/` folder banata hai: manifest.json '
              '(relative paths + SHA-256),\n'
              'inputs ka hash, ffmpeg.log, qa_report.json. Originals kabhi overwrite nahi hote.\n'
              '\n'
              '## Feature status\n'
              '| Phase/Feature | Status | Evidence |\n'
              '|---|---|---|\n'
              '| Phase 0 build_and_check | Verified | files → syntax → env → 21 legacy + 111 '
              'pytest → bundle → sync |\n'
              '| Phase 1 transcription | Verified (CPU), script-level only | Darjeeling + Dwarika '
              '60 s, audit 0 issues (overlap/glyph/gap checks, not word accuracy). 100-word '
              'hand-check sheet: out/dwarika_v6_handcheck.xlsx, pending |\n'
              '| API (app.py /upload) | Verified (mock) | now calls pipeline.api_transcribe = same '
              'stack as kang.py; tests/test_api_parity.py |\n'
              '| Glossary fix (proper nouns) | Verified | Dwarika: 6 fixes (द्वारी का → द्वारका, '
              'कृष्णे → कृष्ण, पैरे → प्यारे) |\n'
              '| F01 animated Hindi subtitles | Generation + cue serialization: tested. Hindi '
              'rendering: user-reviewed (Dwarika 9:16, 2026-10-28). Audio sync vs human-marked '
              'reference (±100 ms, 20 markers): NOT_TESTED | glyph, safe zone, cue-vs-words.json, '
              'export checks |\n'
              '| Phase 2 exit (20 s Hindi promo) | Verified | Darjeeling + Dwarika, 1080×1920, '
              '20.01 s, QA pass |\n'
              '| F02 presets | Verified | 4 presets, schema-checked |\n'
              '| F03 silence/filler cut | Verified | Dwarika 60 s → 56.925 s, 6 cuts, 0 truncated '
              'words |\n'
              '| F04 audio cleanup | Metrics PASS; full acceptance WARN until human listen | '
              'Dwarika −20.0 → −14.3 LUFS, TP −2.0 dBTP. **Human listen pending** |\n'
              '| F05 multi-format | Verified (synthetic) | 9:16/16:9/1:1 tests |\n'
              '| F19-lite palette | Verified | contrast-checked brand_draft.json |\n'
              '| Phase 3 workflow (`kang.py workflow`, `--full`) | Verified (Kaggle T4 GPU) | Full '
              'Dwarika 692 s: 880 s workflow, 0 unfilled gaps, -14.9 LUFS / -2.4 dBTP, 16:9 render '
              'PASS |\n'
              "| Suspect-span + seam repair | Verified | Dwarika: 5/5 spans repaired, 6:06 'शिव के "
              "बारह ज्योतिर्लिंग' fixed; swallowed-word guard |\n"
              '| Spelling voter (default ON) | Verified (Kaggle) | 4 Hindi models (Vaani, '
              'collabora medium, IndicWhisper medium/large), ≥2 must agree; 76 changes on Dwarika; '
              'batched, 1 model per GPU, ~3.3 min |\n'
              '| Speech-gap re-transcription | Verified | Dwarika 60–120 s: Whisper skipped '
              '13.1–29.98 s; 34 words recovered |\n'
              '| Script sanitizer | Verified | Cyrillic/Hebrew hallucinations stripped, '
              'repeated-word loops flagged |\n'
              '| Kaggle bundle + real GPU runs | Verified | Real Kaggle T4 runs: 2026-10-28 (v1, '
              'v2), 2026-10-29 (v3 voter, v6 fast final). Voter models prebuilt by kernel '
              'kang-voter-models |\n'
              '\n'
              '## Master Roadmap v2.0 mapping (honesty rule: PASS / FAIL / WARN / NOT_TESTED / '
              'NOT_APPLICABLE)\n'
              'Old "Verified" above = our own script/test evidence. Under v2.0 a feature is only '
              'PASS with a natural 10–15 min run + qa/ evidence + human checks, so most rows are '
              'still **WARN** or **NOT_TESTED**.\n'
              '\n'
              '| Milestone | Scope (v2.0) | Kang today | Status |\n'
              '|---|---|---|---|\n'
              '| M0 core contracts | IDs, schemas, manifests, logging, recovery | run IDs, '
              'manifest + SHA-256, read-only originals, build gate. hash-bound `approvals.json` '
              '(approver + plan sha256, --yes recorded as auto). Missing: `schema_version`, '
              '`<artifact>.vN.json`, run sub-folders (input/ transcript/ qa/…), cancel/resume '
              'tests, `job_queue.json` | WARN (partial) |\n'
              '| M1 input router | audio / face / no-face / mixed / corrupt | not built '
              '(audio-only assumed); no `input_profile.json` | NOT_TESTED |\n'
              '| M2 transcript intelligence | ASR draft, coverage, targeted retry, correction | '
              'raw layer (words_raw.json), coverage gaps ≥2 s detected + retried (2–3 s leftovers '
              '= needs_review, ≥3 s blocks), suspect-span repair (gaps ≥0.8 s, stretched words), '
              'voter, glossary, diff logs. Missing: 100-word Quick Sample result, ≥500-word long '
              'audit, human correction step, `coverage_report.json` name/shape | WARN (hand-check '
              'pending) |\n'
              '| M3 Workflow G | scenes.json, Gate 1, visuals, captions, preview, Premiere package '
              '(F12) | captions/cuts/audio/16:9 render done. Missing: scene plan, Gate 1 visuals, '
              'package, Premiere import test | NOT_TESTED |\n'
              '| M4 face video | F04, F06, F01, F05 on face clips | F06 not built | NOT_TESTED |\n'
              '| M5 motion + design | F16, F20, F19-lite, F13, F15 | brand.json + F19-lite only | '
              'NOT_TESTED |\n'
              '| M6 extensions | F07–F11, F12 AE, F14, F18, F21, F23 | none (must not start before '
              'M3 exit, risk R6) | NOT_TESTED |\n'
              '| M7 batch | F17 queue | none | NOT_TESTED |\n'
              '\n'
              'Licences: fonts/OFL-{Mukta,Noto,Montserrat}.txt; models + wordlist in '
              'config/model_licenses.json (collabora medium is CC-BY-4.0, attribution needed; '
              'hi_IN.dic is GPL, downloaded at setup, not shipped). D1 = Premiere Pro 24.6.1 (+ '
              'After Effects 24.6.2), user 2026-10-29. D2 = Windows 10 (per user "10th"), Intel '
              'i5, 8 GB RAM, GTX 1650 (4 GB VRAM assumed, typical), 562 GB free SSD. Implication: '
              'Premiere path = FCP7 XML (xmeml) import, test on 24.6.1; local GPU = turbo only '
              '(--no-vote, ENGINE_COMPUTE_TYPE=int8_float16 if OOM), 4-model voter stays on Kaggle '
              'T4. D3–D7 still open (recommended defaults in roadmap notes).\n'
              '\n'
              '## External review fixes (2026-10-29)\n'
              'API cut default OFF (narration preserved); approvals.json bound to plan hash; '
              'setup_kaggle pip runs without shell (the `pydantic>=2.5` redirect bug) + '
              '`KANG_INPUT_AUDIO` explicit input; coverage threshold 2 s; overlap guard no longer '
              'pushes ends past the next start (no word dropped). The 2 s threshold is NOT yet '
              're-run on full Dwarika (Kaggle) — status WARN until then.\n'
              '\n'
              '## Third review fixes (2026-10-29 20:00)\n'
              'transcript.srt/json now built from the FINAL words (raw ASR layer kept in raw/); '
              'API/status report processing_complete, review_required, accepted=False, warnings '
              'separately; approval hash covers preview/ASS/words content + render settings and is '
              're-verified before the final burn; a human "no" = rejected, default-deny = pending, '
              'auto(--yes) = human_approved False; setup same-file copy guard; CLI exit 0 done / 1 '
              'failed / 2 waiting at a gate.\n'
              '\n'
              '## Full Dwarika re-run on new code (Kaggle T4, kernel v7, 2026-10-29 20:22 IST)\n'
              'build_and_check PASS on Kaggle (111 pytest + 21 legacy). workflow --full --yes: 831 '
              's, pass. Coverage at 2.0 s: 8 gaps found, 245 words added, 6/6 suspect spans '
              'repaired, unfilled 0, needs_review(2-3 s) 0. Audit 3 review flags (non-blocking), '
              '172 glossary fixes. Cut 16 (6.3 s) auto-approved via --yes = NOT human-approved. '
              'Loudness -14.9 LUFS, TP -2.4 dBTP. Report: out/kaggle_gpu_full_v7/. Human listening '
              '+ sync still NOT_TESTED.\n'
              '\n'
              '## M3 Workflow G prototype (2026-10-29 21:15 IST)\n'
              '`kang.py package <media> <words.json> [--scenes edited.json] [--cuts approved '
              'cuts.json] [--fps 25] [--win-root E:\\...]`:\n'
              'words -> scenes.json (~30 s scenes at sentence/pause breaks, tiles the timeline, '
              'every word owned once) -> Gate 1 `scene_plan` (hash of scenes.json + words.json + '
              'keep/fps) -> premiere_package/ (sequence.xml FCP7 xmeml on ORIGINAL media, scene '
              'markers, captions.srt, scenes.json, words.json, media/, README_IMPORT.txt, '
              'package_manifest.json sha256). Cuts only from a run whose cut_list gate is '
              'approved; default = narration whole.\n'
              'Status: XML structure/frame QA PASS (10 tests, Dwarika v6 words: 23 scenes, 17309 '
              'frames @25 fps). Gate 1 for that test was auto(--yes) = not human. **Premiere '
              "24.6.1 import: NOT_TESTED** (user's PC). Visuals per scene: NOT_TESTED "
              '(placeholders only).\n'
              '\n'
              '## Known limits\n'
              '- Subtitle font is Mukta ExtraBold (Hindi + English in one face). Noto Sans '
              'Devanagari has no Latin letters, so Hinglish words failed QA.\n'
              '- large-v3 vs turbo compared (A/B 2026-10-28): turbo + seed + spelling voter '
              'chosen; large-v3 3.3x slower, worse timing.\n'
              '- Subtitle sync check compares rendered cues to words.json; it cannot judge whether '
              'ASR timing matches the voice. Watch the preview.\n'
              '- Promo background is a plain brand colour; B-roll/motion graphics are Phase 3+ '
              '(unblocked: subtitle preview signed off 2026-10-28).\n'
              '- ASR still mishears names; extend config/glossary.json per client/topic.\n'
              '\n'
              '## 2026-10-29 22:25 — review-4 (external review of Workflow G)\n'
              '- Transcript acceptance loop: `kang.py review` (review.csv + clips, changes no '
              'words) → user fills action/correct_text → `kang.py apply-review` (new words.json; '
              'transcript gate; transcript_acceptance.json bound to words sha256). Package reads '
              'acceptance: currently **transcript_accepted = False**.\n'
              '- Dwarika v6 review sheet: 54 spans (39 P1 amounts/names incl. 4 reviewer phrases, '
              '4 P2 fragments/low-conf, 11 P3 repeats). Human listen: NOT_TESTED.\n'
              '- Scene planner: no split before continuation words (से/की/का...) or after dangling '
              '(जो/के...), bare pause needs ≥1.0 s, fallback prefers sentence end near target. '
              'Dwarika: 22 scenes. Still needs human scene review.\n'
              '- README: "Target" + "Actual import verification: pending"; states video track '
              'empty, captions separate SRT. Manifest has status block.\n'
              '- v7 run PASS ≠ v6-based package acceptance (separate checks). Premiere import: '
              'NOT_TESTED. Visuals: NOT_STARTED.\n'
              '- 126 pytest + 21 legacy PASS; build_and_check PASS (34 files).\n'
              '\n'
              '## 2026-10-29 23:10 — review-5\n'
              '- apply_review timing: same word count keeps ASR timing; different count = '
              'interpolated, `timing_method=interpolated_after_review`, '
              '`timing_status=unverified`, listed in `timing_unverified`; no fake probability; '
              '`text_status=human_reviewed`.\n'
              '- premiere_package cuts: plan hash of current cuts.json must equal approved hash '
              "AND source media sha256 must match the cut run's input; `cuts_human_approved` "
              'reported separately.\n'
              '- Scenes: break before narrator opener "दोस्तो/दोस्तों", never before a clause '
              'predicate (… | X है). Dwarika 22 scenes; still NOT human-reviewed.\n'
              '- 131 pytest + 21 legacy PASS; build PASS. Packages: dwarika_v6c '
              '(E:\\kang\\packages\\dwarika_v6c), review pack run KANG_004.\n'
              '\n'
              '## 2026-10-31 11:25 Premiere XML fix\n'
              "- H1-H6 (out/kang_xml_diag3.zip) via Open Project: ALL PASS on user's Premiere "
              '24.6.1 (audio, ascii markers, Hindi markers, stills, mp4 loop, full). Old KANG_018 '
              'sequence.xml: FAIL again (same error).\n'
              '- Root cause = old xmeml layout (exact single element not isolated; fix applied as '
              'whole H-layout). core/premiere.py now: GUID uuid (uuid5), media before '
              'timecode/markers, track enabled/locked, audio format, file timecode + width/height, '
              'masterclipid, alphatype. check_xmeml enforces it. tests/test_premiere_layout.py '
              '(6). 194 pytest, bundle 44 files, build_and_check PASS.\n'
              '- New package: runs/2026-10-31_KANG_001 -> out/dwarika_visuals_v2_package.zip '
              '(E:\\kang\\packages\\dwarika_visuals_v2). README says Open Project. Premiere GUI '
              'verify: NOT_TESTED (user).\n'
              '- 12:50 v2 sequence.xml via Open Project: PARSE PASS (no load error); Link Media '
              'dialog showed 4 files missing at E:\\kang\\packages\\dwarika_visuals_v2 -> folder '
              'name/extract path mismatch likely. Timeline content NOT_TESTED until relink.\n'
              "- 12:53 v2 after relink: Premiere timeline PASS (sequence 'KANG 2026-10-31_KANG_001 "
              "visuals', V1 stills+loop playing at 01:02:08, A1 audio, 22 markers, duration "
              '00:11:32:09). Media path mismatch was extract-folder naming; future zips should '
              'contain a top-level folder named as win-root. captions.srt import in Premiere: '
              'NOT_TESTED.\n'
              "- 13:35 user listened 00:13 (N002): extra 'का' after द्वारिका -> N002 = द्वारिका, "
              'human_reviewed. Asked user to confirm N001/N005/N006/N010/N029/N007/N018 same '
              'pattern. SRT sync issue: SRT times correct (0.91s first cue); user to place SRT at '
              '00:00:00:00.\n'
              "- 14:07 captions.srt import at 0:00 = PASS (cue 34 'तो द्वारिका यात्रा की' "
              '00:00:37:09 matches SRT). Premiere package incl. captions now PASS (placeholder '
              'visuals, draft transcript).\n'
              "- 14:15 user's remaining issue = Locate/Link Media. Added FIX_PATHS.bat + "
              'fix_paths.ps1 (rewrites sequence.xml pathurls to the folder it sits in; PASS/WARN '
              'for missing files) to both package writers; zips now have a top-level folder. Run '
              'KANG_002 -> out/dwarika_visuals_v2_1.zip. Python simulation of the ps1 logic PASS '
              '(Downloads path with space); real PowerShell run on Windows NOT_TESTED. 195 pytest, '
              'build_and_check PASS.\n'
              '- 15:17 FIX_PATHS.bat on Windows: PASS (user). Extract-anywhere -> bat -> Ctrl+O '
              'works without Link Media.\n'
              '- 15:25 User approved removing extra का (N001/N005/N006/N007/N010/N018/N029). CSV: '
              'N002 decided_by=listened; 7 rows editorial-policy (user-confirmed pattern, per-row '
              'listening not stated), content_review cleared. apply-review on KANG_012 -> KANG_004 '
              "(2265 words, 0 'द्वारिका का', 0 द्वारका); scenes re-planned, 22 scenes identical "
              'times. Visuals package KANG_005 -> out/dwarika_visuals_v3.zip + '
              'out/captions_v3.srt. Merged-word timing NOT_TESTED. Remaining review: 30 blank '
              'rows, 8 pending.\n'
              '- 15:55 Real assets S01–S03 (Wikimedia Commons, CC BY-SA 4.0, assets/dwarika_real + '
              'CREDITS.txt/json). KANG_006 (words KANG_004) preview + package -> '
              'out/dwarika_visuals_v4_preview_S01_S03.mp4, out/dwarika_visuals_v4.zip. Preview '
              'frames full 16:9, no empty area. Premiere check pending user. Framing list: '
              'notes/framing_checks.md. Transcript pending review kept separate (not a blocker).\n'
              '- 17:55 v5_full: KANG_007, 54 shots S01–S22 (reviews/dwarika_visuals_full.csv, '
              'multi-shot shot_start), preview 692.36 s 1920x1080 25fps + audio, 0 black frames, 0 '
              'silences >3 s, XML 54 visual clipitems + 22 markers check PASS, all 54 mid-frames '
              'inspected (qa/shot_midframes.jpg). out/dwarika_visuals_v5_full.zip + _preview.mp4. '
              'Premiere import of v5 pending user.\n'
              '- 20:25 v5_1 (KANG_008): manifest fixes from user review — CREDITS.txt/json '
              'packaged + hashed (65 files), stats.source_has_video/visual_track added (legacy '
              "'video' kept = source media), visuals.json assets_dir relative. Visuals/XML "
              'identical to v5_full except uuid. Approval flags + premiere_import stay '
              'false/NOT_TESTED until user evidence. Listening spans: '
              'reviews/dwarika_listen_spans_2026-10-31.md. test_visuals 21.\n'
              '- 21:30 Asset Brain Phase 1: core/asset_index.py, kang.py ingest/asset-search, '
              'assets/dwarika_real/index.json (54; 48 BY-SA/3 BY/2 PD/1 CC0), '
              'out/dwarika_assets_contact_sheet.jpg, tests/test_asset_index.py (4). Full pytest '
              '204 PASS. docs/KANG_Extended_Roadmap_v5.2.{md,pdf}. NEXT Phase 2 kang.py match → '
              'broll_suggestions.json.\n'
              '- 21:45 Asset Brain Phase 2 (rule-based): core/broll_match.py, '
              'config/asset_aliases.json, kang.py match / match-approve, tests/test_broll_match.py '
              '(4). Fixed asset_index tokenizer (underscores, danda, s13 ids). 208 pytest PASS. '
              'out/dwarika_broll_suggestions_phase2.zip: hit@3 15/22 (optimistic, file names are '
              'mine). Semantic: NOT_STARTED.\n'
              '- 21:58 Phase 2 v2: match-check (qa) + approve-broll (A/B/C prompt or --choices → '
              'approved_broll.json + visuals.csv) + review_sheet; fallback fill (S22). Structure '
              'PASS 22/22, 66/66, 0 dup. Content review PENDING. 210 pytest. '
              'out/dwarika_broll_suggestions_phase2_v2.zip\n'
              '- 22:05 Phase 2 v3: entity minor-mention rule + scene profiles '
              '(config/scene_profiles.json); S03 fixed, S22 outro. hit@3 19/22 (tuned on this '
              'video, optimistic). 211 pytest (209 full run + 2 broll tests fixed & rerun). '
              'out/dwarika_broll_suggestions_phase2_v3.zip\n'
              '- 2026-11-01 00:33 Hinglish layer v1 IMPLEMENTED (WARN draft): kang.py hinglish '
              '(Yapper romanizer + config/hinglish_names.json, same timings). '
              'out/dwarika_hinglish_v1.zip delivered. 164 review words, 79 Apex diff hints. 220 '
              'pytest PASS, build PASS. Pending: user Dwarka/Dwarika choice, listening, Yapper '
              'data licence.\n'
              '- 2026-11-01 14:56 Hinglish Review Assistant v0 IMPLEMENTED: hinglish-layer / '
              'review-hinglish / apply-hinglish-review. Run 2026-11-01_KANG_001 (45 blocks + clips '
              '+ html), out/dwarika_hinglish_review_v0.zip delivered. 227 pytest PASS, build PASS. '
              'Awaiting user review CSV. Reels Caption Styles NOT_STARTED.\n'
              '\n'
              '## 2026-11-02 07:30 IST fetch v2 (Dwarika)\n'
              '- fetch-plan: multi-word place kept whole + max 2 scene tags; fetch skips '
              '3D/stereo/anaglyph titles. \n'
              '- Query CSV reviews/dwarika_fetch_queries_v2.csv (place "Dwarka Gujarat", 2 per '
              'scene) → assets/dwarika_fetch_v2: 29 stills, all CC BY-SA, ingested; 1 download '
              'error; no result S11,S15,S16,S21,S22.\n'
              '- My eye check (agent, not user): ~25/29 relevant, all Gujarat Dwarka; weak: S19 '
              'egret + district map, S08 lotus, S01_2 estuary egret. v1 errors (Delhi hotels, '
              'salt, 3D) gone. Next: match + approve-broll.\n',
 'app.py': '"""FastAPI service for the AI Video Engine implementing the agreed API contract.\n'
           '\n'
           'The UI source file is not included in this project archive; the public routes and\n'
           'response shapes follow the contract supplied by the project owner.\n'
           '"""\n'
           'from __future__ import annotations\n'
           '\n'
           'import os\n'
           'import re\n'
           'import secrets\n'
           'import threading\n'
           'import time\n'
           'from concurrent.futures import ThreadPoolExecutor\n'
           'from pathlib import Path\n'
           'from typing import Any\n'
           '\n'
           'from fastapi import Depends, FastAPI, File, Form, Header, HTTPException, UploadFile\n'
           'from fastapi.middleware.cors import CORSMiddleware\n'
           'from fastapi.responses import FileResponse\n'
           '\n'
           'import core.engine as engine\n'
           'import core.pipeline as pipeline\n'
           '\n'
           'app = FastAPI(title="AI Video Engine API")\n'
           'app.add_middleware(\n'
           '    CORSMiddleware,\n'
           '    allow_origins=["*"],\n'
           '    allow_methods=["*"],\n'
           '    allow_headers=["*"],\n'
           ')\n'
           '\n'
           '# One worker guarantees at most one engine/GPU pipeline runs at a time.\n'
           '_executor = ThreadPoolExecutor(max_workers=1, thread_name_prefix="ai-video-job")\n'
           '_jobs: dict[str, dict[str, Any]] = {}\n'
           '_jobs_lock = threading.Lock()\n'
           '_JOB_ID_RE = re.compile(r"^[0-9a-fA-F]{8}$")\n'
           'MAX_UPLOAD_BYTES = int(os.environ.get("ENGINE_MAX_UPLOAD_BYTES", str(100 * 1024 * '
           '1024)))\n'
           'UPLOAD_CHUNK_BYTES = 1024 * 1024\n'
           '\n'
           '\n'
           'def _require_token(x_token: str | None = Header(default=None, alias="X-Token")) -> '
           'None:\n'
           '    expected = os.environ.get("ENGINE_TOKEN", "")\n'
           '    if not expected or not x_token or not secrets.compare_digest(x_token, expected):\n'
           '        raise HTTPException(status_code=401, detail="Invalid or missing X-Token")\n'
           '\n'
           '\n'
           'def _job_worker(job_id: str, src: str, cut: bool, lang: str) -> None:\n'
           '    with _jobs_lock:\n'
           '        job = _jobs[job_id]\n'
           '        job["status"] = "running"\n'
           '        job["started_at"] = time.monotonic()\n'
           '    try:\n'
           '        # Same stack as kang.py (gap-fill, suspect-span repair, voter, glossary), not '
           'the old\n'
           '        # engine.run_pipeline; outputs land where /file/{id}/{kind} already looks.\n'
           '        out_dir = Path(engine.BASE) / "storage" / "outputs" / job_id\n'
           '        res = pipeline.api_transcribe(src, str(out_dir), cut=cut, language=lang)\n'
           '        with _jobs_lock:\n'
           '            job["run_id"] = res["run_id"]\n'
           '            # "done" = processing finished; acceptance is separate (Roadmap S4)\n'
           '            job["review_required"] = res["review_required"]\n'
           '            job["accepted"] = res["accepted"]\n'
           '            job["warnings"] = res["warnings"]\n'
           '            job["status"] = "done"\n'
           '            job["secs"] = round(time.monotonic() - job["started_at"], 3)\n'
           '    except Exception as exc:\n'
           '        with _jobs_lock:\n'
           '            job["status"] = "error"\n'
           '            job["error"] = str(exc)\n'
           '            job["secs"] = round(time.monotonic() - job["started_at"], 3)\n'
           '\n'
           '\n'
           '@app.get("/health", dependencies=[Depends(_require_token)])\n'
           'def health():\n'
           '    return {"ok": True}\n'
           '\n'
           '\n'
           '@app.post("/upload", dependencies=[Depends(_require_token)])\n'
           'async def upload(\n'
           '    file: UploadFile = File(...),\n'
           '    cut: bool = Form(False),  # Roadmap v2: narration preserved by default; cuts only '
           'after review\n'
           '    lang: str = Form("hi"),\n'
           '):\n'
           '    job_id = secrets.token_hex(4)\n'
           '    base = Path(engine.BASE)\n'
           '    input_dir = base / "storage" / "inputs"\n'
           '    input_dir.mkdir(parents=True, exist_ok=True)\n'
           '\n'
           '    original_name = Path(file.filename or "upload.bin").name\n'
           '    # Keep a recognizable extension but avoid path traversal and empty names.\n'
           '    safe_name = re.sub(r"[^A-Za-z0-9._-]", "_", original_name).strip("._") or '
           '"upload.bin"\n'
           '    src_path = input_dir / f"{job_id}_{safe_name}"\n'
           '    total_bytes = 0\n'
           '    try:\n'
           '        # Stream to disk in bounded chunks; never load the entire upload into RAM.\n'
           '        with src_path.open("wb") as destination:\n'
           '            while True:\n'
           '                chunk = await file.read(UPLOAD_CHUNK_BYTES)\n'
           '                if not chunk:\n'
           '                    break\n'
           '                total_bytes += len(chunk)\n'
           '                if total_bytes > MAX_UPLOAD_BYTES:\n'
           '                    destination.close()\n'
           '                    src_path.unlink(missing_ok=True)\n'
           '                    raise HTTPException(\n'
           '                        status_code=413,\n'
           '                        detail=f"Upload exceeds the {MAX_UPLOAD_BYTES}-byte limit",\n'
           '                    )\n'
           '                destination.write(chunk)\n'
           '        if total_bytes == 0:\n'
           '            src_path.unlink(missing_ok=True)\n'
           '            raise HTTPException(status_code=400, detail="Uploaded file is empty")\n'
           '    except HTTPException:\n'
           '        raise\n'
           '    except Exception:\n'
           '        src_path.unlink(missing_ok=True)\n'
           '        raise\n'
           '    finally:\n'
           '        await file.close()\n'
           '\n'
           '    job = {\n'
           '        "id": job_id,\n'
           '        "name": original_name,\n'
           '        "status": "queued",\n'
           '        "secs": None,\n'
           '        "error": None,\n'
           '        "started_at": None,\n'
           '        "src": str(src_path),\n'
           '    }\n'
           '    with _jobs_lock:\n'
           '        _jobs[job_id] = job\n'
           '    _executor.submit(_job_worker, job_id, str(src_path), cut, lang)\n'
           '    return {"id": job_id, "name": original_name, "status": "queued"}\n'
           '\n'
           '\n'
           '@app.get("/status/{id}", dependencies=[Depends(_require_token)])\n'
           'def status(id: str):\n'
           '    if not _JOB_ID_RE.fullmatch(id):\n'
           '        raise HTTPException(status_code=404, detail="Job not found")\n'
           '    with _jobs_lock:\n'
           '        job = _jobs.get(id)\n'
           '        if not job:\n'
           '            raise HTTPException(status_code=404, detail="Job not found")\n'
           '        secs = job["secs"]\n'
           '        if job["status"] == "running" and job["started_at"] is not None:\n'
           '            secs = round(time.monotonic() - job["started_at"], 3)\n'
           '        return {\n'
           '            "id": job["id"],\n'
           '            "name": job["name"],\n'
           '            "status": job["status"],\n'
           '            "secs": secs,\n'
           '            "error": job["error"],\n'
           '            # additive fields (old clients ignore them): processing done != transcript '
           'accepted\n'
           '            "run_id": job.get("run_id"),\n'
           '            "review_required": job.get("review_required"),\n'
           '            "accepted": job.get("accepted", False),\n'
           '            "warnings": job.get("warnings", []),\n'
           '        }\n'
           '\n'
           '\n'
           '@app.get("/file/{id}/{kind}", dependencies=[Depends(_require_token)])\n'
           'def download_file(id: str, kind: str):\n'
           '    if not _JOB_ID_RE.fullmatch(id) or kind not in {"srt", "json"}:\n'
           '        raise HTTPException(status_code=404, detail="File not found")\n'
           '    # Resolve the output directory through engine.BASE so local tests and deployments\n'
           '    # can set ENGINE_BASE_DIR without changing the public API.\n'
           '    suffix = "srt" if kind == "srt" else "json"\n'
           '    path = Path(engine.BASE) / "storage" / "outputs" / id / f"transcript.{suffix}"\n'
           '    if not path.is_file():\n'
           '        raise HTTPException(status_code=404, detail="File not found")\n'
           '    media_type = "application/json" if kind == "json" else "text/plain; '
           'charset=utf-8"\n'
           '    return FileResponse(path, media_type=media_type, filename=path.name)\n',
 'brand.json': '{\n'
               '  "client": "Kang",\n'
               '  "colors": {"primary": "#0F172A", "accent": "#2563EB", "highlight": "#FACC15", '
               '"text": "#FFFFFF", "outline": "#000000"},\n'
               '  "fonts": {"heading": "Montserrat-Black.ttf", "body": "Mukta-ExtraBold.ttf", '
               '"hindi": "Mukta-ExtraBold.ttf"},\n'
               '  "logo_path": "assets/logo.png",\n'
               '  "safe_margin_px": 60,\n'
               '  "safe_zones": {"9:16": {"subtitle_bottom_px": 480}, "16:9": '
               '{"subtitle_bottom_px": 60}, "1:1": {"subtitle_bottom_px": 80}, "4:5": '
               '{"subtitle_bottom_px": 100}}\n'
               '}\n',
 'build_and_check.py': '"""Phase 0: build_and_check.py -> one command, PASS/FAIL report.\n'
                       '\n'
                       'Order (roadmap): files -> syntax -> environment (FFmpeg/libass/shaping, '
                       'Python libs,\n'
                       'fonts, brand/presets) -> test_offline.py -> pytest suite -> '
                       'generate_bundle.py -> sync check.\n'
                       'Any failure BLOCKS packaging (generate_bundle is never run after a '
                       'failure).\n'
                       '\n'
                       'Usage:  python build_and_check.py [--no-bundle] [--fetch-fonts] [--quick]\n'
                       'Writes: build_report.json\n'
                       '"""\n'
                       'from __future__ import annotations\n'
                       '\n'
                       'import ast\n'
                       'import datetime as dt\n'
                       'import importlib\n'
                       'import json\n'
                       'import shutil\n'
                       'import os\n'
                       'import subprocess\n'
                       'import sys\n'
                       'from pathlib import Path\n'
                       '\n'
                       'ROOT = Path(__file__).resolve().parent\n'
                       'sys.path.insert(0, str(ROOT))\n'
                       '\n'
                       'REQUIRED_FILES = [\n'
                       '    "app.py", "core/__init__.py", "core/engine.py", "core/gapfill.py", '
                       '"core/voter.py", "core/runs.py", "core/brand.py", "core/edl.py",\n'
                       '    "core/words.py", "core/subtitles.py", "core/render.py", '
                       '"core/cutter.py", "core/audio.py",\n'
                       '    "core/style.py", "core/qa.py", "core/pipeline.py", "kang.py", '
                       '"brand.json", "config/fillers.json",\n'
                       '    "presets/vox_bold_yellow.json", "config/glossary.json", '
                       '"requirements.txt",\n'
                       ']\n'
                       '# Dev-only files: present in the local project, intentionally NOT in the '
                       'Kaggle runtime bundle.\n'
                       'DEV_ONLY_FILES = ["test_offline.py", "generate_bundle.py", '
                       '"tests/test_foundation.py"]\n'
                       'PY_LIBS = ["numpy", "pydantic", "PIL", "fontTools"]\n'
                       'OPTIONAL_LIBS = ["faster_whisper", "fastapi", "pytest"]\n'
                       '\n'
                       '\n'
                       'class Report:\n'
                       '    def __init__(self):\n'
                       '        self.steps = []\n'
                       '\n'
                       '    def add(self, name, ok, detail="", blocking=True):\n'
                       '        self.steps.append({"step": name, "pass": bool(ok), "blocking": '
                       'blocking, "detail": detail})\n'
                       '        mark = "PASS" if ok else ("FAIL" if blocking else "WARN")\n'
                       '        print(f"[{mark}] {name}" + (f" - {detail}" if detail else ""))\n'
                       '        return ok\n'
                       '\n'
                       '    @property\n'
                       '    def ok(self):\n'
                       '        return all(s["pass"] for s in self.steps if s["blocking"])\n'
                       '\n'
                       '\n'
                       'def _run(cmd, timeout=900):\n'
                       '    env = {**os.environ, "PYTHONUTF8": "1", "PYTHONIOENCODING": "utf-8"}  '
                       '# Windows console is cp1252 by default\n'
                       '    r = subprocess.run(cmd, cwd=ROOT, env=env, capture_output=True, '
                       'text=True, encoding="utf-8", errors="replace",\n'
                       '                       timeout=timeout)\n'
                       '    return r.returncode, (r.stdout + r.stderr)\n'
                       '\n'
                       '\n'
                       'def check_files(rep):\n'
                       '    missing = [f for f in REQUIRED_FILES if not (ROOT / f).is_file()]\n'
                       '    dev_missing = [f for f in DEV_ONLY_FILES if not (ROOT / f).is_file()]\n'
                       '    note = f"{len(REQUIRED_FILES)} present" + (f" (runtime tree: dev-only '
                       'absent: {dev_missing})" if dev_missing else "")\n'
                       '    return rep.add("files", not missing, f"missing: {missing}" if missing '
                       'else note)\n'
                       '\n'
                       '\n'
                       'def check_syntax(rep):\n'
                       '    bad = []\n'
                       '    for p in sorted(ROOT.rglob("*.py")):\n'
                       '        if any(part in {"runs", ".venv", "venv", "__pycache__", '
                       '"backup_original"} for part in p.parts):\n'
                       '            continue\n'
                       '        try:\n'
                       '            ast.parse(p.read_text(encoding="utf-8"), str(p))\n'
                       '        except SyntaxError as e:\n'
                       '            bad.append(f"{p.relative_to(ROOT)}:{e.lineno}")\n'
                       '    for p in list(ROOT.glob("*.json")) + list((ROOT / '
                       '"presets").glob("*.json")) + list((ROOT / "config").glob("*.json")):\n'
                       '        try:\n'
                       '            json.loads(p.read_text(encoding="utf-8"))\n'
                       '        except Exception as e:\n'
                       '            bad.append(f"{p.relative_to(ROOT)}: {e}")\n'
                       '    return rep.add("syntax", not bad, ", ".join(bad) if bad else "py + '
                       'json parse OK")\n'
                       '\n'
                       '\n'
                       'def check_env(rep, fetch_fonts=False):\n'
                       '    ok = True\n'
                       '    ff = shutil.which("ffmpeg") and shutil.which("ffprobe")\n'
                       '    ok &= rep.add("ffmpeg binary", bool(ff), shutil.which("ffmpeg") or '
                       '"not on PATH")\n'
                       '    if ff:\n'
                       '        _, conf = _run(["ffmpeg", "-hide_banner", "-buildconf"])\n'
                       '        libs = {k: (f"--enable-{k}" in conf) for k in ("libass", '
                       '"libharfbuzz", "libfreetype", "libfribidi")}\n'
                       '        ok &= rep.add("ffmpeg libass + HarfBuzz", libs["libass"] and '
                       'libs["libharfbuzz"], str(libs))\n'
                       '        _, flt = _run(["ffmpeg", "-hide_banner", "-filters"])\n'
                       '        need = ["ass", "loudnorm", "silencedetect", "afftdn", "ebur128", '
                       '"alimiter"]\n'
                       '        miss = [f for f in need if f" {f} " not in flt]\n'
                       '        ok &= rep.add("ffmpeg filters", not miss, f"missing {miss}" if '
                       'miss else ", ".join(need))\n'
                       '        _, h = _run(["ffmpeg", "-hide_banner", "-h", "filter=ass"])\n'
                       '        ok &= rep.add("ass shaping=complex option", "shaping" in h and '
                       '"complex" in h)\n'
                       '    for lib in PY_LIBS:\n'
                       '        try:\n'
                       '            importlib.import_module(lib)\n'
                       '            rep.add(f"python lib {lib}", True)\n'
                       '        except Exception as e:\n'
                       '            ok &= rep.add(f"python lib {lib}", False, str(e))\n'
                       '    for lib in OPTIONAL_LIBS:\n'
                       '        try:\n'
                       '            importlib.import_module(lib)\n'
                       '            rep.add(f"optional lib {lib}", True, blocking=False)\n'
                       '        except Exception:\n'
                       '            rep.add(f"optional lib {lib}", False, "not installed (needed '
                       'for transcription/API/tests)", blocking=False)\n'
                       '    if fetch_fonts:\n'
                       '        from scripts.fetch_fonts import fetch\n'
                       '        fetch()\n'
                       '    from core.brand import load_brand, resolve_fonts, missing_glyphs\n'
                       '    try:\n'
                       '        brand = load_brand(ROOT / "brand.json")\n'
                       '        rep.add("brand.json schema", True, brand.client)\n'
                       '        fonts, warns = resolve_fonts(brand, ROOT / "fonts")\n'
                       '        rep.add("fonts resolved", not warns, "; ".join(warns) or ", '
                       '".join(p.name for p in set(fonts.values())), blocking=False)\n'
                       '        miss = missing_glyphs(fonts["hindi"], "द्वारका क्षत्रिय श्री ज्ञान '
                       'हृदय Amazing Zip-line 2026")\n'
                       '        ok &= rep.add("Devanagari glyph coverage", not miss, f"missing '
                       '{miss}" if miss else fonts["hindi"].name)\n'
                       '    except Exception as e:\n'
                       '        ok &= rep.add("brand/fonts", False, str(e))\n'
                       '    from core.subtitles import list_presets, load_preset\n'
                       '    try:\n'
                       '        names = list_presets(ROOT / "presets")\n'
                       '        for n in names:\n'
                       '            load_preset(n, ROOT / "presets")\n'
                       '        rep.add("presets schema", bool(names), ", ".join(names))\n'
                       '    except Exception as e:\n'
                       '        ok &= rep.add("presets schema", False, str(e))\n'
                       '    return ok\n'
                       '\n'
                       '\n'
                       'def check_tests(rep, quick=False):\n'
                       '    if not (ROOT / "test_offline.py").is_file() or not (ROOT / '
                       '"tests").is_dir():\n'
                       '        return rep.add("tests", True, "SKIP: dev-only tests not in this '
                       '(Kaggle runtime) tree; run them locally", blocking=False)\n'
                       '    code, out = _run([sys.executable, "test_offline.py"])\n'
                       '    ok = rep.add("test_offline.py (legacy engine, 21 tests)", code == 0, '
                       'next((l for l in out.splitlines() if "PASSED" in l or "FAIL" in l), '
                       'out.strip()[-200:]))\n'
                       '    try:\n'
                       '        importlib.import_module("pytest")\n'
                       '    except Exception:\n'
                       '        return rep.add("pytest suite", False, "pytest not installed: pip '
                       'install pytest") and ok\n'
                       '    args = [sys.executable, "-m", "pytest", "-q", "tests"]\n'
                       '    if quick:\n'
                       '        args += ["-k", "not multi_format and not promo and not '
                       'subtitle_video"]\n'
                       '    code, out = _run(args)\n'
                       '    tail = [l for l in out.strip().splitlines() if "passed" in l or '
                       '"failed" in l or "error" in l]\n'
                       '    return rep.add("pytest suite (tests/)", code == 0, tail[-1] if tail '
                       'else out[-500:]) and ok\n'
                       '\n'
                       '\n'
                       'def check_bundle(rep):\n'
                       '    if not (ROOT / "generate_bundle.py").is_file():\n'
                       '        return rep.add("bundle", True, "SKIP: generate_bundle.py not in '
                       'this (Kaggle runtime) tree", blocking=False)\n'
                       '    code, out = _run([sys.executable, "generate_bundle.py"])\n'
                       '    if not rep.add("generate_bundle.py", code == 0, '
                       'out.strip().splitlines()[-1] if out.strip() else ""):\n'
                       '        return False\n'
                       '    # sync check: every bundled file in the generated bundle matches the '
                       'canonical file\n'
                       '    import generate_bundle as gb\n'
                       '    ns = {}\n'
                       '    src = (ROOT / "kaggle_sync_bundle.py").read_text(encoding="utf-8")\n'
                       '    tree = ast.parse(src)\n'
                       '    for node in tree.body:\n'
                       '        if isinstance(node, ast.Assign) and getattr(node.targets[0], "id", '
                       '"") == "FILES":\n'
                       '            ns["FILES"] = ast.literal_eval(node.value)\n'
                       '    files = ns.get("FILES", {})\n'
                       '    drift = [f for f in gb.BUNDLE_FILES if files.get(f) != (ROOT / '
                       'f).read_text(encoding="utf-8")]\n'
                       '    nb = json.loads((ROOT / '
                       '"ai_video_engine_kaggle_ready.ipynb").read_text(encoding="utf-8"))\n'
                       '    nb_ok = "".join(nb["cells"][1]["source"]) == src\n'
                       '    return rep.add("bundle sync check", not drift and nb_ok,\n'
                       '                   f"drift: {drift}, notebook_match={nb_ok}" if (drift or '
                       'not nb_ok) else f"{len(files)} files in sync")\n'
                       '\n'
                       '\n'
                       'def main(argv=None) -> int:\n'
                       '    argv = sys.argv[1:] if argv is None else argv\n'
                       '    rep = Report()\n'
                       '    print(f"KANG build_and_check  '
                       '{dt.datetime.now().isoformat(timespec=\'seconds\')}\\n" + "-" * 50)\n'
                       '    check_files(rep) and check_syntax(rep)\n'
                       '    if rep.ok:\n'
                       '        check_env(rep, fetch_fonts="--fetch-fonts" in argv)\n'
                       '    if rep.ok:\n'
                       '        check_tests(rep, quick="--quick" in argv)\n'
                       '    if rep.ok and "--no-bundle" not in argv:\n'
                       '        check_bundle(rep)\n'
                       '    elif not rep.ok:\n'
                       '        rep.add("packaging", False, "BLOCKED because an earlier step '
                       'failed")\n'
                       '    (ROOT / "build_report.json").write_text(json.dumps({"pass": rep.ok, '
                       '"steps": rep.steps,\n'
                       '                                                         "at": '
                       'dt.datetime.now().isoformat(timespec="seconds")},\n'
                       '                                                        indent=2, '
                       'ensure_ascii=False), encoding="utf-8")\n'
                       '    print("-" * 50 + f"\\nRESULT: {\'PASS\' if rep.ok else \'FAIL\'}  '
                       '(build_report.json)")\n'
                       '    return 0 if rep.ok else 1\n'
                       '\n'
                       '\n'
                       'if __name__ == "__main__":\n'
                       '    raise SystemExit(main())\n',
 'config/english_script.json': '{\n'
                               '  "_doc": "Caption-only script style for English words (see '
                               'core/script_style.py). Add entries freely; latin phrases must keep '
                               'the same word count.",\n'
                               '  "to_devanagari": {\n'
                               '    "well": "वेल", "connected": "कनेक्टेड", "railway": "रेलवे", '
                               '"station": "स्टेशन",\n'
                               '    "nearest": "नियरेस्ट", "airport": "एयरपोर्ट", "km": "किमी", '
                               '"VIP": "वीआईपी", "AC": "एसी",\n'
                               '    "important": "इम्पॉर्टेंट", "question": "क्वेश्चन", "budget": '
                               '"बजट"\n'
                               '  },\n'
                               '  "to_latin_phrases": [\n'
                               '    {"dev": "टू डेज", "lat": "two days"},\n'
                               '    {"dev": "थ्री डेज", "lat": "three days"},\n'
                               '    {"dev": "वन नाइट", "lat": "one night"},\n'
                               '    {"dev": "टू नाइट", "lat": "two night"},\n'
                               '    {"dev": "बजट", "lat": "budget"}\n'
                               '  ]\n'
                               '}\n',
 'config/fillers.json': '{\n'
                        '  "_note": "Editable filler dictionary (F03). \'auto\' words are proposed '
                        "as cuts; 'review' words are only flagged (e.g. 'matlab' can be real "
                        'content). Nothing is cut until the cut list is approved.",\n'
                        '  "auto": ["उम्म", "अम्म", "हम्म", "आं", "उह", "um", "umm", "uh", "uhh", '
                        '"hmm", "erm"],\n'
                        '  "review": ["मतलब", "तो", "ना", "बस", "वो", "like", "basically", '
                        '"actually", "matlab", "yaani", "यानी"],\n'
                        '  "min_probability_for_keep": 0.0\n'
                        '}\n',
 'config/glossary.json': '{\n'
                         '  "_help": "ASR galat sune shabd -> sahi roop. Multi-word key allowed '
                         "(e.g. 'द्वारी का'). Timing merge hota hai, kuch fabricate nahi hota. Har "
                         'run ke glossary_changes.json me review karein.",\n'
                         '  "द्वारी का": "द्वारका",\n'
                         '  "द्वारी": "द्वारका",\n'
                         '  "पैरे": "प्यारे",\n'
                         '  "कृष्णे": "कृष्ण",\n'
                         '  "द्वारकादीश": "द्वारकाधीश",\n'
                         '  "दार्जलिंग": "दार्जिलिंग",\n'
                         '  "कंचनजंघा": "कंचनजंगा",\n'
                         '  "डार्जलिंग": "दार्जिलिंग",\n'
                         '  "डारजलिंग": "दार्जिलिंग",\n'
                         '  "डाजलिंग": "दार्जिलिंग",\n'
                         '  "डार्जलिं": "दार्जिलिंग",\n'
                         '  "द्वारका दीश": "द्वारकाधीश",\n'
                         '  "द्वारका धीश": "द्वारकाधीश",\n'
                         '  "द्वारकादिश": "द्वारकाधीश",\n'
                         '  "_help_spelling": "Neeche ki entries Dwarika full transcript '
                         '(2026-10-28) se: sirf standard Hindi spelling sudhaar, matlab nahi '
                         'badla. Context-sensitive shabd (बारा, शुरी/शिरी/शी) sirf phrase ke '
                         'saath.",\n'
                         '  "मतूरा": "मथुरा",\n'
                         '  "मतुरा": "मथुरा",\n'
                         '  "बारा जोतिर्लिंग": "बारह ज्योतिर्लिंग",\n'
                         '  "बारा ज्योतिर्लिंग": "बारह ज्योतिर्लिंग",\n'
                         '  "जोतिरलिंग": "ज्योतिर्लिंग",\n'
                         '  "जोतिर्लिंग": "ज्योतिर्लिंग",\n'
                         '  "जोतिर लिंग": "ज्योतिर्लिंग",\n'
                         '  "जो तिरलिंग": "ज्योतिर्लिंग",\n'
                         '  "इस्थापित": "स्थापित",\n'
                         '  "स्तापित": "स्थापित",\n'
                         '  "शुरुवात": "शुरुआत",\n'
                         '  "पहुचेंगे": "पहुँचेंगे",\n'
                         '  "पहुच": "पहुँच",\n'
                         '  "मंदिल": "मंदिर",\n'
                         '  "मंदीर": "मंदिर",\n'
                         '  "मंदीरों": "मंदिरों",\n'
                         '  "मातर": "मात्र",\n'
                         '  "किलोमेटर": "किलोमीटर",\n'
                         '  "किलिमेटर": "किलोमीटर",\n'
                         '  "किलोमेटरी": "किलोमीटर",\n'
                         '  "मिटर": "मीटर",\n'
                         '  "गुदरात": "गुजरात",\n'
                         '  "काठियावार": "काठियावाड़",\n'
                         '  "छेतर": "क्षेत्र",\n'
                         '  "द्वारीका": "द्वारका",\n'
                         '  "दौरिका": "द्वारका",\n'
                         '  "दौरीका": "द्वारका",\n'
                         '  "डौरी का": "द्वारका",\n'
                         '  "द्वारीकादिश": "द्वारकाधीश",\n'
                         '  "द्वारीकाधिश": "द्वारकाधीश",\n'
                         '  "द्वारका धिश": "द्वारकाधीश",\n'
                         '  "द्वारका दिशमंदिर": "द्वारकाधीश मंदिर",\n'
                         '  "शी द्वारकाधीश": "श्री द्वारकाधीश",\n'
                         '  "शुरी कृष्ण": "श्री कृष्ण",\n'
                         '  "शिरी कृष्ण": "श्री कृष्ण",\n'
                         '  "किष्णे": "कृष्ण",\n'
                         '  "भेड़ द्वारका": "बेट द्वारका",\n'
                         '  "भेट द्वारका": "बेट द्वारका",\n'
                         '  "वेट द्वारका": "बेट द्वारका",\n'
                         '  "नारायन": "नारायण",\n'
                         '  "नागेशवर": "नागेश्वर",\n'
                         '  "सिद्धनात": "सिद्धनाथ",\n'
                         '  "भोलेनात": "भोलेनाथ",\n'
                         '  "शिद्धेश्वर": "सिद्धेश्वर",\n'
                         '  "गूपी": "गोपी",\n'
                         '  "रुक्मनी": "रुक्मिणी",\n'
                         '  "शिवराजपूर": "शिवराजपुर",\n'
                         '  "निलकंट": "नीलकंठ",\n'
                         '  "गाहितरी": "गायत्री",\n'
                         '  "इसकान": "इस्कॉन",\n'
                         '  "विश्नु": "विष्णु",\n'
                         '  "बगवान": "भगवान",\n'
                         '  "भगवाण": "भगवान",\n'
                         '  "शीवलिंग": "शिवलिंग",\n'
                         '  "प्राचिन": "प्राचीन",\n'
                         '  "समरपित": "समर्पित",\n'
                         '  "धरम्शाला": "धर्मशाला",\n'
                         '  "धरमशाला": "धर्मशाला",\n'
                         '  "धरमशारा": "धर्मशाला",\n'
                         '  "दरशन": "दर्शन",\n'
                         '  "सक्ते": "सकते",\n'
                         '  "स्थीत": "स्थित",\n'
                         '  "स्तीत": "स्थित",\n'
                         '  "इस्तित": "स्थित",\n'
                         '  "शांदार": "शानदार",\n'
                         '  "शानदा": "शानदार",\n'
                         '  "सपर": "सफ़र",\n'
                         '  "पुरा": "पूरा",\n'
                         '  "बहुती": "बहुत ही",\n'
                         '  "प्रमुग": "प्रमुख",\n'
                         '  "प्रमुक": "प्रमुख",\n'
                         '  "प्रमुकिस्तान": "प्रमुख स्थान",\n'
                         '  "माननेता": "मान्यता",\n'
                         '  "स्थान करने": "स्नान करने",\n'
                         '  "सुबर": "सुबह",\n'
                         '  "साड़े": "साढ़े",\n'
                         '  "पाज": "पाँच",\n'
                         '  "पाच": "पाँच",\n'
                         '  "नाव बजे": "नौ बजे",\n'
                         '  "फैसिलेटिस": "फैसिलिटीज",\n'
                         '  "समी": "सभी",\n'
                         '  "सबी": "सभी",\n'
                         '  "मनकावनाओं": "मनोकामनाओं",\n'
                         '  "विश्वर प्रिशिद्ध": "विश्व प्रसिद्ध",\n'
                         '  "प्रिशिद्ध": "प्रसिद्ध",\n'
                         '  "प्रसीद": "प्रसिद्ध",\n'
                         '  "धाई": "ढाई",\n'
                         '  "पूराना": "पुराना",\n'
                         '  "निर्मान": "निर्माण",\n'
                         '  "पास्तुकला": "वास्तुकला",\n'
                         '  "शिकर": "शिखर",\n'
                         '  "प्रवेशकर": "प्रवेश कर",\n'
                         '  "आटो": "ऑटो",\n'
                         '  "रुपे": "रुपये",\n'
                         '  "स्कुटी": "स्कूटी",\n'
                         '  "स्कूर्टी": "स्कूटी",\n'
                         '  "ओफिस": "ऑफिस",\n'
                         '  "भव्या": "भव्य",\n'
                         '  "आकरशक": "आकर्षक",\n'
                         '  "विराज मान": "विराजमान",\n'
                         '  "महसुस": "महसूस",\n'
                         '  "लगबख": "लगभग",\n'
                         '  "परिस्तित": "पर स्थित",\n'
                         '  "परस्तित": "पर स्थित",\n'
                         '  "परस्ती थे": "पर स्थित है",\n'
                         '  "पर रस्ती थे": "पर स्थित है",\n'
                         '  "आलोकिक": "अलौकिक",\n'
                         '  "अलोकिक": "अलौकिक",\n'
                         '  "स्वयम": "स्वयं",\n'
                         '  "शांती": "शांति",\n'
                         '  "शानती": "शांति",\n'
                         '  "बैटकर": "बैठकर",\n'
                         '  "फोटोशिट": "फोटोशूट",\n'
                         '  "बर्गत": "बरगद",\n'
                         '  "संसट": "सनसेट",\n'
                         '  "गुमने": "घूमने",\n'
                         '  "घुमने": "घूमने",\n'
                         '  "घुमकर": "घूमकर",\n'
                         '  "आईए": "आइए",\n'
                         '  "अराम": "आराम",\n'
                         '  "प्लिसेस": "प्लेसेस",\n'
                         '  "उची": "ऊँची",\n'
                         '  "प्रतेमा": "प्रतिमा",\n'
                         '  "आशिरवात": "आशीर्वाद",\n'
                         '  "विशेश": "विशेष",\n'
                         '  "पुजन": "पूजन",\n'
                         '  "शर्पदोस": "सर्पदोष",\n'
                         '  "महत्वा": "महत्व",\n'
                         '  "गोपिया": "गोपियाँ",\n'
                         '  "लिला": "लीला",\n'
                         '  "तेरते": "तैरते",\n'
                         '  "हैवे": "हाईवे",\n'
                         '  "बजपन": "बचपन",\n'
                         '  "उपहर": "उपहार",\n'
                         '  "स्वरुप": "स्वरूप",\n'
                         '  "टिकेट": "टिकट",\n'
                         '  "टुकडों": "टुकड़ों",\n'
                         '  "मोर पंक": "मोरपंख",\n'
                         '  "चार चान": "चार चाँद",\n'
                         '  "दुरी": "दूरी",\n'
                         '  "निले": "नीले",\n'
                         '  "एक्टिविटिस": "एक्टिविटीज",\n'
                         '  "बोटिं": "बोटिंग",\n'
                         '  "समुद्रे": "समुद्र",\n'
                         '  "आयलेंड": "आइलैंड",\n'
                         '  "माध्याब": "माध्यम",\n'
                         '  "प्रशाद": "प्रसाद",\n'
                         '  "प्रशाद गहन": "प्रसाद ग्रहण",\n'
                         '  "चोटे": "छोटे",\n'
                         '  "अप्रेल": "अप्रैल",\n'
                         '  "टेंपरेशर": "टेंपरेचर",\n'
                         '  "पावान": "पावन",\n'
                         '  "हेल्फुल": "हेल्पफुल",\n'
                         '  "महत्वपुर्ण": "महत्वपूर्ण",\n'
                         '  "बताईएगा": "बताइएगा",\n'
                         '  "दस्व इस्थन": "दसवें स्थान",\n'
                         '  "गर्ब गर्व": "गर्भगृह",\n'
                         '  "श्राब": "श्राप",\n'
                         '  "रिशी": "ऋषि",\n'
                         '  "मंदिला": "मंज़िला",\n'
                         '  "जूता चपल": "जूता चप्पल",\n'
                         '  "जाकिया देकर": "झाँकियाँ देखकर",\n'
                         '  "परपोते बजर नाम": "परपोते वज्रनाभ",\n'
                         '  "मनकामना": "मनोकामना",\n'
                         '  "द्वारीकादीश": "द्वारकाधीश",\n'
                         '  "बारा साल": "बारह साल",\n'
                         '  "रुकमनी": "रुक्मिणी",\n'
                         '  "दुरवासा": "दुर्वासा",\n'
                         '  "गुरू": "गुरु",\n'
                         '  "विश्वर प्रसिद्ध": "विश्व प्रसिद्ध",\n'
                         '  "अचम्बित": "अचंभित",\n'
                         '  "अचंबित": "अचंभित",\n'
                         '  "द्वारिका दीश": "द्वारिकाधीश",\n'
                         '  "द्वारिका धीश": "द्वारिकाधीश",\n'
                         '  "द्वारिका धिश": "द्वारिकाधीश",\n'
                         '  "द्वारिकादीश": "द्वारिकाधीश",\n'
                         '  "द्वारिकाधिश": "द्वारिकाधीश",\n'
                         '  "भेड़ द्वारिका": "बेट द्वारिका",\n'
                         '  "भेट द्वारिका": "बेट द्वारिका",\n'
                         '  "बेड द्वारिका": "बेट द्वारिका",\n'
                         '  "_dwarika_spelling": "No rule converts between the valid spellings '
                         'द्वारका/द्वारिका. Per-video caption spelling = --spelling-policy (e.g. '
                         'config/caption_spelling_policy.json for the Dwarika video)."\n'
                         '}',
 'config/name_suggestions.json': '{\n'
                                 ' "_note": "SUGGESTIONS ONLY for the transcript review (never '
                                 'auto-applied). Each entry: heard variants (normalized match on '
                                 'consecutive words) -> suggested correct name. Add a line '
                                 'whenever review confirms a new mishearing; move it to '
                                 'glossary.json only once it is certain for every video.",\n'
                                 ' "names": [\n'
                                 '  {\n'
                                 '   "suggest": "बेट द्वारका",\n'
                                 '   "heard": [\n'
                                 '    "भेड़ द्वारिका",\n'
                                 '    "भेड द्वारिका",\n'
                                 '    "भेट द्वारिका",\n'
                                 '    "भेंट द्वारिका",\n'
                                 '    "बेड द्वारिका",\n'
                                 '    "बेड द्वारका",\n'
                                 '    "भेड़ दौरी",\n'
                                 '    "भेड़दौरी",\n'
                                 '    "बेट द्वारिका"\n'
                                 '   ]\n'
                                 '  },\n'
                                 '  {\n'
                                 '   "suggest": "वज्रनाभ",\n'
                                 '   "heard": [\n'
                                 '    "भज्रनाम",\n'
                                 '    "भज्र नाम",\n'
                                 '    "वज्रनाम",\n'
                                 '    "बज्रनाभ",\n'
                                 '    "भज्रनाभ"\n'
                                 '   ]\n'
                                 '  },\n'
                                 '  {\n'
                                 '   "suggest": "सर्पदोष",\n'
                                 '   "heard": [\n'
                                 '    "शर्पद्रोस",\n'
                                 '    "शरपदोष",\n'
                                 '    "शर्पदोस",\n'
                                 '    "सर्पद्रोस",\n'
                                 '    "शर्प दोष"\n'
                                 '   ]\n'
                                 '  },\n'
                                 '  {\n'
                                 '   "suggest": "द्वारकाधीश",\n'
                                 '   "heard": [\n'
                                 '    "द्वारकादीश",\n'
                                 '    "द्वारिका दीश",\n'
                                 '    "द्वारिका दिश",\n'
                                 '    "द्वारका दीश"\n'
                                 '   ]\n'
                                 '  },\n'
                                 '  {\n'
                                 '   "suggest": "ज्योतिर्लिंग",\n'
                                 '   "heard": [\n'
                                 '    "जो तिरलिंग",\n'
                                 '    "जो तिरिंग",\n'
                                 '    "ज्योति लिंग",\n'
                                 '    "जोतिर्लिंग"\n'
                                 '   ]\n'
                                 '  },\n'
                                 '  {\n'
                                 '   "suggest": "द्वारका",\n'
                                 '   "heard": [\n'
                                 '    "द्वारी का",\n'
                                 '    "द्वारी की",\n'
                                 '    "दौरिका",\n'
                                 '    "दौरी का",\n'
                                 '    "दौरिक"\n'
                                 '   ]\n'
                                 '  },\n'
                                 '  {\n'
                                 '   "suggest": "भड़केश्वर महादेव",\n'
                                 '   "heard": [\n'
                                 '    "वर्केश्वर महादेव",\n'
                                 '    "भड़केश्वर",\n'
                                 '    "बड़केश्वर"\n'
                                 '   ]\n'
                                 '  },\n'
                                 '  {\n'
                                 '   "suggest": "जीवन लीलाओं",\n'
                                 '   "heard": [\n'
                                 '    "जीवन दिलाओ",\n'
                                 '    "जीवन लीलाओ"\n'
                                 '   ]\n'
                                 '  }\n'
                                 ' ],\n'
                                 ' "protected": [\n'
                                 '  "रुक्मिणी",\n'
                                 '  "रुक्मणी",\n'
                                 '  "द्वारका",\n'
                                 '  "द्वारिका",\n'
                                 '  "द्वारकाधीश",\n'
                                 '  "द्वारिकाधीश",\n'
                                 '  "श्रीकृष्ण",\n'
                                 '  "कृष्ण",\n'
                                 '  "गोमती",\n'
                                 '  "वज्रनाभ",\n'
                                 '  "सुदामा"\n'
                                 ' ],\n'
                                 ' "_protected_note": "Proper names that must NEVER be grouped '
                                 'into a bulk all/none pattern (each occurrence reviewed alone). '
                                 'Spelling choice (e.g. रुक्मिणी standard) is a user policy, not '
                                 'an ASR correction."\n'
                                 '}',
 'config/script_equivalents.json': '{\n'
                                   ' "_doc": "Consensus only: Latin English word <-> Devanagari '
                                   'spellings treated as the SAME word (script difference, not an '
                                   'error). Also uses english_script.json to_devanagari. Units: km '
                                   '= किलोमीटर. Never changes captions.",\n'
                                   ' "latin": {\n'
                                   '  "ac": [\n'
                                   '   "एसी"\n'
                                   '  ],\n'
                                   '  "activities": [\n'
                                   '   "एक्टिविटीज"\n'
                                   '  ],\n'
                                   '  "and": [\n'
                                   '   "एंड"\n'
                                   '  ],\n'
                                   '  "arrange": [\n'
                                   '   "अरेंज"\n'
                                   '  ],\n'
                                   '  "auto": [\n'
                                   '   "ऑटो"\n'
                                   '  ],\n'
                                   '  "bag": [\n'
                                   '   "बैग"\n'
                                   '  ],\n'
                                   '  "banana": [\n'
                                   '   "बनाना"\n'
                                   '  ],\n'
                                   '  "based": [\n'
                                   '   "बेस्ड"\n'
                                   '  ],\n'
                                   '  "boat": [\n'
                                   '   "बोट"\n'
                                   '  ],\n'
                                   '  "boating": [\n'
                                   '   "बोटिंग"\n'
                                   '  ],\n'
                                   '  "box": [\n'
                                   '   "बॉक्स"\n'
                                   '  ],\n'
                                   '  "camera": [\n'
                                   '   "कैमरा"\n'
                                   '  ],\n'
                                   '  "charge": [\n'
                                   '   "चार्ज"\n'
                                   '  ],\n'
                                   '  "comment": [\n'
                                   '   "कमेंट"\n'
                                   '  ],\n'
                                   '  "complete": [\n'
                                   '   "कम्प्लीट",\n'
                                   '   "कंप्लीट"\n'
                                   '  ],\n'
                                   '  "cover": [\n'
                                   '   "कवर"\n'
                                   '  ],\n'
                                   '  "days": [\n'
                                   '   "डेज"\n'
                                   '  ],\n'
                                   '  "deluxe": [\n'
                                   '   "डीलक्स",\n'
                                   '   "डिलक्स"\n'
                                   '  ],\n'
                                   '  "enjoy": [\n'
                                   '   "एंजॉय",\n'
                                   '   "एन्जॉय"\n'
                                   '  ],\n'
                                   '  "enough": [\n'
                                   '   "इनफ"\n'
                                   '  ],\n'
                                   '  "facilities": [\n'
                                   '   "फैसिलिटीज"\n'
                                   '  ],\n'
                                   '  "feet": [\n'
                                   '   "फीट"\n'
                                   '  ],\n'
                                   '  "free": [\n'
                                   '   "फ्री"\n'
                                   '  ],\n'
                                   '  "friend": [\n'
                                   '   "फ्रेंड"\n'
                                   '  ],\n'
                                   '  "hello": [\n'
                                   '   "हेलो",\n'
                                   '   "हैलो"\n'
                                   '  ],\n'
                                   '  "help": [\n'
                                   '   "हेल्प"\n'
                                   '  ],\n'
                                   '  "helpful": [\n'
                                   '   "हेल्पफुल"\n'
                                   '  ],\n'
                                   '  "high": [\n'
                                   '   "हाई"\n'
                                   '  ],\n'
                                   '  "highway": [\n'
                                   '   "हाईवे"\n'
                                   '  ],\n'
                                   '  "hotel": [\n'
                                   '   "होटल"\n'
                                   '  ],\n'
                                   '  "information": [\n'
                                   '   "इंफॉर्मेशन",\n'
                                   '   "इन्फॉर्मेशन",\n'
                                   '   "इनफार्मेशन"\n'
                                   '  ],\n'
                                   '  "informative": [\n'
                                   '   "इन्फॉर्मेटिव",\n'
                                   '   "इंफॉर्मेटिव"\n'
                                   '  ],\n'
                                   '  "island": [\n'
                                   '   "आइलैंड"\n'
                                   '  ],\n'
                                   '  "last": [\n'
                                   '   "लास्ट"\n'
                                   '  ],\n'
                                   '  "like": [\n'
                                   '   "लाइक"\n'
                                   '  ],\n'
                                   '  "line": [\n'
                                   '   "लाइन"\n'
                                   '  ],\n'
                                   '  "locker": [\n'
                                   '   "लॉकर"\n'
                                   '  ],\n'
                                   '  "maximum": [\n'
                                   '   "मैक्सिमम"\n'
                                   '  ],\n'
                                   '  "mobile": [\n'
                                   '   "मोबाइल"\n'
                                   '  ],\n'
                                   '  "motivate": [\n'
                                   '   "मोटिवेट"\n'
                                   '  ],\n'
                                   '  "night": [\n'
                                   '   "नाइट"\n'
                                   '  ],\n'
                                   '  "office": [\n'
                                   '   "ऑफिस"\n'
                                   '  ],\n'
                                   '  "paragliding": [\n'
                                   '   "पैराग्लाइडिंग"\n'
                                   '  ],\n'
                                   '  "phone": [\n'
                                   '   "फोन"\n'
                                   '  ],\n'
                                   '  "places": [\n'
                                   '   "प्लेसेस"\n'
                                   '  ],\n'
                                   '  "plan": [\n'
                                   '   "प्लान"\n'
                                   '  ],\n'
                                   '  "please": [\n'
                                   '   "प्लीज"\n'
                                   '  ],\n'
                                   '  "plus": [\n'
                                   '   "प्लस"\n'
                                   '  ],\n'
                                   '  "point": [\n'
                                   '   "प्वाइंट",\n'
                                   '   "पॉइंट"\n'
                                   '  ],\n'
                                   '  "range": [\n'
                                   '   "रेंज"\n'
                                   '  ],\n'
                                   '  "relative": [\n'
                                   '   "रिलेटिव"\n'
                                   '  ],\n'
                                   '  "rented": [\n'
                                   '   "रेंटेड"\n'
                                   '  ],\n'
                                   '  "ride": [\n'
                                   '   "राइड",\n'
                                   '   "राइट"\n'
                                   '  ],\n'
                                   '  "room": [\n'
                                   '   "रूम"\n'
                                   '  ],\n'
                                   '  "route": [\n'
                                   '   "रूट"\n'
                                   '  ],\n'
                                   '  "share": [\n'
                                   '   "शेयर"\n'
                                   '  ],\n'
                                   '  "side": [\n'
                                   '   "साइड"\n'
                                   '  ],\n'
                                   '  "speed": [\n'
                                   '   "स्पीड"\n'
                                   '  ],\n'
                                   '  "starting": [\n'
                                   '   "स्टार्टिंग"\n'
                                   '  ],\n'
                                   '  "subscribe": [\n'
                                   '   "सब्सक्राइब"\n'
                                   '  ],\n'
                                   '  "sunset": [\n'
                                   '   "सनसेट"\n'
                                   '  ],\n'
                                   '  "taxi": [\n'
                                   '   "टैक्सी"\n'
                                   '  ],\n'
                                   '  "temperature": [\n'
                                   '   "टेम्परेचर",\n'
                                   '   "टेंपरेचर"\n'
                                   '  ],\n'
                                   '  "ticket": [\n'
                                   '   "टिकट"\n'
                                   '  ],\n'
                                   '  "time": [\n'
                                   '   "टाइम"\n'
                                   '  ],\n'
                                   '  "timing": [\n'
                                   '   "टाइमिंग"\n'
                                   '  ],\n'
                                   '  "tour": [\n'
                                   '   "टूर"\n'
                                   '  ],\n'
                                   '  "tours": [\n'
                                   '   "टूर्स"\n'
                                   '  ],\n'
                                   '  "train": [\n'
                                   '   "ट्रेन"\n'
                                   '  ],\n'
                                   '  "travels": [\n'
                                   '   "ट्रैवल्स"\n'
                                   '  ],\n'
                                   '  "video": [\n'
                                   '   "वीडियो"\n'
                                   '  ],\n'
                                   '  "videos": [\n'
                                   '   "वीडियोज"\n'
                                   '  ],\n'
                                   '  "vip": [\n'
                                   '   "वीआईपी"\n'
                                   '  ],\n'
                                   '  "water": [\n'
                                   '   "वाटर"\n'
                                   '  ],\n'
                                   '  "km": [\n'
                                   '   "किलोमीटर",\n'
                                   '   "किमी"\n'
                                   '  ],\n'
                                   '  "budget": [\n'
                                   '   "बजट"\n'
                                   '  ],\n'
                                   '  "important": [\n'
                                   '   "इम्पोर्टेन्ट",\n'
                                   '   "इंपॉर्टेंट"\n'
                                   '  ],\n'
                                   '  "question": [\n'
                                   '   "क्वेश्चन"\n'
                                   '  ],\n'
                                   '  "two": [\n'
                                   '   "टू"\n'
                                   '  ],\n'
                                   '  "three": [\n'
                                   '   "थ्री"\n'
                                   '  ],\n'
                                   '  "one": [\n'
                                   '   "वन"\n'
                                   '  ]\n'
                                   ' },\n'
                                   ' "same": [\n'
                                   '  [\n'
                                   '   "रुपये",\n'
                                   '   "रुपए",\n'
                                   '   "रूपये",\n'
                                   '   "rs"\n'
                                   '  ],\n'
                                   '  [\n'
                                   '   "किलोमीटर",\n'
                                   '   "किमी",\n'
                                   '   "km"\n'
                                   '  ],\n'
                                   '  [\n'
                                   '   "मीटर",\n'
                                   '   "m"\n'
                                   '  ]\n'
                                   ' ]\n'
                                   '}',
 'core/__init__.py': '# AI Video Engine Core Package\n\n',
 'core/audio.py': '"""F04 Audio cleanup + loudness.\n'
                  '\n'
                  'highpass ~80Hz -> denoise (arnndn if a model file is given, else afftdn) -> '
                  'light EQ\n'
                  '-> loudnorm two-pass to -14 LUFS / -1.5 dBTP. Before/after measured with '
                  'ebur128.\n'
                  '"""\n'
                  'from __future__ import annotations\n'
                  '\n'
                  'import json\n'
                  'import re\n'
                  'import subprocess\n'
                  'from pathlib import Path\n'
                  'from typing import Any, Dict, Optional\n'
                  '\n'
                  'from .runs import run_ffmpeg\n'
                  '\n'
                  '\n'
                  'def _ff(args) -> str:\n'
                  '    res = subprocess.run(["ffmpeg", "-hide_banner", "-nostdin", *args], '
                  'capture_output=True, text=True, encoding="utf-8", errors="replace")\n'
                  '    if res.returncode != 0:\n'
                  '        raise RuntimeError(res.stderr[-1500:])\n'
                  '    return res.stderr\n'
                  '\n'
                  '\n'
                  'def measure_loudness(media: str | Path) -> Dict[str, float]:\n'
                  '    """Integrated LUFS, LRA and true peak (dBTP) via ebur128."""\n'
                  '    log = _ff(["-i", str(media), "-vn", "-af", "ebur128=peak=true", "-f", '
                  '"null", "-"])\n'
                  '    summary = log[log.rfind("Summary:"):]\n'
                  '    def grab(pat):\n'
                  '        m = re.search(pat, summary, re.S)\n'
                  '        return float(m.group(1)) if m else float("nan")\n'
                  '    return {\n'
                  '        "integrated_lufs": grab(r"I:\\s*(-?[\\d.]+|-inf)\\s*LUFS"),\n'
                  '        "lra_lu": grab(r"LRA:\\s*(-?[\\d.]+)\\s*LU"),\n'
                  '        "true_peak_dbtp": grab(r"True '
                  'peak:\\s*Peak:\\s*(-?[\\d.]+|-inf)\\s*dBFS"),\n'
                  '    }\n'
                  '\n'
                  '\n'
                  'def cleanup_chain(denoise: bool = True, rnnoise_model: Optional[str] = None) -> '
                  'str:\n'
                  '    chain = ["highpass=f=80"]\n'
                  '    if denoise:\n'
                  '        if rnnoise_model and Path(rnnoise_model).is_file():\n'
                  '            chain.append(f"arnndn=m=\'{Path(rnnoise_model).resolve()}\'")\n'
                  '        else:\n'
                  '            chain.append("afftdn=nr=10:nf=-40")  # gentle: avoids '
                  "'underwater' voice\n"
                  '    chain.append("equalizer=f=250:t=q:w=1:g=-2")   # reduce mud\n'
                  '    chain.append("equalizer=f=3500:t=q:w=1.2:g=2")  # presence\n'
                  '    return ",".join(chain)\n'
                  '\n'
                  '\n'
                  'def loudnorm_two_pass(media: str | Path, out: str | Path, target_lufs: float = '
                  '-14.0, true_peak: float = -1.5,\n'
                  '                      lra: float = 11.0, pre_filter: Optional[str] = None, '
                  'video_copy: bool = False,\n'
                  '                      log: Optional[Path] = None) -> Dict[str, Any]:\n'
                  '    pre = (pre_filter + ",") if pre_filter else ""\n'
                  '    first = _ff(["-i", str(media), "-vn", "-af",\n'
                  '                 '
                  'f"{pre}loudnorm=I={target_lufs}:TP={true_peak}:LRA={lra}:print_format=json", '
                  '"-f", "null", "-"])\n'
                  '    js = json.loads(first[first.rfind("{"): first.rfind("}") + 1])\n'
                  '    second = (f"{pre}loudnorm=I={target_lufs}:TP={true_peak}:LRA={lra}:"\n'
                  '              '
                  'f"measured_I={js[\'input_i\']}:measured_TP={js[\'input_tp\']}:measured_LRA={js[\'input_lra\']}:"\n'
                  '              '
                  'f"measured_thresh={js[\'input_thresh\']}:offset={js[\'target_offset\']}:linear=true,"\n'
                  '              f"aresample=48000")\n'
                  '    args = ["-i", str(media), "-af", second]\n'
                  '    out_s = str(out)\n'
                  '    if video_copy:\n'
                  '        args += ["-c:v", "copy", "-c:a", "aac", "-b:a", "192k"]\n'
                  '    elif out_s.endswith(".wav"):\n'
                  '        args += ["-vn", "-c:a", "pcm_s16le"]\n'
                  '    elif out_s.endswith((".m4a", ".mp4")):\n'
                  '        args += ["-vn", "-c:a", "aac", "-b:a", "192k"]\n'
                  '    else:\n'
                  '        args += ["-vn"]\n'
                  '    run_ffmpeg(args + [out_s], log)\n'
                  '    return {"pass1": js}\n'
                  '\n'
                  '\n'
                  'def clean_audio(media: str | Path, out: str | Path, target_lufs: float = -14.0, '
                  'true_peak: float = -1.5,\n'
                  '                denoise: bool = True, rnnoise_model: Optional[str] = None, '
                  'video_copy: bool = False,\n'
                  '                log: Optional[Path] = None) -> Dict[str, Any]:\n'
                  '    before = measure_loudness(media)\n'
                  '    chain = cleanup_chain(denoise, rnnoise_model)\n'
                  '    info = loudnorm_two_pass(media, out, target_lufs, true_peak, '
                  'pre_filter=chain, video_copy=video_copy, log=log)\n'
                  '    after = measure_loudness(out)\n'
                  '    correction = []\n'
                  '    for _ in range(3):\n'
                  '        delta = target_lufs - after["integrated_lufs"]\n'
                  '        if not (abs(delta) > 0.3 or after["true_peak_dbtp"] > true_peak + '
                  '0.1):\n'
                  '            break\n'
                  '        # loudnorm falls back to dynamic mode when linear gain would clip (e.g. '
                  'a source\n'
                  '        # already peaking at 0 dBTP) and then undershoots. Corrective pass: '
                  'exact gain +\n'
                  '        # true-peak-safe limiter (limit 0.5 dB under the TP target, 4x '
                  'oversampled).\n'
                  '        tmp = Path(out).with_name(Path(out).stem + "_pass2" + '
                  'Path(out).suffix)\n'
                  '        Path(out).replace(tmp)\n'
                  '        limit = 10 ** ((true_peak - 0.5) / 20)\n'
                  '        af = (f"aresample=192000,volume={delta:.2f}dB,"\n'
                  '              '
                  'f"alimiter=limit={limit:.4f}:attack=5:release=50:level=disabled,aresample=48000")\n'
                  '        args = ["-i", str(tmp), "-af", af]\n'
                  '        if video_copy:\n'
                  '            args += ["-c:v", "copy", "-c:a", "aac", "-b:a", "192k"]\n'
                  '        elif str(out).endswith(".wav"):\n'
                  '            args += ["-c:a", "pcm_s16le"]\n'
                  '        run_ffmpeg(args + [str(out)], log)\n'
                  '        tmp.unlink(missing_ok=True)\n'
                  '        correction.append({"gain_db": round(delta, 2), "limiter_linear": '
                  'round(limit, 4)})\n'
                  '        after = measure_loudness(out)\n'
                  '    ok = abs(after["integrated_lufs"] - target_lufs) <= 1.0 and '
                  'after["true_peak_dbtp"] <= true_peak + 0.3\n'
                  '    return {"chain": chain, "target_lufs": target_lufs, "target_true_peak": '
                  'true_peak,\n'
                  '            "before": before, "after": after, "correction": correction, "pass": '
                  'bool(ok), "loudnorm_pass1": info["pass1"],\n'
                  '            "note": "Human A/B listen still required (over-denoise check)."}\n'
                  '\n'
                  '\n'
                  'def ensure_final_true_peak(video: str | Path, limit_db: float = -1.5, log: '
                  'Optional[Path] = None,\n'
                  '                           max_passes: int = 3) -> Dict[str, Any]:\n'
                  '    """Measure the DELIVERED file; if AAC overshoot pushed true peak over '
                  'limit_db, re-limit the\n'
                  "    audio only (video stream copied) with growing margin. Kaggle's FFmpeg 6.1 "
                  'AAC added +1.6 dB\n'
                  '    on Dwarika full (-3.0 -> -1.4 dBTP), local FFmpeg ~+0.9 dB, so headroom '
                  'alone is not enough."""\n'
                  '    video = Path(video)\n'
                  '    m = measure_loudness(video)\n'
                  '    passes = []\n'
                  '    for k in range(max_passes):\n'
                  '        if m["true_peak_dbtp"] <= limit_db:\n'
                  '            break\n'
                  '        excess = m["true_peak_dbtp"] - limit_db\n'
                  '        lim = 10 ** ((limit_db - 1.0 - excess - 0.5 * k) / 20)\n'
                  '        tmp = video.with_name(video.stem + "_tpfix" + video.suffix)\n'
                  '        af = '
                  'f"aresample=192000,alimiter=limit={lim:.4f}:attack=5:release=50:level=disabled,aresample=48000"\n'
                  '        run_ffmpeg(["-i", str(video), "-map", "0", "-c:v", "copy", "-af", af, '
                  '"-c:a", "aac", "-b:a", "160k",\n'
                  '                    "-movflags", "+faststart", str(tmp)], log)\n'
                  '        tmp.replace(video)\n'
                  '        m = measure_loudness(video)\n'
                  '        passes.append({"limiter_linear": round(lim, 4), "true_peak_after": '
                  'm["true_peak_dbtp"]})\n'
                  '    return {"measured": m, "passes": passes, "pass": m["true_peak_dbtp"] <= '
                  'limit_db}\n',
 'core/brand.py': '"""F16 Brand kit engine: brand.json = single source of truth (video + design).\n'
                  '\n'
                  'Rules from the roadmap:\n'
                  '- Validate before every run.\n'
                  '- Missing font -> fallback font + VISIBLE warning (never a silent fail).\n'
                  '- Safe margin per format: max(brand.safe_margin_px, format safe zone).\n'
                  '"""\n'
                  'from __future__ import annotations\n'
                  '\n'
                  'import json\n'
                  'import re\n'
                  'from pathlib import Path\n'
                  'from typing import Dict, List, Optional, Tuple\n'
                  '\n'
                  'from pydantic import BaseModel, Field, field_validator\n'
                  '\n'
                  'HEX_RE = re.compile(r"^#[0-9A-Fa-f]{6}$")\n'
                  '\n'
                  '# Format-level minimums (roadmap F15: never below 40px). Values are px on the '
                  'canvas.\n'
                  'FORMAT_MIN_MARGIN = {"9:16": 40, "16:9": 40, "1:1": 40, "4:5": 40}\n'
                  '\n'
                  'CANVAS = {"9:16": (1080, 1920), "16:9": (1920, 1080), "1:1": (1080, 1080), '
                  '"4:5": (1080, 1350)}\n'
                  '\n'
                  '\n'
                  'class SafeZone(BaseModel):\n'
                  '    subtitle_bottom_px: int = Field(ge=0, le=1000)\n'
                  '\n'
                  '\n'
                  'class Colors(BaseModel):\n'
                  '    primary: str\n'
                  '    accent: str\n'
                  '    highlight: str\n'
                  '    text: str = "#FFFFFF"\n'
                  '    outline: str = "#000000"\n'
                  '\n'
                  '    @field_validator("*")\n'
                  '    @classmethod\n'
                  '    def _hex(cls, v: str) -> str:\n'
                  '        if not HEX_RE.match(v):\n'
                  '            raise ValueError(f"color must be #RRGGBB, got {v!r}")\n'
                  '        return v.upper()\n'
                  '\n'
                  '\n'
                  'class Fonts(BaseModel):\n'
                  '    heading: str\n'
                  '    body: str\n'
                  '    hindi: str\n'
                  '\n'
                  '\n'
                  'class Brand(BaseModel):\n'
                  '    client: str = Field(min_length=1)\n'
                  '    colors: Colors\n'
                  '    fonts: Fonts\n'
                  '    logo_path: Optional[str] = None\n'
                  '    safe_margin_px: int = Field(ge=0, le=400)\n'
                  '    safe_zones: Dict[str, SafeZone]\n'
                  '\n'
                  '    @field_validator("safe_zones")\n'
                  '    @classmethod\n'
                  '    def _formats(cls, v: Dict[str, SafeZone]) -> Dict[str, SafeZone]:\n'
                  '        bad = [k for k in v if k not in CANVAS]\n'
                  '        if bad:\n'
                  '            raise ValueError(f"unknown format(s) in safe_zones: {bad}; allowed '
                  '{list(CANVAS)}")\n'
                  '        if "9:16" not in v:\n'
                  '            raise ValueError("safe_zones must define 9:16 (Workflow A)")\n'
                  '        return v\n'
                  '\n'
                  '    # ---- derived ----------------------------------------------------------\n'
                  '    def margin_for(self, fmt: str) -> int:\n'
                  '        """Rule: max(brand.safe_margin_px, format safe zone minimum)."""\n'
                  '        return max(self.safe_margin_px, FORMAT_MIN_MARGIN.get(fmt, 40))\n'
                  '\n'
                  '    def subtitle_bottom_for(self, fmt: str) -> int:\n'
                  '        zone = self.safe_zones.get(fmt)\n'
                  '        base = zone.subtitle_bottom_px if zone else 0\n'
                  '        return max(base, self.margin_for(fmt))\n'
                  '\n'
                  '\n'
                  'def load_brand(path: str | Path) -> Brand:\n'
                  '    p = Path(path)\n'
                  '    data = json.loads(p.read_text(encoding="utf-8"))\n'
                  '    return Brand.model_validate(data)\n'
                  '\n'
                  '\n'
                  'def resolve_fonts(brand: Brand, fonts_dir: str | Path, fallback: str = '
                  '"NotoSansDevanagari-Bold.ttf") -> Tuple[Dict[str, Path], List[str]]:\n'
                  '    """Return {role: font path} and a list of visible warnings for missing '
                  'fonts."""\n'
                  '    fonts_dir = Path(fonts_dir)\n'
                  '    resolved: Dict[str, Path] = {}\n'
                  '    warnings: List[str] = []\n'
                  '    fb = fonts_dir / fallback\n'
                  '    for role, name in brand.fonts.model_dump().items():\n'
                  '        p = fonts_dir / name\n'
                  '        if p.is_file():\n'
                  '            resolved[role] = p\n'
                  '        elif fb.is_file():\n'
                  '            resolved[role] = fb\n'
                  '            warnings.append(f"FONT MISSING: \'{name}\' ({role}) not in '
                  '{fonts_dir}; using fallback \'{fallback}\'.")\n'
                  '        else:\n'
                  '            raise FileNotFoundError(f"Font \'{name}\' and fallback '
                  '\'{fallback}\' both missing in {fonts_dir}")\n'
                  '    return resolved, warnings\n'
                  '\n'
                  '\n'
                  'def font_family_name(font_path: str | Path) -> str:\n'
                  '    """Family name as libass will see it (name ID 1 first, then 16)."""\n'
                  '    from fontTools.ttLib import TTFont\n'
                  '    f = TTFont(str(font_path), lazy=True)\n'
                  '    name = f["name"]\n'
                  '    for nid in (1, 16):  # libass matches the legacy family (ID1), e.g. "Mukta '
                  'ExtraBold"\n'
                  '        rec = name.getName(nid, 3, 1) or name.getName(nid, 1, 0)\n'
                  '        if rec:\n'
                  '            return str(rec)\n'
                  '    return Path(font_path).stem\n'
                  '\n'
                  '\n'
                  'def missing_glyphs(font_path: str | Path, text: str) -> List[str]:\n'
                  '    """Characters in `text` that the font cannot render (tofu risk)."""\n'
                  '    from fontTools.ttLib import TTFont\n'
                  '    cmap = TTFont(str(font_path), lazy=True).getBestCmap()\n'
                  '    skip = {" ", "\\n", "\\u200c", "\\u200d"}\n'
                  '    return sorted({ch for ch in text if ch not in skip and ord(ch) not in '
                  'cmap})\n'
                  '\n'
                  '\n'
                  'def hex_to_rgb(h: str) -> Tuple[int, int, int]:\n'
                  '    h = h.lstrip("#")\n'
                  '    return int(h[0:2], 16), int(h[2:4], 16), int(h[4:6], 16)\n'
                  '\n'
                  '\n'
                  'def hex_to_ass(h: str, alpha: int = 0) -> str:\n'
                  '    """#RRGGBB -> ASS &HAABBGGRR."""\n'
                  '    r, g, b = hex_to_rgb(h)\n'
                  '    return f"&H{alpha:02X}{b:02X}{g:02X}{r:02X}"\n',
 'core/consensus.py': '"""Three-model consensus for the transcript review (HINTS ONLY: never edits '
                      'words.json).\n'
                      '\n'
                      'Sources: our final words.json (timed), a second timed ASR (Nemotron via '
                      'NeMo-Speech.cpp JSON with word\n'
                      'timestamps) and an optional untimed text (e.g. MAI-Transcribe-2 '
                      'Playground). Each of our words gets what the\n'
                      'other models heard there; spans where anything differs become consensus '
                      'rows:\n'
                      '  others_agree  - both other models agree with each other against ours  -> '
                      'strongest flag, listen first\n'
                      '  all_differ    - three different readings                               -> '
                      'listen\n'
                      '  ours_majority - ours matches one of the two others                     -> '
                      'low priority\n'
                      '  format_only   - same meaning, only digits/number words, Latin/Devanagari '
                      'English, units, punctuation\n'
                      "Kind 'amount' (numbers whose VALUE differs, e.g. 2,500,000 vs दो हजार से "
                      'ढाई हजार) gets priority 0.\n'
                      '  name_hint     - a known mishearing pattern (config/name_suggestions.json) '
                      '-> suggestion only\n'
                      "Timed source is aligned window by window (windows cut at the timed model's "
                      'pauses), so a hint never jumps\n'
                      'to a far-away row. The untimed source is aligned on text over the whole '
                      'file and marked text-only.\n'
                      'Agreement is NOT truth: three models can share a mistake. A human still '
                      'decides every change."""\n'
                      'from __future__ import annotations\n'
                      '\n'
                      'import json\n'
                      'from pathlib import Path\n'
                      'from typing import Any, Dict, List, Optional\n'
                      '\n'
                      'from . import second_opinion as so\n'
                      'from . import hindi_numbers as hn\n'
                      '\n'
                      '_CFG = Path(__file__).resolve().parent.parent / "config"\n'
                      '_EQ: Optional[Dict[str, str]] = None\n'
                      '\n'
                      '\n'
                      'def equivalents() -> Dict[str, str]:\n'
                      '    """norm(token) -> canonical norm, so \'train\' == \'ट्रेन\', \'km\' == '
                      '\'किलोमीटर\', \'रुपए\' == \'रुपये\'."""\n'
                      '    global _EQ\n'
                      '    if _EQ is None:\n'
                      '        eq: Dict[str, str] = {}\n'
                      '        try:\n'
                      '            for lat, dev in json.loads((_CFG / '
                      '"english_script.json").read_text(encoding="utf-8")).get("to_devanagari", '
                      '{}).items():\n'
                      '                eq[so.norm(lat)] = so.norm(dev)\n'
                      '        except (OSError, ValueError):\n'
                      '            pass\n'
                      '        try:\n'
                      '            d = json.loads((_CFG / '
                      '"script_equivalents.json").read_text(encoding="utf-8"))\n'
                      '            for lat, devs in d.get("latin", {}).items():\n'
                      '                for v in devs:\n'
                      '                    eq[so.norm(v)] = so.norm(devs[0])\n'
                      '                eq[so.norm(lat)] = so.norm(devs[0])\n'
                      '            for grp in d.get("same", []):\n'
                      '                for v in grp:\n'
                      '                    eq[so.norm(v)] = so.norm(grp[0])\n'
                      '        except (OSError, ValueError):\n'
                      '            pass\n'
                      '        _EQ = eq\n'
                      '    return _EQ\n'
                      '\n'
                      '\n'
                      'def canon(text: str) -> List[str]:\n'
                      '    """Comparison form: numbers -> \'#value\', English '
                      'script/units/spelling variants -> one form, no punctuation."""\n'
                      '    toks = [t for t in text.split() if so.norm(t)]\n'
                      '    eq = equivalents()\n'
                      '    out = []\n'
                      '    for t in hn.canon_numbers(toks):\n'
                      '        if t.startswith("#"):\n'
                      '            out.append(t)\n'
                      '            continue\n'
                      '        n = so.norm(t)\n'
                      '        out.append(eq.get(n, n))\n'
                      '    return out\n'
                      '\n'
                      'PRIORITY = {"others_agree": 1, "all_differ": 1, "name_hint": 2, '
                      '"ours_majority": 3, "format_only": 3}\n'
                      '\n'
                      '\n'
                      'def timed_words(doc: Any) -> List[Dict[str, Any]]:\n'
                      '    """NeMo-Speech.cpp JSON ({\'words\': [{word,start,end}]}) or a plain '
                      'words list."""\n'
                      '    ws = doc.get("words", []) if isinstance(doc, dict) else doc\n'
                      '    return [{"word": w["word"], "start": float(w["start"]), "end": '
                      'float(w["end"])} for w in ws if str(w.get("word", "")).strip()]\n'
                      '\n'
                      '\n'
                      'def windows(tw: List[Dict[str, Any]], target: float = 25.0, min_gap: float '
                      '= 0.25) -> List[Dict[str, Any]]:\n'
                      '    """Chunks of the timed model\'s words, cut at its pauses near `target` '
                      'seconds."""\n'
                      '    out, cur, t0 = [], [], None\n'
                      '    for i, w in enumerate(tw):\n'
                      '        if t0 is None:\n'
                      '            t0 = w["start"]\n'
                      '        cur.append(w)\n'
                      '        nxt = tw[i + 1] if i + 1 < len(tw) else None\n'
                      '        gap = (nxt["start"] - w["end"]) if nxt else 99\n'
                      '        if nxt is None or (w["end"] - t0 >= target and gap >= min_gap) or '
                      'w["end"] - t0 >= target * 1.6:\n'
                      '            cut = (w["end"] + nxt["start"]) / 2 if nxt else w["end"] + 5\n'
                      '            out.append({"start": out[-1]["end"] if out else 0.0, "end": '
                      'cut, "text": " ".join(x["word"] for x in cur),\n'
                      '                        "toks": list(cur)})\n'
                      '            cur, t0 = [], None\n'
                      '    return out\n'
                      '\n'
                      '\n'
                      'def align_timed(words: List[Dict[str, Any]], wins: List[Dict[str, Any]]) -> '
                      'Dict[str, Any]:\n'
                      '    """Like second_opinion.align, but inside a changed stretch each of the '
                      "other model's tokens goes to OUR\n"
                      '    word nearest in time, so a hint never spills into the neighbouring '
                      'row."""\n'
                      '    import difflib\n'
                      '    mapped: Dict[int, List[str]] = {}\n'
                      '    mid = lambda w: (w["start"] + w["end"]) / 2\n'
                      '    for c in wins:\n'
                      '        idx = [i for i, w in enumerate(words) if c["start"] <= mid(w) < '
                      'c["end"]]\n'
                      '        th = c["toks"]\n'
                      '        a = [" ".join(canon(words[i]["word"])) for i in idx]\n'
                      '        b = [" ".join(canon(t["word"])) for t in th]\n'
                      '        for tag, a0, a1, b0, b1 in difflib.SequenceMatcher(None, a, b, '
                      'autojunk=False).get_opcodes():\n'
                      '            if tag == "equal":\n'
                      '                for k in range(a1 - a0):\n'
                      '                    mapped[idx[a0 + k]] = [th[b0 + k]["word"]]\n'
                      '                continue\n'
                      '            cand = idx[a0:a1] or [x for x in (idx[a0 - 1] if a0 > 0 else '
                      'None, idx[a0] if a0 < len(idx) else None) if x is not None]\n'
                      '            for i in idx[a0:a1]:\n'
                      '                mapped.setdefault(i, [])\n'
                      '            for t in th[b0:b1]:\n'
                      '                if not cand:\n'
                      '                    break\n'
                      '                best = min(cand, key=lambda i: abs(mid(words[i]) - '
                      'mid(t)))\n'
                      '                mapped.setdefault(best, []).append(t["word"])\n'
                      '    return {"mapped": mapped, "ops": []}\n'
                      '\n'
                      '\n'
                      'def align_text(words: List[Dict[str, Any]], text: str) -> Dict[str, Any]:\n'
                      '    end = (words[-1]["end"] + 1) if words else 1\n'
                      '    return so.align(words, [{"start": 0.0, "end": end, "text": text}])\n'
                      '\n'
                      '\n'
                      'def _n(s: str) -> List[str]:\n'
                      '    return canon(s)\n'
                      '\n'
                      '\n'
                      'def name_hits(words: List[Dict[str, Any]], names: List[Dict[str, Any]]) -> '
                      'List[Dict[str, Any]]:\n'
                      '    toks = [so.norm(w["word"]) for w in words]\n'
                      '    hits = []\n'
                      '    for ent in names:\n'
                      '        sug = ent["suggest"]\n'
                      '        for heard in ent["heard"]:\n'
                      '            pat = _n(heard)\n'
                      '            if not pat or pat == _n(sug):\n'
                      '                continue\n'
                      '            k = len(pat)\n'
                      '            for i in range(len(toks) - k + 1):\n'
                      '                if toks[i:i + k] == pat:\n'
                      '                    hits.append({"first": i, "last": i + k - 1, "suggest": '
                      'sug, "heard": heard})\n'
                      '    return hits\n'
                      '\n'
                      '\n'
                      'def consensus(words: List[Dict[str, Any]], timed: Optional[List[Dict[str, '
                      'Any]]] = None,\n'
                      '              text: Optional[str] = None, names: Optional[List[Dict[str, '
                      'Any]]] = None) -> Dict[str, Any]:\n'
                      '    n_al = align_timed(words, windows(timed)) if timed else None\n'
                      '    m_al = align_text(words, text) if text else None\n'
                      '    names = names or []\n'
                      '\n'
                      '    def at(al, i):\n'
                      '        return None if al is None else al["mapped"].get(i, [])\n'
                      '\n'
                      '    def same(i, toks):\n'
                      '        return toks is None or canon(" ".join(toks)) == '
                      'canon(words[i]["word"])\n'
                      '\n'
                      '    bad = [not (same(i, at(n_al, i)) and same(i, at(m_al, i))) for i in '
                      'range(len(words))]\n'
                      '    rows, i = [], 0\n'
                      '    while i < len(words):\n'
                      '        if not bad[i]:\n'
                      '            i += 1\n'
                      '            continue\n'
                      '        j = i\n'
                      '        while j + 1 < len(words) and bad[j + 1]:\n'
                      '            j += 1\n'
                      '        ours = " ".join(words[k]["word"] for k in range(i, j + 1))\n'
                      '        nem = so.hint(n_al, i, j) if n_al else None\n'
                      '        mai = so.hint(m_al, i, j) if m_al else None\n'
                      '        o, a, b = _n(ours), (_n(nem) if nem is not None else None), '
                      '(_n(mai) if mai is not None else None)\n'
                      '        if all(x is None or x == o for x in (a, b)):\n'
                      '            status, sug = "format_only", ours  # digits/words, '
                      'Latin/Devanagari, punctuation: same meaning\n'
                      '        elif a is not None and b is not None and a == b and a != o:\n'
                      '            status, sug = "others_agree", nem\n'
                      '        elif o == a or o == b:\n'
                      '            status, sug = "ours_majority", ours\n'
                      '        else:  # three readings differ (or the only other model differs)\n'
                      '            status, sug = "all_differ", ""\n'
                      '        diff_src = [x for x, c in ((nem, a), (mai, b)) if x is not None and '
                      'c != o]\n'
                      '        other = diff_src[0] if diff_src else (nem if nem is not None else '
                      '(mai or ""))\n'
                      '        ov = hn.values(ours.split())\n'
                      '        amount = any(hn.values(x.split()) != ov for x in diff_src) and (ov '
                      'or any(hn.values(x.split()) for x in diff_src))\n'
                      '        kind = "amount" if amount else ("format" if status == "format_only" '
                      'else so.classify(ours, other))\n'
                      '        rows.append({"first": i, "last": j, "start": words[i]["start"], '
                      '"end": words[j]["end"], "status": status,\n'
                      '                     "kind": kind, "ours": ours, "nemotron": nem or "", '
                      '"mai": mai or "",\n'
                      '                     "suggestion": sug, "mai_alignment": "text-only" if '
                      'm_al else ""})\n'
                      '        i = j + 1\n'
                      '    for h in name_hits(words, names):\n'
                      '        rows.append({"first": h["first"], "last": h["last"], "start": '
                      'words[h["first"]]["start"],\n'
                      '                     "end": words[h["last"]]["end"], "status": "name_hint", '
                      '"kind": "name",\n'
                      '                     "ours": " ".join(words[k]["word"] for k in '
                      'range(h["first"], h["last"] + 1)),\n'
                      '                     "nemotron": so.hint(n_al, h["first"], h["last"]) if '
                      'n_al else "",\n'
                      '                     "mai": so.hint(m_al, h["first"], h["last"]) if m_al '
                      'else "",\n'
                      '                     "suggestion": h["suggest"], "mai_alignment": '
                      '"text-only" if m_al else ""})\n'
                      '    for r in rows:\n'
                      '        r["priority"] = PRIORITY[r["status"]]\n'
                      '        if r["kind"] == "amount" and r["status"] in ("others_agree", '
                      '"all_differ"):\n'
                      '            r["priority"] = 0  # amounts that change meaning: listen first\n'
                      '        elif r["kind"] == "script" and r["status"] != "name_hint":\n'
                      '            r["priority"] = 3  # Latin vs Devanagari English: caption '
                      'style, chosen at package time\n'
                      '        elif r["kind"] == "number" and r["status"] == "ours_majority":\n'
                      '            r["priority"] = 2  # amounts always deserve a listen\n'
                      '    rows.sort(key=lambda r: (r["priority"], r["start"]))\n'
                      '    for k, r in enumerate(rows, 1):\n'
                      '        r["id"], r["pattern"] = f"C{k:03d}", ""\n'
                      '    total = len(words)\n'
                      '    agree = total - sum(bad)\n'
                      '    counts = {s: sum(1 for r in rows if r["status"] == s) for s in '
                      'PRIORITY}\n'
                      '    return {"rows": rows, "counts": counts, "patterns": patterns(rows, '
                      'names=names), "words": total, "words_all_agree": agree,\n'
                      '            "all_agree_pct": round(100.0 * agree / total, 1) if total else '
                      '0.0,\n'
                      '            "sources": {"nemotron_timed": bool(timed), "mai_text": '
                      'bool(text)}}\n'
                      '\n'
                      '\n'
                      'def patterns(rows: List[Dict[str, Any]], min_count: int = 2, names: '
                      'Optional[List[Dict[str, Any]]] = None) -> List[Dict[str, Any]]:\n'
                      '    """Same change suggested by both other models in several places (e.g. '
                      'द्वारका -> द्वारिका x14):\n'
                      '    one decision for the whole pattern instead of N separate listens. Still '
                      'decided by a human."""\n'
                      '    groups: Dict[tuple, List[Dict[str, Any]]] = {}\n'
                      '    for r in rows:\n'
                      '        if r["status"] == "others_agree" and r["kind"] in ("spelling", '
                      '"spacing") and r["suggestion"] \\\n'
                      '                and not (_name_toks(names) & set(_n(r["ours"]) + '
                      '_n(r["suggestion"]))):\n'
                      '            groups.setdefault((" ".join(_n(r["ours"])), " '
                      '".join(_n(r["suggestion"]))), []).append(r)\n'
                      '    out = [{"ours": g[0]["ours"], "suggestion": g[0]["suggestion"], '
                      '"count": len(g),\n'
                      '            "times": " ".join(f"{x[\'start\']:.1f}" for x in g), "_rows": '
                      'g} for g in groups.values() if len(g) >= min_count]\n'
                      '    out.sort(key=lambda p: -p["count"])\n'
                      '    for k, p in enumerate(out, 1):\n'
                      '        p["pattern"] = f"P{k:02d}"\n'
                      '        for r in p.pop("_rows"):\n'
                      '            r["pattern"] = p["pattern"]\n'
                      '    return out\n'
                      '\n'
                      '\n'
                      'def _name_toks(names: Optional[List[Dict[str, Any]]]) -> set:\n'
                      '    """Tokens of known proper names (and their mishearings): never grouped '
                      'into a bulk \'all\' pattern."""\n'
                      '    out: set = set()\n'
                      '    names = list(names or [])\n'
                      '    try:\n'
                      '        cfg = json.loads((Path(__file__).resolve().parent.parent / "config" '
                      '/ "name_suggestions.json").read_text(encoding="utf-8"))\n'
                      '        names += [{"suggest": t, "heard": []} for t in cfg.get("protected", '
                      '[])]\n'
                      '    except (OSError, ValueError):\n'
                      '        pass\n'
                      '    for ent in names:\n'
                      '        for t in [ent["suggest"], *ent["heard"]]:\n'
                      '            out.update(_n(t))\n'
                      '    out.discard("")\n'
                      '    return out\n'
                      '\n'
                      '\n'
                      'def annotate_review(review_rows: List[Dict[str, str]], cons: Dict[str, '
                      'Any],\n'
                      '                    words: Optional[List[Dict[str, Any]]] = None) -> '
                      'List[Dict[str, str]]:\n'
                      '    """Add consensus columns to the user\'s review rows. A consensus part '
                      'is often SMALLER than the review\n'
                      "    row (R038 '35,000 से 4,000 रुपये' vs part '35,000 -> पैंतीस सौ'), so:\n"
                      '      consensus_hint_PART           - the partial suggestion(s) only; never '
                      'paste it as correct_text\n'
                      '      consensus_full_span_UNVERIFIED - the WHOLE row with only those parts '
                      'replaced (amount rows in digits),\n'
                      '                                      left blank when a part needs a listen '
                      '(all_differ) or crosses the row edge\n'
                      '    Both are hints. correct_text must be the full replacement for the whole '
                      'row, chosen after listening."""\n'
                      '    from . import hindi_numbers as hn_\n'
                      '    out = []\n'
                      '    for r in review_rows:\n'
                      '        r = dict(r)\n'
                      '        f, l = int(r["first"]), int(r["last"])\n'
                      '        hit = sorted((c for c in cons["rows"] if c["first"] <= l and '
                      'c["last"] >= f and c["status"] != "format_only"),\n'
                      '                     key=lambda c: c["priority"])\n'
                      '        r["consensus_status"] = hit[0]["status"] if hit else "all_agree"\n'
                      '        r["consensus_hint_PART"] = " | ".join(f"{c[\'ours\']} -> '
                      '{c[\'suggestion\']}" for c in hit if c["suggestion"]\n'
                      '                                              and c["status"] in '
                      '("others_agree", "name_hint"))\n'
                      '        full, note = "", ""\n'
                      '        if words is not None and hit:\n'
                      '            edge = [c for c in hit if c["first"] < f or c["last"] > l]\n'
                      '            listen = [c for c in hit if c["status"] == "all_differ"]\n'
                      '            useful = sorted((c for c in hit if c["status"] in '
                      '("others_agree", "name_hint") and c["suggestion"]),\n'
                      '                            key=lambda c: c["first"])\n'
                      '            if edge:\n'
                      '                note = "span_mismatch: a part crosses this row\'s edge - '
                      'write the full text yourself"\n'
                      '            elif listen:\n'
                      '                note = "needs_listen: models give 3 different readings"\n'
                      '            elif useful:\n'
                      '                toks, k = [], f\n'
                      '                for c in useful:\n'
                      '                    if c["first"] < k:\n'
                      '                        continue\n'
                      '                    toks += [words[x]["word"] for x in range(k, '
                      'c["first"])] + c["suggestion"].split()\n'
                      '                    k = c["last"] + 1\n'
                      '                toks += [words[x]["word"] for x in range(k, l + 1)]\n'
                      '                full = " ".join(toks)\n'
                      '                ours_txt = " ".join(words[x]["word"] for x in range(f, l + '
                      '1))\n'
                      '                if any(c["kind"] == "amount" for c in useful) and '
                      'any(ch.isdigit() for ch in ours_txt):\n'
                      "                    full = hn_.to_digits(full)  # keep the row's own style: "
                      'digits only where ours used digits\n'
                      '        r["consensus_full_span_UNVERIFIED"] = full\n'
                      '        r["consensus_note"] = note\n'
                      '        out.append(r)\n'
                      '    return out\n'
                      '\n'
                      '\n'
                      'def load_names(path: str | Path) -> List[Dict[str, Any]]:\n'
                      '    p = Path(path)\n'
                      '    return json.loads(p.read_text(encoding="utf-8")).get("names", []) if '
                      'p.exists() else []\n',
 'core/cutter.py': '"""F03 Silence + filler cut (non-destructive).\n'
                   '\n'
                   '1) silencedetect=n=-34dB:d=0.45 -> silent ranges\n'
                   "2) filler words (dictionary 'auto') merged with silence as removal "
                   'candidates;\n'
                   "   'review' words only flagged (never auto-cut)\n"
                   '3) padding: 80 ms lead-in, 90 ms lead-out around kept speech\n'
                   '4) cuts.json (keep list = EDL cuts) + cut_list.txt for the human approval '
                   'gate.\n'
                   'Guarantee: a removal range never intersects a kept word (words.json '
                   'boundaries).\n'
                   '"""\n'
                   'from __future__ import annotations\n'
                   '\n'
                   'import json\n'
                   'import re\n'
                   'import subprocess\n'
                   'from pathlib import Path\n'
                   'from typing import Any, Dict, List, Optional, Tuple\n'
                   '\n'
                   'from .runs import filter_script_args, run_ffmpeg\n'
                   '\n'
                   'Range = Tuple[float, float]\n'
                   '\n'
                   '_SIL_START = re.compile(r"silence_start:\\s*(-?[\\d.]+)")\n'
                   '_SIL_END = re.compile(r"silence_end:\\s*(-?[\\d.]+)")\n'
                   '\n'
                   '\n'
                   'def detect_silence(media: str | Path, noise_db: float = -34.0, min_dur: float '
                   '= 0.45,\n'
                   '                   total: Optional[float] = None) -> List[Range]:\n'
                   '    cmd = ["ffmpeg", "-hide_banner", "-nostdin", "-i", str(media), "-vn", '
                   '"-af",\n'
                   '           f"silencedetect=n={noise_db}dB:d={min_dur}", "-f", "null", "-"]\n'
                   '    res = subprocess.run(cmd, capture_output=True, text=True, '
                   'encoding="utf-8", errors="replace")\n'
                   '    if res.returncode != 0:\n'
                   '        raise RuntimeError(res.stderr[-1500:])\n'
                   '    return parse_silencedetect(res.stderr, total)\n'
                   '\n'
                   '\n'
                   'def parse_silencedetect(log: str, total: Optional[float] = None) -> '
                   'List[Range]:\n'
                   '    out: List[Range] = []\n'
                   '    start = None\n'
                   '    for line in log.splitlines():\n'
                   '        m = _SIL_START.search(line)\n'
                   '        if m:\n'
                   '            start = max(0.0, float(m.group(1)))\n'
                   '            continue\n'
                   '        m = _SIL_END.search(line)\n'
                   '        if m and start is not None:\n'
                   '            out.append((round(start, 3), round(float(m.group(1)), 3)))\n'
                   '            start = None\n'
                   '    if start is not None and total is not None and total > start:\n'
                   '        out.append((round(start, 3), round(total, 3)))\n'
                   '    return out\n'
                   '\n'
                   '\n'
                   'def load_fillers(path: str | Path) -> Dict[str, List[str]]:\n'
                   '    d = json.loads(Path(path).read_text(encoding="utf-8"))\n'
                   '    return {"auto": [w.lower() for w in d.get("auto", [])], "review": '
                   '[w.lower() for w in d.get("review", [])]}\n'
                   '\n'
                   '\n'
                   'def _norm(w: str) -> str:\n'
                   '    return re.sub(r"[।?!.,;:\\"\'\\-\\s]", "", w).lower()\n'
                   '\n'
                   '\n'
                   'def merge_ranges(ranges: List[Range], join_gap: float = 0.0) -> List[Range]:\n'
                   '    out: List[Range] = []\n'
                   '    for s, e in sorted(ranges):\n'
                   '        if out and s <= out[-1][1] + join_gap:\n'
                   '            out[-1] = (out[-1][0], max(out[-1][1], e))\n'
                   '        else:\n'
                   '            out.append((s, e))\n'
                   '    return out\n'
                   '\n'
                   '\n'
                   'def tighten_words(words: List[Dict[str, Any]], silences: List[Range], '
                   'min_word: float = 0.06) -> Tuple[List[Dict[str, Any]], List[Dict[str, Any]]]:\n'
                   '    """Whisper word boundaries often swallow the following/preceding pause '
                   '(e.g. a word\n'
                   '    stamped 24.02-24.80 while 24.04-24.60 is measured silence). Trust the '
                   'audio: trim a\n'
                   "    word's edge that lies inside a detected silence. A word fully inside "
                   'silence is\n'
                   '    reported (likely mis-timed) and left untouched. Returns (tightened_words, '
                   'notes)."""\n'
                   '    out, notes = [], []\n'
                   '    for w in words:\n'
                   '        s, e = float(w["start"]), float(w["end"])\n'
                   '        for a, b in silences:\n'
                   '            if b <= s or a >= e:\n'
                   '                continue\n'
                   '            if a <= s and b >= e:\n'
                   '                notes.append({"word": w["word"], "start": s, "end": e, "note": '
                   '"inside_silence_untouched"})\n'
                   '                break\n'
                   '            if a - s < min_word and b < e:   # silence covers (or nearly '
                   'covers) the word start\n'
                   '                if e - b >= min_word:\n'
                   '                    s = b\n'
                   '            elif e - b < min_word and a > s:  # silence covers (or nearly '
                   'covers) the word end\n'
                   '                if a - s >= min_word:\n'
                   '                    e = a\n'
                   '        nw = dict(w)\n'
                   '        if (s, e) != (float(w["start"]), float(w["end"])):\n'
                   '            nw["start"], nw["end"] = round(s, 3), round(e, 3)\n'
                   '            notes.append({"word": w["word"], "from": [w["start"], w["end"]], '
                   '"to": [nw["start"], nw["end"]], "note": "tightened"})\n'
                   '        out.append(nw)\n'
                   '    return out, notes\n'
                   '\n'
                   '\n'
                   'def plan_cuts(words: List[Dict[str, Any]], silences: List[Range], total: '
                   'float,\n'
                   '              fillers: Optional[Dict[str, List[str]]] = None, lead_in: float = '
                   '0.08,\n'
                   '              lead_out: float = 0.09, min_keep: float = 0.10) -> Dict[str, '
                   'Any]:\n'
                   '    """Return {keep:[(in,out)], remove:[...], flagged:[...], stats}.\n'
                   '\n'
                   "    Removal candidates = silences + 'auto' filler words. Each removal is "
                   'shrunk by the\n'
                   '    padding (lead-out after previous speech, lead-in before next speech) and '
                   'then\n'
                   '    clipped so it can never overlap a word we keep.\n'
                   '    """\n'
                   '    fillers = fillers or {"auto": [], "review": []}\n'
                   '    words, tighten_notes = tighten_words(words, silences)\n'
                   '    auto, review = set(map(_norm, fillers["auto"])), set(map(_norm, '
                   'fillers["review"]))\n'
                   '    filler_ranges: List[Range] = []\n'
                   '    flagged: List[Dict[str, Any]] = []\n'
                   '    kept_words: List[Dict[str, Any]] = []\n'
                   '    for i, w in enumerate(words):\n'
                   '        n = _norm(w["word"])\n'
                   '        if n in auto:\n'
                   '            filler_ranges.append((w["start"], w["end"]))\n'
                   '            flagged.append({"i": i, "word": w["word"], "start": w["start"], '
                   '"end": w["end"], "action": "cut_proposed"})\n'
                   '        else:\n'
                   '            kept_words.append(w)\n'
                   '            if n in review:\n'
                   '                flagged.append({"i": i, "word": w["word"], "start": '
                   'w["start"], "end": w["end"], "action": "review_only"})\n'
                   '\n'
                   '    candidates = merge_ranges(list(silences) + filler_ranges, join_gap=0.05)\n'
                   '    removes: List[Range] = []\n'
                   '    for s, e in candidates:\n'
                   '        s2 = s + (lead_out if s > 0 else 0.0)\n'
                   '        e2 = e - (lead_in if e < total else 0.0)\n'
                   '        # never touch a kept word (+ its padding)\n'
                   '        for w in kept_words:\n'
                   '            ws, we = w["start"] - lead_in, w["end"] + lead_out\n'
                   '            if we <= s2 or ws >= e2:\n'
                   '                continue\n'
                   '            # word intersects removal: clip to the side that keeps more '
                   'removal\n'
                   '            if ws - s2 >= e2 - we:\n'
                   '                e2 = min(e2, ws)\n'
                   '            else:\n'
                   '                s2 = max(s2, we)\n'
                   '        if e2 - s2 >= 0.05:\n'
                   '            removes.append((round(s2, 3), round(e2, 3)))\n'
                   '    removes = merge_ranges(removes)\n'
                   '\n'
                   '    keep: List[Range] = []\n'
                   '    cur = 0.0\n'
                   '    for s, e in removes:\n'
                   '        if s - cur >= min_keep:\n'
                   '            keep.append((round(cur, 3), round(s, 3)))\n'
                   '        cur = e\n'
                   '    if total - cur >= min_keep:\n'
                   '        keep.append((round(cur, 3), round(total, 3)))\n'
                   '\n'
                   '    removed = round(sum(e - s for s, e in removes), 3)\n'
                   '    return {\n'
                   '        "params": {"lead_in": lead_in, "lead_out": lead_out, "min_keep": '
                   'min_keep},\n'
                   '        "keep": keep, "remove": removes, "flagged": flagged,\n'
                   '        "words_tightened": words, "tighten_notes": tighten_notes,\n'
                   '        "stats": {"source_dur": round(total, 3), "removed_dur": removed,\n'
                   '                  "kept_dur": round(sum(e - s for s, e in keep), 3), '
                   '"cut_count": len(removes)},\n'
                   '    }\n'
                   '\n'
                   '\n'
                   'def check_no_word_truncated(words: List[Dict[str, Any]], keep: List[Range], '
                   'fillers_auto: Optional[List[str]] = None) -> List[str]:\n'
                   '    """QA: every non-filler word must lie fully inside one keep range."""\n'
                   '    auto = set(map(_norm, fillers_auto or []))\n'
                   '    bad = []\n'
                   '    for w in words:\n'
                   '        if _norm(w["word"]) in auto:\n'
                   '            continue\n'
                   '        if not any(s - 1e-3 <= w["start"] and w["end"] <= e + 1e-3 for s, e in '
                   'keep):\n'
                   '            bad.append(f"{w[\'word\']}@{w[\'start\']}-{w[\'end\']}")\n'
                   '    return bad\n'
                   '\n'
                   '\n'
                   'def remap_words(words: List[Dict[str, Any]], keep: List[Range]) -> '
                   'List[Dict[str, Any]]:\n'
                   '    """Map word timestamps onto the cut timeline (for subtitles after '
                   'cutting)."""\n'
                   '    out = []\n'
                   '    offset_table = []\n'
                   '    t = 0.0\n'
                   '    for s, e in keep:\n'
                   '        offset_table.append((s, e, t))\n'
                   '        t += e - s\n'
                   '    for w in words:\n'
                   '        for s, e, t0 in offset_table:\n'
                   '            if s - 1e-3 <= w["start"] and w["end"] <= e + 1e-3:\n'
                   '                nw = dict(w)\n'
                   '                nw["start"] = round(t0 + max(w["start"], s) - s, 3)\n'
                   '                nw["end"] = round(t0 + min(w["end"], e) - s, 3)\n'
                   '                out.append(nw)\n'
                   '                break\n'
                   '    return out\n'
                   '\n'
                   '\n'
                   'def cut_list_text(plan: Dict[str, Any]) -> str:\n'
                   '    lines = ["KANG cut list (approve before render)", "=" * 40]\n'
                   '    for i, (s, e) in enumerate(plan["remove"], 1):\n'
                   '        lines.append(f"CUT {i:03d}: {s:8.3f} -> {e:8.3f}  ({e - s:.2f}s)")\n'
                   '    for f in plan["flagged"]:\n'
                   '        lines.append(f"{\'FILLER\' if f[\'action\'] == \'cut_proposed\' else '
                   '\'REVIEW\'}: \'{f[\'word\']}\' @ {f[\'start\']:.2f}s ({f[\'action\']})")\n'
                   '    st = plan["stats"]\n'
                   '    lines.append(f"Source {st[\'source_dur\']}s -> kept {st[\'kept_dur\']}s '
                   '(removed {st[\'removed_dur\']}s in {st[\'cut_count\']} cuts)")\n'
                   '    return "\\n".join(lines) + "\\n"\n'
                   '\n'
                   '\n'
                   'def render_cuts(media: str | Path, keep: List[Range], out: str | Path, video: '
                   'bool = True,\n'
                   '                log: Optional[Path] = None, afade_ms: float = 5.0) -> Path:\n'
                   '    """Concatenate keep ranges via trim/atrim + concat. A 5 ms fade at every '
                   'joint\n'
                   '    prevents audible clicks at cut boundaries."""\n'
                   '    if not keep:\n'
                   '        raise ValueError("Nothing to keep")\n'
                   '    f = afade_ms / 1000.0\n'
                   '    parts, labels = [], []\n'
                   '    for i, (s, e) in enumerate(keep):\n'
                   '        d = e - s\n'
                   '        a = (f"[0:a]atrim=start={s:.3f}:end={e:.3f},asetpts=PTS-STARTPTS,"\n'
                   '             f"afade=t=in:st=0:d={f:.3f},afade=t=out:st={max(d - f, '
                   '0):.3f}:d={f:.3f}[a{i}]")\n'
                   '        if video:\n'
                   '            '
                   'parts.append(f"[0:v]trim=start={s:.3f}:end={e:.3f},setpts=PTS-STARTPTS[v{i}]")\n'
                   '            labels.append(f"[v{i}][a{i}]")\n'
                   '        else:\n'
                   '            labels.append(f"[a{i}]")\n'
                   '        parts.append(a)\n'
                   '    n = len(keep)\n'
                   '    if video:\n'
                   '        parts.append(f"{\'\'.join(labels)}concat=n={n}:v=1:a=1[v][a]")\n'
                   '        maps = ["-map", "[v]", "-map", "[a]", "-c:v", "libx264", "-preset", '
                   '"medium", "-crf", "20", "-pix_fmt", "yuv420p"]\n'
                   '    else:\n'
                   '        parts.append(f"{\'\'.join(labels)}concat=n={n}:v=0:a=1[a]")\n'
                   '        maps = ["-map", "[a]"]\n'
                   '    script = Path(out).with_suffix(".filter.txt")\n'
                   '    script.write_text(";\\n".join(parts), encoding="utf-8")\n'
                   '    acodec = ["-c:a", "aac", "-b:a", "192k"] if str(out).endswith((".mp4", '
                   '".m4a", ".mov")) else ["-c:a", "pcm_s16le"] if str(out).endswith(".wav") else '
                   '[]\n'
                   '    run_ffmpeg(["-i", str(media), *filter_script_args(script), *maps, *acodec, '
                   'str(out)], log)\n'
                   '    return Path(out)\n',
 'core/edl.py': '"""Master EDL (kang.edl/1.0): the one brain both engines read.\n'
                '\n'
                'Schema follows roadmap Section 3. Validated with Pydantic before any render.\n'
                '"""\n'
                'from __future__ import annotations\n'
                '\n'
                'import json\n'
                'from pathlib import Path\n'
                'from typing import Any, Dict, List, Literal, Optional\n'
                '\n'
                'from pydantic import BaseModel, Field, model_validator\n'
                '\n'
                'Approval = Literal["pending", "approved", "rejected", "n/a"]\n'
                '\n'
                '\n'
                'class Input(BaseModel):\n'
                '    id: str\n'
                '    path: str\n'
                '    sha256: str = Field(min_length=64, max_length=64)\n'
                '\n'
                '\n'
                'class Cut(BaseModel):\n'
                '    src: str\n'
                '    in_: float = Field(alias="in", ge=0)\n'
                '    out: float = Field(gt=0)\n'
                '    type: Literal["keep", "jumpcut"] = "keep"\n'
                '    reframe_x: float = Field(0.5, ge=0, le=1)\n'
                '\n'
                '    model_config = {"populate_by_name": True}\n'
                '\n'
                '    @model_validator(mode="after")\n'
                '    def _order(self):\n'
                '        if self.out <= self.in_:\n'
                '            raise ValueError(f"cut out ({self.out}) must be > in ({self.in_})")\n'
                '        return self\n'
                '\n'
                '\n'
                'class Subtitles(BaseModel):\n'
                '    preset: str\n'
                '    font: str\n'
                '    ass: str\n'
                '    option: Literal["A", "B"] = "A"\n'
                '\n'
                '\n'
                'class BRoll(BaseModel):\n'
                '    src: str\n'
                '    t_in: float = Field(ge=0)\n'
                '    dur: float = Field(gt=0)\n'
                '    anim: str = "slide_up"\n'
                '    status: Literal["suggested", "approved", "rejected"] = "suggested"\n'
                '    license: str = Field(min_length=1)\n'
                '\n'
                '\n'
                'class Motion(BaseModel):\n'
                '    template: str\n'
                '    params: Dict[str, Any] = {}\n'
                '    t_in: float = Field(ge=0)\n'
                '    dur: float = Field(gt=0)\n'
                '\n'
                '\n'
                'class Audio(BaseModel):\n'
                '    denoise: bool = True\n'
                '    target_lufs: float = Field(-14.0, ge=-30, le=-5)\n'
                '    true_peak_db: float = Field(-1.5, ge=-9, le=0)\n'
                '    music: Optional[str] = None\n'
                '    duck_db: float = -18\n'
                '\n'
                '\n'
                'class Export(BaseModel):\n'
                '    ratio: Literal["9:16", "16:9", "1:1", "4:5"]\n'
                '    codec: Literal["h264"] = "h264"\n'
                '    crf: int = Field(20, ge=0, le=51)\n'
                '\n'
                '\n'
                'class Approvals(BaseModel):\n'
                '    cut_list: Approval = "pending"\n'
                '    preview: Approval = "pending"\n'
                '    paid_api: Approval = "n/a"\n'
                '\n'
                '\n'
                'class EDL(BaseModel):\n'
                '    schema_: Literal["kang.edl/1.0"] = Field("kang.edl/1.0", alias="schema")\n'
                '    run_id: str\n'
                '    brand: str = "brand.json"\n'
                '    canvas: List[int] = Field(default=[1080, 1920], min_length=2, max_length=2)\n'
                '    fps: int = Field(30, ge=1, le=120)\n'
                '    inputs: List[Input]\n'
                '    cuts: List[Cut] = []\n'
                '    subtitles: Optional[Subtitles] = None\n'
                '    b_roll: List[BRoll] = []\n'
                '    motion: List[Motion] = []\n'
                '    audio: Audio = Audio()\n'
                '    export: List[Export] = [Export(ratio="9:16")]\n'
                '    approvals: Approvals = Approvals()\n'
                '    status: Literal["planned", "approved", "rendered", "failed"] = "planned"\n'
                '\n'
                '    model_config = {"populate_by_name": True}\n'
                '\n'
                '    @model_validator(mode="after")\n'
                '    def _refs(self):\n'
                '        ids = {i.id for i in self.inputs}\n'
                '        for c in self.cuts:\n'
                '            if c.src not in ids:\n'
                '                raise ValueError(f"cut references unknown input {c.src!r}")\n'
                '        # cuts per source must not overlap and must be ordered\n'
                '        last: Dict[str, float] = {}\n'
                '        for c in self.cuts:\n'
                '            if c.in_ < last.get(c.src, -1.0) - 1e-6:\n'
                '                raise ValueError(f"cuts overlap / out of order at {c.in_}")\n'
                '            last[c.src] = c.out\n'
                '        return self\n'
                '\n'
                '    def kept_duration(self) -> float:\n'
                '        return round(sum(c.out - c.in_ for c in self.cuts), 3)\n'
                '\n'
                '    def dump(self) -> Dict[str, Any]:\n'
                '        return self.model_dump(by_alias=True, mode="json")\n'
                '\n'
                '\n'
                'def load_edl(path: str | Path) -> EDL:\n'
                '    return '
                'EDL.model_validate(json.loads(Path(path).read_text(encoding="utf-8")))\n'
                '\n'
                '\n'
                'def save_edl(edl: EDL, path: str | Path) -> Path:\n'
                '    p = Path(path)\n'
                '    p.write_text(json.dumps(edl.dump(), ensure_ascii=False, indent=2), '
                'encoding="utf-8")\n'
                '    return p\n',
 'core/engine.py': '"""\n'
                   'AI Video Engine - Canonical Core Engine\n'
                   'Consolidated pipeline for audio processing, Hindi transcription,\n'
                   'timestamp validation, SRT generation, targeted re-transcription, and export.\n'
                   '"""\n'
                   '\n'
                   'from __future__ import annotations\n'
                   '\n'
                   'import json\n'
                   'import logging\n'
                   'import os\n'
                   'import re\n'
                   'import shutil\n'
                   'import subprocess\n'
                   'from pathlib import Path\n'
                   'from typing import Any, Dict, List, Optional, Tuple\n'
                   '\n'
                   'logger = logging.getLogger("ai_video_engine")\n'
                   '\n'
                   '# Configurable project base directory\n'
                   'DEFAULT_BASE = "/kaggle/working/ai-video-engine"\n'
                   'BASE = os.environ.get("ENGINE_BASE_DIR", DEFAULT_BASE)\n'
                   '\n'
                   'DEFAULT_HINDI_PROMPT = (\n'
                   '    "राधे राधे। प्यारे भक्तों, द्वारका, द्वारकाधीश, बेट द्वारका, गोपी तालाब, '
                   '"\n'
                   '    "रुक्मिणी माता मंदिर, नागेश्वर ज्योतिर्लिंग, सुदर्शन सेतु, चार धाम, '
                   'मथुरा।"\n'
                   ')\n'
                   '\n'
                   '# Extra faster-whisper transcribe() options (hotwords, vad_parameters, '
                   'beam_size...). Empty = defaults.\n'
                   '# Set from code or ENGINE_ASR_OPTIONS=\'{"hotwords": "..."}\'. Used by '
                   'tools/kaggle_remote/ab_whisper.py.\n'
                   'ASR_OPTIONS: Dict[str, Any] = {}\n'
                   '# Options used ONLY by repair / gap-fill re-transcription (e.g. a short '
                   'hotword list). The A/B on\n'
                   '# 2026-10-28 showed hotwords on the whole first pass drop ~30% of words, so '
                   'they never go there.\n'
                   'REPAIR_ASR_OPTIONS: Dict[str, Any] = {}\n'
                   '# Fixed sampling seed: faster-whisper falls back to temperature sampling on '
                   'hard windows, and without\n'
                   '# a seed two identical runs differed by 9.4% WER (Kaggle A/B 2026-10-28). '
                   'ENGINE_SEED=-1 disables.\n'
                   'SEED: int = int(os.environ.get("ENGINE_SEED", "1234"))\n'
                   '\n'
                   'MAX_WORD_DURATION_THRESHOLD = 2.5  # seconds\n'
                   'CONTEXT_PADDING_SECONDS = 2.0      # seconds before and after suspicious '
                   'region\n'
                   'FALLBACK_MAX_WORD_DURATION = 1.5   # seconds for unresolved anomalous words\n'
                   '\n'
                   '_cached_model = None\n'
                   '_cached_model_key = None\n'
                   '\n'
                   '\n'
                   'def get_audio_duration(path: str) -> float:\n'
                   '    """Return duration of audio file in seconds using ffprobe."""\n'
                   '    if not os.path.isfile(path):\n'
                   '        raise FileNotFoundError(f"Audio file not found: {path}")\n'
                   '    cmd = [\n'
                   '        "ffprobe", "-v", "error",\n'
                   '        "-show_entries", "format=duration",\n'
                   '        "-of", "default=nw=1:nk=1",\n'
                   '        str(path)\n'
                   '    ]\n'
                   '    res = subprocess.run(cmd, capture_output=True, text=True, '
                   'encoding="utf-8", errors="replace", check=True)\n'
                   '    out = res.stdout.strip()\n'
                   '    if not out:\n'
                   '        raise ValueError(f"Could not determine duration for: {path}")\n'
                   '    return float(out)\n'
                   '\n'
                   '\n'
                   'def cut_silence(src: str, dst: str, margin: str = "0.1s") -> None:\n'
                   '    """Remove silence from audio using auto-editor."""\n'
                   '    if not os.path.isfile(src):\n'
                   '        raise FileNotFoundError(f"Source file not found: {src}")\n'
                   '    os.makedirs(os.path.dirname(dst), exist_ok=True)\n'
                   '    cmd = ["auto-editor", str(src), "--margin", str(margin), "--no-open", '
                   '"-o", str(dst)]\n'
                   '    subprocess.run(cmd, check=True)\n'
                   '    if not os.path.isfile(dst) or os.path.getsize(dst) == 0:\n'
                   '        raise RuntimeError(f"Silence cutting failed to produce output at: '
                   '{dst}")\n'
                   '\n'
                   '\n'
                   'def load_audio(path: str):\n'
                   '    """Decode audio file to 16kHz mono float32 numpy array using ffmpeg."""\n'
                   '    import numpy as np\n'
                   '    if not os.path.isfile(path):\n'
                   '        raise FileNotFoundError(f"Audio file not found: {path}")\n'
                   '    cmd = [\n'
                   '        "ffmpeg", "-v", "error",\n'
                   '        "-i", str(path),\n'
                   '        "-f", "f32le",\n'
                   '        "-ar", "16000",\n'
                   '        "-ac", "1",\n'
                   '        "-"\n'
                   '    ]\n'
                   '    res = subprocess.run(cmd, capture_output=True, check=True)\n'
                   '    audio = np.frombuffer(res.stdout, dtype=np.float32)\n'
                   '    if audio.size == 0:\n'
                   '        raise RuntimeError(f"Audio decoding produced 0 samples for: {path}")\n'
                   '    return audio\n'
                   '\n'
                   '\n'
                   'def get_model(\n'
                   '    model_size: str = "large-v3-turbo",\n'
                   '    device: Optional[str] = None,\n'
                   '    compute_type: Optional[str] = None\n'
                   '):\n'
                   '    """\n'
                   '    Load and cache WhisperModel by model/device/compute configuration.\n'
                   '    GPU is preferred by default. Set ENGINE_DEVICE=cpu or pass device="cpu" '
                   'for testing.\n'
                   '    """\n'
                   '    global _cached_model, _cached_model_key\n'
                   '\n'
                   '    from faster_whisper import WhisperModel\n'
                   '\n'
                   '    if device is None:\n'
                   '        device = os.environ.get("ENGINE_DEVICE", "cuda").strip().lower()\n'
                   '    else:\n'
                   '        device = device.strip().lower()\n'
                   '\n'
                   '    if device not in {"cuda", "cpu"}:\n'
                   '        raise ValueError(f"Unsupported device {device!r}; choose \'cuda\' or '
                   '\'cpu\'.")\n'
                   '\n'
                   '    if device == "cuda":\n'
                   '        try:\n'
                   '            import torch\n'
                   '        except Exception as exc:\n'
                   '            raise RuntimeError(\n'
                   '                "CUDA was requested but PyTorch could not be imported. "\n'
                   '                "Use a Kaggle GPU runtime or explicitly set ENGINE_DEVICE=cpu '
                   'for CPU testing."\n'
                   '            ) from exc\n'
                   '        if not torch.cuda.is_available():\n'
                   '            raise RuntimeError(\n'
                   '                "CUDA was requested, but torch.cuda.is_available() is False. '
                   '"\n'
                   '                "Enable a GPU accelerator in Kaggle. For intentional CPU '
                   'testing only, "\n'
                   '                "set ENGINE_DEVICE=cpu or pass device=\'cpu\'."\n'
                   '            )\n'
                   '\n'
                   '    if compute_type is None:\n'
                   '        compute_type = os.environ.get("ENGINE_COMPUTE_TYPE") or ("float16" if '
                   'device == "cuda" else "int8")\n'
                   '\n'
                   '    cache_key = (model_size, device, compute_type)\n'
                   '    if _cached_model is not None and _cached_model_key == cache_key:\n'
                   '        return _cached_model\n'
                   '\n'
                   '    _cached_model = WhisperModel(model_size, device=device, '
                   'compute_type=compute_type)\n'
                   '    _cached_model_key = cache_key\n'
                   '    return _cached_model\n'
                   '\n'
                   '\n'
                   'def transcribe(\n'
                   '    audio: Any,\n'
                   '    language: str = "hi",\n'
                   '    initial_prompt: Optional[str] = None,\n'
                   '    model_size: str = "large-v3-turbo",\n'
                   '    vad_filter: bool = True,\n'
                   '    model: Any = None,\n'
                   '    extra_opts: Optional[Dict[str, Any]] = None\n'
                   ') -> Tuple[List[Dict[str, Any]], Any]:\n'
                   '    """\n'
                   '    Transcribe audio with faster-whisper.\n'
                   '    Preserves exact spoken Hindi words and timestamps without alteration.\n'
                   '    Uses default Hindi domain prompt if none supplied.\n'
                   '    """\n'
                   '    if model is None:\n'
                   '        model = get_model(model_size=model_size)\n'
                   '\n'
                   '    prompt = initial_prompt if initial_prompt is not None else (\n'
                   '        DEFAULT_HINDI_PROMPT if language == "hi" else None\n'
                   '    )\n'
                   '\n'
                   '    kwargs: Dict[str, Any] = {\n'
                   '        "language": language,\n'
                   '        "vad_filter": vad_filter,\n'
                   '        "word_timestamps": True,\n'
                   '        "condition_on_previous_text": False\n'
                   '    }\n'
                   '    if prompt:\n'
                   '        kwargs["initial_prompt"] = prompt\n'
                   '    env_opts = os.environ.get("ENGINE_ASR_OPTIONS")\n'
                   '    kwargs.update(json.loads(env_opts) if env_opts else {})\n'
                   '    kwargs.update(ASR_OPTIONS)\n'
                   '    kwargs.update(extra_opts or {})\n'
                   '\n'
                   '    if SEED >= 0:\n'
                   '        try:\n'
                   '            import ctranslate2\n'
                   '            ctranslate2.set_random_seed(SEED)\n'
                   '        except Exception:  # older ctranslate2 / mocked engine: run unseeded\n'
                   '            pass\n'
                   '\n'
                   '    segments, info = model.transcribe(audio, **kwargs)\n'
                   '\n'
                   '    rows = []\n'
                   '    for s in segments:\n'
                   '        words = []\n'
                   '        for w in (s.words or []):\n'
                   '            clean_word = w.word.replace("\\ufffd", "").strip()\n'
                   '            if clean_word:\n'
                   '                words.append({\n'
                   '                    "w": clean_word,\n'
                   '                    "start": round(float(w.start), 2),\n'
                   '                    "end": round(float(w.end), 2),\n'
                   '                    "p": round(float(getattr(w, "probability", 1.0) or 0.0), '
                   '3)\n'
                   '                })\n'
                   '\n'
                   '        clean_text = s.text.replace("\\ufffd", "").strip()\n'
                   '        rows.append({\n'
                   '            "start": round(float(s.start), 2),\n'
                   '            "end": round(float(s.end), 2),\n'
                   '            "text": clean_text,\n'
                   '            "words": words\n'
                   '        })\n'
                   '\n'
                   '    return rows, info\n'
                   '\n'
                   '\n'
                   '# '
                   '==============================================================================\n'
                   '# Suspicious Timestamp Detection & Bounded Targeted Re-transcription\n'
                   '# '
                   '==============================================================================\n'
                   '\n'
                   'def detect_suspicious_words(\n'
                   '    rows: List[Dict[str, Any]],\n'
                   '    max_duration: float = MAX_WORD_DURATION_THRESHOLD\n'
                   ') -> List[Dict[str, Any]]:\n'
                   '    """\n'
                   '    Scan transcript rows and flag:\n'
                   '    1. Word timestamps longer than max_duration (default 2.5s)\n'
                   '    2. Clearly invalid timestamps (end <= start, negative timestamps)\n'
                   '    """\n'
                   '    suspicious = []\n'
                   '    for s_idx, row in enumerate(rows):\n'
                   '        words = row.get("words", [])\n'
                   '        for w_idx, w in enumerate(words):\n'
                   '            start = float(w.get("start", 0.0))\n'
                   '            end = float(w.get("end", 0.0))\n'
                   '            dur = end - start\n'
                   '            reasons = []\n'
                   '\n'
                   '            if start < 0 or end < 0:\n'
                   '                reasons.append("negative_timestamp")\n'
                   '            if end <= start:\n'
                   '                reasons.append("non_positive_duration")\n'
                   '            elif dur > max_duration:\n'
                   '                reasons.append("duration_exceeds_threshold")\n'
                   '\n'
                   '            if reasons:\n'
                   '                suspicious.append({\n'
                   '                    "segment_idx": s_idx,\n'
                   '                    "word_idx": w_idx,\n'
                   '                    "word": w.get("w", ""),\n'
                   '                    "start": round(start, 2),\n'
                   '                    "end": round(end, 2),\n'
                   '                    "duration": round(dur, 2),\n'
                   '                    "reasons": reasons\n'
                   '                })\n'
                   '    return suspicious\n'
                   '\n'
                   '\n'
                   'def group_suspicious_windows(\n'
                   '    suspicious_items: List[Dict[str, Any]],\n'
                   '    total_audio_duration: float,\n'
                   '    context_padding: float = CONTEXT_PADDING_SECONDS,\n'
                   '    max_window_duration: float = 35.0\n'
                   ') -> List[Dict[str, Any]]:\n'
                   '    """\n'
                   '    Group clustered suspicious words into bounded [clip_start, clip_end] '
                   'windows\n'
                   '    with context padding before and after.\n'
                   '    """\n'
                   '    if not suspicious_items:\n'
                   '        return []\n'
                   '\n'
                   '    sorted_items = sorted(suspicious_items, key=lambda x: (x["start"], '
                   'x["end"]))\n'
                   '    windows = []\n'
                   '\n'
                   '    cur_start = sorted_items[0]["start"]\n'
                   '    cur_end = sorted_items[0]["end"]\n'
                   '    cur_items = [sorted_items[0]]\n'
                   '\n'
                   '    for item in sorted_items[1:]:\n'
                   '        if item["start"] <= cur_end + (context_padding * 2):\n'
                   '            cur_end = max(cur_end, item["end"])\n'
                   '            cur_items.append(item)\n'
                   '        else:\n'
                   '            clip_start = max(0.0, cur_start - context_padding)\n'
                   '            clip_end = min(total_audio_duration, cur_end + context_padding)\n'
                   '            if clip_end - clip_start > max_window_duration:\n'
                   '                clip_end = clip_start + max_window_duration\n'
                   '            windows.append({\n'
                   '                "clip_start": round(clip_start, 2),\n'
                   '                "clip_end": round(clip_end, 2),\n'
                   '                "suspicious_items": cur_items\n'
                   '            })\n'
                   '            cur_start = item["start"]\n'
                   '            cur_end = item["end"]\n'
                   '            cur_items = [item]\n'
                   '\n'
                   '    clip_start = max(0.0, cur_start - context_padding)\n'
                   '    clip_end = min(total_audio_duration, cur_end + context_padding)\n'
                   '    if clip_end - clip_start > max_window_duration:\n'
                   '        clip_end = clip_start + max_window_duration\n'
                   '    windows.append({\n'
                   '        "clip_start": round(clip_start, 2),\n'
                   '        "clip_end": round(clip_end, 2),\n'
                   '        "suspicious_items": cur_items\n'
                   '    })\n'
                   '    return windows\n'
                   '\n'
                   '\n'
                   'def offset_clip_timestamps(\n'
                   '    clip_rows: List[Dict[str, Any]],\n'
                   '    offset_seconds: float\n'
                   ') -> List[Dict[str, Any]]:\n'
                   '    """\n'
                   '    Offset all segment and word timestamps in clip_rows by offset_seconds.\n'
                   '    """\n'
                   '    offset_rows = []\n'
                   '    for r in clip_rows:\n'
                   '        new_row = {\n'
                   '            "start": round(float(r["start"]) + offset_seconds, 2),\n'
                   '            "end": round(float(r["end"]) + offset_seconds, 2),\n'
                   '            "text": r.get("text", ""),\n'
                   '            "words": []\n'
                   '        }\n'
                   '        for w in r.get("words", []):\n'
                   '            new_row["words"].append({\n'
                   '                "w": w["w"],\n'
                   '                "start": round(float(w["start"]) + offset_seconds, 2),\n'
                   '                "end": round(float(w["end"]) + offset_seconds, 2)\n'
                   '            })\n'
                   '        offset_rows.append(new_row)\n'
                   '    return offset_rows\n'
                   '\n'
                   '\n'
                   'def is_retranscription_reliable(\n'
                   '    clip_rows: List[Dict[str, Any]],\n'
                   '    max_word_duration: float = MAX_WORD_DURATION_THRESHOLD\n'
                   ') -> Tuple[bool, Optional[str]]:\n'
                   '    """\n'
                   '    Verify whether re-transcription output is reliable:\n'
                   '    - Must contain at least one valid word.\n'
                   '    - Must NOT contain any word with duration > max_word_duration.\n'
                   '    - All word start/end times must be non-negative and positive duration.\n'
                   '    """\n'
                   '    words = [w for r in clip_rows for w in r.get("words", [])]\n'
                   '    if not words:\n'
                   '        return False, "retranscription_produced_no_words"\n'
                   '\n'
                   '    for w in words:\n'
                   '        dur = w["end"] - w["start"]\n'
                   '        if dur <= 0:\n'
                   '            return False, f"retranscription_has_non_positive_duration_word: '
                   '{w[\'w\']}"\n'
                   '        if dur > max_word_duration:\n'
                   '            return False, f"retranscription_still_contains_long_word '
                   '({w[\'w\']}: {dur:.2f}s)"\n'
                   '\n'
                   '    return True, None\n'
                   '\n'
                   '\n'
                   'def clean_hindi_token(t: str) -> str:\n'
                   '    """Normalize Hindi token for matching by stripping punctuation and '
                   'whitespace."""\n'
                   '    return re.sub(r"[।?!.,\\s\\-]", "", t)\n'
                   '\n'
                   '\n'
                   'def merge_targeted_retranscription(\n'
                   '    all_words: List[Dict[str, Any]],\n'
                   '    clip_words: List[Dict[str, Any]],\n'
                   '    suspicious_items: List[Dict[str, Any]],\n'
                   '    clip_start: float,\n'
                   '    clip_end: float\n'
                   ') -> Tuple[List[Dict[str, Any]], bool, str]:\n'
                   '    """\n'
                   '    Merge retranscribed clip_words into all_words, replacing the suspicious '
                   'region\n'
                   '    without duplicating boundary words, dropping valid words, or creating '
                   'timestamp overlaps.\n'
                   '    """\n'
                   '    if not clip_words:\n'
                   '        return all_words, False, "no_clip_words"\n'
                   '\n'
                   '    # Identify the index range of suspicious items in all_words\n'
                   '    susp_ids = set()\n'
                   '    for s in suspicious_items:\n'
                   '        for idx, w in enumerate(all_words):\n'
                   '            if (round(float(w["start"]), 2) == round(float(s["start"]), 2) '
                   'and\n'
                   '                round(float(w["end"]), 2) == round(float(s["end"]), 2) and\n'
                   '                w.get("w", "").strip() == s.get("word", "").strip()):\n'
                   '                susp_ids.add(idx)\n'
                   '\n'
                   '    if not susp_ids:\n'
                   '        susp_start = min(s["start"] for s in suspicious_items)\n'
                   '        susp_end = max(s["end"] for s in suspicious_items)\n'
                   '        first_susp_idx = None\n'
                   '        last_susp_idx = None\n'
                   '        for idx, w in enumerate(all_words):\n'
                   '            if w["start"] >= susp_start - 0.05 and w["end"] <= susp_end + '
                   '0.05:\n'
                   '                if first_susp_idx is None:\n'
                   '                    first_susp_idx = idx\n'
                   '                last_susp_idx = idx\n'
                   '    else:\n'
                   '        first_susp_idx = min(susp_ids)\n'
                   '        last_susp_idx = max(susp_ids)\n'
                   '\n'
                   '    if first_susp_idx is None:\n'
                   '        return all_words, False, '
                   '"could_not_locate_suspicious_words_in_transcript"\n'
                   '\n'
                   '    # Guaranteed preservation of all valid words before and after\n'
                   '    prefix_words = all_words[:first_susp_idx]\n'
                   '    suffix_words = all_words[last_susp_idx + 1:]\n'
                   '\n'
                   '    # Deduplicate boundary context from clip_words\n'
                   '    replacement_words = clip_words[:]\n'
                   '\n'
                   '    # Deduplicate against prefix_words\n'
                   '    if prefix_words:\n'
                   '        last_pref = prefix_words[-1]\n'
                   '        last_pref_norm = clean_hindi_token(last_pref["w"])\n'
                   '        trim_prefix_count = 0\n'
                   '        for i, cw in enumerate(replacement_words):\n'
                   '            cw_norm = clean_hindi_token(cw["w"])\n'
                   '            if cw["end"] <= last_pref["end"] + 0.05:\n'
                   '                trim_prefix_count = i + 1\n'
                   '            elif cw["start"] < last_pref["end"] and cw_norm == '
                   'last_pref_norm:\n'
                   '                trim_prefix_count = i + 1\n'
                   '            else:\n'
                   '                break\n'
                   '        replacement_words = replacement_words[trim_prefix_count:]\n'
                   '\n'
                   '    # Deduplicate against suffix_words\n'
                   '    if suffix_words:\n'
                   '        first_suff = suffix_words[0]\n'
                   '        first_suff_norm = clean_hindi_token(first_suff["w"])\n'
                   '        trim_suffix_count = 0\n'
                   '        for i in range(len(replacement_words) - 1, -1, -1):\n'
                   '            cw = replacement_words[i]\n'
                   '            cw_norm = clean_hindi_token(cw["w"])\n'
                   '            if cw["start"] >= first_suff["start"] - 0.05:\n'
                   '                trim_suffix_count += 1\n'
                   '            elif cw["end"] > first_suff["start"] and cw_norm == '
                   'first_suff_norm:\n'
                   '                trim_suffix_count += 1\n'
                   '            else:\n'
                   '                break\n'
                   '        if trim_suffix_count > 0:\n'
                   '            replacement_words = replacement_words[:-trim_suffix_count]\n'
                   '\n'
                   '    if not replacement_words:\n'
                   '        return all_words, False, "no_replacement_words_remaining_after_dedup"\n'
                   '\n'
                   '    # Assemble complete merged words\n'
                   '    merged = prefix_words + replacement_words + suffix_words\n'
                   '\n'
                   '    # Timestamp continuity validation: ensure non-overlapping, strictly '
                   'increasing\n'
                   '    for i in range(len(merged) - 1):\n'
                   '        cur = merged[i]\n'
                   '        nxt = merged[i + 1]\n'
                   '        if cur["end"] > nxt["start"]:\n'
                   '            if nxt["start"] > cur["start"]:\n'
                   '                cur["end"] = nxt["start"]\n'
                   '            else:\n'
                   '                cur["end"] = round(cur["start"] + 0.05, 2)\n'
                   '                nxt["start"] = cur["end"]\n'
                   '                if nxt["end"] <= nxt["start"]:\n'
                   '                    nxt["end"] = round(nxt["start"] + 0.05, 2)\n'
                   '\n'
                   '    return merged, True, "success"\n'
                   '\n'
                   '\n'
                   'def clamp_unresolved_word_durations(\n'
                   '    rows: List[Dict[str, Any]],\n'
                   '    max_duration: float = MAX_WORD_DURATION_THRESHOLD,\n'
                   '    fallback_duration: float = FALLBACK_MAX_WORD_DURATION,\n'
                   '    min_duration: float = 0.05\n'
                   ') -> Tuple[List[Dict[str, Any]], Dict[str, Any]]:\n'
                   '    """\n'
                   '    Safely clamp any remaining word durations that exceed max_duration after '
                   'refinement.\n'
                   '    Guarantees:\n'
                   '    1. Validates non-positive duration (dur <= 0) first; reports anomaly '
                   'without fabricating timestamps.\n'
                   '    2. Validates the affected word and neighboring timestamps before '
                   'clamping.\n'
                   '    3. Never creates an end timestamp <= start, and never creates a new '
                   'overlap.\n'
                   '    4. If space is insufficient or neighbor timestamps are out-of-order, logs '
                   'a warning\n'
                   '       and preserves original timestamps without fabricating invalid timings.\n'
                   '    5. Preserves all original words, metadata, and word order.\n'
                   '    6. Re-syncs row boundaries using validated chronological bounds (min '
                   'start, max end)\n'
                   '       when valid; if word bounds are invalid, reports anomaly without '
                   'inventing timestamps.\n'
                   '    """\n'
                   '    all_words = [w for r in rows for w in r.get("words", [])]\n'
                   '    clamped_count = 0\n'
                   '    anomalies: List[str] = []\n'
                   '\n'
                   '    for i, w in enumerate(all_words):\n'
                   '        start = float(w.get("start", 0.0))\n'
                   '        end = float(w.get("end", 0.0))\n'
                   '        dur = end - start\n'
                   '\n'
                   "        # 1. Validate word's own timestamps: zero or negative duration\n"
                   '        if dur <= 0:\n'
                   '            msg = (\n'
                   '                f"Word \'{w.get(\'w\')}\' has non-positive duration [{start}s '
                   '-> {end}s]. "\n'
                   '                "Preserving original timestamps without fabricating '
                   'duration."\n'
                   '            )\n'
                   '            logger.warning(f"[Clamp Anomaly] {msg}")\n'
                   '            anomalies.append(msg)\n'
                   '            w["warning"] = msg\n'
                   '            w["anomaly"] = "non_positive_word_duration"\n'
                   '            continue\n'
                   '\n'
                   '        # 2. Only process words that exceed max_duration\n'
                   '        if dur <= max_duration:\n'
                   '            continue\n'
                   '\n'
                   '        # Validate preceding neighbor if present\n'
                   '        if i > 0:\n'
                   '            prev = all_words[i - 1]\n'
                   '            prev_end = float(prev.get("end", 0.0))\n'
                   '            if prev_end > start:\n'
                   '                msg = (\n'
                   '                    f"Word \'{w.get(\'w\')}\' at {start}s overlaps preceding '
                   'word \'{prev.get(\'w\')}\' "\n'
                   '                    f"(ends at {prev_end}s). Proceeding with cautious clamp."\n'
                   '                )\n'
                   '                logger.info(f"[Clamp Note] {msg}")\n'
                   '\n'
                   '        target_end = round(start + fallback_duration, 2)\n'
                   '        has_next = (i + 1 < len(all_words))\n'
                   '\n'
                   '        if has_next:\n'
                   '            nxt = all_words[i + 1]\n'
                   '            nxt_start = float(nxt.get("start", 0.0))\n'
                   '\n'
                   '            if nxt_start <= start:\n'
                   '                msg = (\n'
                   '                    f"Word \'{w.get(\'w\')}\' [{start}s -> {end}s] cannot be '
                   'clamped: next word \'{nxt.get(\'w\')}\' "\n'
                   '                    f"starts at or before it ({nxt_start}s <= {start}s). '
                   'Preserving original duration with warning."\n'
                   '                )\n'
                   '                logger.warning(f"[Clamp Anomaly] {msg}")\n'
                   '                anomalies.append(msg)\n'
                   '                w["warning"] = msg\n'
                   '                w["anomaly"] = "cannot_clamp_inverted_or_equal_next_start"\n'
                   '                continue\n'
                   '\n'
                   '            available = nxt_start - start\n'
                   '            if available < min_duration:\n'
                   '                msg = (\n'
                   '                    f"Word \'{w.get(\'w\')}\' [{start}s -> {end}s] cannot be '
                   'clamped: insufficient space before "\n'
                   '                    f"next word \'{nxt.get(\'w\')}\' at {nxt_start}s '
                   '({available:.3f}s < {min_duration}s). Preserving original duration."\n'
                   '                )\n'
                   '                logger.warning(f"[Clamp Anomaly] {msg}")\n'
                   '                anomalies.append(msg)\n'
                   '                w["warning"] = msg\n'
                   '                w["anomaly"] = '
                   '"cannot_clamp_insufficient_space_before_next_word"\n'
                   '                continue\n'
                   '\n'
                   '            safe_end = min(end, target_end, nxt_start)\n'
                   '            if safe_end - start < min_duration:\n'
                   '                safe_end = round(start + min_duration, 3)\n'
                   '                if safe_end > nxt_start:\n'
                   '                    safe_end = nxt_start\n'
                   '        else:\n'
                   '            safe_end = min(end, target_end)\n'
                   '            if safe_end - start < min_duration:\n'
                   '                safe_end = round(start + min_duration, 3)\n'
                   '\n'
                   '        if safe_end - start >= min_duration and safe_end < end:\n'
                   '            w["end"] = round(safe_end, 2)\n'
                   '            clamped_count += 1\n'
                   '            logger.info(f"[Clamp Applied] Clamped word \'{w.get(\'w\')}\' from '
                   '{dur:.2f}s to {w[\'end\'] - start:.2f}s.")\n'
                   '        else:\n'
                   '            msg = f"Word \'{w.get(\'w\')}\' [{start}s -> {end}s] cannot be '
                   'safely clamped to positive duration without overlap. Preserving original '
                   'timestamps."\n'
                   '            logger.warning(f"[Clamp Anomaly] {msg}")\n'
                   '            anomalies.append(msg)\n'
                   '            w["warning"] = msg\n'
                   '\n'
                   '    # Re-sync row boundaries using validated chronological bounds without '
                   'fabricating timestamps\n'
                   '    if clamped_count > 0:\n'
                   '        for r in rows:\n'
                   '            words = r.get("words", [])\n'
                   '            if words:\n'
                   '                w_starts = [float(x["start"]) for x in words if "start" in x]\n'
                   '                w_ends = [float(x["end"]) for x in words if "end" in x]\n'
                   '                if w_starts and w_ends:\n'
                   '                    min_s = round(min(w_starts), 2)\n'
                   '                    max_e = round(max(w_ends), 2)\n'
                   '                    if max_e <= min_s:\n'
                   '                        msg = (\n'
                   '                            f"Row \'{r.get(\'text\', \'\')}\' has invalid word '
                   'boundaries "\n'
                   '                            f"(min start {min_s}s >= max end {max_e}s). "\n'
                   '                            "Preserving original row bounds without '
                   'fabricating timestamps."\n'
                   '                        )\n'
                   '                        logger.warning(f"[Row Anomaly] {msg}")\n'
                   '                        anomalies.append(msg)\n'
                   '                        r["warning"] = msg\n'
                   '                        r["anomaly"] = "invalid_word_boundaries_reported"\n'
                   '                    else:\n'
                   '                        r["start"] = min_s\n'
                   '                        r["end"] = max_e\n'
                   '\n'
                   '    info = {\n'
                   '        "clamped_count": clamped_count,\n'
                   '        "anomalies": anomalies\n'
                   '    }\n'
                   '    return rows, info\n'
                   '\n'
                   '\n'
                   'def refine_transcript(\n'
                   '    rows: List[Dict[str, Any]],\n'
                   '    audio_samples: Any,\n'
                   '    language: str = "hi",\n'
                   '    initial_prompt: Optional[str] = None,\n'
                   '    model_size: str = "large-v3-turbo",\n'
                   '    max_word_duration: float = MAX_WORD_DURATION_THRESHOLD,\n'
                   '    context_padding: float = CONTEXT_PADDING_SECONDS,\n'
                   '    model: Any = None,\n'
                   '    total_duration: Optional[float] = None\n'
                   ') -> Tuple[List[Dict[str, Any]], Dict[str, Any]]:\n'
                   '    """\n'
                   '    Detect suspicious word timestamps (>2.5s) and perform bounded targeted '
                   're-transcription.\n'
                   '    If correction is reliable, replaces suspicious regions without duplicate '
                   'words.\n'
                   '    If correction is uncertain or fails, preserves existing transcript and '
                   'records warnings.\n'
                   '    """\n'
                   '    suspicious_items = detect_suspicious_words(rows, '
                   'max_duration=max_word_duration)\n'
                   '    stats: Dict[str, Any] = {\n'
                   '        "suspicious_count": len(suspicious_items),\n'
                   '        "windows_attempted": 0,\n'
                   '        "corrections_applied": 0,\n'
                   '        "warnings": []\n'
                   '    }\n'
                   '\n'
                   '    if not suspicious_items:\n'
                   '        return rows, stats\n'
                   '\n'
                   '    if total_duration is None:\n'
                   '        try:\n'
                   '            total_duration = float(len(audio_samples) / 16000)\n'
                   '        except Exception:\n'
                   '            total_duration = float(rows[-1]["end"]) if rows else 0.0\n'
                   '\n'
                   '    windows = group_suspicious_windows(\n'
                   '        suspicious_items,\n'
                   '        total_audio_duration=total_duration,\n'
                   '        context_padding=context_padding\n'
                   '    )\n'
                   '\n'
                   '    stats["windows_attempted"] = len(windows)\n'
                   '    current_all_words = [w for r in rows for w in r.get("words", [])]\n'
                   '\n'
                   '    for win in windows:\n'
                   '        c_start = win["clip_start"]\n'
                   '        c_end = win["clip_end"]\n'
                   '        items = win["suspicious_items"]\n'
                   '\n'
                   '        # Slice audio clip for targeted retranscription\n'
                   '        try:\n'
                   '            start_sample = max(0, int(c_start * 16000))\n'
                   '            end_sample = min(len(audio_samples), int(c_end * 16000))\n'
                   '            clip_samples = audio_samples[start_sample:end_sample]\n'
                   '\n'
                   '            if len(clip_samples) == 0:\n'
                   '                raise ValueError("Audio clip slice produced 0 samples")\n'
                   '\n'
                   '            clip_rows, _ = transcribe(\n'
                   '                audio=clip_samples,\n'
                   '                language=language,\n'
                   '                initial_prompt=initial_prompt,\n'
                   '                model_size=model_size,\n'
                   '                model=model\n'
                   '            )\n'
                   '\n'
                   '            offset_rows = offset_clip_timestamps(clip_rows, '
                   'offset_seconds=c_start)\n'
                   '            reliable, reason = is_retranscription_reliable(offset_rows, '
                   'max_word_duration=max_word_duration)\n'
                   '\n'
                   '            if not reliable:\n'
                   '                warn_msg = (\n'
                   '                    f"Suspicious region [{c_start}s - {c_end}s] '
                   're-transcription uncertain: {reason}. "\n'
                   '                    "Original transcript preserved."\n'
                   '                )\n'
                   '                logger.warning(f"[Refine Transcript] {warn_msg}")\n'
                   '                stats["warnings"].append(warn_msg)\n'
                   '                # Attach warning to affected segments\n'
                   '                for it in items:\n'
                   '                    s_idx = it["segment_idx"]\n'
                   '                    if s_idx < len(rows):\n'
                   '                        rows[s_idx]["warning"] = warn_msg\n'
                   '                continue\n'
                   '\n'
                   '            clip_words = [w for r in offset_rows for w in r.get("words", [])]\n'
                   '            new_all_words, merged_ok, merge_msg = '
                   'merge_targeted_retranscription(\n'
                   '                all_words=current_all_words,\n'
                   '                clip_words=clip_words,\n'
                   '                suspicious_items=items,\n'
                   '                clip_start=c_start,\n'
                   '                clip_end=c_end\n'
                   '            )\n'
                   '\n'
                   '            if merged_ok:\n'
                   '                current_all_words = new_all_words\n'
                   '                stats["corrections_applied"] += 1\n'
                   '                logger.info(f"[Refine Transcript] Successfully corrected '
                   'suspicious region [{c_start}s - {c_end}s].")\n'
                   '            else:\n'
                   '                warn_msg = (\n'
                   '                    f"Suspicious region [{c_start}s - {c_end}s] merge skipped: '
                   '{merge_msg}. "\n'
                   '                    "Original transcript preserved."\n'
                   '                )\n'
                   '                logger.warning(f"[Refine Transcript] {warn_msg}")\n'
                   '                stats["warnings"].append(warn_msg)\n'
                   '                for it in items:\n'
                   '                    s_idx = it["segment_idx"]\n'
                   '                    if s_idx < len(rows):\n'
                   '                        rows[s_idx]["warning"] = warn_msg\n'
                   '\n'
                   '        except Exception as exc:\n'
                   '            warn_msg = (\n'
                   '                f"Suspicious region [{c_start}s - {c_end}s] re-transcription '
                   'encountered error: {exc}. "\n'
                   '                "Original transcript preserved."\n'
                   '            )\n'
                   '            logger.warning(f"[Refine Transcript] {warn_msg}")\n'
                   '            stats["warnings"].append(warn_msg)\n'
                   '            for it in items:\n'
                   '                s_idx = it["segment_idx"]\n'
                   '                if s_idx < len(rows):\n'
                   '                    rows[s_idx]["warning"] = warn_msg\n'
                   '\n'
                   '    # If corrections were applied, rebuild rows from the updated word '
                   'sequence\n'
                   '    if stats["corrections_applied"] > 0 and current_all_words:\n'
                   '        # Group words back into segments based on sentence boundaries or '
                   'chunks\n'
                   '        updated_rows = []\n'
                   '        cur_words: List[Dict[str, Any]] = []\n'
                   '\n'
                   '        for w in current_all_words:\n'
                   '            cur_words.append(w)\n'
                   '            is_end = w["w"] and (w["w"][-1] in "।?!.,")\n'
                   '            if is_end or len(cur_words) >= 14:\n'
                   '                seg_text = " ".join(x["w"] for x in cur_words).strip()\n'
                   '                updated_rows.append({\n'
                   '                    "start": cur_words[0]["start"],\n'
                   '                    "end": cur_words[-1]["end"],\n'
                   '                    "text": seg_text,\n'
                   '                    "words": cur_words[:]\n'
                   '                })\n'
                   '                cur_words.clear()\n'
                   '\n'
                   '        if cur_words:\n'
                   '            seg_text = " ".join(x["w"] for x in cur_words).strip()\n'
                   '            updated_rows.append({\n'
                   '                "start": cur_words[0]["start"],\n'
                   '                "end": cur_words[-1]["end"],\n'
                   '                "text": seg_text,\n'
                   '                "words": cur_words[:]\n'
                   '            })\n'
                   '\n'
                   '    active_rows = updated_rows if (stats["corrections_applied"] > 0 and '
                   'current_all_words) else rows\n'
                   '    final_rows, clamp_info = clamp_unresolved_word_durations(\n'
                   '        active_rows,\n'
                   '        max_duration=max_word_duration\n'
                   '    )\n'
                   '    stats["unresolved_clamped"] = clamp_info["clamped_count"]\n'
                   '    if clamp_info["anomalies"]:\n'
                   '        stats["warnings"].extend(clamp_info["anomalies"])\n'
                   '\n'
                   '    return final_rows, stats\n'
                   '\n'
                   '\n'
                   '# '
                   '==============================================================================\n'
                   '# Timestamp Validation, Chunking, and SRT Formatting\n'
                   '# '
                   '==============================================================================\n'
                   '\n'
                   'def validate_and_fix_timestamps(\n'
                   '    chunks: List[Dict[str, Any]],\n'
                   '    min_duration: float = 0.05\n'
                   ') -> List[Dict[str, Any]]:\n'
                   '    """\n'
                   '    Validate timestamps and resolve overlaps while strictly preserving speech '
                   'sync.\n'
                   '    Guarantees:\n'
                   '    1. start >= 0\n'
                   '    2. end > start (positive segment duration)\n'
                   "    3. No overlapping adjacent chunks (chunks[i]['end'] <= "
                   "chunks[i+1]['start'])\n"
                   "    4. Speech onset (nxt['start']) is NEVER shifted forward into the future "
                   'for normal speech.\n'
                   '    5. Anomalous timestamps (equal starts, inverted starts) are explicitly '
                   'flagged with warnings.\n'
                   '    """\n'
                   '    if not chunks:\n'
                   '        return []\n'
                   '\n'
                   '    for i, c in enumerate(chunks):\n'
                   '        c["start"] = max(0.0, round(float(c["start"]), 3))\n'
                   '        raw_end = round(float(c["end"]), 3)\n'
                   '        if raw_end <= c["start"]:\n'
                   '            c["anomaly"] = "zero_or_negative_duration"\n'
                   '            c["end"] = round(c["start"] + min_duration, 3)\n'
                   '            logger.warning(\n'
                   '                f"[Timestamp Warning] Chunk {i} (\'{c.get(\'text\', \'\')}\') '
                   'had non-positive duration "\n'
                   '                f"({c[\'start\']} -> {raw_end}). Clamped end to '
                   '{c[\'end\']}."\n'
                   '            )\n'
                   '        else:\n'
                   '            c["end"] = raw_end\n'
                   '\n'
                   '    for i in range(len(chunks) - 1):\n'
                   '        cur = chunks[i]\n'
                   '        nxt = chunks[i + 1]\n'
                   '\n'
                   '        if nxt["start"] < cur["start"]:\n'
                   '            nxt["anomaly"] = "inverted_start"\n'
                   '            logger.warning(\n'
                   '                f"[Timestamp Anomaly] Chunk {i+1} starts at {nxt[\'start\']}s '
                   'before Chunk {i} starts at {cur[\'start\']}s. "\n'
                   '                f"Flagged anomaly without shifting speech timing."\n'
                   '            )\n'
                   '            if cur["end"] <= cur["start"]:\n'
                   '                cur["end"] = round(cur["start"] + min_duration, 3)\n'
                   '            continue\n'
                   '\n'
                   '        if nxt["start"] == cur["start"]:\n'
                   '            nxt["anomaly"] = "equal_start"\n'
                   '            logger.warning(\n'
                   '                f"[Timestamp Anomaly] Chunk {i} and Chunk {i+1} have identical '
                   'start time ({cur[\'start\']}s). "\n'
                   '                f"Flagged anomaly without shifting speech timing."\n'
                   '            )\n'
                   '            if cur["end"] <= cur["start"]:\n'
                   '                cur["end"] = round(cur["start"] + min_duration, 3)\n'
                   '            continue\n'
                   '\n'
                   '        if cur["end"] > nxt["start"]:\n'
                   '            cur["end"] = nxt["start"]\n'
                   '            if cur["end"] - cur["start"] < min_duration:\n'
                   '                cur["warning"] = "short_duration_segment"\n'
                   '                logger.info(\n'
                   '                    f"[Timestamp Note] Chunk {i} clamped to {cur[\'end\']}s to '
                   'yield to speech onset of Chunk {i+1}. "\n'
                   '                    f"Duration is {round(cur[\'end\'] - cur[\'start\'], '
                   '3)}s."\n'
                   '                )\n'
                   '\n'
                   '    return chunks\n'
                   '\n'
                   '\n'
                   'def chunk_words(\n'
                   '    rows: List[Dict[str, Any]],\n'
                   '    max_words: int = 7,\n'
                   '    max_chars: int = 40,\n'
                   '    gap: float = 0.6,\n'
                   '    min_duration: float = 0.05\n'
                   ') -> List[Dict[str, Any]]:\n'
                   '    """\n'
                   '    Group words into subtitle lines adhering to Hindi punctuation (।?!.,)\n'
                   '    and word limits without altering spoken words.\n'
                   '    Validates timestamps to guarantee positive duration.\n'
                   '    """\n'
                   '    words = [w for r in rows for w in r.get("words", [])]\n'
                   '    if not words:\n'
                   '        fallback = [\n'
                   '            {"start": r["start"], "end": r["end"], "text": r["text"]}\n'
                   '            for r in rows if r.get("text", "").strip()\n'
                   '        ]\n'
                   '        return validate_and_fix_timestamps(fallback, '
                   'min_duration=min_duration)\n'
                   '\n'
                   '    chunks: List[Dict[str, Any]] = []\n'
                   '    cur: List[Dict[str, Any]] = []\n'
                   '\n'
                   '    def flush():\n'
                   '        if cur:\n'
                   '            chunk_text = " ".join(w["w"] for w in cur).strip()\n'
                   '            if chunk_text:\n'
                   '                chunks.append({\n'
                   '                    "start": cur[0]["start"],\n'
                   '                    "end": cur[-1]["end"],\n'
                   '                    "text": chunk_text\n'
                   '                })\n'
                   '            cur.clear()\n'
                   '\n'
                   '    for w in words:\n'
                   '        if cur and (w["start"] - cur[-1]["end"]) > gap:\n'
                   '            flush()\n'
                   '\n'
                   '        cur.append(w)\n'
                   '        current_text = " ".join(x["w"] for x in cur)\n'
                   '\n'
                   '        is_punct = w["w"] and (w["w"][-1] in "।?!.,")\n'
                   '        if len(cur) >= max_words or len(current_text) >= max_chars or '
                   '(is_punct and len(cur) >= 3):\n'
                   '            flush()\n'
                   '\n'
                   '    flush()\n'
                   '    return validate_and_fix_timestamps(chunks, min_duration=min_duration)\n'
                   '\n'
                   '\n'
                   'def format_srt_timestamp(seconds: float) -> str:\n'
                   '    """Format float seconds into SRT timestamp string HH:MM:SS,mmm."""\n'
                   '    ms = round(seconds * 1000)\n'
                   '    h, ms = divmod(ms, 3600000)\n'
                   '    m, ms = divmod(ms, 60000)\n'
                   '    s, ms = divmod(ms, 1000)\n'
                   '    return f"{h:02}:{m:02}:{s:02},{ms:03}"\n'
                   '\n'
                   '\n'
                   'def generate_srt(chunks: List[Dict[str, Any]]) -> str:\n'
                   '    """Generate SRT formatted string from subtitle chunks."""\n'
                   '    blocks = []\n'
                   '    for i, c in enumerate(chunks, 1):\n'
                   '        start_ts = format_srt_timestamp(c["start"])\n'
                   '        end_ts = format_srt_timestamp(c["end"])\n'
                   '        blocks.append(f"{i}\\n{start_ts} --> {end_ts}\\n{c[\'text\']}\\n")\n'
                   '    return "\\n".join(blocks)\n'
                   '\n'
                   '\n'
                   'def save_outputs(\n'
                   '    rows: List[Dict[str, Any]],\n'
                   '    folder: str,\n'
                   '    min_duration: float = 0.05\n'
                   ') -> Tuple[str, str]:\n'
                   '    """Save transcripts as transcript.srt and transcript.json."""\n'
                   '    os.makedirs(folder, exist_ok=True)\n'
                   '    chunks = chunk_words(rows, min_duration=min_duration)\n'
                   '    srt_content = generate_srt(chunks)\n'
                   '\n'
                   '    srt_path = os.path.join(folder, "transcript.srt")\n'
                   '    json_path = os.path.join(folder, "transcript.json")\n'
                   '\n'
                   '    with open(srt_path, "w", encoding="utf-8") as f:\n'
                   '        f.write(srt_content)\n'
                   '\n'
                   '    with open(json_path, "w", encoding="utf-8") as f:\n'
                   '        json.dump(rows, f, ensure_ascii=False, indent=2)\n'
                   '\n'
                   '    return srt_path, json_path\n'
                   '\n'
                   '\n'
                   'def run_pipeline(\n'
                   '    src: str,\n'
                   '    job_name: str,\n'
                   '    cut: bool = False,\n'
                   '    language: str = "hi",\n'
                   '    initial_prompt: Optional[str] = None,\n'
                   '    model_size: str = "large-v3-turbo",\n'
                   '    max_word_duration: float = MAX_WORD_DURATION_THRESHOLD\n'
                   ') -> Dict[str, Any]:\n'
                   '    """Execute transcription and subtitle pipeline for an audio source."""\n'
                   '    if not os.path.isfile(src):\n'
                   '        raise FileNotFoundError(f"Input audio file not found: {src}")\n'
                   '\n'
                   '    out_dir = os.path.join(BASE, "storage", "outputs", job_name)\n'
                   '    os.makedirs(out_dir, exist_ok=True)\n'
                   '\n'
                   '    audio_path = src\n'
                   '    if cut:\n'
                   '        temp_dir = os.path.join(BASE, "storage", "temp")\n'
                   '        os.makedirs(temp_dir, exist_ok=True)\n'
                   '        work = os.path.join(temp_dir, f"{job_name}_cut.mp3")\n'
                   '        cut_silence(src, work)\n'
                   '        audio_path = work\n'
                   '\n'
                   '    shutil.copy2(audio_path, os.path.join(out_dir, "audio_used.mp3"))\n'
                   '\n'
                   '    audio_samples = load_audio(audio_path)\n'
                   '    prompt = initial_prompt if initial_prompt is not None else '
                   'DEFAULT_HINDI_PROMPT\n'
                   '\n'
                   '    rows, info = transcribe(\n'
                   '        audio=audio_samples,\n'
                   '        language=language,\n'
                   '        initial_prompt=prompt,\n'
                   '        model_size=model_size\n'
                   '    )\n'
                   '\n'
                   '    # Apply suspicious timestamp detector and targeted re-transcription\n'
                   '    rows, refine_stats = refine_transcript(\n'
                   '        rows=rows,\n'
                   '        audio_samples=audio_samples,\n'
                   '        language=language,\n'
                   '        initial_prompt=prompt,\n'
                   '        model_size=model_size,\n'
                   '        max_word_duration=max_word_duration\n'
                   '    )\n'
                   '\n'
                   '    srt_path, json_path = save_outputs(rows, out_dir)\n'
                   '    chunks = chunk_words(rows)\n'
                   '\n'
                   '    return {\n'
                   '        "job_name": job_name,\n'
                   '        "audio_used": os.path.join(out_dir, "audio_used.mp3"),\n'
                   '        "srt_path": srt_path,\n'
                   '        "json_path": json_path,\n'
                   '        "segment_count": len(rows),\n'
                   '        "subtitle_line_count": len(chunks),\n'
                   '        "refine_stats": refine_stats\n'
                   '    }\n',
 'core/gapfill.py': '"""Speech-gap detection + bounded re-transcription.\n'
                    '\n'
                    'Whisper sometimes skips 10-20 s of real speech inside one segment (found on '
                    'Dwarika 60-120 s:\n'
                    '13.1 -> 29.98 s had speech at -20 dB mean but zero words). Subtitles then '
                    'vanish.\n'
                    '\n'
                    'find_speech_gaps(): gaps between consecutive words (and at clip edges) longer '
                    'than `min_gap`\n'
                    'whose audio is mostly above `speech_db` (RMS, 100 ms frames).\n'
                    'fill_gaps(): re-transcribe only that window (with small padding), keep words '
                    'that fall inside\n'
                    'the gap, offset back to clip time. Nothing is invented; if the retry returns '
                    'nothing usable the\n'
                    'gap stays and is reported for human review.\n'
                    '"""\n'
                    'from __future__ import annotations\n'
                    '\n'
                    'from typing import Any, Callable, Dict, List, Tuple\n'
                    '\n'
                    'SR = 16000\n'
                    '# Roadmap v2 §6.2: every voiced gap > 2 s is detected, retried and either '
                    'resolved or listed for review.\n'
                    'COVERAGE_GAP_S = 2.0\n'
                    '# A gap still this long after retry blocks the run (unchanged since the '
                    'Dwarika runs); 2-3 s leftovers\n'
                    '# are reported as needs_review (RMS activity can be breath/music, so it is '
                    'not "confirmed speech").\n'
                    'BLOCKING_GAP_S = 3.0\n'
                    '\n'
                    '\n'
                    'def no_overlap(words: List[Dict[str, Any]]) -> List[Dict[str, Any]]:\n'
                    '    """Trim each word\'s end to the next word\'s start. Never pushes an end '
                    'past the next start (the old\n'
                    '    max(start+0.04, next) guard could). A word squeezed to zero length is not '
                    'dropped and gets no\n'
                    '    invented duration: its text merges into the neighbour '
                    '(words.merge_zero_duration); a squeezed but\n'
                    '    still positive word keeps its tiny duration and is flagged '
                    'timing_squeezed for review."""\n'
                    '    from .words import merge_zero_duration\n'
                    '    out = sorted((dict(w) for w in words), key=lambda w: (w["start"], '
                    'w["end"]))\n'
                    '    for a, b in zip(out, out[1:]):\n'
                    '        if b["start"] < a["end"]:\n'
                    '            a["end"] = max(a["start"], b["start"])\n'
                    '            if a["end"] - a["start"] < 0.04:\n'
                    '                a["timing_squeezed"] = True\n'
                    '    return merge_zero_duration(out)\n'
                    '\n'
                    '\n'
                    'def _active_ratio(samples, t0: float, t1: float, speech_db: float, frame: '
                    'float = 0.1) -> float:\n'
                    '    import numpy as np\n'
                    '    a = samples[int(t0 * SR):int(t1 * SR)]\n'
                    '    n = int(frame * SR)\n'
                    '    if len(a) < n:\n'
                    '        return 0.0\n'
                    '    a = a[: len(a) // n * n].reshape(-1, n)\n'
                    '    rms = np.sqrt((a.astype("float64") ** 2).mean(axis=1)) + 1e-12\n'
                    '    db = 20 * np.log10(rms)\n'
                    '    return float((db > speech_db).mean())\n'
                    '\n'
                    '\n'
                    'def find_speech_gaps(words: List[Dict[str, Any]], samples, min_gap: float = '
                    'COVERAGE_GAP_S,\n'
                    '                     speech_db: float = -35.0, min_active: float = 0.4) -> '
                    'List[Dict[str, Any]]:\n'
                    '    total = len(samples) / SR\n'
                    '    edges = [0.0] + [x for w in words for x in (w["start"], w["end"])] + '
                    '[total]\n'
                    '    gaps = []\n'
                    '    for i in range(0, len(edges), 2):\n'
                    '        g0, g1 = edges[i], edges[i + 1]\n'
                    '        if g1 - g0 >= min_gap:\n'
                    '            r = _active_ratio(samples, g0, g1, speech_db)\n'
                    '            if r >= min_active:\n'
                    '                gaps.append({"start": round(g0, 3), "end": round(g1, 3), '
                    '"active": round(r, 2)})\n'
                    '    return gaps\n'
                    '\n'
                    '\n'
                    'def fill_gaps(words: List[Dict[str, Any]], samples, gaps: List[Dict[str, '
                    'Any]],\n'
                    '              transcribe_fn: Callable[[Any], List[Dict[str, Any]]], pad: '
                    'float = 0.3\n'
                    '              ) -> Tuple[List[Dict[str, Any]], List[Dict[str, Any]]]:\n'
                    '    """transcribe_fn(samples_slice) -> engine rows. Returns (words, '
                    'report)."""\n'
                    '    from .words import rows_to_words\n'
                    '    out = [dict(w) for w in words]\n'
                    '    report = []\n'
                    '    for g in gaps:\n'
                    '        c0 = max(0.0, g["start"] - pad)\n'
                    '        c1 = min(len(samples) / SR, g["end"] + pad)\n'
                    '        try:\n'
                    '            rows = transcribe_fn(samples[int(c0 * SR):int(c1 * SR)])\n'
                    '        except Exception as exc:  # pragma: no cover\n'
                    '            report.append({**g, "filled": 0, "error": str(exc)})\n'
                    '            continue\n'
                    '        new = []\n'
                    '        for w in rows_to_words(rows):\n'
                    '            s, e = round(w["start"] + c0, 3), round(w["end"] + c0, 3)\n'
                    '            s, e = max(s, g["start"]), min(e, g["end"])\n'
                    '            if e - s >= 0.04:\n'
                    '                new.append({**w, "start": s, "end": e, "gap_fill": True})\n'
                    '        report.append({**g, "filled": len(new)})\n'
                    '        out.extend(new)\n'
                    '    # guard: no overlaps introduced at the joins\n'
                    '    return no_overlap(out), report\n'
                    '\n'
                    '\n'
                    '# --- Suspect-span repair '
                    '---------------------------------------------------------------\n'
                    '# Found on Dwarika full (Kaggle T4, 2026-10-28): Whisper compressed whole '
                    'sentences into one\n'
                    '# stretched token ("ले करने के" = 11.56 s at 571.5 s; "द" = 1.5 s + 2.2 s '
                    'silence-free gap at\n'
                    '# 176 s). find_speech_gaps() never saw them: the stretched word *covers* the '
                    'missing speech and\n'
                    '# the leftover gaps were below the old 3 s detector. A bounded '
                    're-transcription of a ~10 s window around each span\n'
                    '# recovered every missing word in tests. Nothing is invented: if the retry is '
                    'not clearly better\n'
                    '# the original words stay and the span is reported for review.\n'
                    'LONG_WORD_S = 1.2       # a single Hindi word longer than this almost always '
                    'hides dropped words\n'
                    'SHORT_GAP_S = 0.8       # inter-word gap that is mostly speech energy\n'
                    '# Dwarika v3 (2026-10-29, user review at 6:06): a gap-fill window ended '
                    'exactly on a first-pass\n'
                    "# word, so 'जो भगवान शिव के' (369.3-370.2 s) survived as the first-pass "
                    "hallucination 'टैक्सी के'\n"
                    '# (के p=0.10). Coverage looked fine, so neither trigger above fired. A very '
                    'low-confidence word\n'
                    '# right at a gap-fill seam is now a suspect span, accepted on confidence '
                    'instead of coverage.\n'
                    'SEAM_LOW_PROB = 0.2\n'
                    'SEAM_WINDOW_S = 1.5\n'
                    'CONTEXT_S = 3.0\n'
                    '\n'
                    '\n'
                    'def _uncovered_speech(words, samples, t0: float, t1: float, speech_db: float '
                    '= -35.0, long_word: float = LONG_WORD_S) -> float:\n'
                    '    """Seconds of speech-energy audio in [t0, t1] not covered by a '
                    'plausible-length word."""\n'
                    '    import numpy as np\n'
                    '    frame = 0.05\n'
                    '    n = max(0, int((t1 - t0) / frame))\n'
                    '    if n == 0:\n'
                    '        return 0.0\n'
                    '    cov = np.zeros(n, bool)\n'
                    '    for w in words:\n'
                    '        if w["end"] - w["start"] > long_word:\n'
                    '            s, e = w["start"], w["start"] + 0.6  # only the plausible head of '
                    'a stretched word counts\n'
                    '        else:\n'
                    '            s, e = w["start"], w["end"]\n'
                    '        a, b = int((s - t0) / frame), int((e - t0) / frame) + 1\n'
                    '        cov[max(0, a):max(0, min(n, b))] = True\n'
                    '    a = samples[int(t0 * SR):int(t0 * SR) + n * int(frame * SR)]\n'
                    '    if len(a) < n * int(frame * SR):\n'
                    '        n = len(a) // int(frame * SR)\n'
                    '        cov = cov[:n]\n'
                    '        a = a[: n * int(frame * SR)]\n'
                    '    if n == 0:\n'
                    '        return 0.0\n'
                    '    rms = np.sqrt((a.astype("float64").reshape(n, -1) ** 2).mean(axis=1)) + '
                    '1e-12\n'
                    '    speech = 20 * np.log10(rms) > speech_db\n'
                    '    return round(float((speech & ~cov).sum() * frame), 2)\n'
                    '\n'
                    '\n'
                    'def _drop_edge_duplicates(new: List[Dict[str, Any]], neighbours: '
                    'List[Dict[str, Any]], tol: float = 0.25) -> List[Dict[str, Any]]:\n'
                    '    """Retry windows overlap the kept words by a few frames, so the edge word '
                    'can come back twice\n'
                    "    (Dwarika 169.1 s: 'को को'). Drop a new edge word whose text matches the "
                    'adjacent kept word."""\n'
                    '    if not new or not neighbours:\n'
                    '        return new\n'
                    '    from .words import _bare\n'
                    '    left = max((w for w in neighbours if w["end"] <= new[0]["start"] + tol), '
                    'key=lambda w: w["end"], default=None)\n'
                    '    right = min((w for w in neighbours if w["start"] >= new[-1]["end"] - '
                    'tol), key=lambda w: w["start"], default=None)\n'
                    '    if left is not None and _bare(left["word"]) == _bare(new[0]["word"]) and '
                    'new[0]["start"] - left["end"] <= tol:\n'
                    '        new = new[1:]\n'
                    '    if new and right is not None and _bare(right["word"]) == '
                    '_bare(new[-1]["word"]) and right["start"] - new[-1]["end"] <= tol:\n'
                    '        new = new[:-1]\n'
                    '    return new\n'
                    '\n'
                    '\n'
                    'def find_suspect_spans(words: List[Dict[str, Any]], samples, speech_db: float '
                    '= -35.0,\n'
                    '                       long_word: float = LONG_WORD_S, short_gap: float = '
                    'SHORT_GAP_S,\n'
                    '                       merge_within: float = 2.0) -> List[Dict[str, Any]]:\n'
                    '    spans = []\n'
                    '    for i, w in enumerate(words):\n'
                    '        if w["end"] - w["start"] > long_word:\n'
                    '            spans.append({"start": w["start"], "end": w["end"], "why": '
                    'f"long_word {w[\'word\']!r} {w[\'end\'] - w[\'start\']:.2f}s"})\n'
                    '    for a, b in zip(words, words[1:]):\n'
                    '        if b["start"] - a["end"] >= short_gap and _active_ratio(samples, '
                    'a["end"], b["start"], speech_db) >= 0.5:\n'
                    '            spans.append({"start": a["end"], "end": b["start"], "why": '
                    'f"speech_gap {b[\'start\'] - a[\'end\']:.2f}s"})\n'
                    '    for i, w in enumerate(words):\n'
                    '        if w.get("gap_fill") or w.get("probability", 1.0) >= SEAM_LOW_PROB:\n'
                    '            continue\n'
                    '        near = [x for x in words[max(0, i - 4):i + 5]\n'
                    '                if x.get("gap_fill") and (abs(x["end"] - w["start"]) <= '
                    'SEAM_WINDOW_S or abs(w["end"] - x["start"]) <= SEAM_WINDOW_S)]\n'
                    '        if near:\n'
                    '            first = next((x for x in words[i::-1] if x.get("gap_fill")), '
                    'None)  # seam start = last filled word before\n'
                    '            s0 = first["end"] if first and first["end"] <= w["start"] else '
                    'w["start"]\n'
                    '            spans.append({"start": s0, "end": w["end"], "why": '
                    'f"seam_low_conf {w[\'word\']!r} p={w[\'probability\']:.2f}",\n'
                    '                          "seam": True})\n'
                    '    spans.sort(key=lambda s: s["start"])\n'
                    '    merged: List[Dict[str, Any]] = []\n'
                    '    for s in spans:\n'
                    '        if merged and s["start"] - merged[-1]["end"] <= merge_within:\n'
                    '            merged[-1]["end"] = max(merged[-1]["end"], s["end"])\n'
                    '            merged[-1]["why"] += "; " + s["why"]\n'
                    '            merged[-1]["seam"] = merged[-1].get("seam", False) or '
                    's.get("seam", False)\n'
                    '        else:\n'
                    '            merged.append(dict(s))\n'
                    '    return [{**s, "start": round(s["start"], 3), "end": round(s["end"], 3)} '
                    'for s in merged]\n'
                    '\n'
                    '\n'
                    'def _conf(ws: List[Dict[str, Any]]) -> Tuple[float, float]:\n'
                    '    ps = [w.get("probability", 1.0) for w in ws] or [0.0]\n'
                    '    return min(ps), sum(ps) / len(ps)\n'
                    '\n'
                    '\n'
                    'SWALLOW_S = 0.9     # a repaired word this long ...\n'
                    'SWALLOW_PROB = 0.7  # ... and this unsure has usually swallowed several '
                    'words\n'
                    "                    # (Dwarika 369.8 s: 'शिव के बारह ज्योतिर्लिंग' -> one 1.2 "
                    "s 'श्योतिर्लिंग', p 0.57)\n"
                    '\n'
                    '\n'
                    'def _swallowed(w: Dict[str, Any]) -> bool:\n'
                    '    p = w.get("probability", w.get("p"))\n'
                    '    return p is not None and w["end"] - w["start"] > SWALLOW_S and p < '
                    'SWALLOW_PROB\n'
                    '\n'
                    '\n'
                    'def repair_suspect_spans(words: List[Dict[str, Any]], samples, transcribe_fn: '
                    'Callable[[Any], List[Dict[str, Any]]],\n'
                    '                         spans: List[Dict[str, Any]] | None = None, context: '
                    'float = CONTEXT_S,\n'
                    '                         speech_db: float = -35.0, fallback_fns: '
                    'List[Callable[[Any], List[Dict[str, Any]]]] | None = None\n'
                    '                         ) -> Tuple[List[Dict[str, Any]], List[Dict[str, '
                    'Any]]]:\n'
                    '    """Re-transcribe a word-aligned window around each suspect span; swap in '
                    'the new words only\n'
                    "    when they cover the span's speech better and contain no stretched word. "
                    'If an attempt is not\n'
                    '    accepted, each fallback_fn (e.g. same model without the domain prompt) is '
                    'tried in order.\n'
                    '    Returns (words, report)."""\n'
                    '    from .words import rows_to_words\n'
                    '    out = [dict(w) for w in words]\n'
                    '    spans = find_suspect_spans(out, samples, speech_db) if spans is None else '
                    'spans\n'
                    '    total = len(samples) / SR\n'
                    '    report = []\n'
                    '    for sp in spans:\n'
                    '        rec: Dict[str, Any] = {**sp, "attempts": []}\n'
                    '        ok = False\n'
                    '        # Whisper is window-sensitive (Dwarika 571 s: a 568.3 s start merged '
                    'garbage into one 4 s\n'
                    '        # token, 566.5 s came back clean), so try a wider window too, each '
                    'with every transcriber.\n'
                    '        for ctx in (context, context + 2.0):\n'
                    '            inside = [w for w in out if w["start"] >= sp["start"] - ctx and '
                    'w["end"] <= sp["end"] + ctx]\n'
                    '            w0 = min([w["start"] for w in inside] + [sp["start"]])\n'
                    '            w1 = max([w["end"] for w in inside] + [sp["end"]])\n'
                    '            c0, c1 = max(0.0, w0 - 0.3), min(total, w1 + 0.3)\n'
                    '            old = [w for w in out if w["start"] >= w0 - 1e-6 and w["end"] <= '
                    'w1 + 1e-6]\n'
                    '            before = _uncovered_speech(old, samples, w0, w1, speech_db)\n'
                    '            for k, fn in enumerate([transcribe_fn] + list(fallback_fns or '
                    '[])):\n'
                    '                try:\n'
                    '                    rows = fn(samples[int(c0 * SR):int(c1 * SR)])\n'
                    '                except Exception as exc:  # pragma: no cover\n'
                    '                    rec["attempts"].append({"context": ctx, "fn": k, "error": '
                    'str(exc)})\n'
                    '                    continue\n'
                    '                new = []\n'
                    '                for w in rows_to_words(rows):\n'
                    '                    s, e = round(w["start"] + c0, 3), round(w["end"] + c0, '
                    '3)\n'
                    '                    if w0 - 0.05 <= (s + e) / 2 <= w1 + 0.05:\n'
                    '                        new.append({**w, "start": max(s, w0), "end": min(e, '
                    'w1), "repaired": True})\n'
                    '                new = [w for w in new if w["end"] - w["start"] >= 0.04]\n'
                    '                new = _drop_edge_duplicates(new, [w for w in out if w["end"] '
                    '<= w0 + 1e-6 or w["start"] >= w1 - 1e-6])\n'
                    '                after = _uncovered_speech(new, samples, w0, w1, speech_db)\n'
                    '                stretched = [w["word"] for w in new if w["end"] - w["start"] '
                    '> LONG_WORD_S\n'
                    '                             or _swallowed(w)]\n'
                    '                ok = bool(new) and not stretched and after < before and '
                    'len(new) >= len(old) * 0.8\n'
                    '                if not ok and sp.get("seam") and new and not stretched and '
                    'len(new) >= len(old) and after <= before + 0.1:\n'
                    '                    (omin, omean), (nmin, nmean) = _conf(old), _conf(new)\n'
                    '                    ok = nmin >= max(omin + 0.15, SEAM_LOW_PROB) and nmean >= '
                    'omean  # clearly more confident\n'
                    '                rec["attempts"].append({"context": ctx, "fn": k, "window": '
                    '[round(w0, 3), round(w1, 3)],\n'
                    '                                        "old_words": len(old), "new_words": '
                    'len(new), "uncovered_before": before,\n'
                    '                                        "uncovered_after": after, '
                    '"stretched": stretched, "accepted": ok})\n'
                    '                if ok:\n'
                    '                    keep = [w for w in out if not (w["start"] >= w0 - 1e-6 '
                    'and w["end"] <= w1 + 1e-6)]\n'
                    '                    out = sorted(keep + new, key=lambda w: (w["start"], '
                    'w["end"]))\n'
                    '                    break\n'
                    '            if ok:\n'
                    '                break\n'
                    '        rec["accepted"] = ok\n'
                    '        report.append(rec)\n'
                    '    return no_overlap(out), report\n',
 'core/hindi_numbers.py': '"""Hindi number words / digits -> values, so \'दो किलोमीटर\', \'2 km\' '
                          "and '2 किलोमीटर' compare as the same,\n"
                          "while '2,500,000 रुपये' vs 'दो हजार से ढाई हजार रुपये' is flagged as a "
                          'real amount mismatch."""\n'
                          'from __future__ import annotations\n'
                          '\n'
                          'import re\n'
                          'from typing import List, Optional, Tuple\n'
                          '\n'
                          '_ONES = ("शून्य एक दो तीन चार पांच छह सात आठ नौ दस ग्यारह बारह तेरह '
                          'चौदह पंद्रह सोलह सत्रह अठारह उन्नीस बीस "\n'
                          '         "इक्कीस बाईस तेईस चौबीस पच्चीस छब्बीस सत्ताईस अट्ठाईस उनतीस '
                          'तीस इकतीस बत्तीस तैंतीस चौंतीस पैंतीस छत्तीस "\n'
                          '         "सैंतीस अड़तीस उनतालीस चालीस इकतालीस बयालीस तैंतालीस चवालीस '
                          'पैंतालीस छियालीस सैंतालीस अड़तालीस उनचास पचास "\n'
                          '         "इक्यावन बावन तिरपन चौवन पचपन छप्पन सत्तावन अट्ठावन उनसठ साठ '
                          'इकसठ बासठ तिरसठ चौंसठ पैंसठ छियासठ सड़सठ "\n'
                          '         "अड़सठ उनहत्तर सत्तर इकहत्तर बहत्तर तिहत्तर चौहत्तर पचहत्तर '
                          'छिहत्तर सतहत्तर अठहत्तर उन्यासी अस्सी इक्यासी "\n'
                          '         "बयासी तिरासी चौरासी पचासी छियासी सत्तासी अट्ठासी नवासी नब्बे '
                          'इक्यानवे बानवे तिरानवे चौरानवे पचानवे छियानवे "\n'
                          '         "सत्तानवे अट्ठानवे निन्यानवे").split()\n'
                          'BASE = {w: float(i) for i, w in enumerate(_ONES)}\n'
                          'BASE.update({"पाँच": 5.0, "छः": 6.0, "छे": 6.0, "छ": 6.0, "ढाई": 2.5, '
                          '"डेढ़": 1.5, "डेढ": 1.5})\n'
                          'MULT_SMALL = {"सौ": 100.0}\n'
                          'MULT_BIG = {"हजार": 1e3, "हज़ार": 1e3, "लाख": 1e5, "करोड़": 1e7, '
                          '"करोड": 1e7}\n'
                          'PRE = {"साढ़े": 0.5, "साढे": 0.5, "सवा": 0.25, "पौने": -0.25}\n'
                          '_DIG = re.compile(r"^\\d[\\d,]*(\\.\\d+)?$")\n'
                          '\n'
                          '\n'
                          'def _clean(t: str) -> str:\n'
                          '    return re.sub(r"[\\u0964\\u0965.,!?;:\\"\'()\\-]+$", "", t.strip()) '
                          'if not _DIG.match(t.strip()) else t.strip()\n'
                          '\n'
                          '\n'
                          'def is_num_token(t: str) -> bool:\n'
                          '    c = _clean(t)\n'
                          '    return bool(_DIG.match(c)) or c in BASE or c in MULT_SMALL or c in '
                          'MULT_BIG or c in PRE\n'
                          '\n'
                          '\n'
                          'def parse_run(toks: List[str]) -> Optional[float]:\n'
                          '    """Value of one run of number tokens (\'एक सौ पैंतीस\' -> 135, '
                          '\'पैंतीस सौ\' -> 3500, \'2,500\' -> 2500)."""\n'
                          '    total, cur, pre = 0.0, 0.0, 0.0\n'
                          '    seen = False\n'
                          '    for raw in toks:\n'
                          '        t = _clean(raw)\n'
                          '        if _DIG.match(t):\n'
                          '            cur += float(t.replace(",", "")); seen = True\n'
                          '        elif t in PRE:\n'
                          '            pre = PRE[t]\n'
                          '        elif t in BASE:\n'
                          '            cur += BASE[t] + pre; pre = 0.0; seen = True\n'
                          '        elif t in MULT_SMALL:\n'
                          '            cur = (cur or 1.0) * MULT_SMALL[t]; seen = True\n'
                          '        elif t in MULT_BIG:\n'
                          '            total += (cur or 1.0) * MULT_BIG[t]; cur = 0.0; seen = '
                          'True\n'
                          '        else:\n'
                          '            return None\n'
                          '    return total + cur if seen else None\n'
                          '\n'
                          '\n'
                          'def canon_numbers(toks: List[str]) -> List[str]:\n'
                          '    """Replace each run of number tokens with \'#<value>\' (other '
                          'tokens unchanged)."""\n'
                          '    out: List[str] = []\n'
                          '    i = 0\n'
                          '    while i < len(toks):\n'
                          '        if is_num_token(toks[i]) and _clean(toks[i]) not in PRE or '
                          '(_clean(toks[i]) in PRE and i + 1 < len(toks) and is_num_token(toks[i + '
                          '1])):\n'
                          '            j = i\n'
                          '            while j < len(toks) and is_num_token(toks[j]):\n'
                          '                j += 1\n'
                          '            v = parse_run(toks[i:j])\n'
                          '            if v is None:\n'
                          '                out.extend(toks[i:j])\n'
                          '            else:\n'
                          '                out.append("#" + (str(int(v)) if v == int(v) else '
                          'str(v)))\n'
                          '            i = j\n'
                          '        else:\n'
                          '            out.append(toks[i]); i += 1\n'
                          '    return out\n'
                          '\n'
                          '\n'
                          'def values(toks: List[str]) -> List[float]:\n'
                          '    return [float(t[1:]) for t in canon_numbers(toks) if '
                          't.startswith("#")]\n'
                          '\n'
                          '\n'
                          'def to_digits(text: str) -> str:\n'
                          '    """Write every number run as plain digits (\'पैंतीस सौ से 4,000 '
                          'रुपये\' -> \'3500 से 4000 रुपये\')."""\n'
                          '    toks = text.split()\n'
                          '    out = []\n'
                          '    for t in canon_numbers(toks):\n'
                          '        out.append(t[1:] if t.startswith("#") else t)\n'
                          '    return " ".join(out)\n',
 'core/package_readiness.py': '"""Separate package generation from human acceptance. No '
                              'media/model execution."""\n'
                              'import json\n'
                              'from pathlib import Path\n'
                              '\n'
                              '\n'
                              'def timing_pending(words):\n'
                              '    return [i for i, word in enumerate(words)\n'
                              '            if word.get("timing_status") == "unverified"\n'
                              '            or (word.get("timing_method") == '
                              '"interpolated_after_review"\n'
                              '                and word.get("timing_status") not in {"verified", '
                              '"human_verified", "aligned_verified"})]\n'
                              '\n'
                              '\n'
                              'def classify_package(status, words, qa_pass):\n'
                              '    pending = timing_pending(words)\n'
                              '    if not qa_pass:\n'
                              '        state = "qa_failed"\n'
                              '    elif (not status.get("transcript_accepted", False)\n'
                              '          or not status.get("scene_plan_human_approved", False) or '
                              'pending):\n'
                              '        state = "draft_package_ready"\n'
                              '    else:\n'
                              '        state = "package_generated_pending_verification"\n'
                              '    return {\n'
                              '        "package_status": state,\n'
                              '        "text_accepted": bool(status.get("transcript_accepted", '
                              'False)),\n'
                              '        "changed_word_timing_review_required": bool(pending),\n'
                              '        "timing_unverified_word_indices": pending,\n'
                              '        "full_audio_sync": "NOT_TESTED",\n'
                              '        "accepted": False,\n'
                              '        "acceptance_scope": "generation_only; final '
                              'QA/import/visual evidence required",\n'
                              '    }\n'
                              '\n'
                              '\n'
                              'def finalize_package_readiness(run, result, package_dir, words, '
                              'report):\n'
                              '    manifest_path = Path(package_dir) / "package_manifest.json"\n'
                              '    manifest = '
                              'json.loads(manifest_path.read_text(encoding="utf-8"))\n'
                              '    status = manifest.setdefault("status", {})\n'
                              '    readiness = classify_package(status, words, '
                              'bool(report.get("pass", False)))\n'
                              '    status.update(readiness)\n'
                              '    tmp = manifest_path.with_name(manifest_path.name + ".tmp")\n'
                              '    tmp.write_text(json.dumps(manifest, ensure_ascii=False, '
                              'indent=2), encoding="utf-8")\n'
                              '    tmp.replace(manifest_path)\n'
                              '    result.update(readiness)\n'
                              '    result["processing_complete"] = True\n'
                              '    run.manifest["package_readiness"] = readiness\n'
                              '    run.set_status(readiness["package_status"])\n'
                              '    run.add_output(manifest_path, "package_manifest")\n',
 'core/pipeline.py': '"""Planner + workflow orchestration (Layers 3-7).\n'
                     '\n'
                     'Every function creates/uses a Run (unique folder + manifest), never touches '
                     'originals,\n'
                     'and stops at approval gates unless approve=True is passed explicitly.\n'
                     '"""\n'
                     'from __future__ import annotations\n'
                     '\n'
                     'import json\n'
                     'import os\n'
                     'import shutil\n'
                     'import time\n'
                     'from pathlib import Path, PureWindowsPath\n'
                     'from typing import Any, Callable, Dict, List, Optional\n'
                     '\n'
                     'from . import audio as audio_mod\n'
                     'from . import cutter, qa, render\n'
                     'from .brand import CANVAS, load_brand, resolve_fonts\n'
                     'from .edl import EDL, Approvals, Audio, Cut, Export, Input, Subtitles, '
                     'save_edl\n'
                     'from .runs import Run, media_summary, write_json\n'
                     'from .subtitles import build_ass, build_srt, load_preset, make_layout\n'
                     'from .gapfill import BLOCKING_GAP_S, COVERAGE_GAP_S, fill_gaps, '
                     'find_speech_gaps, repair_suspect_spans\n'
                     'from .words import join_split_tokens, repeated_runs, sanitize_script, '
                     'apply_glossary, audit_words, load_glossary, load_words, review_sample, '
                     'rows_to_words\n'
                     '\n'
                     'ROOT = Path(__file__).resolve().parent.parent\n'
                     'Approver = Callable[[str, str], bool]\n'
                     '\n'
                     '\n'
                     'def _deny(gate: str, info: str) -> bool:\n'
                     '    return False\n'
                     '\n'
                     '\n'
                     '_deny.approver_name = "default-deny"\n'
                     '\n'
                     '\n'
                     'def cli_approver(gate: str, info: str) -> bool:\n'
                     '    print(f"\\n=== APPROVAL GATE: {gate} ===\\n{info}")\n'
                     '    try:\n'
                     '        return input("Approve? [y/N]: ").strip().lower() in {"y", "yes", '
                     '"haan", "ha"}\n'
                     '    except EOFError:\n'
                     '        return False\n'
                     '\n'
                     '\n'
                     'cli_approver.approver_name = "cli-user"\n'
                     '\n'
                     '\n'
                     'def plan_hash(info: str, artifacts=(), settings: Optional[Dict[str, Any]] = '
                     'None) -> str:\n'
                     '    """sha256 over the text shown + the CONTENT of every artifact file + the '
                     'settings dict, so a changed\n'
                     '    preview/words/ASS/settings behind the same path gives a different '
                     'hash."""\n'
                     '    import hashlib\n'
                     '    from .runs import sha256_file\n'
                     '    h = hashlib.sha256(info.encode("utf-8"))\n'
                     '    for a in artifacts:\n'
                     '        h.update(f"\\n{Path(a).name}:{sha256_file(a)}".encode("utf-8"))\n'
                     '    if settings is not None:\n'
                     '        h.update(json.dumps(settings, sort_keys=True, ensure_ascii=False, '
                     'default=str).encode("utf-8"))\n'
                     '    return h.hexdigest()\n'
                     '\n'
                     '\n'
                     'AUTO_APPROVERS = {"auto(--yes)", "default-deny"}\n'
                     '\n'
                     '\n'
                     'def _gate(run: Run, gate: str, info: str, approver: Approver, artifacts=(), '
                     'settings=None) -> bool:\n'
                     '    """Every decision is recorded in approvals.json bound to plan_hash(info, '
                     'artifacts, settings) = the exact\n'
                     "    plan/preview shown, with the approver's identity (cli-user, auto(--yes), "
                     "...). A 'no' from a real\n"
                     "    approver is 'rejected'; the non-interactive default-deny leaves the gate "
                     "'pending'. auto(--yes) is\n"
                     '    recorded as automation, never as a human review (manifest '
                     'human_approved=False)."""\n'
                     '    ok = approver(gate, info)\n'
                     '    h = plan_hash(info, artifacts, settings)\n'
                     '    who = getattr(approver, "approver_name", getattr(approver, "__name__", '
                     '"callable"))\n'
                     '    value = "approved" if ok else ("pending" if who == "default-deny" else '
                     '"rejected")\n'
                     '    run.set_approval(gate, value, plan_hash=h, approver=who)\n'
                     '    run.manifest.setdefault("human_approved", {})[gate] = bool(ok) and who '
                     'not in AUTO_APPROVERS\n'
                     '    run.save()\n'
                     '    run.log_step(f"gate:{gate}", approved=ok, plan_hash=h[:12], '
                     'approver=who)\n'
                     '    return ok\n'
                     '\n'
                     '\n'
                     '# '
                     '--------------------------------------------------------------------------- '
                     'Phase 1\n'
                     'def transcribe_to_words(audio_path: str, run: Optional[Run] = None, '
                     'language: str = "hi",\n'
                     '                        model_size: str = "large-v3-turbo", initial_prompt: '
                     'Optional[str] = None,\n'
                     '                        device: Optional[str] = None, transcriber: '
                     'Optional[Callable] = None,\n'
                     '                        voters: Optional[Dict[str, Any]] = None, wordlist: '
                     'Optional[str] = None,\n'
                     '                        vote_agree: bool = True, spelling_policy: '
                     'Optional[str] = None) -> Dict[str, Any]:\n'
                     '    """Fresh run, cut=False: transcript JSON + SRT + words.json + anomaly '
                     'audit + review sample.\n'
                     '    `transcriber(audio_samples) -> rows` can be injected for offline tests.\n'
                     '    `voters` {name: ct2 model dir | callable(samples)->rows} turns on the '
                     'spelling voter (core/voter.py):\n'
                     '    second models only correct the spelling of words already there, never '
                     'add/drop words or timing.\n'
                     '    Default off; also ENGINE_VOTERS="name=dir,..." + '
                     'ENGINE_WORDLIST=<hunspell .dic>."""\n'
                     '    from . import engine\n'
                     '    run = run or Run()\n'
                     '    run.add_input(audio_path, "A1")\n'
                     '    samples = engine.load_audio(audio_path)\n'
                     '    if transcriber is not None:\n'
                     '        rows, stats = transcriber(samples), {"mocked": True}\n'
                     '        retry = transcriber\n'
                     '        fallbacks = []\n'
                     '    else:\n'
                     '        model = engine.get_model(model_size, device=device)\n'
                     '        rows, _ = engine.transcribe(samples, language=language, '
                     'initial_prompt=initial_prompt, model=model)\n'
                     '        rows, stats = engine.refine_transcript(rows, samples, '
                     'language=language, initial_prompt=initial_prompt, model=model)\n'
                     '        retry = lambda clip: engine.transcribe(clip, language=language, '
                     'initial_prompt=initial_prompt, model=model,\n'
                     '                                               '
                     'extra_opts=engine.REPAIR_ASR_OPTIONS)[0]\n'
                     '        # Dwarika 571 s: with the domain prompt the retry still merged a '
                     'stretched token; without\n'
                     '        # it the same window came back word-perfect. Second chance only, '
                     'never the first pass.\n'
                     '        fallbacks = [lambda clip: engine.transcribe(clip, language=language, '
                     'initial_prompt="", model=model,\n'
                     '                                                    '
                     'extra_opts=engine.REPAIR_ASR_OPTIONS)[0]]\n'
                     '    run.path("raw/.keep")  # raw ASR layer (Roadmap §6.1): written once, '
                     'never edited afterwards\n'
                     '    raw_srt, raw_js = engine.save_outputs(rows, str(run.dir / "raw"))\n'
                     '    raw_words = rows_to_words(rows)\n'
                     '    write_json(run.path("words_raw.json"), raw_words)  # untouched ASR '
                     'output\n'
                     '    gaps = find_speech_gaps(raw_words, samples)\n'
                     '    filled, gap_report = fill_gaps(raw_words, samples, gaps, retry) if gaps '
                     'else (raw_words, [])\n'
                     '    filled, repair_report = repair_suspect_spans(filled, samples, retry, '
                     'fallback_fns=fallbacks)\n'
                     '    leftover = find_speech_gaps(filled, samples)  # >= COVERAGE_GAP_S (2 s, '
                     'Roadmap v2)\n'
                     '    unfilled = [g for g in leftover if g["end"] - g["start"] >= '
                     'BLOCKING_GAP_S]\n'
                     '    needs_review = [g for g in leftover if g["end"] - g["start"] < '
                     'BLOCKING_GAP_S]\n'
                     '    # a word still stretched past BLOCKING_WORD_S hides dropped speech: same '
                     'verdict as an unfilled gap\n'
                     '    unfilled += [{"start": w["start"], "end": w["end"], "stretched_word": '
                     'w["word"]}\n'
                     '                 for w in filled if w["end"] - w["start"] > '
                     'BLOCKING_WORD_S]\n'
                     '    write_json(run.path("speech_gaps.json"), {"coverage_threshold_s": '
                     'COVERAGE_GAP_S, "blocking_threshold_s": BLOCKING_GAP_S,\n'
                     '                                              "found": gap_report, '
                     '"suspect_spans": repair_report,\n'
                     '                                              "still_unfilled": unfilled, '
                     '"needs_review": needs_review})\n'
                     '    words, sanitized = sanitize_script(filled)\n'
                     '    words, joins = join_split_tokens(words)\n'
                     '    write_json(run.path("sanitized.json"), sanitized)\n'
                     '    write_json(run.path("joined_tokens.json"), joins)\n'
                     '    glossary = load_glossary(ROOT / "config" / "glossary.json")\n'
                     '    votes = _spelling_vote(words, samples, voters, wordlist, vote_agree, '
                     'glossary, device, language, run, audio_path)\n'
                     '    if votes is not None:\n'
                     '        words, vote_changes = votes\n'
                     '        write_json(run.path("vote_changes.json"), vote_changes)\n'
                     '    words, changes = apply_glossary(words, glossary)\n'
                     '    write_json(run.path("glossary_changes.json"), changes)\n'
                     '    words = _apply_spelling_policy(words, spelling_policy, run)\n'
                     '    wpath = write_json(run.path("words.json"), words)\n'
                     '    # final transcript.srt/json come from the SAME final words (after '
                     'gap-fill, repair, voter, glossary)\n'
                     '    srt, js = engine.save_outputs(words_to_rows(words), str(run.dir))\n'
                     '    audit = audit_words(words)\n'
                     '    audit["speech_gaps_unfilled"] = unfilled\n'
                     '    audit["speech_gaps_needs_review"] = needs_review\n'
                     '    audit["sanitized"] = sanitized\n'
                     '    reps = repeated_runs(words)\n'
                     '    audit["issues"] += reps\n'
                     '    audit["issue_count"] += len(reps)\n'
                     '    write_json(run.path("anomaly_audit.json"), audit)\n'
                     '    run.path("review_sample.txt").write_text(review_sample(words, 40), '
                     'encoding="utf-8")\n'
                     '    for p, k in ((raw_srt, "srt_raw"), (raw_js, "transcript_raw"), (srt, '
                     '"srt"), (js, "transcript"), (wpath, "words")):\n'
                     '        run.add_output(Path(p), k)\n'
                     '    run.log_step("transcribe", cut=False, model=model_size, '
                     'words=len(words), refine=stats, audit_issues=audit["issue_count"],\n'
                     '                 glossary_fixes=len(changes), gaps_found=len(gap_report),\n'
                     '                 gap_words_added=sum(g["filled"] for g in gap_report), '
                     'gaps_unfilled=len(unfilled),\n'
                     '                 spans_repaired=sum(r["accepted"] for r in repair_report), '
                     'spans_checked=len(repair_report),\n'
                     '                 spelling_votes=None if votes is None else len(votes[1]))\n'
                     '    run.set_status("transcribed")\n'
                     '    return {"run_id": run.run_id, "dir": str(run.dir), "words": len(words), '
                     '"audit": audit,\n'
                     '            "glossary_fixes": changes, "vote_changes": None if votes is None '
                     'else votes[1], "speech_gaps": gap_report, "suspect_spans": repair_report, '
                     '"gaps_unfilled": unfilled,\n'
                     '            "gaps_needs_review": needs_review}\n'
                     '\n'
                     '\n'
                     'def words_to_rows(words: List[Dict[str, Any]], gap: float = 1.0) -> '
                     'List[Dict[str, Any]]:\n'
                     '    """Final words.json -> engine row shape '
                     "({start,end,text,words:[{w,start,end,p}]}) so the API's\n"
                     '    transcript.json/.srt keep their old contract. New row at a pause >= gap '
                     's or after । ? !"""\n'
                     '    rows: List[Dict[str, Any]] = []\n'
                     '    cur: List[Dict[str, Any]] = []\n'
                     '\n'
                     '    def flush():\n'
                     '        if cur:\n'
                     '            rows.append({"start": cur[0]["start"], "end": cur[-1]["end"], '
                     '"text": " ".join(w["w"] for w in cur),\n'
                     '                         "words": list(cur)})\n'
                     '            cur.clear()\n'
                     '    for w in words:\n'
                     '        if cur and w["start"] - cur[-1]["end"] >= gap:\n'
                     '            flush()\n'
                     '        cur.append({"w": w["word"], "start": w["start"], "end": w["end"], '
                     '"p": w.get("probability", 1.0)})\n'
                     '        if w["word"].rstrip().endswith(("।", "?", "!")):\n'
                     '            flush()\n'
                     '    flush()\n'
                     '    return rows\n'
                     '\n'
                     '\n'
                     'def api_transcribe(src: str, out_dir: str, cut: bool = False, language: str '
                     '= "hi",\n'
                     '                   transcriber: Optional[Callable] = None) -> Dict[str, '
                     'Any]:\n'
                     '    """app.py /upload job: the SAME transcription stack as kang.py (refine, '
                     'gap-fill, suspect-span\n'
                     '    repair, sanitize, joins, spelling voter, glossary) in its own run '
                     'folder; then\n'
                     '    transcript.srt/json are written to out_dir so the public API contract is '
                     'unchanged."""\n'
                     '    from . import engine\n'
                     '    if not os.path.isfile(src):\n'
                     '        raise FileNotFoundError(f"Input audio file not found: {src}")\n'
                     '    out = Path(out_dir)\n'
                     '    out.mkdir(parents=True, exist_ok=True)\n'
                     '    audio_path = src\n'
                     '    if cut:  # same pre-trim the old engine.run_pipeline did for cut=True\n'
                     '        audio_path = str(out / "cut_input.mp3")\n'
                     '        engine.cut_silence(src, audio_path)\n'
                     '    shutil.copy2(audio_path, out / "audio_used.mp3")\n'
                     '    res = transcribe_to_words(audio_path, language=language, '
                     'transcriber=transcriber)\n'
                     '    words = load_words(Path(res["dir"]) / "words.json")\n'
                     '    shutil.copy2(Path(res["dir"]) / "transcript.srt", out / '
                     '"transcript.srt")  # same final revision as words.json\n'
                     '    shutil.copy2(Path(res["dir"]) / "transcript.json", out / '
                     '"transcript.json")\n'
                     '    srt_path, json_path = str(out / "transcript.srt"), str(out / '
                     '"transcript.json")\n'
                     '    if res["gaps_unfilled"]:\n'
                     '        raise RuntimeError(f"{len(res[\'gaps_unfilled\'])} speech gap(s) '
                     'still unfilled; see run {res[\'run_id\']}")\n'
                     '    review = res.get("gaps_needs_review", [])\n'
                     '    return {"run_id": res["run_id"], "run_dir": res["dir"], "srt_path": '
                     'srt_path, "json_path": json_path,\n'
                     '            # processing finished != transcript accepted: acceptance is a '
                     'human step (Roadmap S4)\n'
                     '            "processing_complete": True, "review_required": bool(review), '
                     '"accepted": False,\n'
                     '            "gaps_needs_review": review, "audit_issues": '
                     'res["audit"]["issue_count"],\n'
                     '            "warnings": [f"{len(review)} voiced gap(s) of 2-3 s need '
                     'review"] if review else [],\n'
                     '            "words": res["words"], "glossary_fixes": '
                     'len(res["glossary_fixes"]),\n'
                     '            "spelling_votes": None if res["vote_changes"] is None else '
                     'len(res["vote_changes"])}\n'
                     '\n'
                     '\n'
                     '_ALT_CACHE: Dict[tuple, List[Dict[str, Any]]] = {}\n'
                     '\n'
                     '\n'
                     'VOTE_BATCH = int(os.environ.get("ENGINE_VOTE_BATCH", "8"))  # 0 = sequential '
                     '(old, ~4x slower on T4)\n'
                     '\n'
                     '\n'
                     '# batched passes of the Hindi fine-tunes return ~57-59% of the base word '
                     'count on Dwarika (normal:\n'
                     '# they merge/skip fillers); only a near-empty pass means the batched mode '
                     'broke for that model.\n'
                     'VOTE_MIN_FRAC = 0.3\n'
                     '\n'
                     '\n'
                     'def _cuda_count() -> int:\n'
                     '    try:\n'
                     '        import ctranslate2\n'
                     '        return max(1, ctranslate2.get_cuda_device_count())\n'
                     '    except Exception:\n'
                     '        return 1\n'
                     '\n'
                     '\n'
                     'def _alt_pass(src: str, samples, language: str, dev: str, device_index: int '
                     '= 0, min_words: int = 0):\n'
                     '    """Full-file pass of one Hindi second model (spelling only: its words '
                     'never add/drop/move ours).\n'
                     '    Batched inference (faster-whisper BatchedInferencePipeline, VAD chunks) '
                     'is ~3-4x faster on GPU.\n'
                     '    Hindi fine-tunes were trained without a prompt."""\n'
                     '    from faster_whisper import WhisperModel\n'
                     '    from . import engine\n'
                     '    m = WhisperModel(src, device=dev, device_index=device_index,\n'
                     '                     compute_type="float16" if dev == "cuda" else "int8")\n'
                     '    try:\n'
                     '        if VOTE_BATCH > 0:\n'
                     '            from faster_whisper import BatchedInferencePipeline\n'
                     '            segs, _ = BatchedInferencePipeline(model=m).transcribe(\n'
                     '                samples, language=language, batch_size=VOTE_BATCH, '
                     'word_timestamps=True, beam_size=5)\n'
                     '            got = [{"word": w.word.strip(), "start": round(float(w.start), '
                     '3), "end": round(float(w.end), 3),\n'
                     '                    "probability": round(float(w.probability), 3)}\n'
                     '                   for sg in segs for w in (sg.words or []) if '
                     'w.word.strip() and w.end > w.start]\n'
                     '            # Dwarika v4: IndicWhisper medium gave no usable batched words '
                     '-> 0 votes. Fall back.\n'
                     '            if len(got) >= min_words:\n'
                     '                return got\n'
                     '            print(f"voter {Path(src).name}: batched gave {len(got)} words (< '
                     '{min_words}); sequential pass")\n'
                     '        return rows_to_words(engine.transcribe(samples, language=language, '
                     'initial_prompt="", model=m)[0])\n'
                     '    finally:\n'
                     '        del m\n'
                     '\n'
                     '\n'
                     'def _default_voters(wordlist=None):\n'
                     '    import os\n'
                     '    cfg_p = ROOT / "config" / "voters.json"\n'
                     '    if not cfg_p.exists():\n'
                     '        return {}, wordlist\n'
                     '    cfg = json.loads(cfg_p.read_text(encoding="utf-8"))\n'
                     '    d = Path(os.environ.get("KANG_VOTER_DIR") or (ROOT / '
                     'cfg.get("default_dir", "models/voters")))\n'
                     '    found = {n: str(d / sub) for n, sub in cfg.get("voters", {}).items() if '
                     '(d / sub / "model.bin").exists()}\n'
                     '    wl = wordlist or os.environ.get("ENGINE_WORDLIST") or str(d / '
                     'cfg.get("wordlist", "hi_IN.dic"))\n'
                     '    return found, wl\n'
                     '\n'
                     '\n'
                     'def _spelling_vote(words, samples, voters, wordlist, agree, glossary, '
                     'device, language, run, audio_path=""):\n'
                     '    """None when the voter is off; else (words, changes). Missing '
                     'wordlist/model = warning, never a crash."""\n'
                     '    import os\n'
                     '    from .voter import load_wordlist, spelling_vote\n'
                     '    if os.environ.get("ENGINE_VOTE", "1") == "0":\n'
                     '        return None\n'
                     '    if voters is None and os.environ.get("ENGINE_VOTERS"):\n'
                     '        voters = dict(x.split("=", 1) for x in '
                     'os.environ["ENGINE_VOTERS"].split(",") if "=" in x)\n'
                     '    if voters is None:  # default ON: whatever config/voters.json models are '
                     'set up\n'
                     '        voters, wordlist = _default_voters(wordlist)\n'
                     '        if not voters:\n'
                     '            run.warn("spelling voter not set up (python '
                     'tools/setup_voters.py); continuing without it")\n'
                     '            return None\n'
                     '    if not voters:\n'
                     '        return None\n'
                     '    wl = wordlist or os.environ.get("ENGINE_WORDLIST")\n'
                     '    vocab = load_wordlist(wl, extra=glossary.values()) if wl and '
                     'Path(wl).exists() else set()\n'
                     '    if not vocab:\n'
                     '        run.warn(f"spelling voter skipped: wordlist not found ({wl})")\n'
                     '        return None\n'
                     '    alts = {}\n'
                     '    jobs = {}\n'
                     '    for name, src in voters.items():\n'
                     '        try:\n'
                     '            if callable(src):\n'
                     '                alts[name] = rows_to_words(src(samples))\n'
                     '                continue\n'
                     '            key = (str(src), str(Path(audio_path).resolve()) if audio_path '
                     'else id(samples), len(samples), language)\n'
                     '            if key in _ALT_CACHE:\n'
                     '                alts[name] = _ALT_CACHE[key]\n'
                     '            else:\n'
                     '                jobs[name] = (key, str(src))\n'
                     '        except Exception as e:  # one broken second model must not stop the '
                     'transcript\n'
                     '            run.warn(f"spelling voter: {name} failed: {type(e).__name__}: '
                     '{e}")\n'
                     '    if jobs:\n'
                     '        import time as _t\n'
                     '        dev = device or os.environ.get("ENGINE_DEVICE", "cuda")\n'
                     '        n_dev = _cuda_count() if dev == "cuda" else 1\n'
                     '        t0 = _t.time()\n'
                     '        from concurrent.futures import ThreadPoolExecutor\n'
                     '        names = list(jobs)\n'
                     '        import queue\n'
                     '        free = queue.Queue()\n'
                     '        for i in range(max(1, n_dev)):\n'
                     '            free.put(i)\n'
                     '\n'
                     '        def _on_free_gpu(src):  # exactly one model per GPU at any time (v5: '
                     '2 on GPU 0 -> CUDA invalid argument)\n'
                     '            gi = free.get()\n'
                     '            try:\n'
                     '                return _alt_pass(src, samples, language, dev, gi, '
                     'int(len(words) * VOTE_MIN_FRAC))\n'
                     '            finally:\n'
                     '                free.put(gi)\n'
                     '        with ThreadPoolExecutor(max_workers=max(1, min(n_dev, len(names)))) '
                     'as ex:\n'
                     '            futs = {n: ex.submit(_on_free_gpu, jobs[n][1]) for n in names}\n'
                     '            for n, f in futs.items():\n'
                     '                try:\n'
                     '                    _ALT_CACHE[jobs[n][0]] = alts[n] = f.result()\n'
                     '                except Exception as e:\n'
                     '                    run.warn(f"spelling voter: {n} failed: '
                     '{type(e).__name__}: {e}")\n'
                     '        run.log_step("voter_alt_passes", models=names, gpus=n_dev, '
                     'seconds=round(_t.time() - t0, 1),\n'
                     '                     batch=VOTE_BATCH)\n'
                     '    if not alts:\n'
                     '        return None\n'
                     '    voted, ch = spelling_vote(words, alts, vocab, require_agree=agree and '
                     'len(alts) >= 2)\n'
                     '    return voted, ch\n'
                     '\n'
                     '\n'
                     'def _apply_spelling_policy(words, policy_path: Optional[str], run: Run):\n'
                     '    """Opt-in per-video caption spelling (core/spelling_policy.py). Logged, '
                     'never default."""\n'
                     '    if not policy_path:\n'
                     '        return words\n'
                     '    from .spelling_policy import apply_policy, load_policy\n'
                     '    pol = load_policy(policy_path)\n'
                     '    run.add_input(policy_path, "SPELLING_POLICY")\n'
                     '    words_pre = write_json(run.path("words_pre_policy.json"), words)\n'
                     '    words, ch = apply_policy(words, pol)\n'
                     '    write_json(run.path("spelling_policy_changes.json"), {"policy": '
                     'str(policy_path), "video": pol.get("video"),\n'
                     '                                                          "decided_by": '
                     '"editorial-policy", "changes": ch})\n'
                     '    run.log_step("spelling_policy", policy=str(policy_path), '
                     'changes=len(ch), pre_policy=str(words_pre.name))\n'
                     '    return words\n'
                     '\n'
                     '\n'
                     'def fix_words(words_path: str, run: Optional[Run] = None, glossary_path: '
                     'Optional[str] = None,\n'
                     '              spelling_policy: Optional[str] = None) -> Dict[str, Any]:\n'
                     '    """Apply config/glossary.json (+ optional per-video spelling policy) to '
                     'an existing words.json\n'
                     '    in a NEW run (original untouched)."""\n'
                     '    run = run or Run()\n'
                     '    run.add_input(words_path, "W1")\n'
                     '    words, changes = apply_glossary(load_words(words_path),\n'
                     '                                    load_glossary(glossary_path or ROOT / '
                     '"config" / "glossary.json"))\n'
                     '    words = _apply_spelling_policy(words, spelling_policy, run)\n'
                     '    wpath = write_json(run.path("words.json"), words)\n'
                     '    write_json(run.path("glossary_changes.json"), changes)\n'
                     '    run.path("review_sample.txt").write_text(review_sample(words, 40), '
                     'encoding="utf-8")\n'
                     '    run.add_output(wpath, "words")\n'
                     '    run.log_step("fix_words", fixes=len(changes))\n'
                     '    run.set_status("words_fixed")\n'
                     '    return {"run_id": run.run_id, "dir": str(run.dir), "words_json": '
                     'str(wpath), "fixes": changes}\n'
                     '\n'
                     '\n'
                     '# '
                     '--------------------------------------------------------------------------- '
                     'F01 / Phase 2\n'
                     'def make_subtitles(run: Run, words: List[Dict[str, Any]], fmt: str = "9:16", '
                     'preset_name: str = "vox_bold_yellow",\n'
                     '                   brand_path: Optional[str] = None, option: Optional[str] = '
                     'None) -> Dict[str, Any]:\n'
                     '    brand = load_brand(brand_path or ROOT / "brand.json")\n'
                     '    fonts, warnings = resolve_fonts(brand, ROOT / "fonts")\n'
                     '    for w in warnings:\n'
                     '        run.warn(w)\n'
                     '        print("WARNING:", w)\n'
                     '    preset = load_preset(preset_name, ROOT / "presets")\n'
                     '    font = fonts[preset.font_role]\n'
                     '    layout = make_layout(brand, preset, fmt, font)\n'
                     '    ass, events = build_ass(words, brand, preset, layout, font, '
                     'option=option)\n'
                     '    ass_p = run.path(f"subs_{fmt.replace(\':\', \'x\')}.ass")\n'
                     '    ass_p.write_text(ass, encoding="utf-8")\n'
                     '    srt_p = run.path("subs.srt")\n'
                     '    srt_p.write_text(build_srt(events), encoding="utf-8")\n'
                     '    run.add_output(ass_p, "ass")\n'
                     '    run.add_output(srt_p, "srt")\n'
                     '    checks = [qa.check_no_overlap(ass), qa.check_glyphs(font, [w["word"] for '
                     'w in words])]\n'
                     '    if (option or ("B" if preset.highlight == "pop" else "A")) == "A":\n'
                     '        checks.append(qa.check_sync(ass, words))\n'
                     '    if events:\n'
                     '        t = (events[0]["start"] + events[0]["end"]) / 2\n'
                     '        png = run.path(f"qa/subtitle_frame_{fmt.replace(\':\', '
                     '\'x\')}.png")\n'
                     '        qa.render_ass_on_black(ass_p, ROOT / "fonts", CANVAS[fmt], t, png)\n'
                     '        checks.append(qa.check_safe_zone(png, CANVAS[fmt], layout.margin_lr, '
                     'layout.margin_v))\n'
                     '    # Glyph/shaping test on fixed conjunct/matra sentences, same style as '
                     'the real subtitles\n'
                     '    test_words = [{"word": w, "start": 0.2 + i * 0.4, "end": 0.55 + i * '
                     '0.4}\n'
                     '                  for i, w in enumerate(" '
                     '".join(qa.GLYPH_TEST_SENTENCES[:1]).split())]\n'
                     '    gass, _ = build_ass(test_words, brand, preset, layout, font, '
                     'option="A")\n'
                     '    gpath = run.path(f"qa/glyph_test_{fmt.replace(\':\', \'x\')}.ass")\n'
                     '    gpath.write_text(gass, encoding="utf-8")\n'
                     '    checks.append(qa.check_glyphs(font, qa.GLYPH_TEST_SENTENCES))\n'
                     '    checks.append(qa.check_shaping(gpath, ROOT / "fonts", CANVAS[fmt], 0.9, '
                     'run.path("qa")))\n'
                     '    return {"ass": ass_p, "srt": srt_p, "events": events, "layout": layout, '
                     '"brand": brand,\n'
                     '            "preset": preset, "font": font, "checks": checks}\n'
                     '\n'
                     '\n'
                     'def subtitle_video(video: str, words_path: str, run: Optional[Run] = None, '
                     'fmt: str = "9:16",\n'
                     '                   preset_name: str = "vox_bold_yellow", preview_start: '
                     'float = 0.0,\n'
                     '                   approver: Approver = _deny, option: Optional[str] = None) '
                     '-> Dict[str, Any]:\n'
                     '    """F01: subs.ass + subs.srt + preview_10s.mp4 -> (approve) -> final.mp4 '
                     '+ render_settings.json."""\n'
                     '    run = run or Run()\n'
                     '    run.add_input(video, "A1")\n'
                     '    words = load_words(words_path)\n'
                     '    shutil.copy2(words_path, run.path("words.json"))\n'
                     '    sub = make_subtitles(run, words, fmt, preset_name, option=option)\n'
                     '    log = run.path("ffmpeg.log")\n'
                     '    src = media_summary(video)\n'
                     '    work_video = video\n'
                     '    if not src["has_video"]:\n'
                     '        work_video = str(render.audio_to_video(video, run.path("bg.mp4"), '
                     'fmt, sub["brand"].colors.primary, log=log))\n'
                     '    elif (src["width"], src["height"]) != CANVAS[fmt]:\n'
                     '        work_video = str(render.render_format(video, fmt, '
                     'run.path("reframed.mp4"), src["width"], src["height"], log=log))\n'
                     '    prev = render.make_preview(work_video, sub["ass"], ROOT / "fonts", '
                     'run.path("preview_10s.mp4"), start=preview_start, log=log)\n'
                     '    run.add_output(prev, "preview")\n'
                     '    settings = {"fmt": fmt, "preset": sub["preset"].model_dump(), "layout": '
                     'sub["layout"].model_dump(),\n'
                     '                "font": Path(sub["font"]).name, "filter": "ass (libass, '
                     'shaping=complex)", "crf": 20}\n'
                     '    write_json(run.path("render_settings.json"), settings)\n'
                     '    report = qa.qa_report(sub["checks"])\n'
                     '    write_json(run.path("qa_report.json"), report)\n'
                     '    result: Dict[str, Any] = {"run_id": run.run_id, "dir": str(run.dir), '
                     '"preview": str(prev), "qa": report, "final": None}\n'
                     '    if not report["pass"]:\n'
                     '        run.set_status("qa_failed")\n'
                     '        return result\n'
                     '    gate_art = [prev, sub["ass"], run.path("words.json")]\n'
                     '    gate_info = f"Preview: {prev}\\nQA: PASS"\n'
                     '    if _gate(run, "preview", gate_info, approver, artifacts=gate_art, '
                     'settings=settings):\n'
                     '        if not run.approval_valid("preview", plan_hash(gate_info, gate_art, '
                     'settings)):  # changed since approval\n'
                     '            run.set_status("approval_invalidated")\n'
                     '            result["error"] = "preview/words/style changed after approval; '
                     're-approve"\n'
                     '            return result\n'
                     '        final = render.burn_subtitles(work_video, sub["ass"], ROOT / '
                     '"fonts", run.path("final.mp4"), log=log)\n'
                     '        dur = media_summary(work_video)["duration"]\n'
                     '        chk = qa.check_export(final, dur, CANVAS[fmt])\n'
                     '        report = qa.qa_report(sub["checks"] + [chk])\n'
                     '        write_json(run.path("qa_report.json"), report)\n'
                     '        run.add_output(final, "final")\n'
                     '        result.update(final=str(final), qa=report)\n'
                     '        run.set_status("done" if report["pass"] else "qa_failed")\n'
                     '    else:\n'
                     '        run.set_status("awaiting_preview_approval")\n'
                     '    return result\n'
                     '\n'
                     '\n'
                     'DELIVERY_LUFS = -14.0  # measured on the delivered MP4, not the intermediate '
                     'WAV\n'
                     'DELIVERY_TP = -1.5\n'
                     'AAC_SAFE_TP = -3.0\n'
                     "BLOCKING_WORD_S = 3.0  # one 'word' this long = whole sentences dropped "
                     '(Dwarika 571 s: 11.56 s)\n'
                     'FULL_MAX_S = 3600.0  # --full mode: whole file, up to 1 hour\n'
                     '\n'
                     '\n'
                     'def promo_clip(audio_path: str, words_path: str, start: float, duration: '
                     'float, run: Optional[Run] = None,\n'
                     '               fmt: str = "9:16", preset_name: str = "vox_bold_yellow", '
                     'approver: Approver = _deny,\n'
                     '               max_duration: float = 60.0) -> Dict[str, Any]:\n'
                     '    """Phase 2 exit proof: voiceover slice -> 15-30s branded promo MP4 with '
                     'word-highlight subtitles.\n'
                     '    max_duration is raised only by the workflow\'s --full mode."""\n'
                     '    if not 1 <= duration <= max_duration:\n'
                     '        raise ValueError(f"promo duration must be 1-{max_duration:g} s '
                     '(roadmap: 15-30 s)")\n'
                     '    run = run or Run()\n'
                     '    run.add_input(audio_path, "A1")\n'
                     '    words = [dict(w) for w in load_words(words_path) if w["start"] >= start '
                     'and w["end"] <= start + duration]\n'
                     '    for w in words:\n'
                     '        w["start"] = round(w["start"] - start, 3)\n'
                     '        w["end"] = round(w["end"] - start, 3)\n'
                     '    if not words:\n'
                     '        raise ValueError("No words inside the selected promo window")\n'
                     '    sliced_words = write_json(run.path("promo_words.json"), words)\n'
                     '    brand = load_brand(ROOT / "brand.json")\n'
                     '    log = run.path("ffmpeg.log")\n'
                     '    bg = render.audio_to_video(audio_path, run.path("promo_bg.mp4"), fmt, '
                     'brand.colors.primary, start=start,\n'
                     '                               duration=duration, log=log)\n'
                     '    return subtitle_video(str(bg), str(sliced_words), run=run, fmt=fmt, '
                     'preset_name=preset_name, approver=approver)\n'
                     '\n'
                     '\n'
                     '# '
                     '--------------------------------------------------------------------------- '
                     'M3 Workflow G (F12)\n'
                     'def _to_timeline(t: float, keep: List[List[float]]) -> float:\n'
                     '    """Source time -> time on the cut timeline (a time inside a removed '
                     'range snaps to the next kept start)."""\n'
                     '    acc = 0.0\n'
                     '    for s, e in keep:\n'
                     '        if t < s:\n'
                     '            return acc\n'
                     '        if t <= e:\n'
                     '            return acc + (t - s)\n'
                     '        acc += e - s\n'
                     '    return acc\n'
                     '\n'
                     '\n'
                     'def premiere_package(media: str, words_path: str, run: Optional[Run] = None, '
                     'approver: Approver = _deny,\n'
                     '                     scenes_path: Optional[str] = None, cuts_path: '
                     'Optional[str] = None, fps: int = 25,\n'
                     '                     win_root: Optional[str] = None, preset_name: str = '
                     '"vox_bold_yellow",\n'
                     '                     target_scene_s: float = 30.0, english_script: str = '
                     '"asr") -> Dict[str, Any]:\n'
                     '    """Accepted transcript -> scenes.json -> Gate 1 (scene_plan) -> Premiere '
                     'package:\n'
                     '    sequence.xml (FCP7 xmeml, original media, editable cuts, scene markers) '
                     '+ captions.srt + scenes.json\n'
                     '    + words.json + media/ + README_IMPORT.txt + package_manifest.json '
                     '(sha256 of every file).\n'
                     '    Narration is kept whole unless an APPROVED cut plan (cuts.json from a '
                     'run whose cut_list gate is\n'
                     '    approved) is passed with cuts_path."""\n'
                     '    from . import premiere, scenes as scenes_mod\n'
                     '    from .runs import sha256_file\n'
                     '    run = run or Run()\n'
                     '    inp = run.add_input(media, "A1")\n'
                     '    ms = media_summary(media)\n'
                     '    words = load_words(words_path)\n'
                     '    shutil.copy2(words_path, run.path("words.json"))\n'
                     '    result: Dict[str, Any] = {"run_id": run.run_id, "dir": str(run.dir), '
                     '"package": None}\n'
                     '    # transcript accepted only via apply_review (writes '
                     'transcript_acceptance.json next to the words)\n'
                     '    acc_p = Path(words_path).resolve().parent / '
                     '"transcript_acceptance.json"\n'
                     '    transcript_accepted = False\n'
                     '    if acc_p.is_file():\n'
                     '        from .runs import sha256_file as _sha\n'
                     '        acc = json.loads(acc_p.read_text(encoding="utf-8"))\n'
                     '        transcript_accepted = bool(acc.get("accepted")) and '
                     'acc.get("words_sha256") == _sha(Path(words_path))\n'
                     '    result["transcript_accepted"] = transcript_accepted\n'
                     '\n'
                     '    keep = [[0.0, ms["duration"]]]\n'
                     '    if cuts_path:\n'
                     '        cdir = Path(cuts_path).resolve().parent\n'
                     '        rec = json.loads((cdir / '
                     '"approvals.json").read_text(encoding="utf-8")) if (cdir / '
                     '"approvals.json").is_file() else []\n'
                     '        last = [r for r in rec if r["gate"] == "cut_list"]\n'
                     '        if not last or last[-1]["value"] != "approved":\n'
                     '            run.set_status("blocked_unapproved_cuts")\n'
                     '            result["error"] = "cuts.json has no approved cut_list gate; '
                     'narration kept whole is the default"\n'
                     '            return result\n'
                     '        # the approval must be for THIS cuts.json content (plan hash) and '
                     'THIS source media\n'
                     '        cplan = json.loads(Path(cuts_path).read_text(encoding="utf-8"))\n'
                     '        h_now = plan_hash(cutter.cut_list_text(cplan), [cuts_path])\n'
                     '        if last[-1].get("plan_hash") != h_now:\n'
                     '            run.set_status("blocked_unapproved_cuts")\n'
                     '            result["error"] = "cuts.json changed after its cut_list approval '
                     '(plan hash mismatch); re-run the cut gate"\n'
                     '            return result\n'
                     '        src_man = cdir / "manifest.json"\n'
                     '        src_inputs = '
                     'json.loads(src_man.read_text(encoding="utf-8")).get("inputs", []) if '
                     'src_man.is_file() else []\n'
                     '        if not any(i.get("sha256") == inp["sha256"] for i in src_inputs):\n'
                     '            run.set_status("blocked_unapproved_cuts")\n'
                     '            result["error"] = "cuts.json was approved for different source '
                     'media (sha256 mismatch)"\n'
                     '            return result\n'
                     '        keep = cplan["keep"]\n'
                     '        result["cuts_approved_by"] = last[-1]["approver"]\n'
                     '        result["cuts_human_approved"] = last[-1]["approver"] not in '
                     'AUTO_APPROVERS\n'
                     '\n'
                     "    # Step 1: scene plan (generated, or the user's edited scenes.json)\n"
                     '    if scenes_path:\n'
                     '        plan = json.loads(Path(scenes_path).read_text(encoding="utf-8"))\n'
                     '    else:\n'
                     '        plan = scenes_mod.plan_scenes(words, ms["duration"], '
                     'target_s=target_scene_s)\n'
                     '    errs = scenes_mod.validate_scenes(plan, n_words=len(words))\n'
                     '    sp = write_json(run.path("scenes.json"), plan)\n'
                     '    run.add_output(sp, "scenes")\n'
                     '    result["scenes"] = len(plan.get("scenes", []))\n'
                     '    if errs:\n'
                     '        run.set_status("qa_failed")\n'
                     '        result["error"] = "scene plan invalid"\n'
                     '        result["problems"] = errs\n'
                     '        return result\n'
                     '\n'
                     '    # Gate 1: scene plan, bound to scenes.json + words.json content\n'
                     '    info = scenes_mod.scene_plan_text(plan)\n'
                     '    art = [sp, run.path("words.json")]\n'
                     '    if not _gate(run, "scene_plan", info, approver, artifacts=art, '
                     'settings={"keep": keep, "fps": fps}):\n'
                     '        run.set_status("awaiting_scene_approval")\n'
                     '        result["scene_plan"] = info\n'
                     '        return result\n'
                     '    result["human_approved"] = run.manifest.get("human_approved", '
                     '{}).get("scene_plan", False)\n'
                     '\n'
                     '    # Step 2: package\n'
                     '    pkg = run.path("premiere_package")\n'
                     '    (pkg / "media").mkdir(parents=True, exist_ok=True)\n'
                     '    mname = Path(media).name\n'
                     '    shutil.copy2(media, pkg / "media" / mname)\n'
                     '    xml_media_path = (str(PureWindowsPath(win_root) / "media" / mname) if '
                     'win_root\n'
                     '                      else str((pkg / "media" / mname).resolve()))\n'
                     '    markers = [{"at": _to_timeline(s["start"], keep), "name": f"{s[\'id\']} '
                     '{s[\'title\']}"[:80],\n'
                     '                "comment": s["text"][:500]} for s in plan["scenes"]]\n'
                     '    w, h = (ms.get("width") or 1920, ms.get("height") or 1080)\n'
                     '    xml, stats = premiere.build_xmeml(\n'
                     '        f"KANG {run.run_id}", {"path": xml_media_path, "name": mname, '
                     '"duration": ms["duration"],\n'
                     '                              "has_video": ms["has_video"], "channels": '
                     'ms.get("channels", 2),\n'
                     '                              "sample_rate": ms.get("sample_rate", 48000)},\n'
                     '        keep, fps=fps, width=w, height=h, markers=markers)\n'
                     '    (pkg / "sequence.xml").write_text(xml, encoding="utf-8")\n'
                     '    words_tl = cutter.remap_words(words, [tuple(k) for k in keep]) if '
                     'cuts_path else words\n'
                     '    from . import script_style\n'
                     '    cap_words, script_rep = script_style.apply(\n'
                     '        words_tl, english_script, script_style.load_table(ROOT / "config" / '
                     '"english_script.json"))\n'
                     '    write_json(run.path("english_script_changes.json"), script_rep)\n'
                     '    for u in script_rep["unmapped"]:\n'
                     '        run.warn(f"english_script=devanagari: no mapping for \'{u}\' (left '
                     'as transcribed)")\n'
                     '    sub = make_subtitles(run, cap_words, "16:9", preset_name)\n'
                     '    shutil.copy2(sub["srt"], pkg / "captions.srt")\n'
                     '    shutil.copy2(sp, pkg / "scenes.json")\n'
                     '    write_json(pkg / "words.json", words_tl)\n'
                     '    (pkg / "README_IMPORT.txt").write_text(\n'
                     '        "KANG Premiere package (" + run.run_id + ")\\n"\n'
                     '        "Target: Premiere Pro 24.6.1\\n"\n'
                     '        "Actual import verification: pending (NOT_TESTED)\\n"\n'
                     '        "Contents: narration audio + scene markers. Video track is EMPTY (no '
                     'visuals yet).\\n"\n'
                     '        "Captions: separate captions.srt (NOT embedded in the XML).\\n"\n'
                     '        "Transcript: " + ("ACCEPTED after human review" if '
                     'transcript_accepted else\n'
                     '                          "NOT accepted yet - names/amounts need a listen '
                     '(python kang.py review)") + "\\n"\n'
                     '        "Paths: XML points to " + (str(PureWindowsPath(win_root) / "media" / '
                     'mname) if win_root else "media/" + mname)\n'
                     '        + ". Put the folder exactly there or use Link Media.\\n\\n"\n'
                     '        "1. Copy this whole folder to " + (win_root or "your PC") + " (keep '
                     'media/ next to sequence.xml).\\n"\n'
                     '        "2. Premiere: File > Open Project (Ctrl+O), file type \'All Files\', '
                     'pick sequence.xml. It opens as its own project (File > Import of the XML '
                     'failed on the test PC).\\n"\n'
                     '        "3. Media offline? Double-click FIX_PATHS.bat first (points the XML '
                     'at this folder), or Link Media > media/" + mname + ".\\n"\n'
                     '        "4. File > Import > captions.srt, then drag it onto the sequence '
                     '(caption track).\\n"\n'
                     '        "5. Scene markers S01... sit on the sequence timeline; scenes.json '
                     'has each scene\'s text.\\n"\n'
                     '        f"Sequence: {stats[\'sequence_s\']} s at {fps} fps, '
                     '{stats[\'clips\']} clip(s), {stats[\'markers\']} markers.\\n",\n'
                     '        encoding="utf-8")\n'
                     '    premiere.write_fix_paths(pkg)\n'
                     '    checks = [premiere.check_xmeml(xml, fps)]\n'
                     '    exp = round(sum(e - s for s, e in keep), 3)\n'
                     '    checks.append({"name": "sequence_duration", "pass": '
                     'abs(stats["sequence_s"] - exp) <= 1.5 / fps * max(1, len(keep)),\n'
                     '                   "detail": {"expected_s": exp, "sequence_s": '
                     'stats["sequence_s"]}})\n'
                     '    files = sorted(p for p in pkg.rglob("*") if p.is_file())\n'
                     '    write_json(pkg / "package_manifest.json", {"run_id": run.run_id, '
                     '"source_sha256": inp["sha256"], "stats": stats,\n'
                     '               "status": {"transcript_accepted": transcript_accepted, '
                     '"words_source": str(words_path),\n'
                     '                          "scene_plan_human_approved": '
                     'result["human_approved"], "visuals": "NOT_STARTED",\n'
                     '                          "captions": "separate_srt", '
                     '"caption_english_script": english_script,\n'
                     '                          "premiere_import": "NOT_TESTED"},\n'
                     '               "files": {str(p.relative_to(pkg)).replace("\\\\", "/"): '
                     'sha256_file(p) for p in files}})\n'
                     '    report = qa.qa_report(checks)\n'
                     '    write_json(run.path("qa_report.json"), report)\n'
                     '    run.add_output(pkg / "sequence.xml", "premiere_xml")\n'
                     '    run.log_step("premiere_package", **stats)\n'
                     '    run.set_status("package_ready" if report["pass"] else "qa_failed")\n'
                     '    result.update(package=str(pkg), stats=stats, qa=report)\n'
                     '    # generation != acceptance: draft_package_ready / '
                     'package_generated_pending_verification / qa_failed\n'
                     '    from .package_readiness import finalize_package_readiness\n'
                     '    finalize_package_readiness(run, result, pkg, words_tl, report)\n'
                     '    return result\n'
                     '\n'
                     '\n'
                     '# '
                     '--------------------------------------------------------------------------- '
                     'M3 visuals\n'
                     'def visuals_template(scenes_path: str, out_csv: Optional[str] = None, run: '
                     'Optional[Run] = None) -> Dict[str, Any]:\n'
                     '    """scenes.json -> visuals.csv (one row per scene) for the user to fill: '
                     'brief, asset, licence."""\n'
                     '    from . import scenes as scenes_mod, visuals as vis\n'
                     '    run = run or Run()\n'
                     '    run.add_input(scenes_path, "SCENES")\n'
                     '    plan = json.loads(Path(scenes_path).read_text(encoding="utf-8"))\n'
                     '    errs = scenes_mod.validate_scenes(plan)\n'
                     '    if errs:\n'
                     '        run.set_status("qa_failed")\n'
                     '        return {"run_id": run.run_id, "ok": False, "problems": errs}\n'
                     '    out = vis.write_template(plan, out_csv or run.path("visuals.csv"))\n'
                     '    run.add_output(out, "visuals_template")\n'
                     '    run.set_status("visuals_template_ready")\n'
                     '    return {"run_id": run.run_id, "dir": str(run.dir), "ok": True, '
                     '"visuals_csv": str(out), "scenes": len(plan["scenes"])}\n'
                     '\n'
                     '\n'
                     'def visual_preview(media: str, words_path: str, scenes_path: str, '
                     'visuals_csv: str, assets_dir: str,\n'
                     '                   scene_ids: Optional[List[str]] = None, run: Optional[Run] '
                     '= None, approver: Approver = _deny,\n'
                     '                   fps: int = 25, preset_name: str = "vox_bold_yellow", '
                     'win_root: Optional[str] = None) -> Dict[str, Any]:\n'
                     '    """Approved local visuals -> (a) integration preview mp4 of the chosen '
                     'contiguous scenes (visuals +\n'
                     '    narration + burned captions, 16:9) and (b) Premiere package whose '
                     'sequence carries every approved\n'
                     '    visual on its own video track at scene times. Narration kept whole (no '
                     'cuts in this step).\n'
                     '    Nothing is accepted here: transcript/timing/import status travel in '
                     'package_manifest.json."""\n'
                     '    from . import premiere, scenes as scenes_mod, visuals as vis\n'
                     '    from .runs import run_ffmpeg, sha256_file\n'
                     '    run = run or Run()\n'
                     '    inp = run.add_input(media, "A1")\n'
                     '    run.add_input(words_path, "W1")\n'
                     '    run.add_input(scenes_path, "SCENES")\n'
                     '    run.add_input(visuals_csv, "VISUALS_CSV")\n'
                     '    ms = media_summary(media)\n'
                     '    words = load_words(words_path)\n'
                     '    plan = json.loads(Path(scenes_path).read_text(encoding="utf-8"))\n'
                     '    result: Dict[str, Any] = {"run_id": run.run_id, "dir": str(run.dir), '
                     '"accepted": False}\n'
                     '    errs = scenes_mod.validate_scenes(plan, n_words=len(words))\n'
                     '    if abs(plan.get("duration", 0) - ms["duration"]) > 0.05:\n'
                     '        errs.append(f"scenes.json duration {plan.get(\'duration\')} != media '
                     '{ms[\'duration\']:.3f}")\n'
                     '    if errs:\n'
                     '        run.set_status("qa_failed")\n'
                     '        return result | {"error": "scene plan invalid for this media/words", '
                     '"problems": errs}\n'
                     '    sp = write_json(run.path("scenes.json"), plan)\n'
                     '    shutil.copy2(words_path, run.path("words.json"))\n'
                     '    chk = vis.validate(vis.read_csv(visuals_csv), plan, assets_dir)\n'
                     '    adir_abs = Path(chk["visuals"]["assets_dir"])\n'
                     '    try:  # portable: store assets_dir relative to the KANG root when it '
                     'lives inside it\n'
                     '        chk["visuals"]["assets_dir"] = '
                     'adir_abs.relative_to(ROOT.resolve()).as_posix()\n'
                     '    except ValueError:\n'
                     '        chk["visuals"]["assets_dir"] = adir_abs.name\n'
                     '    vp = write_json(run.path("visuals.json"), chk["visuals"])\n'
                     '    for w_ in chk["warnings"]:\n'
                     '        run.warn(w_)\n'
                     '    result.update(visuals=len(chk["visuals"]["items"]), '
                     'warnings=chk["warnings"])\n'
                     '    if not chk["ok"]:\n'
                     '        run.set_status("qa_failed")\n'
                     '        return result | {"error": "visuals.csv invalid", "problems": '
                     'chk["problems"]}\n'
                     '    if not chk["visuals"]["items"]:\n'
                     '        run.set_status("qa_failed")\n'
                     '        return result | {"error": "no scene has an asset yet"}\n'
                     '    info = vis.selection_text(chk["visuals"])\n'
                     '    if not _gate(run, "visual_selection", info, approver, artifacts=[vp, '
                     'sp], settings={"fps": fps, "scenes": scene_ids}):\n'
                     '        run.set_status("awaiting_visual_approval")\n'
                     '        return result | {"visual_selection": info}\n'
                     '    result["visuals_human_approved"] = run.manifest.get("human_approved", '
                     '{}).get("visual_selection", False)\n'
                     '    adir = adir_abs\n'
                     '    log = run.path("ffmpeg.log")\n'
                     '    checks: List[Dict[str, Any]] = []\n'
                     '    acc_p = Path(words_path).resolve().parent / '
                     '"transcript_acceptance.json"\n'
                     '    transcript_accepted = False\n'
                     '    if acc_p.is_file():\n'
                     '        acc = json.loads(acc_p.read_text(encoding="utf-8"))\n'
                     '        transcript_accepted = bool(acc.get("accepted")) and '
                     'acc.get("words_sha256") == sha256_file(Path(words_path))\n'
                     '\n'
                     '    # (b) Premiere package: narration whole + visual track + scene markers + '
                     'full captions.srt\n'
                     '    pkg = run.path("premiere_package")\n'
                     '    (pkg / "media").mkdir(parents=True, exist_ok=True)\n'
                     '    (pkg / "visuals").mkdir(exist_ok=True)\n'
                     '    mname = Path(media).name\n'
                     '    shutil.copy2(media, pkg / "media" / mname)\n'
                     '    base = PureWindowsPath(win_root) if win_root else None\n'
                     '    vclips = []\n'
                     '    for it in chk["visuals"]["items"]:\n'
                     '        dst = pkg / "visuals" / Path(it["asset"]).name\n'
                     '        if not dst.exists():\n'
                     '            shutil.copy2(adir / it["asset"], dst)\n'
                     '        vinfo = media_summary(dst)\n'
                     '        vclips.append({"path": str(base / "visuals" / dst.name) if base else '
                     'str(dst.resolve()), "name": dst.name,\n'
                     '                       "width": vinfo.get("width"), "height": '
                     'vinfo.get("height"),\n'
                     '                       "kind": it["kind"], "start": it["start"], "end": '
                     'it["end"], "in": it["asset_in"],\n'
                     '                       "src_duration": it.get("asset_duration", '
                     'it["duration"]), "loop": it["fit"] == "loop"})\n'
                     '    markers = [{"at": s["start"], "name": f"{s[\'id\']} '
                     '{s[\'title\']}"[:80], "comment": s["text"][:500]} for s in plan["scenes"]]\n'
                     '    xml, stats = premiere.build_xmeml(\n'
                     '        f"KANG {run.run_id} visuals", {"path": str(base / "media" / mname) '
                     'if base else str((pkg / "media" / mname).resolve()),\n'
                     '                                       "name": mname, "duration": '
                     'ms["duration"], "has_video": ms["has_video"],\n'
                     '                                       "channels": ms.get("channels", 2), '
                     '"sample_rate": ms.get("sample_rate", 48000),\n'
                     '                                       "width": ms.get("width"), "height": '
                     'ms.get("height")},\n'
                     '        [[0.0, ms["duration"]]], fps=fps, markers=markers, '
                     'visual_clips=vclips)\n'
                     '    (pkg / "sequence.xml").write_text(xml, encoding="utf-8")\n'
                     '    checks.append(premiere.check_xmeml(xml, fps))\n'
                     '    full_sub = make_subtitles(run, words, "16:9", preset_name)\n'
                     '    checks.extend(c for c in full_sub.get("checks", []) if isinstance(c, '
                     'dict))\n'
                     '    shutil.copy2(full_sub["srt"], pkg / "captions.srt")\n'
                     '    shutil.copy2(sp, pkg / "scenes.json")\n'
                     '    shutil.copy2(vp, pkg / "visuals.json")\n'
                     '    write_json(pkg / "words.json", words)\n'
                     '\n'
                     '    # (a) integration preview of the chosen contiguous scenes\n'
                     '    preview = None\n'
                     '    if scene_ids:\n'
                     '        sel = vis.pick(chk["visuals"], plan, scene_ids)\n'
                     '        t0, t1 = sel[0]["start"], sel[-1]["end"]\n'
                     '        F = lambda t: int(round((t - t0) * fps))\n'
                     '        parts = run.path("preview_parts")\n'
                     '        parts.mkdir(exist_ok=True)\n'
                     '        lst = []\n'
                     '        for it in sel:\n'
                     '            seg = parts / f"{it.get(\'shot\', it[\'scene\'])}.mp4"\n'
                     '            nfr = F(it["end"]) - F(it["start"])\n'
                     '            run_ffmpeg(vis.segment_args(it, adir / it["asset"], seg, nfr / '
                     'fps, 1920, 1080, fps), log)\n'
                     '            lst.append(f"file \'{seg.resolve().as_posix()}\'")\n'
                     '        (parts / "concat.txt").write_text("\\n".join(lst) + "\\n", '
                     'encoding="utf-8")\n'
                     '        vonly = run.path("preview_visuals_only.mp4")\n'
                     '        run_ffmpeg(["-f", "concat", "-safe", "0", "-i", str(parts / '
                     '"concat.txt"), "-c", "copy", str(vonly)], log)\n'
                     '        dur = F(t1) / fps\n'
                     '        wsl = [dict(w, start=round(w["start"] - t0, 3), end=round(w["end"] - '
                     't0, 3))\n'
                     '               for w in words if w["start"] >= t0 - 1e-6 and w["end"] <= t1 '
                     '+ 1e-6]\n'
                     '        sub = make_subtitles(run, wsl, "16:9", preset_name)\n'
                     '        checks.extend(c for c in sub.get("checks", []) if isinstance(c, '
                     'dict))\n'
                     '        preview = run.path("preview_" + "_".join(s["scene"] for s in '
                     '(sel[0], sel[-1])) + ".mp4")\n'
                     '        # +faststart rewrites the file at the end; on network filesystems '
                     '(JuiceFS sandbox) that\n'
                     '        # fails with "Error writing trailer: Input/output error" -> encode '
                     'on local temp, then move\n'
                     '        import tempfile\n'
                     '        tmp_out = Path(tempfile.mkdtemp(prefix="kang_prev_")) / '
                     'preview.name\n'
                     '        run_ffmpeg(["-i", str(vonly), "-ss", f"{t0:.3f}", "-t", '
                     'f"{dur:.3f}", "-i", str(media), "-map", "0:v", "-map", "1:a",\n'
                     '                    "-vf", render.subtitles_filter(sub["ass"], ROOT / '
                     '"fonts"), "-c:v", "libx264", "-preset", "veryfast",\n'
                     '                    "-crf", "21", "-pix_fmt", "yuv420p", "-c:a", "aac", '
                     '"-b:a", "160k", "-t", f"{dur:.3f}",\n'
                     '                    "-movflags", "+faststart", str(tmp_out)], log)\n'
                     '        shutil.move(str(tmp_out), str(preview))\n'
                     '        pm = media_summary(preview)\n'
                     '        checks.append({"name": "preview_duration", "pass": '
                     'abs(pm["duration"] - dur) <= 0.1,\n'
                     '                       "detail": {"expected_s": round(dur, 3), "actual_s": '
                     'round(pm["duration"], 3)}})\n'
                     '        checks.append({"name": "preview_has_audio_video", "pass": '
                     'pm["has_video"] and pm["has_audio"]})\n'
                     '        run.add_output(preview, "visual_preview")\n'
                     '        result.update(preview=str(preview), preview_range_s=[round(t0, 3), '
                     'round(t1, 3)],\n'
                     '                      preview_scenes=[i["scene"] for i in sel])\n'
                     '\n'
                     '    (pkg / "README_IMPORT.txt").write_text(\n'
                     '        "KANG Premiere package with scene visuals (" + run.run_id + ")\\n"\n'
                     '        "Target: Premiere Pro 24.6.1. Actual import verification: pending '
                     '(NOT_TESTED)\\n"\n'
                     '        "Video track: " + str(len(vclips)) + " approved scene visual(s) from '
                     'visuals/; scenes without a visual stay empty.\\n"\n'
                     '        "Narration: whole (no cuts). Captions: separate captions.srt.\\n"\n'
                     '        "Transcript: " + ("ACCEPTED" if transcript_accepted else "DRAFT - '
                     'not accepted (listening review pending)") + "\\n"\n'
                     '        "Paths: XML points to " + (str(base) if base else "this folder") + " '
                     '(media/ and visuals/ next to sequence.xml).\\n\\n"\n'
                     '        "1. Extract anywhere (or to " + (str(base) if base else "your PC") + '
                     '"), then double-click FIX_PATHS.bat\\n"\n'
                     '        "   (rewrites sequence.xml to this folder, says PASS/WARN). No Link '
                     'Media needed.\\n"\n'
                     '        "2. Premiere: File > Open Project (Ctrl+O), file type \'All Files\', '
                     'pick sequence.xml (not File > Import).\\n"\n'
                     '        "3. Home key (playhead 00:00:00:00), select captions.srt in the '
                     'Project panel after File > Import, press \'.\' (Overwrite).\\n"\n'
                     '        "4. If anything still shows offline: right-click > Link Media > '
                     'media/ or visuals/.\\n", encoding="utf-8")\n'
                     '    credits = [n for n in ("CREDITS.txt", "CREDITS.json") if (adir / '
                     'n).is_file()]\n'
                     '    for n in credits:  # rights evidence travels with the visuals and is '
                     'hashed in the manifest\n'
                     '        shutil.copy2(adir / n, pkg / n)\n'
                     '    if vclips and not credits:\n'
                     '        run.warn("assets dir has no CREDITS.txt/CREDITS.json; licence '
                     'evidence not packaged")\n'
                     '    premiere.write_fix_paths(pkg)\n'
                     '    files = sorted(p for p in pkg.rglob("*") if p.is_file())\n'
                     '    report = qa.qa_report(checks)\n'
                     '    write_json(pkg / "package_manifest.json", {\n'
                     '        "run_id": run.run_id, "source_sha256": inp["sha256"], "stats": '
                     'stats,\n'
                     '        "status": {"transcript_accepted": transcript_accepted, '
                     '"words_source": str(words_path),\n'
                     '                   "scene_plan_human_approved": False, "visuals": '
                     '"selected_by_user",\n'
                     '                   "visuals_human_approved": '
                     'result["visuals_human_approved"],\n'
                     '                   "captions": "separate_srt", "premiere_import": '
                     '"NOT_TESTED",\n'
                     '                   "preview": "integration_preview" if preview else '
                     '"none"},\n'
                     '        "files": {str(p.relative_to(pkg)).replace("\\\\", "/"): '
                     'sha256_file(p) for p in files}})\n'
                     '    write_json(run.path("qa_report.json"), report)\n'
                     '    run.add_output(pkg / "sequence.xml", "premiere_xml")\n'
                     '    run.log_step("visual_preview", visual_items=len(vclips), '
                     'preview=bool(preview), **stats)\n'
                     '    result.update(package=str(pkg), stats=stats, qa=report)\n'
                     '    from .package_readiness import finalize_package_readiness\n'
                     '    finalize_package_readiness(run, result, pkg, words, report)\n'
                     '    return result\n'
                     '\n'
                     '\n'
                     '# '
                     '--------------------------------------------------------------------------- '
                     'F03\n'
                     'def review_transcript(media: str, words_path: str, run: Optional[Run] = '
                     'None, manual: Optional[List[str]] = None,\n'
                     '                      clips: bool = True, pad_s: float = 1.5) -> Dict[str, '
                     'Any]:\n'
                     '    """Make review.csv (+ one mp3 clip per span) for the human listen. '
                     'Changes NO words."""\n'
                     '    from . import review as rv, scenes as scenes_mod\n'
                     '    run = run or Run()\n'
                     '    run.add_input(media, "A1")\n'
                     '    words = load_words(words_path)\n'
                     '    shutil.copy2(words_path, run.path("words.json"))\n'
                     '    spans = rv.review_spans(words)\n'
                     '    man, missing = [], []\n'
                     '    for ph in manual or []:\n'
                     '        m = rv.find_phrase(words, ph)\n'
                     '        (man.append(m) if m else missing.append(ph))\n'
                     '    spans = rv.merge_manual(spans, man, words)\n'
                     '    rdir = run.path("review")\n'
                     '    rdir.mkdir(parents=True, exist_ok=True)\n'
                     '    if clips:\n'
                     '        cdir = rdir / "clips"\n'
                     '        cdir.mkdir(exist_ok=True)\n'
                     '        for s in spans:\n'
                     '            out = cdir / f"{s[\'id\']}_{int(s[\'start\'] // '
                     '60):02d}m{int(s[\'start\'] % 60):02d}s.mp3"\n'
                     '            ss = max(0.0, s["start"] - pad_s)\n'
                     '            from .runs import run_ffmpeg\n'
                     '            run_ffmpeg(["-ss", f"{ss:.2f}", "-t", f"{s[\'end\'] + pad_s - '
                     'ss:.2f}", "-i", media,\n'
                     '                               "-vn", "-ac", "1", "-b:a", "96k", str(out)])\n'
                     '            s["clip"] = "clips/" + out.name\n'
                     '    plan = scenes_mod.plan_scenes(words, media_summary(media)["duration"])\n'
                     '    csv_p = rv.write_review_csv(rdir / "review.csv", spans, plan)\n'
                     '    run.add_output(csv_p, "review_csv")\n'
                     '    pri = {k: sum(1 for s in spans if s["priority"] == k) for k in (1, 2, '
                     '3)}\n'
                     '    run.log_step("review_transcript", spans=len(spans), **{f"p{k}": v for k, '
                     'v in pri.items()})\n'
                     '    run.set_status("review_required")\n'
                     '    return {"run_id": run.run_id, "dir": str(run.dir), "review_csv": '
                     'str(csv_p), "spans": len(spans),\n'
                     '            "by_priority": pri, "manual_not_found": missing, '
                     '"transcript_accepted": False}\n'
                     '\n'
                     '\n'
                     'TEXT_SOURCES = {"human": "human_reviewed", "auto-extraction": '
                     '"auto_extraction_recommended"}\n'
                     '\n'
                     '\n'
                     'def apply_transcript_review(words_path: str, review_csv: str, run: '
                     'Optional[Run] = None,\n'
                     '                            approver: Approver = _deny, source: str = '
                     '"human") -> Dict[str, Any]:\n'
                     '    """Apply the human\'s review.csv -> new words.json. Acceptance '
                     '(transcript_acceptance.json) is\n'
                     '    written only when every priority-1 row has an action AND a human '
                     'approves the transcript gate."""\n'
                     '    from . import review as rv\n'
                     '    from .runs import sha256_file\n'
                     '    run = run or Run()\n'
                     '    words = load_words(words_path)\n'
                     '    rows = rv.read_review_csv(review_csv)\n'
                     '    status = TEXT_SOURCES[source]\n'
                     '    res = rv.apply_review(words, rows, text_status=status)\n'
                     '    result: Dict[str, Any] = {"run_id": run.run_id, "dir": str(run.dir), '
                     '"ok": res["ok"], "problems": res["problems"],\n'
                     '                              "applied": res["applied"], '
                     '"transcript_accepted": False, "text_status": status}\n'
                     '    if not res["ok"]:\n'
                     '        run.set_status("qa_failed")\n'
                     '        return result\n'
                     '    shutil.copy2(review_csv, run.path("review.csv"))\n'
                     '    wp = write_json(run.path("words.json"), res["words"])\n'
                     '    run.add_output(wp, "words_reviewed")\n'
                     '    write_json(run.path("review_changes.json"), res["applied"])\n'
                     '    open_p1 = [r["id"] for r in rows if str(r.get("priority")) == "1" and '
                     'not (r.get("action") or "").strip()]\n'
                     '    result["words"] = str(wp)\n'
                     '    result["open_priority1"] = open_p1\n'
                     '    result["timing_unverified"] = res.get("timing_unverified", [])\n'
                     '    not_listened = [a["id"] for a in res["applied"] if a["text_status"] != '
                     '"human_reviewed" or a["content_review_pending"]]\n'
                     '    result["not_listened"] = not_listened\n'
                     '    if source != "human" or not_listened:  # auto-extraction / '
                     'editorial-policy / pending rows: a DRAFT, never accepted\n'
                     '        run.path("transcript.txt").write_text(" ".join(w["word"] for w in '
                     'res["words"]), encoding="utf-8")\n'
                     '        run.log_step("apply_review_draft", source=source, '
                     'fixes=len(res["applied"]))\n'
                     '        run.set_status("draft_corrected_not_accepted")\n'
                     '        result["status"] = "draft_corrected_not_accepted"\n'
                     '        return result\n'
                     '    if open_p1:\n'
                     '        run.set_status("review_required")\n'
                     '        return result\n'
                     '    info = f"Transcript review: {len(res[\'applied\'])} fix(es) applied, '
                     '{len(rows)} rows reviewed. Accept transcript?"\n'
                     '    if _gate(run, "transcript", info, approver, artifacts=[wp, '
                     'run.path("review.csv")]):\n'
                     '        human = run.manifest.get("human_approved", {}).get("transcript", '
                     'False)\n'
                     '        who = getattr(approver, "approver_name", getattr(approver, '
                     '"__name__", "callable"))\n'
                     '        write_json(run.path("transcript_acceptance.json"), {"accepted": '
                     'human, "words_sha256": sha256_file(wp),\n'
                     '                                                            "approver": who, '
                     '"text_accepted": human,\n'
                     '                                                            '
                     '"changed_word_timing_review_required": bool(res.get("timing_unverified", '
                     '[])),\n'
                     '                                                            '
                     '"timing_unverified": res.get("timing_unverified", []),\n'
                     '                                                            '
                     '"full_audio_sync": "NOT_TESTED"})\n'
                     '        result["transcript_accepted"] = human\n'
                     '        run.set_status("transcript_accepted" if human else '
                     '"review_required")\n'
                     '    else:\n'
                     '        run.set_status("awaiting_transcript_approval")\n'
                     '    return result\n'
                     '\n'
                     '\n'
                     'def plan_and_cut(media: str, words_path: str, run: Optional[Run] = None, '
                     'approver: Approver = _deny,\n'
                     '                 noise_db: float = -34.0, min_silence: float = 0.45) -> '
                     'Dict[str, Any]:\n'
                     '    run = run or Run()\n'
                     '    inp = run.add_input(media, "A1")\n'
                     '    words = load_words(words_path)\n'
                     '    total = media_summary(media)["duration"]\n'
                     '    sil = cutter.detect_silence(media, noise_db, min_silence, total)\n'
                     '    fill = cutter.load_fillers(ROOT / "config" / "fillers.json")\n'
                     '    plan = cutter.plan_cuts(words, sil, total, fill)\n'
                     '    plan["silences"] = sil\n'
                     '    words = plan.pop("words_tightened")  # silence-aware word bounds (used '
                     'for QA + remap)\n'
                     '    run.path("cut_list.txt").write_text(cutter.cut_list_text(plan), '
                     'encoding="utf-8")\n'
                     '    write_json(run.path("words_tightened.json"), words)\n'
                     '    bad = cutter.check_no_word_truncated(words, plan["keep"], fill["auto"])\n'
                     '    write_json(run.path("cuts.json"), plan)\n'
                     '    has_video = media_summary(media)["has_video"]\n'
                     '    edl = EDL(run_id=run.run_id, inputs=[Input(id="A1", path=inp["path"], '
                     'sha256=inp["sha256"])],\n'
                     '              cuts=[Cut(src="A1", **{"in": s}, out=e, type="jumpcut") for s, '
                     'e in plan["keep"]],\n'
                     '              approvals=Approvals(cut_list="pending", preview="pending"))\n'
                     '    save_edl(edl, run.path("edl.json"))\n'
                     '    result: Dict[str, Any] = {"run_id": run.run_id, "dir": str(run.dir), '
                     '"plan": plan["stats"], "truncated_words": bad, "output": None}\n'
                     '    run.log_step("cut_plan", **plan["stats"], truncated=len(bad))\n'
                     '    if bad:\n'
                     '        run.set_status("qa_failed")\n'
                     '        return result\n'
                     '    if _gate(run, "cut_list", cutter.cut_list_text(plan), approver, '
                     'artifacts=[run.path("cuts.json")]):\n'
                     '        ext = ".mp4" if has_video else ".wav"\n'
                     '        out = cutter.render_cuts(media, plan["keep"], run.path(f"cut{ext}"), '
                     'video=has_video, log=run.path("ffmpeg.log"))\n'
                     '        new_words = cutter.remap_words(words, plan["keep"])\n'
                     '        write_json(run.path("words_cut.json"), new_words)\n'
                     '        dur = media_summary(out)["duration"]\n'
                     '        result["output"] = str(out)\n'
                     '        result["duration_check"] = {"expected": plan["stats"]["kept_dur"], '
                     '"actual": dur,\n'
                     '                                    "pass": abs(dur - '
                     'plan["stats"]["kept_dur"]) <= 1 / 30 + 0.05}\n'
                     '        edl.approvals.cut_list = "approved"\n'
                     '        save_edl(edl, run.path("edl.json"))\n'
                     '        run.add_output(out, "cut")\n'
                     '        run.set_status("cut_done")\n'
                     '    else:\n'
                     '        run.set_status("awaiting_cut_approval")\n'
                     '    return result\n'
                     '\n'
                     '\n'
                     '# '
                     '--------------------------------------------------------------------------- '
                     'F04\n'
                     'def clean_audio(media: str, run: Optional[Run] = None, target_lufs: float = '
                     '-14.0, true_peak: float = -1.5,\n'
                     '                rnnoise_model: Optional[str] = None) -> Dict[str, Any]:\n'
                     '    run = run or Run()\n'
                     '    run.add_input(media, "A1")\n'
                     '    has_video = media_summary(media)["has_video"]\n'
                     '    out = run.path("clean.mp4" if has_video else "clean.wav")\n'
                     '    rep = audio_mod.clean_audio(media, out, target_lufs, true_peak, '
                     'rnnoise_model=rnnoise_model,\n'
                     '                                video_copy=has_video, '
                     'log=run.path("ffmpeg.log"))\n'
                     '    write_json(run.path("loudness_report.json"), rep)\n'
                     '    run.add_output(out, "clean_audio")\n'
                     '    run.log_step("audio_cleanup", passed=rep["pass"], after=rep["after"])\n'
                     '    run.set_status("audio_done" if rep["pass"] else "qa_failed")\n'
                     '    return {"run_id": run.run_id, "dir": str(run.dir), "output": str(out), '
                     '"report": rep}\n'
                     '\n'
                     '\n'
                     '# '
                     '--------------------------------------------------------------------------- '
                     'F05\n'
                     'def multi_format(video: str, words_path: Optional[str], formats: List[str], '
                     'run: Optional[Run] = None,\n'
                     '                 preset_name: str = "vox_bold_yellow", reframe_x: float = '
                     '0.5) -> Dict[str, Any]:\n'
                     '    run = run or Run()\n'
                     '    run.add_input(video, "A1")\n'
                     '    src = media_summary(video)\n'
                     '    words = load_words(words_path) if words_path else None\n'
                     '    outs, checks = {}, []\n'
                     '    for fmt in formats:\n'
                     '        ass = make_subtitles(run, words, fmt, preset_name)["ass"] if words '
                     'else None\n'
                     '        out = render.render_format(video, fmt, '
                     'run.path(f"out_{fmt.replace(\':\', \'x\')}.mp4"), src["width"], '
                     'src["height"],\n'
                     '                                   ass_path=ass, fonts_dir=ROOT / "fonts", '
                     'reframe_x=reframe_x, log=run.path("ffmpeg.log"))\n'
                     '        c = qa.check_export(out, src["duration"], CANVAS[fmt])\n'
                     '        c["name"] = f"export_{fmt}"\n'
                     '        checks.append(c)\n'
                     '        outs[fmt] = str(out)\n'
                     '        run.add_output(out, f"export_{fmt}")\n'
                     '    report = qa.qa_report(checks)\n'
                     '    write_json(run.path("qa_report.json"), report)\n'
                     '    run.set_status("done" if report["pass"] else "qa_failed")\n'
                     '    return {"run_id": run.run_id, "dir": str(run.dir), "outputs": outs, '
                     '"qa": report}\n'
                     '\n'
                     '\n'
                     '# '
                     '--------------------------------------------------------------------------- '
                     'F19-lite\n'
                     'def palette_from_screenshot(image: str, run: Optional[Run] = None, k: int = '
                     '5) -> Dict[str, Any]:\n'
                     '    from .style import extract_palette, palette_to_brand_draft\n'
                     '    run = run or Run()\n'
                     '    run.add_input(image, "IMG1")\n'
                     '    draft = palette_to_brand_draft(extract_palette(image, k))\n'
                     '    p = write_json(run.path("brand_draft.json"), draft)\n'
                     '    run.add_output(p, "brand_draft")\n'
                     '    run.set_status("draft_ready")\n'
                     '    return {"run_id": run.run_id, "dir": str(run.dir), "draft": draft}\n'
                     '\n'
                     '\n'
                     '# '
                     '--------------------------------------------------------------------------- '
                     'Phase 3 workflow\n'
                     'def workflow_clip(source: str, start: float = 0.0, duration: float = 60.0,\n'
                     '                  formats: Optional[List[str]] = None, preset_name: str = '
                     '"vox_bold_yellow",\n'
                     '                  approver: Approver = _deny, language: str = "hi", device: '
                     'Optional[str] = None,\n'
                     '                  model_size: str = "large-v3-turbo", initial_prompt: '
                     'Optional[str] = None,\n'
                     '                  transcriber: Optional[Callable] = None, full: bool = '
                     'False,\n'
                     '                  spelling_policy: Optional[str] = None) -> Dict[str, Any]:\n'
                     '    """Phase 3 exit proof, one call, no code edits:\n'
                     '    full=True: process the whole file from `start` (duration ignored, max 1 '
                     'h); default formats 16:9.\n'
                     '    slice -> transcribe (+glossary) -> F03 cut [gate] -> F04 clean -> '
                     'F01/F02 subtitles per format [gate].\n'
                     '    Each stage is its own child run; the parent run records child IDs and '
                     'the combined verdict."""\n'
                     '    from .runs import run_ffmpeg\n'
                     '    if full:\n'
                     '        duration = round(media_summary(source)["duration"] - start, 3)\n'
                     '        formats = formats or ["16:9"]\n'
                     '        if not 5 <= duration <= FULL_MAX_S:\n'
                     '            raise ValueError(f"full mode: source after --start must be '
                     '5-{FULL_MAX_S:g} s, got {duration}")\n'
                     '    elif not 5 <= duration <= 60:\n'
                     '        raise ValueError("workflow clip duration must be 5-60 s (use '
                     'full=True / --full for the whole file)")\n'
                     '    formats = formats or ["9:16", "16:9", "1:1"]\n'
                     '    max_render = FULL_MAX_S if full else 60.0\n'
                     '    t0 = time.time()\n'
                     '    wf = Run()\n'
                     '    wf.add_input(source, "SRC")\n'
                     '    sl = wf.path("source_slice.wav")\n'
                     '    run_ffmpeg(["-ss", f"{start:.3f}", "-t", f"{duration:.3f}", "-i", '
                     'str(source), "-vn", "-ac", "1",\n'
                     '                "-ar", "48000", str(sl)], log_path=wf.path("ffmpeg.log"))\n'
                     '    stages: Dict[str, Any] = {}\n'
                     '    verdict: List[Dict[str, Any]] = []\n'
                     '\n'
                     '    def done(status: str) -> Dict[str, Any]:\n'
                     '        summary = {"run_id": wf.run_id, "dir": str(wf.dir), "source": '
                     'str(source), "start": start,\n'
                     '                   "duration": duration, "full": full, "seconds_taken": '
                     'round(time.time() - t0, 1), "stages": stages, "checks": verdict,\n'
                     '                   "pass": bool(verdict) and all(c["pass"] for c in verdict) '
                     'and status == "done"}\n'
                     '        write_json(wf.path("workflow_report.json"), summary)\n'
                     '        wf.set_status(status)\n'
                     '        return summary\n'
                     '\n'
                     '    t = transcribe_to_words(str(sl), language=language, '
                     'model_size=model_size, initial_prompt=initial_prompt,\n'
                     '                            spelling_policy=spelling_policy,\n'
                     '                            device=device, transcriber=transcriber)\n'
                     '    stages["transcribe"] = t["run_id"]\n'
                     '    words_path = Path(t["dir"]) / "words.json"\n'
                     '    verdict.append({"name": "subtitle_coverage", "pass": not '
                     't["gaps_unfilled"],\n'
                     '                    "detail": {"gaps_found": len(t["speech_gaps"]),\n'
                     '                               "words_added": sum(g["filled"] for g in '
                     't["speech_gaps"]),\n'
                     '                               "suspect_spans": len(t.get("suspect_spans", '
                     '[])),\n'
                     '                               "spans_repaired": sum(r["accepted"] for r in '
                     't.get("suspect_spans", [])),\n'
                     '                               "unfilled": t["gaps_unfilled"],\n'
                     '                               "needs_review_2_3s": '
                     't.get("gaps_needs_review", []),\n'
                     '                               "coverage_threshold_s": COVERAGE_GAP_S, '
                     '"blocking_threshold_s": BLOCKING_GAP_S}})\n'
                     '    verdict.append({"name": "transcript_audit", "pass": True, "blocking": '
                     'False,  # review flags, not a failure\n'
                     '                    "review_flags": t["audit"]["issue_count"], "detail": '
                     '{"issues": t["audit"]["issue_count"], "glossary_fixes": '
                     'len(t["glossary_fixes"])}})\n'
                     '\n'
                     '    c = plan_and_cut(str(sl), str(words_path), approver=approver)\n'
                     '    stages["cut"] = c["run_id"]\n'
                     '    if not c["output"]:\n'
                     '        verdict.append({"name": "cut", "pass": False, "detail": '
                     '{"truncated": c["truncated_words"]}})\n'
                     '        return done("awaiting_cut_approval" if not c["truncated_words"] else '
                     '"qa_failed")\n'
                     '    verdict.append({"name": "cut", "pass": c["duration_check"]["pass"] and '
                     'not c["truncated_words"],\n'
                     '                    "detail": {**c["plan"], "duration_check": '
                     'c["duration_check"]}})\n'
                     '\n'
                     '    # Internal true-peak target sits 1 dB under the -1.5 dBTP delivery '
                     'target: the one AAC\n'
                     '    # encode in render adds ~0.8 dB. The delivered file itself is measured '
                     'below (final_audio).\n'
                     '    a = clean_audio(c["output"], true_peak=AAC_SAFE_TP)\n'
                     '    stages["audio"] = a["run_id"]\n'
                     '    rep = a["report"]\n'
                     '    verdict.append({"name": "loudness", "pass": rep["pass"],\n'
                     '                    "detail": {"before": rep["before"]["integrated_lufs"], '
                     '"after": rep["after"]["integrated_lufs"],\n'
                     '                               "true_peak": '
                     'rep["after"]["true_peak_dbtp"]}})\n'
                     '\n'
                     '    words_cut = Path(c["dir"]) / "words_cut.json"\n'
                     '    clean_dur = media_summary(a["output"])["duration"]\n'
                     '    finals = {}\n'
                     '    for fmt in formats:\n'
                     '        p = promo_clip(a["output"], str(words_cut), 0.0, min(clean_dur, '
                     'max_render), fmt=fmt,\n'
                     '                       preset_name=preset_name, approver=approver, '
                     'max_duration=max_render)\n'
                     '        stages[f"render_{fmt}"] = p["run_id"]\n'
                     '        verdict.append({"name": f"render_{fmt}", "pass": bool(p["final"]) '
                     'and p["qa"]["pass"],\n'
                     '                        "detail": {"failed": p["qa"]["failed"], "final": '
                     'bool(p["final"])}})\n'
                     '        if p["final"]:\n'
                     '            dst = wf.path(f"final_{fmt.replace(\':\', \'x\')}.mp4")\n'
                     '            shutil.copy2(p["final"], dst)\n'
                     '            wf.add_output(dst, f"final_{fmt}")\n'
                     '            finals[fmt] = str(dst)\n'
                     '            tp = audio_mod.ensure_final_true_peak(dst, DELIVERY_TP, '
                     'log=wf.path("ffmpeg.log"))\n'
                     '            fl = tp["measured"]\n'
                     '            verdict.append({"name": f"final_audio_{fmt}",\n'
                     '                            "pass": fl["true_peak_dbtp"] <= DELIVERY_TP and '
                     'abs(fl["integrated_lufs"] - DELIVERY_LUFS) <= 1.0,\n'
                     '                            "detail": {"integrated_lufs": '
                     'fl["integrated_lufs"], "true_peak": fl["true_peak_dbtp"],\n'
                     '                                       "tp_fix_passes": tp["passes"],\n'
                     '                                       "limits": {"true_peak_max": '
                     'DELIVERY_TP, "lufs": DELIVERY_LUFS}}})\n'
                     '    stages["finals"] = finals\n'
                     '    return done("done" if len(finals) == len(formats) else '
                     '"awaiting_preview_approval")\n'
                     '\n'
                     '\n'
                     'def second_opinion(words_path: str, chunks_path: str, review_csv: '
                     'Optional[str] = None,\n'
                     '                   run: Optional[Run] = None, label: str = "nemotron") -> '
                     'Dict[str, Any]:\n'
                     '    """Second-model hints for review (changes no words): review.csv + '
                     'nemotron_hint/nemotron_agrees,\n'
                     '    and disagreements.json (every place the two models differ, classed)."""\n'
                     '    from . import second_opinion as so, review as rv\n'
                     '    run = run or Run()\n'
                     '    run.add_input(words_path, "W1")\n'
                     '    run.add_input(chunks_path, "SECOND_ASR")\n'
                     '    words = load_words(words_path)\n'
                     '    chunks = json.loads(Path(chunks_path).read_text(encoding="utf-8"))\n'
                     '    chunks = chunks["chunks"] if isinstance(chunks, dict) else chunks\n'
                     '    al = so.align(words, chunks)\n'
                     '    dis = so.disagreements(al)\n'
                     '    kinds = {k: sum(1 for d in dis if d["kind"] == k) for k in ("content", '
                     '"spelling", "spacing", "number", "script")}\n'
                     '    write_json(run.path("disagreements.json"), {"label": label, "counts": '
                     'kinds, "items": dis})\n'
                     '    run.add_output(run.path("disagreements.json"), "second_opinion")\n'
                     '    res: Dict[str, Any] = {"run_id": run.run_id, "dir": str(run.dir), '
                     '"counts": kinds, "review_csv": None}\n'
                     '    if review_csv:\n'
                     '        rows = so.annotate_rows(rv.read_review_csv(review_csv), al, words)\n'
                     '        outp = so.write_csv(run.path("review_with_hints.csv"), rows)\n'
                     '        run.add_output(outp, "review_csv")\n'
                     '        res["review_csv"] = str(outp)\n'
                     '        res["rows"] = len(rows)\n'
                     '        res["rows_disagree"] = sum(1 for r in rows if r["nemotron_agrees"] '
                     '== "no")\n'
                     '    run.log_step("second_opinion", **kinds)\n'
                     '    run.set_status("hints_ready")\n'
                     '    return res\n'
                     '\n'
                     '\n'
                     'def consensus_review(words_path: str, timed_path: Optional[str] = None, '
                     'text_path: Optional[str] = None,\n'
                     '                     review_csv: Optional[str] = None, names_path: '
                     'Optional[str] = None,\n'
                     '                     run: Optional[Run] = None, media: Optional[str] = None, '
                     'pad_s: float = 1.5) -> Dict[str, Any]:\n'
                     '    """3-model consensus hints (ours + timed 2nd ASR + untimed text). '
                     'Changes no words: writes consensus.csv,\n'
                     '    consensus.json and, with review_csv, review_consensus.csv '
                     '(consensus_status / consensus_suggestion)."""\n'
                     '    from . import consensus as cs, second_opinion as so, review as rv\n'
                     '    run = run or Run()\n'
                     '    run.add_input(words_path, "W1")\n'
                     '    words = load_words(words_path)\n'
                     '    timed = text = None\n'
                     '    if timed_path:\n'
                     '        run.add_input(timed_path, "TIMED_ASR")\n'
                     '        timed = '
                     'cs.timed_words(json.loads(Path(timed_path).read_text(encoding="utf-8")))\n'
                     '    if text_path:\n'
                     '        run.add_input(text_path, "TEXT_ASR")\n'
                     '        text = Path(text_path).read_text(encoding="utf-8")\n'
                     '    names_path = names_path or str(Path(__file__).resolve().parent.parent / '
                     '"config" / "name_suggestions.json")\n'
                     '    c = cs.consensus(words, timed, text, cs.load_names(names_path))\n'
                     '    write_json(run.path("consensus.json"), c)\n'
                     '    run.add_output(run.path("consensus.json"), "consensus")\n'
                     '    if media:  # one mp3 clip per priority-1/2 row that no pattern already '
                     'covers\n'
                     '        from .runs import run_ffmpeg\n'
                     '        cdir = run.path("clips")\n'
                     '        cdir.mkdir(exist_ok=True)\n'
                     '        for r in c["rows"]:\n'
                     '            r["clip"] = ""\n'
                     '            if r["priority"] <= 2 and not r["pattern"]:  # 0 amount, 1 '
                     'disagreement, 2 name/number\n'
                     '                out = cdir / f"{r[\'id\']}_{int(r[\'start\'] // '
                     '60):02d}m{int(r[\'start\'] % 60):02d}s.mp3"\n'
                     '                ss = max(0.0, r["start"] - pad_s)\n'
                     '                run_ffmpeg(["-ss", f"{ss:.2f}", "-t", f"{r[\'end\'] + pad_s '
                     '- ss:.2f}", "-i", media,\n'
                     '                            "-vn", "-ac", "1", "-b:a", "96k", str(out)])\n'
                     '                r["clip"] = "clips/" + out.name\n'
                     '    cols = ["id", "priority", "status", "kind", "pattern", "start", "end", '
                     '"first", "last", "ours", "nemotron",\n'
                     '            "mai", "suggestion", "mai_alignment"] + (["clip"] if media else '
                     '[])\n'
                     '    so.write_csv(run.path("consensus.csv"), [{k: r[k] for k in cols} | '
                     '{"action": "", "correct_text": ""}\n'
                     '                                              for r in c["rows"]])\n'
                     '    run.add_output(run.path("consensus.csv"), "consensus_csv")\n'
                     '    if c["patterns"]:\n'
                     '        so.write_csv(run.path("consensus_patterns.csv"), [p | {"action": ""} '
                     'for p in c["patterns"]])\n'
                     '        run.add_output(run.path("consensus_patterns.csv"), '
                     '"consensus_patterns")\n'
                     '    res: Dict[str, Any] = {"run_id": run.run_id, "dir": str(run.dir), '
                     '"counts": c["counts"],\n'
                     '                           "words": c["words"], "all_agree_pct": '
                     'c["all_agree_pct"], "review_csv": None,\n'
                     '                           "note": "hints only; agreement is not truth; '
                     'nothing applied"}\n'
                     '    if review_csv:\n'
                     '        rows = cs.annotate_review(rv.read_review_csv(review_csv), c, words)\n'
                     '        outp = so.write_csv(run.path("review_consensus.csv"), rows)\n'
                     '        run.add_output(outp, "review_csv")\n'
                     '        res["review_csv"] = str(outp)\n'
                     '        res["review_rows_by_status"] = {s: sum(1 for r in rows if '
                     'r["consensus_status"] == s)\n'
                     '                                        for s in ("others_agree", '
                     '"all_differ", "name_hint", "ours_majority", "format_only", "all_agree")}\n'
                     '    res["patterns"] = len(c["patterns"])\n'
                     '    res["rows_priority_1"] = sum(1 for r in c["rows"] if r["priority"] == '
                     '1)\n'
                     '    res["rows_amount_p0"] = sum(1 for r in c["rows"] if r["priority"] == 0)\n'
                     '    res["listen_rows"] = sum(1 for r in c["rows"] if r["priority"] <= 2 and '
                     'not r["pattern"])\n'
                     '    res["pattern_rows_covered"] = sum(p["count"] for p in c["patterns"])\n'
                     '    run.log_step("consensus", **c["counts"])\n'
                     '    run.set_status("hints_ready")\n'
                     '    return res\n',
 'core/premiere.py': '"""M3 Workflow G, F12: Premiere Pro package as Final Cut Pro 7 XML (xmeml '
                     'v4).\n'
                     '\n'
                     'Premiere Pro 24.x opens xmeml via File > Open Project (Ctrl+O, "All Files") '
                     "- verified on the user's PC 2026-10-31.\n"
                     'Element layout mirrors the H1-H6 diagnostic files that Premiere accepted '
                     '(GUID uuid, media before timecode/markers,\n'
                     'track enabled/locked, sequence audio format, file timecode + video '
                     'width/height, masterclipid, alphatype); the\n'
                     'previous layout (non-GUID uuid, markers before media, bare tracks) was '
                     'rejected with "project could not be loaded". The sequence references the '
                     'ORIGINAL\n'
                     'media (non-destructive): every kept range is one clipitem, so jump cuts stay '
                     'editable in Premiere.\n'
                     'Scenes become sequence markers. Captions travel as an SRT sidecar (File > '
                     'Import -> caption track).\n'
                     'Integer timebase only (25 / 30 / 50 / 60 fps, ntsc FALSE) so frame maths is '
                     'exact.\n'
                     '"""\n'
                     'from __future__ import annotations\n'
                     '\n'
                     'import uuid as _uuid\n'
                     'import xml.etree.ElementTree as ET\n'
                     'from pathlib import Path, PureWindowsPath\n'
                     'from typing import Any, Dict, List, Optional, Sequence, Tuple\n'
                     'from urllib.parse import quote\n'
                     '\n'
                     'Range = Tuple[float, float]\n'
                     '\n'
                     '\n'
                     'def frames(t: float, fps: int) -> int:\n'
                     '    return int(round(t * fps))\n'
                     '\n'
                     '\n'
                     'def pathurl(path: str) -> str:\n'
                     '    """file://localhost/E%3a/kang/x.mp3 style URL (what Premiere itself '
                     'writes) for Windows or POSIX paths."""\n'
                     '    if len(path) > 1 and path[1] == ":":  # Windows drive path\n'
                     '        p = PureWindowsPath(path).as_posix()\n'
                     '        drive, rest = p[0], p[2:]\n'
                     '        return "file://localhost/" + drive + "%3a" + quote(rest, safe="/")\n'
                     '    return "file://localhost" + quote(str(Path(path).as_posix()), safe="/")\n'
                     '\n'
                     '\n'
                     'def _rate(parent: ET.Element, fps: int) -> None:\n'
                     '    r = ET.SubElement(parent, "rate")\n'
                     '    ET.SubElement(r, "timebase").text = str(fps)\n'
                     '    ET.SubElement(r, "ntsc").text = "FALSE"\n'
                     '\n'
                     '\n'
                     'def _timecode(parent: ET.Element, fps: int) -> None:\n'
                     '    tc = ET.SubElement(parent, "timecode")\n'
                     '    _rate(tc, fps)\n'
                     '    _txt(tc, "string", "00:00:00:00")\n'
                     '    _txt(tc, "frame", 0)\n'
                     '    _txt(tc, "displayformat", "NDF")\n'
                     '\n'
                     '\n'
                     'def _video_chars(parent: ET.Element, fps: int, width: int, height: int) -> '
                     'None:\n'
                     '    sc = ET.SubElement(parent, "samplecharacteristics")\n'
                     '    _rate(sc, fps)\n'
                     '    _txt(sc, "width", width)\n'
                     '    _txt(sc, "height", height)\n'
                     '    _txt(sc, "anamorphic", "FALSE")\n'
                     '    _txt(sc, "pixelaspectratio", "square")\n'
                     '    _txt(sc, "fielddominance", "none")\n'
                     '\n'
                     '\n'
                     'def _close_track(track: ET.Element) -> None:\n'
                     '    _txt(track, "enabled", "TRUE")\n'
                     '    _txt(track, "locked", "FALSE")\n'
                     '\n'
                     '\n'
                     'def sequence_uuid(name: str) -> str:\n'
                     '    """Deterministic RFC-4122 GUID per sequence name (Premiere expects a '
                     'GUID, not free text)."""\n'
                     '    return str(_uuid.uuid5(_uuid.NAMESPACE_URL, "kang-sequence:" + name))\n'
                     '\n'
                     '\n'
                     'def _txt(parent: ET.Element, tag: str, value: Any) -> ET.Element:\n'
                     '    e = ET.SubElement(parent, tag)\n'
                     '    e.text = str(value)\n'
                     '    return e\n'
                     '\n'
                     '\n'
                     'def build_xmeml(name: str, media: Dict[str, Any], keep: Sequence[Range], '
                     'fps: int = 25,\n'
                     '                width: int = 1920, height: int = 1080, markers: '
                     'Sequence[Dict[str, Any]] = (),\n'
                     '                visual_clips: Sequence[Dict[str, Any]] = ()) -> Tuple[str, '
                     'Dict[str, Any]]:\n'
                     '    """media: {path (as Premiere should see it), name, duration, has_video, '
                     'channels, sample_rate}.\n'
                     '    keep: source ranges in seconds, in order.\n'
                     '    visual_clips: scene visuals [{path, name, kind still|video, start, end '
                     '(timeline s), in (s), src_duration (s,\n'
                     '    video), loop}] placed on their own video track (V1 when the narration '
                     'has no picture, else V2).\n'
                     '    Returns (xml text, stats)."""\n'
                     '    if fps not in (24, 25, 30, 50, 60):\n'
                     '        raise ValueError("fps must be an integer timebase '
                     '(24/25/30/50/60)")\n'
                     '    keep = [(float(a), float(b)) for a, b in keep if b > a]\n'
                     '    if not keep:\n'
                     '        raise ValueError("nothing to place on the timeline")\n'
                     '    for (a1, b1), (a2, b2) in zip(keep, keep[1:]):\n'
                     '        if a2 < b1:\n'
                     '            raise ValueError(f"keep ranges overlap: {a1}-{b1} / {a2}-{b2}")\n'
                     '    src_frames = frames(media["duration"], fps)\n'
                     '    channels = 1 if int(media.get("channels", 2)) == 1 else 2\n'
                     '    has_video = bool(media.get("has_video"))\n'
                     '\n'
                     "    # timeline positions in whole frames; each clip's length = its source "
                     'length in frames\n'
                     '    clips, pos = [], 0\n'
                     '    for a, b in keep:\n'
                     '        i_, o_ = frames(a, fps), min(frames(b, fps), src_frames)\n'
                     '        if o_ <= i_:\n'
                     '            continue\n'
                     '        clips.append({"in": i_, "out": o_, "start": pos, "end": pos + (o_ - '
                     'i_)})\n'
                     '        pos += o_ - i_\n'
                     '    seq_frames = pos\n'
                     '\n'
                     '    root = ET.Element("xmeml", version="4")\n'
                     '    seq = ET.SubElement(root, "sequence", id="sequence-1")\n'
                     '    _txt(seq, "uuid", sequence_uuid(name))\n'
                     '    _txt(seq, "duration", seq_frames)\n'
                     '    _rate(seq, fps)\n'
                     '    _txt(seq, "name", name)\n'
                     '    mt = ET.SubElement(seq, "media")\n'
                     '    vid = ET.SubElement(mt, "video")\n'
                     '    _video_chars(ET.SubElement(vid, "format"), fps, width, height)\n'
                     '    vtrack = ET.SubElement(vid, "track")\n'
                     '    aud = ET.SubElement(mt, "audio")\n'
                     '    _txt(aud, "numOutputChannels", 2)\n'
                     '    afs = ET.SubElement(ET.SubElement(aud, "format"), '
                     '"samplecharacteristics")\n'
                     '    _txt(afs, "depth", 16)\n'
                     '    _txt(afs, "samplerate", int(media.get("sample_rate", 48000)))\n'
                     '    atracks = [ET.SubElement(aud, "track") for _ in range(channels)]\n'
                     '\n'
                     '    file_written = False\n'
                     '\n'
                     '    def file_el(parent: ET.Element) -> None:\n'
                     '        nonlocal file_written\n'
                     '        if file_written:\n'
                     '            ET.SubElement(parent, "file", id="file-1")\n'
                     '            return\n'
                     '        file_written = True\n'
                     '        f = ET.SubElement(parent, "file", id="file-1")\n'
                     '        _txt(f, "name", media["name"])\n'
                     '        _txt(f, "pathurl", pathurl(media["path"]))\n'
                     '        _rate(f, fps)\n'
                     '        _txt(f, "duration", src_frames)\n'
                     '        _timecode(f, fps)\n'
                     '        fm = ET.SubElement(f, "media")\n'
                     '        if has_video:\n'
                     '            _video_chars(ET.SubElement(fm, "video"), fps, '
                     'int(media.get("width") or width),\n'
                     '                         int(media.get("height") or height))\n'
                     '        fa = ET.SubElement(fm, "audio")\n'
                     '        sa = ET.SubElement(fa, "samplecharacteristics")\n'
                     '        _txt(sa, "depth", 16)\n'
                     '        _txt(sa, "samplerate", int(media.get("sample_rate", 48000)))\n'
                     '        _txt(fa, "channelcount", channels)\n'
                     '\n'
                     '    def clipitem(track: ET.Element, cid: str, c: Dict[str, int], kind: str, '
                     'tindex: int) -> None:\n'
                     '        ci = ET.SubElement(track, "clipitem", id=cid)\n'
                     '        _txt(ci, "masterclipid", "masterclip-1")\n'
                     '        _txt(ci, "name", media["name"])\n'
                     '        _txt(ci, "enabled", "TRUE")\n'
                     '        _txt(ci, "duration", src_frames)\n'
                     '        _rate(ci, fps)\n'
                     '        _txt(ci, "start", c["start"])\n'
                     '        _txt(ci, "end", c["end"])\n'
                     '        _txt(ci, "in", c["in"])\n'
                     '        _txt(ci, "out", c["out"])\n'
                     '        if kind == "video":\n'
                     '            _txt(ci, "alphatype", "none")\n'
                     '        file_el(ci)\n'
                     '        if kind == "audio":\n'
                     '            st = ET.SubElement(ci, "sourcetrack")\n'
                     '            _txt(st, "mediatype", "audio")\n'
                     '            _txt(st, "trackindex", tindex)\n'
                     '\n'
                     '    for n, c in enumerate(clips, 1):\n'
                     '        if has_video:\n'
                     '            clipitem(vtrack, f"clipitem-v{n}", c, "video", 1)\n'
                     '        for t, tr in enumerate(atracks, 1):\n'
                     '            clipitem(tr, f"clipitem-a{t}-{n}", c, "audio", t)\n'
                     '\n'
                     '    n_visual = 0\n'
                     '    if visual_clips:\n'
                     '        vis_track = ET.SubElement(vid, "track") if has_video else vtrack\n'
                     '        files_done: Dict[str, str] = {}\n'
                     '        for k, v in enumerate(sorted(visual_clips, key=lambda x: '
                     'x["start"]), 1):\n'
                     '            s0, s1 = frames(v["start"], fps), min(frames(v["end"], fps), '
                     'seq_frames)\n'
                     '            if s1 <= s0:\n'
                     '                continue\n'
                     '            still = v["kind"] == "still"\n'
                     '            fdur = (s1 - s0) if still else frames(v["src_duration"], fps)\n'
                     '            a_in = 0 if still else frames(v.get("in", 0.0), fps)\n'
                     '            pieces, pos_ = [], s0\n'
                     '            while pos_ < s1:  # a looped video repeats as consecutive '
                     'clipitems\n'
                     '                ln = min(s1 - pos_, fdur - a_in)\n'
                     '                if ln <= 0:\n'
                     '                    break\n'
                     '                pieces.append((pos_, pos_ + ln, a_in, a_in + ln))\n'
                     '                pos_ += ln\n'
                     '                if not v.get("loop"):\n'
                     '                    break\n'
                     '                a_in = 0\n'
                     '            for j, (st, en, i_, o_) in enumerate(pieces, 1):\n'
                     '                ci = ET.SubElement(vis_track, "clipitem", '
                     'id=f"visual-{k}-{j}")\n'
                     '                vkey = files_done.get(v["path"]) or f"file-v{len(files_done) '
                     '+ 1}"\n'
                     '                _txt(ci, "masterclipid", "masterclip-" + vkey)\n'
                     '                _txt(ci, "name", v["name"])\n'
                     '                _txt(ci, "enabled", "TRUE")\n'
                     '                _txt(ci, "duration", fdur)\n'
                     '                _rate(ci, fps)\n'
                     '                for tag, val in (("start", st), ("end", en), ("in", i_), '
                     '("out", o_)):\n'
                     '                    _txt(ci, tag, val)\n'
                     '                _txt(ci, "alphatype", "none")\n'
                     '                fid = files_done.get(v["path"])\n'
                     '                if fid:\n'
                     '                    ET.SubElement(ci, "file", id=fid)\n'
                     '                else:\n'
                     '                    fid = f"file-v{len(files_done) + 1}"\n'
                     '                    files_done[v["path"]] = fid\n'
                     '                    f = ET.SubElement(ci, "file", id=fid)\n'
                     '                    _txt(f, "name", v["name"])\n'
                     '                    _txt(f, "pathurl", pathurl(v["path"]))\n'
                     '                    _rate(f, fps)\n'
                     '                    _txt(f, "duration", fdur)\n'
                     '                    _timecode(f, fps)\n'
                     '                    _video_chars(ET.SubElement(ET.SubElement(f, "media"), '
                     '"video"), fps,\n'
                     '                                 int(v.get("width") or width), '
                     'int(v.get("height") or height))\n'
                     '                n_visual += 1\n'
                     '    for tr in [t for t in vid.findall("track")] + atracks:\n'
                     '        _close_track(tr)\n'
                     '    _timecode(seq, fps)  # after <media>, then markers (Premiere-accepted '
                     'order)\n'
                     '    for m in markers:\n'
                     '        mk = ET.SubElement(seq, "marker")\n'
                     '        _txt(mk, "comment", m.get("comment", ""))\n'
                     '        _txt(mk, "name", m["name"])\n'
                     '        _txt(mk, "in", frames(m["at"], fps))\n'
                     '        _txt(mk, "out", -1)\n'
                     '    ET.indent(root, space="  ")\n'
                     '    xml = \'<?xml version="1.0" encoding="UTF-8"?>\\n<!DOCTYPE xmeml>\\n\' + '
                     'ET.tostring(root, encoding="unicode") + "\\n"\n'
                     '    stats = {"fps": fps, "clips": len(clips), "sequence_frames": seq_frames, '
                     '"sequence_s": round(seq_frames / fps, 3),\n'
                     '             "source_frames": src_frames, "audio_tracks": channels, "video": '
                     'has_video, "source_has_video": has_video,\n'
                     '             "visual_track": n_visual > 0, "markers": len(markers), '
                     '"visual_clipitems": n_visual}\n'
                     '    return xml, stats\n'
                     '\n'
                     '\n'
                     'def check_xmeml(xml: str, fps: int) -> Dict[str, Any]:\n'
                     '    """Structural QA: parses, clips contiguous on the timeline (no '
                     'gap/overlap), in<out, one full file element,\n'
                     '    sequence duration = sum of clip lengths. Returns {pass, problems}."""\n'
                     '    problems: List[str] = []\n'
                     '    root = ET.fromstring(xml.split("\\n", 2)[2] if xml.startswith("<?xml") '
                     'else xml)\n'
                     '    seq = root.find("sequence")\n'
                     '    dur = int(seq.findtext("duration"))\n'
                     '    if int(seq.findtext("rate/timebase")) != fps:\n'
                     '        problems.append("sequence timebase mismatch")\n'
                     '    full_files = [f for f in root.iter("file") if f.find("pathurl") is not '
                     'None]\n'
                     '    ids = {f.get("id") for f in root.iter("file")}\n'
                     '    if len(full_files) != len(ids) or len({f.get("id") for f in full_files}) '
                     '!= len(full_files):\n'
                     '        problems.append(f"every <file> id needs exactly one full definition '
                     '({len(full_files)} full, {len(ids)} ids)")\n'
                     '    # Premiere-accepted layout (2026-10-31 H1-H6): GUID uuid, media before '
                     'timecode/markers, closed tracks\n'
                     '    try:\n'
                     '        _uuid.UUID(seq.findtext("uuid") or "")\n'
                     '    except ValueError:\n'
                     '        problems.append("sequence uuid is not a GUID")\n'
                     '    tags = [c.tag for c in seq]\n'
                     '    if "media" not in tags or "timecode" not in tags or '
                     'tags.index("timecode") < tags.index("media"):\n'
                     '        problems.append("sequence <timecode> must follow <media>")\n'
                     '    elif any(t == "marker" and i < tags.index("media") for i, t in '
                     'enumerate(tags)):\n'
                     '        problems.append("markers must follow <media>")\n'
                     '    for tr in seq.findall("media/video/track") + '
                     'seq.findall("media/audio/track"):\n'
                     '        if [c.tag for c in tr][-2:] != ["enabled", "locked"]:\n'
                     '            problems.append("track missing trailing <enabled>/<locked>")\n'
                     '            break\n'
                     '    if "file-1" not in {f.get("id") for f in full_files}:\n'
                     '        problems.append("narration media file-1 missing")\n'
                     '    for kind in ("video", "audio"):\n'
                     '        for ti, tr in enumerate(seq.findall(f"media/{kind}/track"), 1):\n'
                     '            vis = [ci for ci in tr.findall("clipitem") if ci.get("id", '
                     '"").startswith("visual-")]\n'
                     '            if vis:  # scene visuals: ordered, no overlap, inside the '
                     'sequence (gaps allowed = scene without visual)\n'
                     '                end_prev = 0\n'
                     '                for ci in vis:\n'
                     '                    s, e, i, o = (int(ci.findtext(k)) for k in ("start", '
                     '"end", "in", "out"))\n'
                     '                    if s < end_prev or e > dur or o <= i or e - s != o - i:\n'
                     '                        problems.append(f"{kind} track {ti} '
                     '{ci.get(\'id\')}: overlap/length/out of sequence")\n'
                     '                    end_prev = e\n'
                     '                if len(vis) != len(tr.findall("clipitem")):\n'
                     '                    problems.append(f"{kind} track {ti}: visuals mixed with '
                     'narration clips")\n'
                     '                continue\n'
                     '            pos = 0\n'
                     '            for ci in tr.findall("clipitem"):\n'
                     '                s, e, i, o = (int(ci.findtext(k)) for k in ("start", "end", '
                     '"in", "out"))\n'
                     '                if s != pos:\n'
                     '                    problems.append(f"{kind} track {ti} {ci.get(\'id\')}: '
                     'starts {s}, expected {pos}")\n'
                     '                if o <= i or e - s != o - i:\n'
                     '                    problems.append(f"{kind} track {ti} {ci.get(\'id\')}: '
                     'length mismatch")\n'
                     '                pos = e\n'
                     '            if tr.findall("clipitem") and pos != dur:\n'
                     '                problems.append(f"{kind} track {ti}: ends {pos}, sequence '
                     '{dur}")\n'
                     '    return {"name": "premiere_xml", "pass": not problems, "problems": '
                     'problems}\n'
                     '\n'
                     '\n'
                     '# --- "extract anywhere" relink helper (2026-10-31: user hit Link Media '
                     'after extracting to Downloads) ---\n'
                     'PATH_PREFIX_RE = '
                     'r"<pathurl>(file://localhost/[^<]*?/)(?:media|visuals)/[^/<]+</pathurl>"\n'
                     '\n'
                     'FIX_PATHS_PS1 = r"""# KANG: point sequence.xml at THIS folder (wherever you '
                     'extracted it). Safe to run again.\n'
                     "$ErrorActionPreference = 'Stop'\n"
                     '$here = Split-Path -Parent $MyInvocation.MyCommand.Path\n'
                     "$p = Join-Path $here 'sequence.xml'\n"
                     '$x = [IO.File]::ReadAllText($p, [Text.Encoding]::UTF8)\n'
                     '$m = [regex]::Match($x, '
                     "'<pathurl>(file://localhost/[^<]*?/)(?:media|visuals)/[^/<]+</pathurl>')\n"
                     "if (-not $m.Success) { Write-Host 'FAIL: no media path found in "
                     "sequence.xml'; exit 1 }\n"
                     '$cur = $m.Groups[1].Value\n'
                     "$segs = ($here -replace '\\\\', '/').Split('/') | Where-Object { $_ -ne '' "
                     '}\n'
                     'if ($segs[0] -notmatch \'^[A-Za-z]:$\') { Write-Host "FAIL: folder must be '
                     'on a drive letter (got $here)"; exit 1 }\n'
                     '$drive = $segs[0].Substring(0, 1)\n'
                     '$rest = ($segs | Select-Object -Skip 1 | ForEach-Object { '
                     "[uri]::EscapeDataString($_) }) -join '/'\n"
                     '$new = if ($rest) { "file://localhost/$drive%3a/$rest/" } else { '
                     '"file://localhost/$drive%3a/" }\n'
                     '$x = $x.Replace($cur, $new)\n'
                     '[IO.File]::WriteAllText($p, $x, (New-Object Text.UTF8Encoding($false)))\n'
                     '$missing = @()\n'
                     'foreach ($u in [regex]::Matches($x, '
                     "'<pathurl>file://localhost/([^<]+)</pathurl>')) {\n"
                     "  $f = [uri]::UnescapeDataString($u.Groups[1].Value) -replace '/', '\\'\n"
                     '  if (-not (Test-Path -LiteralPath $f)) { $missing += $f }\n'
                     '}\n'
                     "if ($missing.Count) { Write-Host 'WARN: missing files:'; $missing | "
                     'Sort-Object -Unique | ForEach-Object { Write-Host "  $_" } }\n'
                     'else { Write-Host "PASS: sequence.xml now points to $here" }\n'
                     '"""\n'
                     '\n'
                     "FIX_PATHS_BAT = ('@echo off\\r\\npowershell -NoProfile -ExecutionPolicy "
                     'Bypass -File "%~dp0fix_paths.ps1"\\r\\n\'\n'
                     "                 'echo.\\r\\necho Ab Premiere mein Ctrl+O se sequence.xml "
                     "kholo.\\r\\npause\\r\\n')\n"
                     '\n'
                     '\n'
                     'def write_fix_paths(pkg_dir: Path) -> None:\n'
                     '    """Drop FIX_PATHS.bat + fix_paths.ps1 next to sequence.xml (Windows, '
                     'ASCII, CRLF)."""\n'
                     '    (Path(pkg_dir) / '
                     '"fix_paths.ps1").write_bytes(FIX_PATHS_PS1.replace("\\r\\n", '
                     '"\\n").replace("\\n", "\\r\\n").encode("ascii"))\n'
                     '    (Path(pkg_dir) / '
                     '"FIX_PATHS.bat").write_bytes(FIX_PATHS_BAT.encode("ascii"))\n',
 'core/qa.py': '"""Layer 5 QA: automatic checks from roadmap Section 9.\n'
               '\n'
               'Each check returns {"name", "pass", "detail"}; qa_report() aggregates them.\n'
               '"""\n'
               'from __future__ import annotations\n'
               '\n'
               'from pathlib import Path\n'
               'from typing import Any, Dict, List, Optional, Tuple\n'
               '\n'
               'from .runs import media_summary, run_ffmpeg\n'
               'from .subtitles import ass_events, ass_word_onsets\n'
               '\n'
               '# Conjunct / matra test sentences (F01 glyph test): Dwarka, Kshatriya, Shri, etc.\n'
               'GLYPH_TEST_SENTENCES = [\n'
               '    "द्वारिका क्षत्रिय श्री",\n'
               '    "ज्ञान त्र्यंबकेश्वर कृष्ण",\n'
               '    "हृदय श्रृंगार द्वारिकाधीश",\n'
               '    "रुक्मिणी दर्शन प्रसाद",\n'
               ']\n'
               '\n'
               '\n'
               'def _res(name: str, ok: bool, **detail: Any) -> Dict[str, Any]:\n'
               '    return {"name": name, "pass": bool(ok), "detail": detail}\n'
               '\n'
               '\n'
               'def check_export(path: str | Path, expected_dur: Optional[float] = None, size: '
               'Optional[Tuple[int, int]] = None,\n'
               '                 fps: float = 30.0, tol_frames: float = 1.0) -> Dict[str, Any]:\n'
               '    """ffprobe clean, resolution exact, duration within +/- tol frames of '
               'expectation."""\n'
               '    try:\n'
               '        m = media_summary(path)\n'
               '    except Exception as exc:\n'
               '        return _res("export_integrity", False, error=str(exc))\n'
               '    ok = m.get("has_video", False)\n'
               '    d: Dict[str, Any] = {"duration": m["duration"], "width": m.get("width"), '
               '"height": m.get("height")}\n'
               '    if size:\n'
               '        ok &= (m.get("width"), m.get("height")) == tuple(size)\n'
               '    if expected_dur is not None:\n'
               '        tol = tol_frames / fps + 0.025  # + AAC frame padding allowance\n'
               '        d["expected"] = expected_dur\n'
               '        d["delta"] = round(m["duration"] - expected_dur, 4)\n'
               '        ok &= abs(m["duration"] - expected_dur) <= tol\n'
               '    return _res("export_integrity", ok, **d)\n'
               '\n'
               '\n'
               'def check_sync(ass_text: str, words: List[Dict[str, Any]], tol_ms: float = 100.0) '
               '-> Dict[str, Any]:\n'
               '    """Option A: every karaoke word onset vs words.json start (+/- tol_ms)."""\n'
               '    onsets = ass_word_onsets(ass_text)\n'
               '    if len(onsets) != len(words):\n'
               '        return _res("subtitle_sync", False, error="word count mismatch", '
               'ass=len(onsets), words=len(words))\n'
               '    deltas = [abs(t - w["start"]) * 1000 for (_, t), w in zip(onsets, words)]\n'
               '    worst = max(deltas) if deltas else 0.0\n'
               '    return _res("subtitle_sync", worst <= tol_ms, worst_ms=round(worst, 1), '
               'tol_ms=tol_ms, words=len(words))\n'
               '\n'
               '\n'
               'def check_no_overlap(ass_text: str) -> Dict[str, Any]:\n'
               '    ev = sorted(ass_events(ass_text))\n'
               '    bad = [(a, b) for a, b in zip(ev, ev[1:]) if a[1] > b[0] + 1e-6]\n'
               '    return _res("no_overlap", not bad, overlaps=len(bad))\n'
               '\n'
               '\n'
               'def check_glyphs(font_path: str | Path, texts: List[str]) -> Dict[str, Any]:\n'
               '    from .brand import missing_glyphs\n'
               '    miss = sorted({c for t in texts for c in missing_glyphs(font_path, t)})\n'
               '    return _res("glyph_coverage", not miss, missing=[f"U+{ord(c):04X}" for c in '
               'miss])\n'
               '\n'
               '\n'
               'def text_bbox(png: str | Path, threshold: int = 40, bg: Tuple[int, int, int] = (0, '
               '0, 0)) -> Optional[Tuple[int, int, int, int]]:\n'
               '    """Bounding box of pixels that differ from a flat background."""\n'
               '    import numpy as np\n'
               '    from PIL import Image\n'
               '    a = np.asarray(Image.open(png).convert("RGB")).astype(int)\n'
               '    diff = np.abs(a - np.array(bg)).sum(-1) > threshold\n'
               '    ys, xs = np.nonzero(diff)\n'
               '    if len(xs) == 0:\n'
               '        return None\n'
               '    return int(xs.min()), int(ys.min()), int(xs.max()), int(ys.max())\n'
               '\n'
               '\n'
               'def check_safe_zone(png: str | Path, canvas: Tuple[int, int], margin_lr: int, '
               'bottom_px: int,\n'
               '                    bg: Tuple[int, int, int] = (0, 0, 0)) -> Dict[str, Any]:\n'
               '    """Subtitle pixels must stay inside: x in [margin, W-margin], bottom edge <= H '
               '- bottom_px."""\n'
               '    box = text_bbox(png, bg=bg)\n'
               '    if box is None:\n'
               '        return _res("safe_zone", False, error="no subtitle pixels found")\n'
               '    W, H = canvas\n'
               '    x0, y0, x1, y1 = box\n'
               '    ok = x0 >= margin_lr - 2 and x1 <= W - margin_lr + 2 and y1 <= H - bottom_px + '
               '4\n'
               '    return _res("safe_zone", ok, bbox=box, limits={"x_min": margin_lr, "x_max": W '
               '- margin_lr, "y_max": H - bottom_px})\n'
               '\n'
               '\n'
               'def check_contrast(fg: str, bg: str, target: float = 7.0, hard_fail: float = 4.5) '
               '-> Dict[str, Any]:\n'
               '    from .style import contrast_ratio\n'
               '    r = contrast_ratio(fg, bg)\n'
               '    return _res("contrast", r >= hard_fail, ratio=r, target=target, meets_target=r '
               '>= target, hard_fail_below=hard_fail)\n'
               '\n'
               '\n'
               'def render_ass_on_black(ass_path: str | Path, fonts_dir: str | Path, canvas: '
               'Tuple[int, int], t: float,\n'
               '                        out_png: str | Path, log: Optional[Path] = None, shaping: '
               'str = "complex") -> Path:\n'
               '    """Render a single subtitle frame on pure black (for bbox + glyph human '
               'check)."""\n'
               '    from .render import subtitles_filter\n'
               '    W, H = canvas\n'
               '    run_ffmpeg(["-f", "lavfi", "-i", f"color=c=black:s={W}x{H}:r=30:d={t + '
               '0.5:.3f}",\n'
               '                "-vf", subtitles_filter(ass_path, fonts_dir, shaping) + '
               'f",trim=start={t:.3f},setpts=PTS-STARTPTS",\n'
               '                "-frames:v", "1", str(out_png)], log)\n'
               '    return Path(out_png)\n'
               '\n'
               '\n'
               'def qa_report(checks: List[Dict[str, Any]]) -> Dict[str, Any]:\n'
               '    return {"pass": all(c["pass"] for c in checks), "checks": checks,\n'
               '            "failed": [c["name"] for c in checks if not c["pass"]]}\n'
               '\n'
               '\n'
               'def check_shaping(ass_path: str | Path, fonts_dir: str | Path, canvas: Tuple[int, '
               'int], t: float,\n'
               '                  work_dir: str | Path) -> Dict[str, Any]:\n'
               '    """Automatic broken-glyph detector: render the same frame with complex and '
               'with simple\n'
               '    shaping. Correct conjunct shaping is visibly narrower than the broken '
               '(halant-visible)\n'
               '    simple rendering. Pipeline must use complex and the widths must differ."""\n'
               '    wd = Path(work_dir)\n'
               '    good = render_ass_on_black(ass_path, fonts_dir, canvas, t, wd / '
               '"shaping_complex.png", shaping="complex")\n'
               '    bad = render_ass_on_black(ass_path, fonts_dir, canvas, t, wd / '
               '"shaping_simple.png", shaping="simple")\n'
               '    bg, bb = text_bbox(good), text_bbox(bad)\n'
               '    if not bg or not bb:\n'
               '        return _res("hindi_shaping", False, error="no text rendered")\n'
               '    wg, wb = bg[2] - bg[0], bb[2] - bb[0]\n'
               '    return _res("hindi_shaping", wg < wb * 0.97, complex_width=wg, '
               'simple_width=wb, frame=str(good))\n',
 'core/render.py': '"""Layer 4 Executor: FFmpeg renders.\n'
                   '\n'
                   '- burn_subtitles: F01 step 4 (subtitles=subs.ass:fontsdir=./fonts)\n'
                   '- make_preview: 10s low-res preview (approval gate)\n'
                   '- audio_to_video: voiceover-only input -> branded background video (Workflow '
                   'promo)\n'
                   '- reframe_filter / render_format: F05 multi-format (crop/pad per export, uses '
                   'EDL reframe_x)\n'
                   '"""\n'
                   'from __future__ import annotations\n'
                   '\n'
                   'from pathlib import Path\n'
                   'from typing import List, Optional\n'
                   '\n'
                   'from .brand import CANVAS\n'
                   'from .runs import run_ffmpeg\n'
                   '\n'
                   '\n'
                   'def _filter_path(p: str | Path) -> str:\n'
                   '    """Escape a path for use inside an ffmpeg filtergraph option."""\n'
                   '    s = str(Path(p).resolve()).replace("\\\\", "/")\n'
                   '    return s.replace(":", r"\\:").replace("\'", r"\\\'").replace(",", r"\\,")\n'
                   '\n'
                   '\n'
                   'def subtitles_filter(ass_path: str | Path, fonts_dir: str | Path, shaping: str '
                   '= "complex") -> str:\n'
                   '    """libass burn-in via the `ass` filter with COMPLEX (HarfBuzz) shaping.\n'
                   '\n'
                   '    Verified on FFmpeg 7.1: the default `subtitles` filter (auto/simple '
                   'shaping) breaks\n'
                   '    Devanagari conjuncts (द्वा -> द्\u200cवा, क्ष -> क्\u200cष). `ass` + '
                   'shaping=complex renders them correctly.\n'
                   '    """\n'
                   '    return '
                   '(f"ass=filename=\'{_filter_path(ass_path)}\':fontsdir=\'{_filter_path(fonts_dir)}\'"\n'
                   '            f":shaping={shaping}")\n'
                   '\n'
                   '\n'
                   'def audio_to_video(audio: str | Path, out: str | Path, fmt: str = "9:16", '
                   'color: str = "#0F172A",\n'
                   '                   fps: int = 30, start: float = 0.0, duration: '
                   'Optional[float] = None,\n'
                   '                   log: Optional[Path] = None) -> Path:\n'
                   '    """Voiceover/MP3 -> solid branded background video (for subtitle-only '
                   'promos)."""\n'
                   '    w, h = CANVAS[fmt]\n'
                   '    args: List[str] = []\n'
                   '    if start:\n'
                   '        args += ["-ss", f"{start:.3f}"]\n'
                   '    if duration:\n'
                   '        args += ["-t", f"{duration:.3f}"]\n'
                   '    args += ["-i", str(audio), "-f", "lavfi", "-i", '
                   'f"color=c=0x{color.lstrip(\'#\')}:s={w}x{h}:r={fps}",\n'
                   '             "-map", "1:v", "-map", "0:a", "-shortest", *(["-t", '
                   'f"{duration:.3f}"] if duration else []), "-c:v", "libx264", "-preset", '
                   '"veryfast", "-crf", "23",\n'
                   '             "-pix_fmt", "yuv420p", "-c:a", "aac", "-b:a", "160k", str(out)]\n'
                   '    run_ffmpeg(args, log)\n'
                   '    return Path(out)\n'
                   '\n'
                   '\n'
                   'def burn_subtitles(video: str | Path, ass_path: str | Path, fonts_dir: str | '
                   'Path, out: str | Path,\n'
                   '                   start: float = 0.0, duration: Optional[float] = None, '
                   'scale_h: Optional[int] = None,\n'
                   '                   crf: int = 20, log: Optional[Path] = None) -> Path:\n'
                   '    """Burn ASS subtitles. start/duration slice the source while keeping '
                   'subtitle sync\n'
                   '    (setpts keeps original timestamps for libass, then resets)."""\n'
                   '    vf = []\n'
                   '    if start:\n'
                   '        vf.append(f"setpts=PTS+{start:.3f}/TB")\n'
                   '    vf.append(subtitles_filter(ass_path, fonts_dir))\n'
                   '    if start:\n'
                   '        vf.append("setpts=PTS-STARTPTS")\n'
                   '    if scale_h:\n'
                   '        vf.append(f"scale=-2:{scale_h}")\n'
                   '    args: List[str] = []\n'
                   '    if start:\n'
                   '        args += ["-ss", f"{start:.3f}"]\n'
                   '    args += ["-i", str(video)]\n'
                   '    if duration:\n'
                   '        args += ["-t", f"{duration:.3f}"]\n'
                   '    # Full-length burn: copy the already-encoded audio. A second AAC '
                   'generation raised true peak\n'
                   '    # by ~1.9 dB on Dwarika full (-2.0 -> -0.1 dBTP). Slices/previews still '
                   're-encode.\n'
                   '    acodec = ["-c:a", "copy"] if not start and not duration else ["-c:a", '
                   '"aac", "-b:a", "160k"]\n'
                   '    args += ["-vf", ",".join(vf), "-c:v", "libx264", "-preset", "veryfast" if '
                   'scale_h else "medium",\n'
                   '             "-crf", str(crf), "-pix_fmt", "yuv420p", *acodec, "-movflags", '
                   '"+faststart", str(out)]\n'
                   '    run_ffmpeg(args, log)\n'
                   '    return Path(out)\n'
                   '\n'
                   '\n'
                   'def make_preview(video: str | Path, ass_path: str | Path, fonts_dir: str | '
                   'Path, out: str | Path,\n'
                   '                 start: float = 0.0, seconds: float = 10.0, log: '
                   'Optional[Path] = None) -> Path:\n'
                   '    """10-second low-res (960 px tall) preview for the approval gate."""\n'
                   '    return burn_subtitles(video, ass_path, fonts_dir, out, start=start, '
                   'duration=seconds, scale_h=960,\n'
                   '                          crf=28, log=log)\n'
                   '\n'
                   '\n'
                   'def reframe_filter(src_w: int, src_h: int, fmt: str, reframe_x: float = 0.5) '
                   '-> str:\n'
                   '    """Crop to target aspect around reframe_x (0..1 of source width), then '
                   'scale to canvas.\n'
                   '    Taller-than-target sources are cropped vertically (centered)."""\n'
                   '    tw, th = CANVAS[fmt]\n'
                   '    target = tw / th\n'
                   '    src = src_w / src_h\n'
                   '    if abs(src - target) < 1e-3:\n'
                   '        return f"scale={tw}:{th}"\n'
                   '    if src > target:  # too wide -> crop width\n'
                   '        cw = int(round(src_h * target)) // 2 * 2\n'
                   '        x = int(round(reframe_x * src_w - cw / 2))\n'
                   '        x = max(0, min(src_w - cw, x))\n'
                   '        return f"crop={cw}:{src_h}:{x}:0,scale={tw}:{th}"\n'
                   '    ch = int(round(src_w / target)) // 2 * 2\n'
                   '    y = max(0, (src_h - ch) // 2)\n'
                   '    return f"crop={src_w}:{ch}:0:{y},scale={tw}:{th}"\n'
                   '\n'
                   '\n'
                   'def render_format(video: str | Path, fmt: str, out: str | Path, src_w: int, '
                   'src_h: int,\n'
                   '                  ass_path: Optional[str | Path] = None, fonts_dir: '
                   'Optional[str | Path] = None,\n'
                   '                  reframe_x: float = 0.5, crf: int = 20, log: Optional[Path] = '
                   'None) -> Path:\n'
                   '    vf = reframe_filter(src_w, src_h, fmt, reframe_x) + ",setsar=1"\n'
                   '    if ass_path:\n'
                   '        vf += "," + subtitles_filter(ass_path, fonts_dir)\n'
                   '    run_ffmpeg(["-i", str(video), "-vf", vf, "-c:v", "libx264", "-preset", '
                   '"medium", "-crf", str(crf),\n'
                   '                "-pix_fmt", "yuv420p", "-c:a", "aac", "-b:a", "160k", '
                   '"-movflags", "+faststart", str(out)], log)\n'
                   '    return Path(out)\n'
                   '\n'
                   '\n'
                   'def extract_frame(video: str | Path, t: float, out: str | Path, log: '
                   'Optional[Path] = None) -> Path:\n'
                   '    run_ffmpeg(["-ss", f"{t:.3f}", "-i", str(video), "-frames:v", "1", '
                   'str(out)], log)\n'
                   '    return Path(out)\n',
 'core/runs.py': '"""Layer 1 (Intake) + Layer 7 (Archive).\n'
                 '\n'
                 '- Unique run IDs (YYYY-MM-DD_KANG_NNN), one run = one folder.\n'
                 '- Media integrity: sha256 + ffprobe.\n'
                 '- manifest.json written for every run.\n'
                 '- Originals are never overwritten: every write goes inside the run folder.\n'
                 '"""\n'
                 'from __future__ import annotations\n'
                 '\n'
                 'import datetime as _dt\n'
                 'import hashlib\n'
                 'import json\n'
                 'import os\n'
                 'import re\n'
                 'import subprocess\n'
                 'from pathlib import Path\n'
                 'from typing import Any, Dict, List, Optional\n'
                 '\n'
                 'RUN_ID_RE = re.compile(r"^(\\d{4}-\\d{2}-\\d{2})_KANG_(\\d{3,})$")\n'
                 '\n'
                 '\n'
                 'def runs_root() -> Path:\n'
                 '    root = Path(os.environ.get("KANG_RUNS_DIR", '
                 'Path(__file__).resolve().parent.parent / "runs"))\n'
                 '    root.mkdir(parents=True, exist_ok=True)\n'
                 '    return root\n'
                 '\n'
                 '\n'
                 'def sha256_file(path: str | Path, chunk: int = 1 << 20) -> str:\n'
                 '    h = hashlib.sha256()\n'
                 '    with open(path, "rb") as f:\n'
                 '        while True:\n'
                 '            b = f.read(chunk)\n'
                 '            if not b:\n'
                 '                break\n'
                 '            h.update(b)\n'
                 '    return h.hexdigest()\n'
                 '\n'
                 '\n'
                 'def ffprobe(path: str | Path) -> Dict[str, Any]:\n'
                 '    """Return ffprobe JSON (format + streams). Raises RuntimeError on unreadable '
                 'media."""\n'
                 '    p = Path(path)\n'
                 '    if not p.is_file() or p.stat().st_size == 0:\n'
                 '        raise FileNotFoundError(f"Media missing or empty: {p}")\n'
                 '    cmd = ["ffprobe", "-v", "error", "-print_format", "json", "-show_format", '
                 '"-show_streams", str(p)]\n'
                 '    res = subprocess.run(cmd, capture_output=True, text=True, encoding="utf-8", '
                 'errors="replace")\n'
                 '    if res.returncode != 0:\n'
                 '        raise RuntimeError(f"ffprobe failed for {p}: {res.stderr.strip()}")\n'
                 '    return json.loads(res.stdout)\n'
                 '\n'
                 '\n'
                 'def media_summary(path: str | Path) -> Dict[str, Any]:\n'
                 '    info = ffprobe(path)\n'
                 '    v = next((s for s in info.get("streams", []) if s.get("codec_type") == '
                 '"video"), None)\n'
                 '    a = next((s for s in info.get("streams", []) if s.get("codec_type") == '
                 '"audio"), None)\n'
                 '    out: Dict[str, Any] = {\n'
                 '        "duration": float(info.get("format", {}).get("duration", 0.0) or 0.0),\n'
                 '        "has_video": v is not None,\n'
                 '        "has_audio": a is not None,\n'
                 '    }\n'
                 '    if v:\n'
                 '        out["width"] = int(v.get("width", 0))\n'
                 '        out["height"] = int(v.get("height", 0))\n'
                 '        num, _, den = str(v.get("avg_frame_rate", "0/1")).partition("/")\n'
                 '        try:\n'
                 '            out["fps"] = round(float(num) / float(den or 1), 3)\n'
                 '        except ZeroDivisionError:\n'
                 '            out["fps"] = 0.0\n'
                 '        out["vcodec"] = v.get("codec_name")\n'
                 '        if v.get("nb_frames"):\n'
                 '            out["nb_frames"] = int(v["nb_frames"])\n'
                 '    if a:\n'
                 '        out["acodec"] = a.get("codec_name")\n'
                 '        out["sample_rate"] = int(a.get("sample_rate", 0) or 0)\n'
                 '        out["channels"] = int(a.get("channels", 0) or 0)\n'
                 '    return out\n'
                 '\n'
                 '\n'
                 'def new_run_id(date: Optional[_dt.date] = None, root: Optional[Path] = None) -> '
                 'str:\n'
                 '    date = date or _dt.date.today()\n'
                 '    root = root or runs_root()\n'
                 '    prefix = date.isoformat()\n'
                 '    nums = [int(m.group(2)) for d in root.iterdir() if (m := '
                 'RUN_ID_RE.match(d.name)) and m.group(1) == prefix]\n'
                 '    return f"{prefix}_KANG_{(max(nums) + 1 if nums else 1):03d}"\n'
                 '\n'
                 '\n'
                 'class Run:\n'
                 '    """One run = one folder with manifest.json, logs and all outputs."""\n'
                 '\n'
                 '    def __init__(self, run_id: Optional[str] = None, root: Optional[Path] = '
                 'None):\n'
                 '        self.root = root or runs_root()\n'
                 '        if run_id is None:\n'
                 '            # create atomically (retry on race)\n'
                 '            for _ in range(50):\n'
                 '                rid = new_run_id(root=self.root)\n'
                 '                try:\n'
                 '                    (self.root / rid).mkdir(parents=False, exist_ok=False)\n'
                 '                    run_id = rid\n'
                 '                    break\n'
                 '                except FileExistsError:\n'
                 '                    continue\n'
                 '            else:  # pragma: no cover\n'
                 '                raise RuntimeError("Could not allocate a unique run id")\n'
                 '        if not RUN_ID_RE.match(run_id):\n'
                 '            raise ValueError(f"Invalid run id: {run_id}")\n'
                 '        self.run_id = run_id\n'
                 '        self.dir = self.root / run_id\n'
                 '        self.dir.mkdir(parents=True, exist_ok=True)\n'
                 '        self.manifest_path = self.dir / "manifest.json"\n'
                 '        if self.manifest_path.is_file():\n'
                 '            self.manifest = '
                 'json.loads(self.manifest_path.read_text(encoding="utf-8"))\n'
                 '        else:\n'
                 '            self.manifest = {\n'
                 '                "run_id": run_id,\n'
                 '                "created_at": _dt.datetime.now().isoformat(timespec="seconds"),\n'
                 '                "inputs": [],\n'
                 '                "outputs": [],\n'
                 '                "steps": [],\n'
                 '                "warnings": [],\n'
                 '                "approvals": {},\n'
                 '                "status": "created",\n'
                 '            }\n'
                 '            self.save()\n'
                 '\n'
                 '    # ---- helpers -------------------------------------------------------\n'
                 '    def path(self, name: str) -> Path:\n'
                 '        p = (self.dir / name).resolve()\n'
                 '        if self.dir.resolve() not in p.parents and p != self.dir.resolve():\n'
                 '            raise ValueError(f"Refusing to write outside run folder: {name}")\n'
                 '        p.parent.mkdir(parents=True, exist_ok=True)\n'
                 '        return p\n'
                 '\n'
                 '    def add_input(self, path: str | Path, input_id: str = "A1") -> Dict[str, '
                 'Any]:\n'
                 '        p = Path(path).resolve()\n'
                 '        entry = {"id": input_id, "path": str(p), "sha256": sha256_file(p)}\n'
                 '        try:\n'
                 '            entry["media"] = media_summary(p)\n'
                 '        except Exception as exc:  # non-media inputs (json, png) are fine\n'
                 '            entry["media_error"] = str(exc)\n'
                 '        self.manifest["inputs"] = [i for i in self.manifest["inputs"] if i["id"] '
                 '!= input_id] + [entry]\n'
                 '        self.save()\n'
                 '        return entry\n'
                 '\n'
                 '    def add_output(self, path: str | Path, kind: str) -> Dict[str, Any]:\n'
                 '        p = Path(path).resolve()\n'
                 '        try:\n'
                 '            rel = str(p.relative_to(self.dir.resolve()))\n'
                 '        except ValueError:\n'
                 '            rel = str(p)\n'
                 '        entry = {"kind": kind, "path": rel, "sha256": sha256_file(p)}\n'
                 '        self.manifest["outputs"] = [o for o in self.manifest["outputs"] if '
                 'o["path"] != entry["path"]] + [entry]\n'
                 '        self.save()\n'
                 '        return entry\n'
                 '\n'
                 '    def log_step(self, name: str, **data: Any) -> None:\n'
                 '        self.manifest["steps"].append({"step": name, "at": '
                 '_dt.datetime.now().isoformat(timespec="seconds"), **data})\n'
                 '        self.save()\n'
                 '\n'
                 '    def warn(self, msg: str) -> None:\n'
                 '        self.manifest["warnings"].append(msg)\n'
                 '        self.save()\n'
                 '\n'
                 '    def set_approval(self, gate: str, value: str, plan_hash: Optional[str] = '
                 'None,\n'
                 '                     approver: Optional[str] = None) -> None:\n'
                 '        """manifest.approvals[gate] = value (old contract) + an append-only '
                 'record in approvals.json\n'
                 '        binding the decision to the exact plan it saw (sha256), who/what '
                 'approved it, and when."""\n'
                 '        if value not in {"pending", "approved", "rejected", "n/a"}:\n'
                 '            raise ValueError(value)\n'
                 '        self.manifest["approvals"][gate] = value\n'
                 '        if plan_hash is not None:\n'
                 '            p = self.path("approvals.json")\n'
                 '            recs = json.loads(p.read_text(encoding="utf-8")) if p.exists() else '
                 '[]\n'
                 '            recs.append({"gate": gate, "value": value, "plan_hash": plan_hash, '
                 '"approver": approver or "unknown",\n'
                 '                         "at": '
                 '_dt.datetime.now().astimezone().isoformat(timespec="seconds")})\n'
                 '            p.write_text(json.dumps(recs, ensure_ascii=False, indent=2), '
                 'encoding="utf-8")\n'
                 '        self.save()\n'
                 '\n'
                 '    def approval_valid(self, gate: str, plan_hash: str) -> bool:\n'
                 '        """True only if the LAST record for gate is \'approved\' for this exact '
                 'plan hash."""\n'
                 '        p = self.path("approvals.json")\n'
                 '        recs = [r for r in (json.loads(p.read_text(encoding="utf-8")) if '
                 'p.exists() else []) if r["gate"] == gate]\n'
                 '        return bool(recs) and recs[-1]["value"] == "approved" and '
                 'recs[-1]["plan_hash"] == plan_hash\n'
                 '\n'
                 '    def set_status(self, status: str) -> None:\n'
                 '        self.manifest["status"] = status\n'
                 '        self.save()\n'
                 '\n'
                 '    def save(self) -> None:\n'
                 '        tmp = self.manifest_path.with_suffix(".tmp")\n'
                 '        tmp.write_text(json.dumps(self.manifest, ensure_ascii=False, indent=2), '
                 'encoding="utf-8")\n'
                 '        tmp.replace(self.manifest_path)\n'
                 '\n'
                 '\n'
                 'def write_json(path: str | Path, data: Any) -> Path:\n'
                 '    p = Path(path)\n'
                 '    p.parent.mkdir(parents=True, exist_ok=True)\n'
                 '    p.write_text(json.dumps(data, ensure_ascii=False, indent=2), '
                 'encoding="utf-8")\n'
                 '    return p\n'
                 '\n'
                 '\n'
                 '_FF_MAJOR: Optional[int] = None\n'
                 '\n'
                 '\n'
                 'def ffmpeg_major() -> int:\n'
                 '    """Major version of the ffmpeg on PATH (git/nightly builds count as new). '
                 'Cached."""\n'
                 '    global _FF_MAJOR\n'
                 '    if _FF_MAJOR is None:\n'
                 '        try:\n'
                 '            out = subprocess.run(["ffmpeg", "-hide_banner", "-version"], '
                 'capture_output=True,\n'
                 '                                 text=True, encoding="utf-8", '
                 'errors="replace").stdout\n'
                 '            m = re.search(r"ffmpeg version n?(\\d+)\\.", out)\n'
                 '            _FF_MAJOR = int(m.group(1)) if m else 99\n'
                 '        except OSError:\n'
                 '            _FF_MAJOR = 99\n'
                 '    return _FF_MAJOR\n'
                 '\n'
                 '\n'
                 'def filter_script_args(script: Path) -> List[str]:\n'
                 '    """FFmpeg 9 removed -filter_complex_script; 7.x+ reads an option from a file '
                 'with -/filter_complex."""\n'
                 '    return ["-/filter_complex", str(script)] if ffmpeg_major() >= 7 else '
                 '["-filter_complex_script", str(script)]\n'
                 '\n'
                 '\n'
                 'def run_ffmpeg(args: List[str], log_path: Optional[Path] = None) -> None:\n'
                 '    cmd = ["ffmpeg", "-hide_banner", "-nostdin", "-y", *args]\n'
                 '    res = subprocess.run(cmd, capture_output=True, text=True, encoding="utf-8", '
                 'errors="replace")\n'
                 '    if log_path:\n'
                 '        with open(log_path, "a", encoding="utf-8") as f:\n'
                 '            f.write("$ " + " ".join(cmd) + "\\n" + res.stderr[-20000:] + "\\n")\n'
                 '    if res.returncode != 0:\n'
                 '        raise RuntimeError(f"ffmpeg failed ({res.returncode}): '
                 '{res.stderr.strip()[-2000:]}")\n',
 'core/scenes.py': '"""M3 Workflow G, step 1: accepted transcript -> scenes.json (scene plan for '
                   'Gate 1).\n'
                   '\n'
                   'A scene is a contiguous stretch of narration (~15-45 s) split at sentence ends '
                   '(। ? !) or pauses.\n'
                   'Scenes tile the timeline from 0 to the media duration with no gap and no '
                   'overlap, so every\n'
                   'spoken word belongs to exactly one scene. Visual fields are placeholders the '
                   'user fills or\n'
                   'approves; nothing here invents visuals.\n'
                   '"""\n'
                   'from __future__ import annotations\n'
                   '\n'
                   'from typing import Any, Dict, List, Optional\n'
                   '\n'
                   'SCHEMA = "kang.scenes/1.0"\n'
                   'SENT_END = ("।", "?", "!", "॥", ".")\n'
                   '\n'
                   '\n'
                   '# A scene must not start with a word that continues the previous phrase '
                   '("शुरुआत | से", "135 km | की दूरी")\n'
                   'CONTINUES = {"जो", "जिसे", "जिसका", "जिसकी", "जिसके", "जहाँ", "जहां", "से", '
                   '"की", "का", "के", "को", "में", "पर", "ने", "तक", "भी", "ही", "है", "हैं", '
                   '"था", "थी", "थे",\n'
                   '             "वाला", "वाली", "वाले", "लिए", "द्वारा", "और", "या", "तो", "कि", '
                   '"जो", "गया", "गई", "गए", "रहा",\n'
                   '             "रही", "रहे", "सकते", "सकता", "सकती", "जाता", "जाती", "जाते", '
                   '"करके", "बाद", "पहले", "साथ", "चाहिए", "होगा", "होगी", "होंगे", "पड़ेगा",\n'
                   '             "पड़ता", "ज्योतिर्लिंग", "मंदिर", "जी", "महादेव", "नगरी", '
                   '"दर्शन"}\n'
                   '\n'
                   '\n'
                   'def _sent_end(w: Dict[str, Any]) -> bool:\n'
                   '    return w["word"].rstrip().endswith(SENT_END)\n'
                   '\n'
                   '\n'
                   '# ...and must not end on a word that opens a phrase ("... में है जो | द्वारका '
                   'का")\n'
                   'DANGLING = {"जो", "कि", "और", "या", "तो", "के", "की", "का", "ने", "को", "एक", '
                   '"यह", "वह", "इस", "उस", "ये",\n'
                   '            "वो", "जिस", "जहाँ", "जहां", "आप", "हम", "मैं", "भगवान", "श्री", '
                   '"स्वामी", "संत", "माता", "बेट",\n'
                   '            "बहुत", "सबसे"}\n'
                   '\n'
                   '\n'
                   'def _clean(t: str) -> str:\n'
                   '    return t.strip().strip(",.।?!")\n'
                   '\n'
                   '\n'
                   'def _continues(words: List[Dict[str, Any]], i: int) -> bool:\n'
                   '    """True when words[i+1] grammatically continues words[i]\'s phrase (or '
                   'words[i] dangles)."""\n'
                   '    if words[i]["word"].rstrip().endswith((",",) + SENT_END):\n'
                   '        return i + 1 < len(words) and _clean(words[i + 1]["word"]) in {"से", '
                   '"की", "का", "के", "को", "में", "ने",\n'
                   '                                                                          '
                   '"जो", "जिसे", "जिसका", "जिसकी", "जिसके"}\n'
                   '    if i > 0 and _clean(words[i]["word"]) == "से" and _clean(words[i - '
                   '1]["word"]) == "में":\n'
                   '        return True  # "... में से | एक"\n'
                   '    if i + 2 < len(words) and _clean(words[i + 2]["word"]) in {"है", "हैं", '
                   '"था", "थी", "थे", "हो", "होता", "होती"}:\n'
                   '        return True  # "... भी | स्थापित है": next two words are this '
                   "clause's predicate\n"
                   '    return i + 1 < len(words) and (_clean(words[i + 1]["word"]) in CONTINUES '
                   'or _clean(words[i]["word"]) in DANGLING)\n'
                   '\n'
                   '\n'
                   'TOPIC_START = {"दोस्तो", "दोस्तों"}\n'
                   '\n'
                   '\n'
                   'def _topic_next(words: List[Dict[str, Any]], i: int) -> bool:\n'
                   '    """Narrator\'s topic opener ("दोस्तो ...") right after i: a natural scene '
                   'start."""\n'
                   '    return i + 1 < len(words) and _clean(words[i + 1]["word"]) in TOPIC_START\n'
                   '\n'
                   '\n'
                   'def _is_break(words: List[Dict[str, Any]], i: int, pause_s: float) -> bool:\n'
                   '    w = words[i]\n'
                   '    if _sent_end(w):\n'
                   '        return True\n'
                   '    if i + 1 >= len(words):\n'
                   '        return False\n'
                   '    if (_topic_next(words, i) and words[i + 1]["start"] - w["end"] >= 0.2\n'
                   '            and _clean(w["word"]) not in DANGLING):  # "... लगाते हैं, तो | '
                   'दोस्तों" still dangles\n'
                   '        return True\n'
                   '    if _continues(words, i):\n'
                   '        return False\n'
                   '    gap = words[i + 1]["start"] - w["end"]\n'
                   '    # a bare pause (no comma) needs to be long; a comma pause can be short\n'
                   '    return gap >= (pause_s if w["word"].rstrip().endswith(",") else '
                   'max(pause_s, 1.0))\n'
                   '\n'
                   '\n'
                   'def plan_scenes(words: List[Dict[str, Any]], duration: float, target_s: float '
                   '= 30.0,\n'
                   '                min_s: float = 15.0, max_s: float = 45.0, pause_s: float = '
                   '0.6) -> Dict[str, Any]:\n'
                   '    """Greedy: close a scene at the first break after target_s; if none by '
                   'max_s, close at the\n'
                   '    widest pause inside (min_s, max_s]. Boundaries sit mid-pause so no word is '
                   'ever split."""\n'
                   '    if not words:\n'
                   '        return {"schema": SCHEMA, "duration": round(duration, 3), "scenes": '
                   '[]}\n'
                   '    groups: List[List[int]] = []\n'
                   '    cur: List[int] = []\n'
                   '    seg_start = 0.0\n'
                   '    for i, w in enumerate(words):\n'
                   '        cur.append(i)\n'
                   '        length = w["end"] - seg_start\n'
                   '        last = i == len(words) - 1\n'
                   '        if last:\n'
                   '            break\n'
                   '        if length >= target_s and _is_break(words, i, pause_s):\n'
                   '            groups.append(cur)\n'
                   '            cur, seg_start = [], (w["end"] + words[i + 1]["start"]) / 2\n'
                   '        elif length >= max_s:\n'
                   '            # widest pause between words whose end is past min_s\n'
                   '            # best boundary inside (min_s, max_s]: sentence end > clean pause '
                   '> any pause; widest wins\n'
                   '            best, best_score = len(cur) - 1, -1.0\n'
                   '            for k in range(len(cur) - 1):\n'
                   '                a, b = words[cur[k]], words[cur[k + 1]]\n'
                   '                if a["end"] - seg_start < min_s:\n'
                   '                    continue\n'
                   '                gap = b["start"] - a["end"]\n'
                   '                length = a["end"] - seg_start\n'
                   '                score = (100.0 if _sent_end(a) else 20.0 if '
                   'a["word"].rstrip().endswith(",") else 0.0) \\\n'
                   '                    + (50.0 if _topic_next(words, cur[k]) else 0.0) \\\n'
                   '                    - (80.0 if _clean(a["word"]) in DANGLING else 0.0) \\\n'
                   '                    + (10.0 if not _continues(words, cur[k]) else -40.0) + 5.0 '
                   '* min(gap, 1.5) \\\n'
                   '                    - abs(length - target_s) / 5.0\n'
                   '                if score > best_score:\n'
                   '                    best, best_score = k, score\n'
                   '            groups.append(cur[: best + 1])\n'
                   '            j = cur[best]\n'
                   '            seg_start = (words[j]["end"] + words[j + 1]["start"]) / 2\n'
                   '            cur = cur[best + 1:]\n'
                   '    if cur:\n'
                   '        if groups and words[cur[-1]]["end"] - words[cur[0]]["start"] < min_s / '
                   '2:\n'
                   '            groups[-1].extend(cur)  # tiny tail joins the previous scene\n'
                   '        else:\n'
                   '            groups.append(cur)\n'
                   '    scenes = []\n'
                   '    for n, g in enumerate(groups):\n'
                   '        start = 0.0 if n == 0 else (words[groups[n - 1][-1]]["end"] + '
                   'words[g[0]]["start"]) / 2\n'
                   '        end = duration if n == len(groups) - 1 else (words[g[-1]]["end"] + '
                   'words[groups[n + 1][0]]["start"]) / 2\n'
                   '        text = " ".join(words[i]["word"] for i in g)\n'
                   '        scenes.append({\n'
                   '            "id": f"S{n + 1:02d}", "start": round(start, 3), "end": '
                   'round(max(end, start), 3),\n'
                   '            "first_word": g[0], "last_word": g[-1], "word_count": len(g),\n'
                   '            "title": " ".join(words[i]["word"] for i in g[:6]).rstrip("।?!,. '
                   '"),\n'
                   '            "text": text,\n'
                   '            "visual": {"type": "tbd", "brief": "", "source": "", "licence": '
                   '""},\n'
                   '            "status": "proposed",\n'
                   '        })\n'
                   '    return {"schema": SCHEMA, "duration": round(duration, 3), "scenes": '
                   'scenes}\n'
                   '\n'
                   '\n'
                   'def validate_scenes(plan: Dict[str, Any], n_words: Optional[int] = None, tol: '
                   'float = 0.002) -> List[str]:\n'
                   '    """Problems as plain strings; [] = valid. Tiling + word ownership + '
                   'required fields."""\n'
                   '    errs: List[str] = []\n'
                   '    if plan.get("schema") != SCHEMA:\n'
                   '        errs.append(f"schema must be {SCHEMA}")\n'
                   '    sc = plan.get("scenes") or []\n'
                   '    if not sc:\n'
                   '        return errs + ["no scenes"]\n'
                   '    if abs(sc[0]["start"]) > tol:\n'
                   '        errs.append(f"first scene starts at {sc[0][\'start\']} not 0")\n'
                   '    if abs(sc[-1]["end"] - plan["duration"]) > tol:\n'
                   '        errs.append(f"last scene ends at {sc[-1][\'end\']} not '
                   '{plan[\'duration\']}")\n'
                   '    seen = set()\n'
                   '    for a, b in zip(sc, sc[1:]):\n'
                   '        if abs(a["end"] - b["start"]) > tol:\n'
                   '            errs.append(f"{a[\'id\']}->{b[\'id\']}: gap/overlap {a[\'end\']} '
                   'vs {b[\'start\']}")\n'
                   '        if b["first_word"] != a["last_word"] + 1:\n'
                   '            errs.append(f"{a[\'id\']}->{b[\'id\']}: words not contiguous")\n'
                   '    for s in sc:\n'
                   '        if s["id"] in seen:\n'
                   '            errs.append(f"duplicate id {s[\'id\']}")\n'
                   '        seen.add(s["id"])\n'
                   '        if s["end"] <= s["start"]:\n'
                   '            errs.append(f"{s[\'id\']}: end <= start")\n'
                   '        for k in ("title", "text", "visual"):\n'
                   '            if k not in s:\n'
                   '                errs.append(f"{s[\'id\']}: missing {k}")\n'
                   '    if n_words is not None and (sc[0]["first_word"] != 0 or '
                   'sc[-1]["last_word"] != n_words - 1):\n'
                   '        errs.append("scenes do not cover every word")\n'
                   '    return errs\n'
                   '\n'
                   '\n'
                   'def scene_plan_text(plan: Dict[str, Any]) -> str:\n'
                   '    def ts(t: float) -> str:\n'
                   '        return f"{int(t // 60):02d}:{t % 60:04.1f}"\n'
                   '    lines = [f"{len(plan[\'scenes\'])} scenes, {plan[\'duration\']:.1f} s"]\n'
                   '    for s in plan["scenes"]:\n'
                   '        lines.append(f"{s[\'id\']}  {ts(s[\'start\'])}-{ts(s[\'end\'])}  '
                   '{s[\'word_count\']:3d} words  {s[\'title\']}")\n'
                   '    return "\\n".join(lines)\n',
 'core/script_style.py': '"""Caption script style for English words inside Hindi narration '
                         '(display layer only).\n'
                         '\n'
                         'Modes:\n'
                         "  asr        -> captions exactly as the transcript (mixed: 'railway "
                         "station' + 'थ्री डेज').  DEFAULT\n"
                         '  latin      -> listed Devanagari-English words/phrases shown in Latin '
                         "('three days और two night')\n"
                         "  devanagari -> listed Latin words shown in Devanagari ('रेलवे स्टेशन')\n"
                         '\n'
                         'Only entries in config/english_script.json are changed; anything '
                         'unlisted stays as transcribed and\n'
                         'is reported in `unmapped`. Replacement is word-for-word (same count), so '
                         'every word keeps its own\n'
                         'start/end timing. words.json / the transcript are NEVER modified: this '
                         'only feeds captions.\n'
                         '"""\n'
                         'from __future__ import annotations\n'
                         '\n'
                         'import copy\n'
                         'import json\n'
                         'import re\n'
                         'from pathlib import Path\n'
                         'from typing import Any, Dict, List, Tuple\n'
                         '\n'
                         'MODES = ("asr", "latin", "devanagari")\n'
                         '_PUNCT = re.compile(r"^(.*?)([,.!?।;:\\"\'\\u2019\\u201d]*)$", re.S)\n'
                         '_LATIN = re.compile(r"[A-Za-z]")\n'
                         '\n'
                         '\n'
                         'def load_table(path: str | Path) -> Dict[str, Any]:\n'
                         '    data = json.loads(Path(path).read_text(encoding="utf-8"))\n'
                         '    for key in ("to_devanagari", "to_latin_phrases"):\n'
                         '        data.setdefault(key, {} if key == "to_devanagari" else [])\n'
                         '    for ph in data["to_latin_phrases"]:\n'
                         '        if len(ph["dev"].split()) != len(ph["lat"].split()):\n'
                         '            raise ValueError(f"english_script phrase must keep word '
                         'count: {ph}")\n'
                         '    return data\n'
                         '\n'
                         '\n'
                         'def _split(tok: str) -> Tuple[str, str]:\n'
                         '    m = _PUNCT.match(tok)\n'
                         '    return (m.group(1), m.group(2)) if m else (tok, "")\n'
                         '\n'
                         '\n'
                         'def apply(words: List[Dict[str, Any]], mode: str, table: Dict[str, Any]) '
                         '-> Tuple[List[Dict[str, Any]], Dict[str, Any]]:\n'
                         '    if mode not in MODES:\n'
                         '        raise ValueError(f"english script mode must be one of {MODES}, '
                         'got {mode!r}")\n'
                         '    out = copy.deepcopy(words)\n'
                         '    changes: List[Dict[str, Any]] = []\n'
                         '    if mode == "devanagari":\n'
                         '        tab = {k.lower(): v for k, v in table["to_devanagari"].items()}\n'
                         '        for i, w in enumerate(out):\n'
                         '            core, p = _split(w["word"])\n'
                         '            new = tab.get(core.lower())\n'
                         '            if new:\n'
                         '                changes.append({"i": i, "from": w["word"], "to": new + '
                         'p})\n'
                         '                w["word"] = new + p\n'
                         '    elif mode == "latin":\n'
                         '        phrases = sorted(table["to_latin_phrases"], key=lambda ph: '
                         '-len(ph["dev"].split()))\n'
                         '        i = 0\n'
                         '        while i < len(out):\n'
                         '            hit = None\n'
                         '            for ph in phrases:\n'
                         '                dev = ph["dev"].split()\n'
                         '                seg = [_split(out[j]["word"])[0] for j in range(i, min(i '
                         '+ len(dev), len(out)))]\n'
                         '                if seg == dev:\n'
                         '                    hit = ph\n'
                         '                    break\n'
                         '            if not hit:\n'
                         '                i += 1\n'
                         '                continue\n'
                         '            for k, lat in enumerate(hit["lat"].split()):\n'
                         '                core, p = _split(out[i + k]["word"])\n'
                         '                changes.append({"i": i + k, "from": out[i + k]["word"], '
                         '"to": lat + p})\n'
                         '                out[i + k]["word"] = lat + p\n'
                         '            i += len(hit["dev"].split())\n'
                         '    unmapped = []\n'
                         '    if mode == "devanagari":\n'
                         '        unmapped = sorted({_split(w["word"])[0] for w in out if '
                         '_LATIN.search(w["word"])})\n'
                         '    return out, {"mode": mode, "changed": len(changes), "changes": '
                         'changes, "unmapped": unmapped}\n',
 'core/second_opinion.py': '"""Second-opinion ASR hints (e.g. Nemotron 3.5 ASR) for the human '
                           'transcript review.\n'
                           '\n'
                           "HINTS ONLY: nothing here changes words.json. The second model's text "
                           'is aligned to our words chunk by\n'
                           'chunk (difflib on normalized tokens), so each review row gets '
                           '`nemotron_hint` (what the other model\n'
                           'heard for that span) and `nemotron_agrees` (yes/no after '
                           'normalization). `disagreements()` lists every\n'
                           'other place the two models differ, classed content / spelling / '
                           'spacing / number / script. `number` is NOT\n'
                           'formatting-only: digits vs number words can hide a real amount error '
                           '(2,500,000 vs दो हजार से ढाई हजार),\n'
                           'so number rows always need a listen; only `script` (Latin vs '
                           'Devanagari) is usually cosmetic. Neither model is ground truth; a '
                           'human still decides.\n'
                           '"""\n'
                           'from __future__ import annotations\n'
                           '\n'
                           'import csv\n'
                           'import difflib\n'
                           'import re\n'
                           'import unicodedata\n'
                           'from pathlib import Path\n'
                           'from typing import Any, Dict, List, Tuple\n'
                           '\n'
                           '_P = '
                           're.compile(r"[\\u0964\\u0965.,!?;:\\"\'()\\[\\]\\u2018\\u2019\\u201c\\u201d\\-\\u2013\\u2014]")\n'
                           '_DIGIT = re.compile(r"\\d")\n'
                           '_LATIN = re.compile(r"[A-Za-z]")\n'
                           '_NUMWORDS = {"एक", "दो", "तीन", "चार", "पांच", "पाँच", "छह", "छे", '
                           '"सात", "आठ", "नौ", "दस", "बारह", "सत्रह",\n'
                           '             "बाईस", "पच्चीस", "बावन", "सौ", "हजार", "हज़ार", "लाख", '
                           '"ढाई", "डेढ़", "पंद्रह", "पैंतीस",\n'
                           '             "चालीस", "सवा", "साढ़े"}\n'
                           '\n'
                           '\n'
                           'def norm(tok: str) -> str:\n'
                           '    t = unicodedata.normalize("NFC", tok.lower())\n'
                           '    t = _P.sub("", t).replace("\\u093c", "").replace("\\u0901", '
                           '"\\u0902")\n'
                           '    return t\n'
                           '\n'
                           '\n'
                           'def _toks(text: str) -> List[str]:\n'
                           '    return [t for t in text.split() if norm(t)]\n'
                           '\n'
                           '\n'
                           'def align(words: List[Dict[str, Any]], chunks: List[Dict[str, Any]]) '
                           '-> Dict[str, Any]:\n'
                           '    """Map each of our word indices to the other model\'s surface '
                           'tokens."""\n'
                           '    mapped: Dict[int, List[str]] = {}\n'
                           '    ops: List[Dict[str, Any]] = []\n'
                           '    for c in chunks:\n'
                           '        idx = [i for i, w in enumerate(words) if c["start"] <= '
                           '(w["start"] + w["end"]) / 2 < c["end"]]\n'
                           '        ours = [norm(words[i]["word"]) for i in idx]\n'
                           '        theirs_s = _toks(c.get("text", ""))\n'
                           '        theirs = [norm(t) for t in theirs_s]\n'
                           '        sm = difflib.SequenceMatcher(None, ours, theirs, '
                           'autojunk=False)\n'
                           '        for tag, a0, a1, b0, b1 in sm.get_opcodes():\n'
                           '            if tag == "equal":\n'
                           '                for k in range(a1 - a0):\n'
                           '                    mapped[idx[a0 + k]] = [theirs_s[b0 + k]]\n'
                           '                continue\n'
                           '            seg = theirs_s[b0:b1]\n'
                           '            if a1 > a0:\n'
                           '                mapped[idx[a0]] = list(seg)\n'
                           '                for k in range(a0 + 1, a1):\n'
                           '                    mapped[idx[k]] = []\n'
                           '                first, last = idx[a0], idx[a1 - 1]\n'
                           '            else:  # insertion: attach to the previous word of ours '
                           '(or next if none)\n'
                           '                anchor = idx[a0 - 1] if a0 > 0 else (idx[0] if idx '
                           'else None)\n'
                           '                if anchor is None:\n'
                           '                    continue\n'
                           '                mapped.setdefault(anchor, []).extend(seg)\n'
                           '                first = last = anchor\n'
                           '            ops.append({"tag": tag, "first": first, "last": last, '
                           '"start": words[first]["start"],\n'
                           '                        "ours": " ".join(words[i]["word"] for i in '
                           'idx[a0:a1]), "theirs": " ".join(seg)})\n'
                           '    return {"mapped": mapped, "ops": ops}\n'
                           '\n'
                           '\n'
                           'def hint(al: Dict[str, Any], first: int, last: int) -> str:\n'
                           '    return " ".join(t for i in range(first, last + 1) for t in '
                           'al["mapped"].get(i, []))\n'
                           '\n'
                           '\n'
                           'def classify(ours: str, theirs: str) -> str:\n'
                           '    o, t = ours.split(), theirs.split()\n'
                           '    if o and t and norm("".join(o)) == norm("".join(t)):\n'
                           '        return "spacing"\n'
                           '    if _DIGIT.search(ours) and (any(norm(x) in _NUMWORDS for x in t) '
                           'or _DIGIT.search(theirs)):\n'
                           '        return "number"\n'
                           '    if _LATIN.search(ours) and not _LATIN.search(theirs):\n'
                           '        return "script"\n'
                           '    if o and t and len(o) == len(t) and '
                           'all(difflib.SequenceMatcher(None, norm(a), norm(b)).ratio() >= 0.75\n'
                           '                                            for a, b in zip(o, t)):\n'
                           '        return "spelling"\n'
                           '    return "content"\n'
                           '\n'
                           '\n'
                           'def disagreements(al: Dict[str, Any]) -> List[Dict[str, Any]]:\n'
                           '    return [dict(op, kind=classify(op["ours"], op["theirs"])) for op '
                           'in al["ops"]]\n'
                           '\n'
                           '\n'
                           'def annotate_rows(rows: List[Dict[str, str]], al: Dict[str, Any], '
                           'words: List[Dict[str, Any]]) -> List[Dict[str, str]]:\n'
                           '    out = []\n'
                           '    for r in rows:\n'
                           '        r = dict(r)\n'
                           '        f, l = int(r["first"]), int(r["last"])\n'
                           '        h = hint(al, f, l)\n'
                           '        ours = " ".join(words[i]["word"] for i in range(f, l + 1))\n'
                           '        r["nemotron_hint"] = h\n'
                           '        same = [norm(x) for x in ours.split()] == [norm(x) for x in '
                           'h.split()]\n'
                           '        r["nemotron_agrees"] = "yes" if same else "no"\n'
                           '        r["nemotron_diff"] = "" if same else classify(ours, h)\n'
                           '        out.append(r)\n'
                           '    return out\n'
                           '\n'
                           '\n'
                           'def write_csv(path: str | Path, rows: List[Dict[str, str]]) -> Path:\n'
                           '    path = Path(path)\n'
                           '    fields = list(rows[0].keys()) if rows else []\n'
                           '    with path.open("w", encoding="utf-8-sig", newline="") as fh:\n'
                           '        w = csv.DictWriter(fh, fieldnames=fields)\n'
                           '        w.writeheader()\n'
                           '        w.writerows(rows)\n'
                           '    return path\n',
 'core/spelling_policy.py': '"""Per-video / per-topic caption spelling policy (opt-in, never '
                            'global).\n'
                            '\n'
                            'A policy file (e.g. config/caption_spelling_policy.json) is an '
                            'EDITORIAL choice of display\n'
                            'spelling between valid variants (द्वारका vs द्वारिका). It is not an '
                            'ASR fix and not a claim\n'
                            'that anything was listened to. It runs only when a policy is passed '
                            'explicitly, after the\n'
                            'glossary, and every change is logged (raw ASR stays in words_raw.json '
                            '/ raw/).\n'
                            '\n'
                            'Format: {"video": "...", "replace": {"द्वारका": "द्वारिका", ...}, '
                            '"never": {"रुक्मिणी": "रुक्मणी"}}\n'
                            'Replacement is whole-token only (a glossary-merged entry like "बेट '
                            'द्वारका" is matched token by\n'
                            'token), longest key first, timing untouched, word count unchanged.\n'
                            '"""\n'
                            'from __future__ import annotations\n'
                            '\n'
                            'import json\n'
                            'import re\n'
                            'from pathlib import Path\n'
                            'from typing import Any, Dict, List, Tuple\n'
                            '\n'
                            '_TRAIL = re.compile(r"^(.*?)([\\s,.!?।॥\\"\'”’…:;-]*)$", re.S)\n'
                            '\n'
                            '\n'
                            'def load_policy(path: str | Path) -> Dict[str, Any]:\n'
                            '    data = json.loads(Path(path).read_text(encoding="utf-8"))\n'
                            '    rep = data.get("replace")\n'
                            '    if not isinstance(rep, dict) or not rep:\n'
                            '        raise ValueError(f"spelling policy {path}: \'replace\' map '
                            'missing or empty")\n'
                            '    for src, dst in rep.items():\n'
                            '        for bad_src, bad_dst in (data.get("never") or {}).items():\n'
                            '            if dst == bad_dst or (src == bad_src and dst == '
                            'bad_dst):\n'
                            '                raise ValueError(f"spelling policy {path}: replace '
                            '{src}->{dst} breaks never {bad_src}->{bad_dst}")\n'
                            '    data["_path"] = str(path)\n'
                            '    return data\n'
                            '\n'
                            '\n'
                            'def _apply_text(text: str, rules: List[Tuple[List[str], str]]) -> '
                            'str:\n'
                            '    toks = text.split(" ")\n'
                            '    out, i = [], 0\n'
                            '    while i < len(toks):\n'
                            '        for src, dst in rules:\n'
                            '            n = len(src)\n'
                            '            seg = toks[i:i + n]\n'
                            '            if len(seg) == n:\n'
                            '                core = [_TRAIL.match(t).group(1) for t in seg]\n'
                            '                if core == src:\n'
                            '                    tail = _TRAIL.match(seg[-1]).group(2)\n'
                            '                    out.append(dst + tail)\n'
                            '                    i += n\n'
                            '                    break\n'
                            '        else:\n'
                            '            out.append(toks[i])\n'
                            '            i += 1\n'
                            '    return " ".join(out)\n'
                            '\n'
                            '\n'
                            'def apply_policy(words: List[Dict[str, Any]], policy: Dict[str, Any]) '
                            '-> Tuple[List[Dict[str, Any]], List[Dict[str, Any]]]:\n'
                            '    """Returns (new_words, changes). Input never mutated; one change '
                            'entry per word changed."""\n'
                            '    rules = sorted(((k.split(), v) for k, v in '
                            'policy["replace"].items()), key=lambda r: -len(r[0]))\n'
                            '    out, changes = [], []\n'
                            '    for i, w in enumerate(words):\n'
                            '        new = _apply_text(w["word"], rules)\n'
                            '        if new != w["word"]:\n'
                            '            changes.append({"index": i, "from": w["word"], "to": new, '
                            '"start": w.get("start"),\n'
                            '                            "end": w.get("end"), "decided_by": '
                            '"editorial-policy",\n'
                            '                            "policy": policy.get("video") or '
                            'policy.get("_path")})\n'
                            '            w = {**w, "word": new}\n'
                            '        out.append(w)\n'
                            '    return out, changes\n',
 'core/style.py': '"""F19-lite Screenshot style cloner (Phase 3): K-Means palette -> '
                  'brand_draft.json.\n'
                  '\n'
                  'Output is a DRAFT. It is merged into brand.json only after user approval.\n'
                  "Label: 'close, identical nahi'.\n"
                  '"""\n'
                  'from __future__ import annotations\n'
                  '\n'
                  'from pathlib import Path\n'
                  'from typing import Any, Dict, List, Tuple\n'
                  '\n'
                  'import numpy as np\n'
                  '\n'
                  '\n'
                  'def _hex(rgb) -> str:\n'
                  '    return "#{:02X}{:02X}{:02X}".format(*[int(round(c)) for c in rgb])\n'
                  '\n'
                  '\n'
                  'def rel_luminance(hex_color: str) -> float:\n'
                  '    h = hex_color.lstrip("#")\n'
                  '    rgb = [int(h[i:i + 2], 16) / 255 for i in (0, 2, 4)]\n'
                  '    lin = [c / 12.92 if c <= 0.03928 else ((c + 0.055) / 1.055) ** 2.4 for c in '
                  'rgb]\n'
                  '    return 0.2126 * lin[0] + 0.7152 * lin[1] + 0.0722 * lin[2]\n'
                  '\n'
                  '\n'
                  'def contrast_ratio(a: str, b: str) -> float:\n'
                  '    """WCAG 2.x contrast ratio."""\n'
                  '    la, lb = sorted((rel_luminance(a), rel_luminance(b)), reverse=True)\n'
                  '    return round((la + 0.05) / (lb + 0.05), 2)\n'
                  '\n'
                  '\n'
                  'def kmeans(pixels: np.ndarray, k: int = 5, iters: int = 25, seed: int = 0) -> '
                  'Tuple[np.ndarray, np.ndarray]:\n'
                  '    """Deterministic k-means (k-means++ init, fixed seed)."""\n'
                  '    rng = np.random.default_rng(seed)\n'
                  '    X = pixels.astype(np.float64)\n'
                  '    centers = [X[rng.integers(len(X))]]\n'
                  '    for _ in range(1, k):\n'
                  '        d = np.min(((X[:, None, :] - np.array(centers)[None]) ** 2).sum(-1), '
                  'axis=1)\n'
                  '        if d.sum() == 0:\n'
                  '            centers.append(X[rng.integers(len(X))])\n'
                  '        else:\n'
                  '            centers.append(X[rng.choice(len(X), p=d / d.sum())])\n'
                  '    C = np.array(centers)\n'
                  '    for _ in range(iters):\n'
                  '        labels = np.argmin(((X[:, None, :] - C[None]) ** 2).sum(-1), axis=1)\n'
                  '        newC = np.array([X[labels == j].mean(0) if np.any(labels == j) else '
                  'C[j] for j in range(k)])\n'
                  '        if np.allclose(newC, C):\n'
                  '            break\n'
                  '        C = newC\n'
                  '    labels = np.argmin(((X[:, None, :] - C[None]) ** 2).sum(-1), axis=1)\n'
                  '    return C, np.bincount(labels, minlength=k)\n'
                  '\n'
                  '\n'
                  'def _saturation(rgb) -> float:\n'
                  '    mx, mn = max(rgb) / 255, min(rgb) / 255\n'
                  '    return 0.0 if mx == 0 else (mx - mn) / mx\n'
                  '\n'
                  '\n'
                  'def extract_palette(image_path: str | Path, k: int = 5, sample: int = 20000) -> '
                  'List[Dict[str, Any]]:\n'
                  '    from PIL import Image\n'
                  '    img = Image.open(image_path).convert("RGB")\n'
                  '    img.thumbnail((400, 400))\n'
                  '    px = np.asarray(img).reshape(-1, 3)\n'
                  '    if len(px) > sample:\n'
                  '        px = px[np.random.default_rng(0).choice(len(px), sample, '
                  'replace=False)]\n'
                  '    C, counts = kmeans(px, k=min(k, len(np.unique(px, axis=0))))\n'
                  '    total = counts.sum()\n'
                  '    pal = [{"hex": _hex(c), "share": round(float(n) / total, 4), "saturation": '
                  'round(_saturation(c), 3),\n'
                  '            "luminance": round(rel_luminance(_hex(c)), 4)} for c, n in zip(C, '
                  'counts) if n > 0]\n'
                  '    return sorted(pal, key=lambda p: -p["share"])\n'
                  '\n'
                  '\n'
                  'def palette_to_brand_draft(palette: List[Dict[str, Any]], client: str = '
                  '"draft") -> Dict[str, Any]:\n'
                  '    """background = most common; accent = most saturated of the rest; highlight '
                  '= the\n'
                  '    remaining color with best contrast against background. Text = white/black '
                  'by contrast."""\n'
                  '    bg = palette[0]["hex"]\n'
                  '    rest = palette[1:] or palette\n'
                  '    accent = max(rest, key=lambda p: (p["saturation"], p["share"]))["hex"]\n'
                  '    others = [p for p in rest if p["hex"] != accent] or rest\n'
                  '    highlight = max(others, key=lambda p: contrast_ratio(p["hex"], bg))["hex"]\n'
                  '    text = "#FFFFFF" if contrast_ratio("#FFFFFF", bg) >= '
                  'contrast_ratio("#000000", bg) else "#000000"\n'
                  '    checks = {\n'
                  '        "text_vs_background": contrast_ratio(text, bg),\n'
                  '        "highlight_vs_background": contrast_ratio(highlight, bg),\n'
                  '        "accent_vs_background": contrast_ratio(accent, bg),\n'
                  '    }\n'
                  '    return {\n'
                  '        "status": "draft (approve before merging into brand.json)",\n'
                  '        "label": "close, identical nahi",\n'
                  '        "client": client,\n'
                  '        "colors": {"primary": bg, "accent": accent, "highlight": highlight, '
                  '"text": text,\n'
                  '                   "outline": "#000000" if text == "#FFFFFF" else "#FFFFFF"},\n'
                  '        "palette": palette,\n'
                  '        "contrast": checks,\n'
                  '        "contrast_ok": checks["text_vs_background"] >= 4.5,\n'
                  '    }\n'
                  '\n'
                  '\n'
                  'def merge_brand_draft(brand: Dict[str, Any], draft: Dict[str, Any]) -> '
                  'Dict[str, Any]:\n'
                  '    """Return a new brand dict with draft colors (only after user '
                  'approval)."""\n'
                  '    out = dict(brand)\n'
                  '    out["colors"] = {**brand.get("colors", {}), **draft["colors"]}\n'
                  '    return out\n',
 'core/subtitles.py': '"""F01 Animated subtitles (word highlight) + F02 subtitle presets.\n'
                      '\n'
                      'Option A (Phase 2): one ASS event per chunk with \\\\kf karaoke tags.\n'
                      '    SecondaryColour = text (white, not yet spoken), PrimaryColour = '
                      'highlight.\n'
                      'Option B (Phase 3): one ASS event per word; the active word gets highlight '
                      'colour\n'
                      '    plus a \\\\t(\\\\fscx108\\\\fscy108) scale pop.\n'
                      '"""\n'
                      'from __future__ import annotations\n'
                      '\n'
                      'import json\n'
                      'import re\n'
                      'from pathlib import Path\n'
                      'from typing import Any, Dict, List, Optional, Tuple\n'
                      '\n'
                      'from pydantic import BaseModel, Field\n'
                      '\n'
                      'from .brand import CANVAS, Brand, font_family_name, hex_to_ass\n'
                      '\n'
                      'SENTENCE_END = "।?!.॥"\n'
                      'SOFT_BREAK = ",;:"\n'
                      '\n'
                      '\n'
                      '# '
                      '--------------------------------------------------------------------------- '
                      'F02\n'
                      'class Preset(BaseModel):\n'
                      '    name: str\n'
                      '    description: str = ""\n'
                      '    font_role: str = Field("hindi", pattern="^(heading|body|hindi)$")\n'
                      '    size_pct: float = Field(gt=1, le=15, description="font size as % of '
                      'canvas height")\n'
                      '    outline_px: float = Field(6, ge=0, le=20)\n'
                      '    shadow_px: float = Field(2, ge=0, le=20)\n'
                      '    uppercase: bool = False\n'
                      '    bold: bool = True\n'
                      '    box: bool = False\n'
                      '    max_chars: int = Field(32, ge=8, le=80)\n'
                      '    max_words: int = Field(4, ge=1, le=10)\n'
                      '    min_words: int = Field(2, ge=1, le=10)\n'
                      '    max_dur: float = Field(2.0, gt=0.3, le=8)\n'
                      '    pause_break: float = Field(0.35, ge=0.05, le=3)\n'
                      '    highlight: str = Field("kf", pattern="^(kf|pop)$")\n'
                      '    position: str = Field("bottom", pattern="^(bottom|middle)$")\n'
                      '\n'
                      '\n'
                      'def load_preset(name_or_path: str, presets_dir: str | Path = "presets") -> '
                      'Preset:\n'
                      '    p = Path(name_or_path)\n'
                      '    if not p.suffix:\n'
                      '        p = Path(presets_dir) / f"{name_or_path}.json"\n'
                      '    return '
                      'Preset.model_validate(json.loads(p.read_text(encoding="utf-8")))\n'
                      '\n'
                      '\n'
                      'def list_presets(presets_dir: str | Path = "presets") -> List[str]:\n'
                      '    return sorted(p.stem for p in Path(presets_dir).glob("*.json"))\n'
                      '\n'
                      '\n'
                      '# '
                      '--------------------------------------------------------------------------- '
                      'chunking\n'
                      'def _text_len(words: List[Dict[str, Any]]) -> int:\n'
                      '    return len(" ".join(w["word"] for w in words))\n'
                      '\n'
                      '\n'
                      'def chunk_words(words: List[Dict[str, Any]], max_words: int = 4, min_words: '
                      'int = 2,\n'
                      '                max_chars: int = 32, max_dur: float = 2.0, pause_break: '
                      'float = 0.35,\n'
                      '                fits=None) -> List[List[Dict[str, Any]]]:\n'
                      '    """Split words into 2-4 word chunks.\n'
                      '\n'
                      '    Break when: adding the word exceeds max_words/max_chars/max_dur, a '
                      'pause > pause_break\n'
                      '    precedes the word, or the previous word ends a sentence. `fits(text)` '
                      '(optional) is a\n'
                      '    pixel-width check; a chunk that would not fit the safe width is broken '
                      'early.\n'
                      '    """\n'
                      '    chunks: List[List[Dict[str, Any]]] = []\n'
                      '    cur: List[Dict[str, Any]] = []\n'
                      '    for w in words:\n'
                      '        if cur:\n'
                      '            prev = cur[-1]\n'
                      '            pause = w["start"] - prev["end"]\n'
                      '            cand = cur + [w]\n'
                      '            too_many = len(cand) > max_words\n'
                      '            too_long = _text_len(cand) > max_chars\n'
                      '            too_slow = (w["end"] - cur[0]["start"]) > max_dur\n'
                      '            no_fit = fits is not None and not fits(" ".join(x["word"] for x '
                      'in cand))\n'
                      '            sentence = prev["word"][-1:] in SENTENCE_END\n'
                      '            soft = prev["word"][-1:] in SOFT_BREAK and len(cur) >= '
                      'min_words\n'
                      '            if too_many or too_long or too_slow or no_fit or pause > '
                      'pause_break or sentence or soft:\n'
                      '                chunks.append(cur)\n'
                      '                cur = []\n'
                      '        cur.append(w)\n'
                      '    if cur:\n'
                      '        chunks.append(cur)\n'
                      '    return chunks\n'
                      '\n'
                      '\n'
                      'def chunk_timings(chunks: List[List[Dict[str, Any]]], merge_gap: float = '
                      '0.08) -> List[Tuple[float, float]]:\n'
                      '    """Event start/end per chunk. Gap < merge_gap -> previous end extended '
                      '(no flicker);\n'
                      '    overlap -> previous end trimmed to next start."""\n'
                      '    times = [(c[0]["start"], c[-1]["end"]) for c in chunks]\n'
                      '    out: List[Tuple[float, float]] = []\n'
                      '    for i, (s, e) in enumerate(times):\n'
                      '        if i + 1 < len(times):\n'
                      '            ns = times[i + 1][0]\n'
                      '            if e > ns:            # overlap -> trim\n'
                      '                e = ns\n'
                      '            elif ns - e < merge_gap:  # tiny gap -> merge\n'
                      '                e = ns\n'
                      '        out.append((round(s, 3), round(max(e, s + 0.01), 3)))\n'
                      '    return out\n'
                      '\n'
                      '\n'
                      '# '
                      '--------------------------------------------------------------------------- '
                      'ASS\n'
                      'def ass_time(t: float) -> str:\n'
                      '    cs = int(round(max(t, 0) * 100))\n'
                      '    h, cs = divmod(cs, 360000)\n'
                      '    m, cs = divmod(cs, 6000)\n'
                      '    s, cs = divmod(cs, 100)\n'
                      '    return f"{h}:{m:02d}:{s:02d}.{cs:02d}"\n'
                      '\n'
                      '\n'
                      'def _escape(text: str) -> str:\n'
                      '    return text.replace("\\\\", "\\\\\\\\").replace("{", "(").replace("}", '
                      '")").replace("\\n", " ")\n'
                      '\n'
                      '\n'
                      'class Layout(BaseModel):\n'
                      '    fmt: str\n'
                      '    width: int\n'
                      '    height: int\n'
                      '    font_size: int\n'
                      '    margin_lr: int\n'
                      '    margin_v: int\n'
                      '    font_family: str\n'
                      '\n'
                      '\n'
                      'def make_layout(brand: Brand, preset: Preset, fmt: str, font_path: str | '
                      'Path) -> Layout:\n'
                      '    w, h = CANVAS[fmt]\n'
                      '    size = int(round(h * preset.size_pct / 100))\n'
                      '    if fmt == "16:9":\n'
                      '        size = int(round(size * 1.35))  # landscape: height is small, keep '
                      'legible\n'
                      '    margin_v = brand.subtitle_bottom_for(fmt) if preset.position == '
                      '"bottom" else int(h * 0.42)\n'
                      '    return Layout(fmt=fmt, width=w, height=h, font_size=size, '
                      'margin_lr=brand.margin_for(fmt),\n'
                      '                  margin_v=margin_v, '
                      'font_family=font_family_name(font_path))\n'
                      '\n'
                      '\n'
                      'def make_fit_checker(layout: Layout, font_path: str | Path, outline_px: '
                      'float):\n'
                      '    """Pixel-width check with the real font (HarfBuzz shaping via Pillow '
                      'raqm when available)."""\n'
                      '    from PIL import ImageFont\n'
                      '    font = ImageFont.truetype(str(font_path), layout.font_size)\n'
                      '    max_w = layout.width - 2 * layout.margin_lr - 2 * outline_px\n'
                      '\n'
                      '    def fits(text: str) -> bool:\n'
                      '        l, t, r, b = font.getbbox(text)\n'
                      '        return (r - l) <= max_w * 1.08  # libass scale-pop headroom handled '
                      'by 108% factor\n'
                      '    return fits\n'
                      '\n'
                      '\n'
                      'def build_ass(words: List[Dict[str, Any]], brand: Brand, preset: Preset, '
                      'layout: Layout,\n'
                      '              font_path: Optional[str | Path] = None, option: Optional[str] '
                      '= None) -> Tuple[str, List[Dict[str, Any]]]:\n'
                      '    """Return (ASS text, events). events = [{start,end,text,words:[...]}] '
                      'for QA."""\n'
                      '    option = option or ("B" if preset.highlight == "pop" else "A")\n'
                      '    c = brand.colors\n'
                      '    primary = hex_to_ass(c.highlight)\n'
                      '    secondary = hex_to_ass(c.text)\n'
                      '    outline_col = hex_to_ass(c.outline)\n'
                      '    back = hex_to_ass(c.primary, alpha=0x40) if preset.box else '
                      'hex_to_ass("#000000", alpha=0x80)\n'
                      '    border_style = 3 if preset.box else 1\n'
                      '    bold = -1 if preset.bold else 0\n'
                      '    align = 2  # bottom-center\n'
                      '\n'
                      '    if option == "A":\n'
                      '        style_primary, style_secondary = primary, secondary\n'
                      '    else:\n'
                      '        style_primary, style_secondary = secondary, secondary\n'
                      '\n'
                      '    header = f"""[Script Info]\n'
                      '; Generated by Kang AI (F01/F02) preset={preset.name} option={option}\n'
                      'ScriptType: v4.00+\n'
                      'PlayResX: {layout.width}\n'
                      'PlayResY: {layout.height}\n'
                      'WrapStyle: 2\n'
                      'ScaledBorderAndShadow: yes\n'
                      'YCbCr Matrix: TV.709\n'
                      '\n'
                      '[V4+ Styles]\n'
                      'Format: Name, Fontname, Fontsize, PrimaryColour, SecondaryColour, '
                      'OutlineColour, BackColour, Bold, Italic, Underline, StrikeOut, ScaleX, '
                      'ScaleY, Spacing, Angle, BorderStyle, Outline, Shadow, Alignment, MarginL, '
                      'MarginR, MarginV, Encoding\n'
                      'Style: '
                      'Kang,{layout.font_family},{layout.font_size},{style_primary},{style_secondary},{outline_col},{back},{bold},0,0,0,100,100,0,0,{border_style},{preset.outline_px},{preset.shadow_px},{align},{layout.margin_lr},{layout.margin_lr},{layout.margin_v},1\n'
                      '\n'
                      '[Events]\n'
                      'Format: Layer, Start, End, Style, Name, MarginL, MarginR, MarginV, Effect, '
                      'Text\n'
                      '"""\n'
                      '    fits = make_fit_checker(layout, font_path, preset.outline_px) if '
                      'font_path else None\n'
                      '    chunks = chunk_words(words, preset.max_words, preset.min_words, '
                      'preset.max_chars, preset.max_dur,\n'
                      '                         preset.pause_break, fits=fits)\n'
                      '    timings = chunk_timings(chunks)\n'
                      '    lines: List[str] = []\n'
                      '    events: List[Dict[str, Any]] = []\n'
                      '\n'
                      '    def disp(w: str) -> str:\n'
                      '        return _escape(w.upper() if preset.uppercase else w)\n'
                      '\n'
                      '    for chunk, (cs, ce) in zip(chunks, timings):\n'
                      '        if option == "A":\n'
                      '            parts = []\n'
                      '            pos = int(round(cs * 100))  # integer centiseconds -> no drift\n'
                      '            for i, w in enumerate(chunk):\n'
                      '                s_cs, e_cs = int(round(w["start"] * 100)), '
                      'int(round(w["end"] * 100))\n'
                      '                if s_cs > pos:\n'
                      '                    parts.append(f"{{\\\\k{s_cs - pos}}}")\n'
                      '                    pos = s_cs\n'
                      '                d = max(e_cs - pos, 1)\n'
                      '                parts.append(f"{{\\\\kf{d}}}{disp(w[\'word\'])}" + (" " if '
                      'i + 1 < len(chunk) else ""))\n'
                      '                pos += d\n'
                      '            text = "".join(parts)\n'
                      '            lines.append(f"Dialogue: '
                      '0,{ass_time(cs)},{ass_time(ce)},Kang,,0,0,0,,{text}")\n'
                      '            events.append({"start": cs, "end": ce, "text": " '
                      '".join(w["word"] for w in chunk), "words": chunk})\n'
                      '        else:  # Option B: per-word events with scale pop\n'
                      '            for i, w in enumerate(chunk):\n'
                      '                ws = w["start"] if i else cs\n'
                      '                we = chunk[i + 1]["start"] if i + 1 < len(chunk) else ce\n'
                      '                if we <= ws:\n'
                      '                    continue\n'
                      '                segs = []\n'
                      '                for j, x in enumerate(chunk):\n'
                      '                    t = disp(x["word"])\n'
                      '                    if j == i:\n'
                      '                        '
                      'segs.append(f"{{\\\\c{primary}\\\\fscx100\\\\fscy100\\\\t(0,80,\\\\fscx108\\\\fscy108)\\\\t(80,160,\\\\fscx104\\\\fscy104)}}{t}{{\\\\c{secondary}\\\\fscx100\\\\fscy100}}")\n'
                      '                    else:\n'
                      '                        segs.append(t)\n'
                      '                lines.append(f"Dialogue: '
                      '0,{ass_time(ws)},{ass_time(we)},Kang,,0,0,0,,{\' \'.join(segs)}")\n'
                      '            events.append({"start": cs, "end": ce, "text": " '
                      '".join(w["word"] for w in chunk), "words": chunk})\n'
                      '    return header + "\\n".join(lines) + "\\n", events\n'
                      '\n'
                      '\n'
                      '# '
                      '--------------------------------------------------------------------------- '
                      'SRT\n'
                      'def srt_time(t: float) -> str:\n'
                      '    ms = int(round(max(t, 0) * 1000))\n'
                      '    h, ms = divmod(ms, 3600000)\n'
                      '    m, ms = divmod(ms, 60000)\n'
                      '    s, ms = divmod(ms, 1000)\n'
                      '    return f"{h:02d}:{m:02d}:{s:02d},{ms:03d}"\n'
                      '\n'
                      '\n'
                      'def build_srt(events: List[Dict[str, Any]]) -> str:\n'
                      '    return "\\n".join(f"{i}\\n{srt_time(e[\'start\'])} --> '
                      '{srt_time(e[\'end\'])}\\n{e[\'text\']}\\n"\n'
                      '                     for i, e in enumerate(events, 1))\n'
                      '\n'
                      '\n'
                      '# '
                      '--------------------------------------------------------------------------- '
                      'QA helpers\n'
                      '_DIALOGUE_RE = re.compile(r"^Dialogue: \\d+,([\\d:.]+),([\\d:.]+),")\n'
                      '_KF_RE = re.compile(r"\\{\\\\k(f?)(\\d+)\\}([^{]*)")\n'
                      '\n'
                      '\n'
                      'def parse_ass_time(s: str) -> float:\n'
                      '    h, m, rest = s.split(":")\n'
                      '    return int(h) * 3600 + int(m) * 60 + float(rest)\n'
                      '\n'
                      '\n'
                      'def ass_word_onsets(ass_text: str) -> List[Tuple[str, float]]:\n'
                      '    """Recover (word, onset seconds) from Option-A \\\\kf karaoke '
                      'events."""\n'
                      '    out: List[Tuple[str, float]] = []\n'
                      '    for line in ass_text.splitlines():\n'
                      '        m = _DIALOGUE_RE.match(line)\n'
                      '        if not m:\n'
                      '            continue\n'
                      '        t = parse_ass_time(m.group(1))\n'
                      '        body = line.split(",", 9)[9]\n'
                      '        for kf, cs, word in _KF_RE.findall(body):\n'
                      '            if kf == "f" and word.strip():\n'
                      '                out.append((word.strip(), round(t, 3)))\n'
                      '            t += int(cs) / 100.0\n'
                      '    return out\n'
                      '\n'
                      '\n'
                      'def ass_events(ass_text: str) -> List[Tuple[float, float]]:\n'
                      '    ev = []\n'
                      '    for line in ass_text.splitlines():\n'
                      '        m = _DIALOGUE_RE.match(line)\n'
                      '        if m:\n'
                      '            ev.append((parse_ass_time(m.group(1)), '
                      'parse_ass_time(m.group(2))))\n'
                      '    return ev\n',
 'core/visuals.py': '"""M3 Workflow G, step 2: approved LOCAL visuals per scene (manual selection, '
                    'nothing invented).\n'
                    '\n'
                    'scenes.json -> visuals.csv template (one row per scene: brief / asset / '
                    'licence filled by the user)\n'
                    '-> validate against the assets folder -> visuals.json (sha256 per asset) -> '
                    'Gate "visual_selection"\n'
                    '-> FFmpeg integration preview (selected scenes: visuals + narration + burned '
                    'captions)\n'
                    '-> Premiere: the same assets as clips on a visual track at scene times '
                    '(core/premiere.py).\n'
                    '\n'
                    'Rules: an asset must exist inside the assets folder, have a non-empty brief '
                    'and licence, and be an\n'
                    'image or a video. A video shorter than its scene is an error unless fit=loop. '
                    'No downloads, no AI\n'
                    'generation here; a scene without an asset stays empty (not a placeholder) and '
                    'is reported.\n'
                    '"""\n'
                    'from __future__ import annotations\n'
                    '\n'
                    'import csv\n'
                    'import hashlib\n'
                    'from pathlib import Path\n'
                    'from typing import Any, Dict, List, Optional, Sequence\n'
                    '\n'
                    'SCHEMA = "kang.visuals/1.0"\n'
                    'STILL_EXT = {".jpg", ".jpeg", ".png", ".webp", ".bmp", ".tif", ".tiff"}\n'
                    'VIDEO_EXT = {".mp4", ".mov", ".m4v", ".mkv", ".webm", ".avi", ".mxf"}\n'
                    'FITS = ("cover", "contain", "loop")  # loop = video repeats to fill the scene '
                    '(cover framing)\n'
                    'COLUMNS = ["scene", "start", "end", "duration_s", "narration", "brief", '
                    '"asset", "asset_in_s", "fit", "licence", "notes", "shot_start"]\n'
                    'MIN_SHOT_S = 2.0  # a shot shorter than this is a flash, not a visual\n'
                    '\n'
                    '\n'
                    'def _sha(p: Path) -> str:\n'
                    '    h = hashlib.sha256()\n'
                    '    with open(p, "rb") as f:\n'
                    '        for b in iter(lambda: f.read(1 << 20), b""):\n'
                    '            h.update(b)\n'
                    '    return h.hexdigest()\n'
                    '\n'
                    '\n'
                    'def write_template(plan: Dict[str, Any], out_csv: str | Path) -> Path:\n'
                    '    """One row per scene; brief/asset/licence blank for the user. Narration '
                    'shown for context only."""\n'
                    '    out = Path(out_csv)\n'
                    '    with open(out, "w", encoding="utf-8-sig", newline="") as fh:\n'
                    '        w = csv.DictWriter(fh, COLUMNS)\n'
                    '        w.writeheader()\n'
                    '        for s in plan["scenes"]:\n'
                    '            w.writerow({"scene": s["id"], "start": f"{s[\'start\']:.3f}", '
                    '"end": f"{s[\'end\']:.3f}",\n'
                    '                        "duration_s": f"{s[\'end\'] - s[\'start\']:.2f}", '
                    '"narration": s["text"][:220],\n'
                    '                        "brief": "", "asset": "", "asset_in_s": "0", "fit": '
                    '"cover", "licence": "", "notes": "", "shot_start": ""})\n'
                    '    return out\n'
                    '\n'
                    '\n'
                    'def read_csv(path: str | Path) -> List[Dict[str, str]]:\n'
                    '    with open(path, encoding="utf-8-sig", newline="") as fh:\n'
                    '        return [{k: (v or "").strip() for k, v in r.items()} for r in '
                    'csv.DictReader(fh)]\n'
                    '\n'
                    '\n'
                    'def asset_kind(p: Path) -> Optional[str]:\n'
                    '    ext = p.suffix.lower()\n'
                    '    return "still" if ext in STILL_EXT else "video" if ext in VIDEO_EXT else '
                    'None\n'
                    '\n'
                    '\n'
                    'def validate(rows: Sequence[Dict[str, str]], plan: Dict[str, Any], '
                    'assets_dir: str | Path,\n'
                    '             probe=None) -> Dict[str, Any]:\n'
                    '    """-> {ok, problems, warnings, visuals: kang.visuals/1.0 dict}. '
                    '`probe(path)->{duration}` for videos\n'
                    '    (defaults to runs.media_summary). Scene times always come from '
                    'scenes.json, never from the CSV."""\n'
                    '    if probe is None:\n'
                    '        from .runs import media_summary as probe\n'
                    '    adir = Path(assets_dir).resolve()\n'
                    '    scenes = {s["id"]: s for s in plan["scenes"]}\n'
                    '    problems: List[str] = []\n'
                    '    warnings: List[str] = []\n'
                    '    items: List[Dict[str, Any]] = []\n'
                    '    # Several rows per scene = shots. Row 1 starts at the scene start; each '
                    'further row needs\n'
                    '    # shot_start (timeline seconds, inside the scene, increasing). A shot '
                    'ends where the next begins.\n'
                    '    groups: Dict[str, List[Dict[str, str]]] = {}\n'
                    '    for r in rows:\n'
                    '        sid = r.get("scene", "")\n'
                    '        if sid not in scenes:\n'
                    '            problems.append(f"{sid or \'?\'}: scene not in scenes.json")\n'
                    '            continue\n'
                    '        if r.get("asset"):\n'
                    '            groups.setdefault(sid, []).append(r)\n'
                    '    for sid, grp in groups.items():\n'
                    '        sc = scenes[sid]\n'
                    '        starts: List[float] = []\n'
                    '        bad = False\n'
                    '        for k, r in enumerate(grp):\n'
                    '            if k == 0:\n'
                    '                if r.get("shot_start") and abs(float(r["shot_start"]) - '
                    'sc["start"]) > 0.002:\n'
                    '                    problems.append(f"{sid}: first shot must start at the '
                    'scene start ({sc[\'start\']:.3f})")\n'
                    '                    bad = True\n'
                    '                starts.append(sc["start"])\n'
                    '                continue\n'
                    '            try:\n'
                    '                t = float(r.get("shot_start") or "nan")\n'
                    '            except ValueError:\n'
                    '                t = float("nan")\n'
                    '            if t != t:\n'
                    '                problems.append(f"{sid}: shot {k + 1} needs shot_start '
                    '(seconds on the timeline)")\n'
                    '                bad = True\n'
                    '                break\n'
                    '            if not (starts[-1] + MIN_SHOT_S <= t <= sc["end"] - MIN_SHOT_S):\n'
                    '                problems.append(f"{sid}: shot {k + 1} shot_start {t:.3f} must '
                    'be >= previous+{MIN_SHOT_S}s and "\n'
                    '                                f"<= scene end-{MIN_SHOT_S}s '
                    '({sc[\'start\']:.3f}-{sc[\'end\']:.3f})")\n'
                    '                bad = True\n'
                    '                break\n'
                    '            starts.append(t)\n'
                    '        if bad:\n'
                    '            continue\n'
                    '        ends = starts[1:] + [sc["end"]]\n'
                    '        for k, (r, t0, t1) in enumerate(zip(grp, starts, ends)):\n'
                    '            shot = sid if len(grp) == 1 else f"{sid}.{k + 1}"\n'
                    '            dur = round(t1 - t0, 3)\n'
                    '            p = (adir / r["asset"]).resolve()\n'
                    '            try:\n'
                    '                p.relative_to(adir)\n'
                    '            except ValueError:\n'
                    '                problems.append(f"{shot}: asset outside the assets folder: '
                    '{r[\'asset\']}")\n'
                    '                continue\n'
                    '            if not p.is_file():\n'
                    '                problems.append(f"{shot}: asset not found: {r[\'asset\']}")\n'
                    '                continue\n'
                    '            kind = asset_kind(p)\n'
                    '            if kind is None:\n'
                    '                problems.append(f"{shot}: unsupported asset type '
                    '{p.suffix}")\n'
                    '                continue\n'
                    '            if not r.get("brief"):\n'
                    '                problems.append(f"{shot}: brief is empty")\n'
                    '            if not r.get("licence"):\n'
                    '                problems.append(f"{shot}: licence is empty (own / CC-BY ... / '
                    'licensed - write it)")\n'
                    '            fit = (r.get("fit") or "cover").lower()\n'
                    '            if fit not in FITS:\n'
                    '                problems.append(f"{shot}: fit must be one of {FITS}")\n'
                    '                continue\n'
                    '            try:\n'
                    '                a_in = float(r.get("asset_in_s") or 0)\n'
                    '            except ValueError:\n'
                    '                problems.append(f"{shot}: asset_in_s is not a number")\n'
                    '                continue\n'
                    '            item = {"scene": sid, "shot": shot, "start": t0, "end": t1, '
                    '"duration": dur, "asset": r["asset"],\n'
                    '                    "kind": kind, "fit": fit, "asset_in": a_in, "brief": '
                    'r.get("brief", ""),\n'
                    '                    "licence": r.get("licence", ""), "sha256": _sha(p)}\n'
                    '            if kind == "video":\n'
                    '                alen = float(probe(p).get("duration") or 0)\n'
                    '                item["asset_duration"] = round(alen, 3)\n'
                    '                if a_in < 0 or a_in >= alen:\n'
                    '                    problems.append(f"{shot}: asset_in_s {a_in} outside clip '
                    '(0-{alen:.2f} s)")\n'
                    '                elif alen - a_in < dur - 0.04 and fit != "loop":\n'
                    '                    problems.append(f"{shot}: clip has {alen - a_in:.2f} s '
                    'from asset_in_s, scene needs {dur:.2f} s "\n'
                    '                                    f"(trim scene, pick a longer clip, or '
                    'fit=loop)")\n'
                    '            elif fit == "loop":\n'
                    '                problems.append(f"{shot}: fit=loop is for video clips only")\n'
                    '            items.append(item)\n'
                    '    dup = sorted({i["asset"] for i in items if sum(j["asset"] == i["asset"] '
                    'for j in items) > 1})\n'
                    '    if dup:\n'
                    '        warnings.append(f"asset used in more than one shot: {\', '
                    '\'.join(dup[:6])}")\n'
                    '    missing = [s for s in scenes if s not in {i["scene"] for i in items}]\n'
                    '    if missing:\n'
                    '        warnings.append(f"{len(missing)} scene(s) without a visual: {\', '
                    '\'.join(missing[:12])}"\n'
                    '                        + (" ..." if len(missing) > 12 else ""))\n'
                    '    visuals = {"schema": SCHEMA, "assets_dir": str(adir), "items": items, '
                    '"scenes_without_visual": missing}\n'
                    '    return {"ok": not problems, "problems": problems, "warnings": warnings, '
                    '"visuals": visuals}\n'
                    '\n'
                    '\n'
                    'def selection_text(visuals: Dict[str, Any]) -> str:\n'
                    '    """What the approver sees at the visual_selection gate."""\n'
                    '    lines = [f"{len(visuals[\'items\'])} scene visual(s); without visual: '
                    '{len(visuals[\'scenes_without_visual\'])}"]\n'
                    '    for i in visuals["items"]:\n'
                    '        lines.append(f"{i.get(\'shot\', i[\'scene\'])} '
                    "{i['start']:.1f}-{i['end']:.1f}s {i['kind']} {i['asset']} [{i['fit']}] "
                    '"\n'
                    '                     f"licence={i[\'licence\']} | {i[\'brief\'][:80]}")\n'
                    '    return "\\n".join(lines)\n'
                    '\n'
                    '\n'
                    'def pick(visuals: Dict[str, Any], plan: Dict[str, Any], scene_ids: '
                    'Sequence[str]) -> List[Dict[str, Any]]:\n'
                    '    """Selected scenes, in order, contiguous, each with a visual (the preview '
                    'never shows gaps)."""\n'
                    '    order = [s["id"] for s in plan["scenes"]]\n'
                    '    bad = [s for s in scene_ids if s not in order]\n'
                    '    if bad:\n'
                    '        raise ValueError(f"unknown scene(s): {bad}")\n'
                    '    idx = sorted(order.index(s) for s in scene_ids)\n'
                    '    if idx != list(range(idx[0], idx[-1] + 1)):\n'
                    '        raise ValueError("preview scenes must be contiguous (e.g. '
                    'S01,S02,S03)")\n'
                    '    sel = [order[k] for k in idx]\n'
                    '    have = {i["scene"] for i in visuals["items"]}\n'
                    '    no = [s for s in sel if s not in have]\n'
                    '    if no:\n'
                    '        raise ValueError(f"scene(s) without an approved visual: {no}")\n'
                    '    return sorted((i for i in visuals["items"] if i["scene"] in sel), '
                    'key=lambda i: i["start"])\n'
                    '\n'
                    '\n'
                    'def _vf(fit: str, w: int, h: int, fps: int) -> str:\n'
                    '    if fit == "contain":\n'
                    '        geo = '
                    'f"scale={w}:{h}:force_original_aspect_ratio=decrease,pad={w}:{h}:(ow-iw)/2:(oh-ih)/2:color=black"\n'
                    '    else:\n'
                    '        geo = '
                    'f"scale={w}:{h}:force_original_aspect_ratio=increase,crop={w}:{h}"\n'
                    '    return f"{geo},setsar=1,fps={fps},format=yuv420p"\n'
                    '\n'
                    '\n'
                    'def segment_args(item: Dict[str, Any], src: Path, out: Path, dur: float, w: '
                    'int, h: int, fps: int) -> List[str]:\n'
                    '    """ffmpeg args for one scene segment: exact frame count, same codec '
                    'params for every segment\n'
                    '    (so the concat demuxer can join them without re-encoding)."""\n'
                    '    nframes = max(1, int(round(dur * fps)))\n'
                    '    if item["kind"] == "still":\n'
                    '        inp = ["-loop", "1", "-framerate", str(fps), "-i", str(src)]\n'
                    '    else:\n'
                    '        inp = (["-stream_loop", "-1"] if item["fit"] == "loop" else []) + '
                    '["-ss", f"{item[\'asset_in\']:.3f}", "-i", str(src)]\n'
                    '    return [*inp, "-an", "-vf", _vf(item["fit"], w, h, fps), "-frames:v", '
                    'str(nframes),\n'
                    '            "-c:v", "libx264", "-preset", "veryfast", "-crf", "21", '
                    '"-pix_fmt", "yuv420p",\n'
                    '            "-r", str(fps), "-video_track_timescale", str(fps * 1000), '
                    'str(out)]\n',
 'core/voter.py': '"""Spelling voter: keep the base model\'s words and timing, borrow spelling '
                  'from a second model.\n'
                  '\n'
                  'Why: Kaggle A/B 2026-10-28 on Dwarika (692 s). large-v3 had the most complete '
                  'first pass but spelled\n'
                  'hard words badly (ज्योतिर्लिंग, किलोमीटर: 0 correct); Hindi fine-tunes (vaani, '
                  'IndicWhisper) spelled\n'
                  'them right but dropped words. So: never take words FROM the second model, only '
                  'fix the spelling of a\n'
                  'base word that sits at the same time.\n'
                  '\n'
                  'A base word is replaced only when ALL hold:\n'
                  '  * its core token is Devanagari, not a number, and NOT in the wordlist (likely '
                  'misspelled)\n'
                  '  * an alt word overlaps it in time (>= min_overlap of the shorter one)\n'
                  '  * the alt token IS in the wordlist and looks similar (difflib ratio >= '
                  'min_ratio)\n'
                  '  * the alt token is not just the base with its ending cut off (protects '
                  'inflections: मंदिरों -> मंदिर)\n'
                  '  * with require_agree, at least two alt models propose the same fix and none '
                  'proposes another\n'
                  'Punctuation around the base word is preserved. Every change is reported for '
                  'human review.\n'
                  '"""\n'
                  'from __future__ import annotations\n'
                  '\n'
                  'import difflib\n'
                  'import re\n'
                  'from pathlib import Path\n'
                  'from typing import Any, Dict, Iterable, List, Optional, Sequence, Set, Tuple\n'
                  '\n'
                  '_DEV = re.compile(r"[\\u0900-\\u097F]")\n'
                  '_NUM = re.compile(r"^[0-9०-९,\\.]+$")\n'
                  '_EDGE = '
                  're.compile(r"^([^\\u0900-\\u0963\\u0966-\\u097F0-9A-Za-z]*)(.*?)([^\\u0900-\\u0963\\u0966-\\u097F0-9A-Za-z]*)$", '
                  're.S)\n'
                  '_ZW = str.maketrans("", "", "\\u200c\\u200d")\n'
                  '\n'
                  '\n'
                  'def split_core(word: str) -> Tuple[str, str, str]:\n'
                  '    m = _EDGE.match(word)\n'
                  '    return (m.group(1), m.group(2).translate(_ZW), m.group(3)) if m else ("", '
                  'word, "")\n'
                  '\n'
                  '\n'
                  'def load_wordlist(path: Optional[str | Path], extra: Iterable[str] = ()) -> '
                  'Set[str]:\n'
                  '    """hunspell .dic (first line = count, \'word/FLAGS\') or one word per line. '
                  'Empty set if missing."""\n'
                  '    words: Set[str] = set()\n'
                  '    if path and Path(path).exists():\n'
                  '        for i, line in enumerate(Path(path).read_text(encoding="utf-8", '
                  'errors="ignore").splitlines()):\n'
                  '            t = line.split("/")[0].strip().translate(_ZW)\n'
                  '            if t and not (i == 0 and t.isdigit()):\n'
                  '                words.add(t)\n'
                  '    for e in extra:\n'
                  '        for t in re.split(r"\\s+", e):\n'
                  '            c = split_core(t)[1]\n'
                  '            if c:\n'
                  '                words.add(c)\n'
                  '    return words\n'
                  '\n'
                  '\n'
                  'def _overlap(a: Dict[str, Any], b: Dict[str, Any]) -> float:\n'
                  '    inter = min(a["end"], b["end"]) - max(a["start"], b["start"])\n'
                  '    shorter = max(min(a["end"] - a["start"], b["end"] - b["start"]), 0.05)\n'
                  '    return max(inter, 0.0) / shorter\n'
                  '\n'
                  '\n'
                  'def _is_trimmed_inflection(base: str, alt: str) -> bool:\n'
                  '    """alt is base with a piece cut off: an inflection (मंदिरों->मंदिर) or a '
                  'compound half (पावनधाम->धाम)."""\n'
                  '    return len(alt) < len(base) and (base.startswith(alt) or '
                  'base.endswith(alt))\n'
                  '\n'
                  '\n'
                  'def _same_nasal(a: str, b: str) -> bool:\n'
                  '    """यहाँ / यहां are both fine: chandrabindu vs anusvara is not a spelling '
                  'error."""\n'
                  '    return a.replace("\\u0901", "\\u0902") == b.replace("\\u0901", "\\u0902")\n'
                  '\n'
                  '\n'
                  'def _best_alt(bw: Dict[str, Any], alt: Sequence[Dict[str, Any]], vocab: '
                  'Set[str], core: str,\n'
                  '              min_overlap: float, min_ratio: float, lo: int) -> '
                  'Tuple[Optional[str], int, bool]:\n'
                  '    """Return (token, new_lo, had_any_overlap). `alt` is sorted by start; lo is '
                  'a moving cursor."""\n'
                  '    while lo < len(alt) and alt[lo]["end"] < bw["start"] - 1.0:\n'
                  '        lo += 1\n'
                  '    best, best_r, seen = None, 0.0, False\n'
                  '    j = lo\n'
                  '    while j < len(alt) and alt[j]["start"] <= bw["end"] + 0.5:\n'
                  '        aw = alt[j]\n'
                  '        j += 1\n'
                  '        if _overlap(bw, aw) < min_overlap:\n'
                  '            continue\n'
                  '        seen = True\n'
                  '        t = split_core(aw["word"])[1]\n'
                  '        if (not t or t == core or t not in vocab or _NUM.match(t) or '
                  '_is_trimmed_inflection(core, t)\n'
                  '                or _same_nasal(core, t)):\n'
                  '            continue\n'
                  '        r = difflib.SequenceMatcher(None, core, t).ratio()\n'
                  '        if r >= min_ratio and r > best_r:\n'
                  '            best, best_r = t, r\n'
                  '    return best, lo, seen\n'
                  '\n'
                  '\n'
                  'def spelling_vote(base: List[Dict[str, Any]], alts: Dict[str, List[Dict[str, '
                  'Any]]], vocab: Set[str],\n'
                  '                  min_overlap: float = 0.5, min_ratio: float = 0.6, '
                  'require_agree: bool = False\n'
                  '                  ) -> Tuple[List[Dict[str, Any]], List[Dict[str, Any]]]:\n'
                  '    if not vocab or not alts:\n'
                  '        return [dict(w) for w in base], []\n'
                  '    sorted_alts = {k: sorted(v, key=lambda w: w["start"]) for k, v in '
                  'alts.items()}\n'
                  '    cursors = {k: 0 for k in alts}\n'
                  '    out, changes = [], []\n'
                  '    for i, bw in enumerate(base):\n'
                  '        w = dict(bw)\n'
                  '        pre, core, post = split_core(w["word"])\n'
                  '        # a glossary phrase merged into one entry ("बेट द्वारका") is already '
                  'curated: leave it\n'
                  '        if core and " " not in core and _DEV.search(core) and not '
                  '_NUM.match(core) and core not in vocab:\n'
                  '            picks, seen_any = {}, {}\n'
                  '            for k, alt in sorted_alts.items():\n'
                  '                tok, cursors[k], seen = _best_alt(w, alt, vocab, core, '
                  'min_overlap, min_ratio, cursors[k])\n'
                  '                seen_any[k] = seen\n'
                  '                if tok:\n'
                  '                    picks[k] = tok\n'
                  '            choice = None\n'
                  '            if picks:\n'
                  '                vals = set(picks.values())\n'
                  '                if require_agree:  # >= 2 models propose the same fix and none '
                  'proposes another\n'
                  '                    if len(vals) == 1 and len(picks) >= 2:\n'
                  '                        choice = vals.pop()\n'
                  '                elif len(vals) == 1:\n'
                  '                    choice = vals.pop()\n'
                  '            if choice:\n'
                  '                w["word"] = pre + choice + post\n'
                  '                changes.append({"index": i, "start": w["start"], "end": '
                  'w["end"], "from": bw["word"],\n'
                  '                                "to": w["word"], "by": sorted(picks)})\n'
                  '        out.append(w)\n'
                  '    return out, changes\n',
 'core/words.py': '"""Phase 1: words.json (word-level transcript) + anomaly audit.\n'
                  '\n'
                  'words.json format (list):\n'
                  '  {"word": str, "start": float, "end": float, "probability": float}\n'
                  '"""\n'
                  'from __future__ import annotations\n'
                  '\n'
                  'import json\n'
                  'from pathlib import Path\n'
                  'from typing import Any, Dict, List\n'
                  '\n'
                  '\n'
                  'def rows_to_words(rows: List[Dict[str, Any]]) -> List[Dict[str, Any]]:\n'
                  '    """Flatten engine transcript rows into words.json entries (order '
                  'preserved)."""\n'
                  '    out: List[Dict[str, Any]] = []\n'
                  '    for r in rows:\n'
                  '        for w in r.get("words", []):\n'
                  '            text = str(w.get("w", w.get("word", ""))).strip()\n'
                  '            if not text:\n'
                  '                continue\n'
                  '            out.append({\n'
                  '                "word": text,\n'
                  '                "start": round(float(w["start"]), 3),\n'
                  '                "end": round(float(w["end"]), 3),\n'
                  '                "probability": round(float(w.get("p", w.get("probability", '
                  '1.0))), 3),\n'
                  '            })\n'
                  '    return merge_zero_duration(out)\n'
                  '\n'
                  '\n'
                  'def merge_zero_duration(words: List[Dict[str, Any]]) -> List[Dict[str, Any]]:\n'
                  '    """Words with end <= start (ASR clamp anomalies) get no invented duration: '
                  'their text is\n'
                  '    attached to the previous word (or the next one if first), whose real timing '
                  'is kept.\n'
                  '    Merged entries carry `merged` so the review sample can show them."""\n'
                  '    out: List[Dict[str, Any]] = []\n'
                  '    pending: List[str] = []\n'
                  '    for w in words:\n'
                  '        if w["end"] <= w["start"]:\n'
                  '            if out:\n'
                  '                prev = out[-1]\n'
                  '                prev["word"] = f\'{prev["word"]} {w["word"]}\'\n'
                  '                prev["end"] = max(prev["end"], w["end"])\n'
                  '                prev.setdefault("merged", []).append(w["word"])\n'
                  '            else:\n'
                  '                pending.append(w["word"])\n'
                  '            continue\n'
                  '        w = dict(w)\n'
                  '        if pending:\n'
                  '            w["word"] = " ".join(pending + [w["word"]])\n'
                  '            w["merged"] = pending\n'
                  '            pending = []\n'
                  '        out.append(w)\n'
                  '    return out\n'
                  '\n'
                  '\n'
                  'def audit_words(words: List[Dict[str, Any]], max_word_dur: float = 2.5, '
                  'low_prob: float = 0.4) -> Dict[str, Any]:\n'
                  '    """Anomaly audit (Phase 1): long words, non-positive durations, overlaps, '
                  'low confidence."""\n'
                  '    issues: List[Dict[str, Any]] = []\n'
                  '    for i, w in enumerate(words):\n'
                  '        d = w["end"] - w["start"]\n'
                  '        if w["start"] < 0:\n'
                  '            issues.append({"i": i, "word": w["word"], "issue": '
                  '"negative_start"})\n'
                  '        if d <= 0:\n'
                  '            issues.append({"i": i, "word": w["word"], "issue": '
                  '"non_positive_duration"})\n'
                  '        elif d > max_word_dur:\n'
                  '            issues.append({"i": i, "word": w["word"], "issue": "too_long", '
                  '"dur": round(d, 3)})\n'
                  '        if i and words[i - 1]["end"] > w["start"] + 1e-6:\n'
                  '            issues.append({"i": i, "word": w["word"], "issue": '
                  '"overlaps_previous"})\n'
                  '        if w.get("probability", 1.0) < low_prob:\n'
                  '            issues.append({"i": i, "word": w["word"], "issue": '
                  '"low_confidence", "p": w["probability"]})\n'
                  '    return {"word_count": len(words), "issue_count": len(issues), "issues": '
                  'issues,\n'
                  '            "ok": not any(x["issue"] in {"negative_start", '
                  '"non_positive_duration", "overlaps_previous"} for x in issues)}\n'
                  '\n'
                  '\n'
                  'def validate_words(words: Any) -> None:\n'
                  '    if not isinstance(words, list) or not words:\n'
                  '        raise ValueError("words.json must be a non-empty list")\n'
                  '    for i, w in enumerate(words):\n'
                  '        for k in ("word", "start", "end"):\n'
                  '            if k not in w:\n'
                  '                raise ValueError(f"words[{i}] missing \'{k}\'")\n'
                  '        if float(w["end"]) <= float(w["start"]):\n'
                  '            raise ValueError(f"words[{i}] (\'{w[\'word\']}\') end <= start")\n'
                  '\n'
                  '\n'
                  'def load_words(path: str | Path) -> List[Dict[str, Any]]:\n'
                  '    words = json.loads(Path(path).read_text(encoding="utf-8"))\n'
                  '    validate_words(words)\n'
                  '    for w in words:\n'
                  '        w.setdefault("probability", 1.0)\n'
                  '    return words\n'
                  '\n'
                  '\n'
                  'def review_sample(words: List[Dict[str, Any]], n: int = 25) -> str:\n'
                  '    """Human-review sample: first n words with timings (Phase 1 exit '
                  'proof)."""\n'
                  '    lines = [f"{w[\'start\']:7.2f} - {w[\'end\']:7.2f}  '
                  'p={w.get(\'probability\', 1):.2f}  {w[\'word\']}" for w in words[:n]]\n'
                  '    return "\\n".join(lines)\n'
                  '\n'
                  '\n'
                  '# --- Glossary correction (proper nouns ASR keeps getting wrong) '
                  '--------------------------\n'
                  '_PUNCT = "।,.!?;:\\"\'”“‘’()"\n'
                  '\n'
                  '\n'
                  'def _bare(token: str) -> str:\n'
                  '    return token.strip().strip(_PUNCT)\n'
                  '\n'
                  '\n'
                  'def load_glossary(path: str | Path) -> Dict[str, str]:\n'
                  '    """config/glossary.json: {"wrong form (may span tokens)": "correct form"}. '
                  'Keys starting with \'_\' ignored."""\n'
                  '    p = Path(path)\n'
                  '    if not p.is_file():\n'
                  '        return {}\n'
                  '    data = json.loads(p.read_text(encoding="utf-8"))\n'
                  '    return {k: v for k, v in data.items() if not k.startswith("_") and '
                  'isinstance(v, str)}\n'
                  '\n'
                  '\n'
                  'def apply_glossary(words: List[Dict[str, Any]], glossary: Dict[str, str]) -> '
                  'tuple[List[Dict[str, Any]], List[Dict[str, Any]]]:\n'
                  '    """Run glossary passes until stable (max 3), so a fix can feed another '
                  'rule:\n'
                  '    \'द्वारी का दीश\' -> \'द्वारका दीश\' -> \'द्वारकाधीश\'."""\n'
                  '    changes: List[Dict[str, Any]] = []\n'
                  '    for _ in range(3):\n'
                  '        words, ch = _glossary_pass(words, glossary)\n'
                  '        if not ch:\n'
                  '            break\n'
                  '        changes += ch\n'
                  '    return words, changes\n'
                  '\n'
                  '\n'
                  'def _glossary_pass(words: List[Dict[str, Any]], glossary: Dict[str, str]) -> '
                  'tuple[List[Dict[str, Any]], List[Dict[str, Any]]]:\n'
                  '    """Replace mis-heard words/phrases. Multi-token matches merge into one word '
                  'spanning\n'
                  '    first.start -> last.end (timing preserved, nothing fabricated). Trailing '
                  'punctuation of the\n'
                  '    last matched token is kept. Returns (new_words, changes). Input is never '
                  'mutated."""\n'
                  '    rules = sorted(((k.split(), v) for k, v in glossary.items()), key=lambda r: '
                  '-len(r[0]))\n'
                  '    out, changes, i = [], [], 0\n'
                  '    while i < len(words):\n'
                  '        hit = None\n'
                  '        for src, dst in rules:\n'
                  '            n = len(src)\n'
                  '            if i + n <= len(words) and [_bare(w["word"]) for w in words[i:i + '
                  'n]] == src:\n'
                  '                hit = (n, dst)\n'
                  '                break\n'
                  '        if not hit:\n'
                  '            out.append(dict(words[i])); i += 1\n'
                  '            continue\n'
                  '        n, dst = hit\n'
                  '        span = words[i:i + n]\n'
                  '        last = span[-1]["word"].strip()\n'
                  '        tail = last[len(last.rstrip(_PUNCT)):]\n'
                  '        new = dict(span[0])\n'
                  '        new.update(word=dst + tail, end=span[-1]["end"],\n'
                  '                   probability=min(w.get("probability", 1.0) for w in span), '
                  'corrected_from=" ".join(w["word"] for w in span))\n'
                  '        changes.append({"at": new["start"], "from": new["corrected_from"], '
                  '"to": new["word"]})\n'
                  '        out.append(new); i += n\n'
                  '    return out, changes\n'
                  '\n'
                  '\n'
                  'def join_split_tokens(words: List[Dict[str, Any]]) -> tuple[List[Dict[str, '
                  'Any]], List[Dict[str, Any]]]:\n'
                  '    """Whisper splits hyphen/comma joins into their own tokens: \'साथ\' '
                  "'-साथ', '10' ',000',\n"
                  "    '400' '-500'. A token that starts with '-' or ',' is glued onto the "
                  'previous word (timing\n'
                  '    first.start -> last.end, nothing invented). Returns (words, joins)."""\n'
                  '    out: List[Dict[str, Any]] = []\n'
                  '    joins: List[Dict[str, Any]] = []\n'
                  '    for w in words:\n'
                  '        t = w["word"].strip()\n'
                  '        if out and len(t) > 1 and t[0] in "-," and (t[0] == "-" or '
                  't[1:2].isdigit()):\n'
                  '            prev = out[-1]\n'
                  '            new = prev["word"].rstrip() + t\n'
                  '            joins.append({"at": prev["start"], "from": f\'{prev["word"]} '
                  '{w["word"]}\', "to": new})\n'
                  '            prev.update(word=new, end=max(prev["end"], w["end"]),\n'
                  '                        probability=min(prev.get("probability", 1.0), '
                  'w.get("probability", 1.0)))\n'
                  '            continue\n'
                  '        out.append(dict(w))\n'
                  '    return out, joins\n'
                  '\n'
                  '\n'
                  '# --- Script sanity (Whisper hallucinations like "हैассור" / "¿ neutral") '
                  '------------------\n'
                  'def _allowed_char(ch: str) -> bool:\n'
                  '    o = ord(ch)\n'
                  '    return (0x0900 <= o <= 0x097F or 0xA8E0 <= o <= 0xA8FF or ch.isascii() or '
                  'ch in "।॥–—‘’“”…₹ " or o in (0x200C, 0x200D))\n'
                  '\n'
                  '\n'
                  'def sanitize_script(words: List[Dict[str, Any]]) -> tuple[List[Dict[str, Any]], '
                  'List[Dict[str, Any]]]:\n'
                  '    """Strip characters outside Devanagari + ASCII from each word '
                  '(Cyrillic/Hebrew/Arabic\n'
                  '    fragments are ASR hallucinations for Hindi/Hinglish). A word left with no '
                  'letters is dropped.\n'
                  '    ASCII-only words with p < 0.1 next to stripped garbage are dropped too. '
                  'Returns (words, log)."""\n'
                  '    out, log = [], []\n'
                  '    for w in words:\n'
                  '        text = w["word"]\n'
                  '        clean = "".join(ch for ch in text if _allowed_char(ch)).strip()\n'
                  '        clean = clean.replace("¿", "").replace("¡", "").strip()\n'
                  '        if clean != text.strip():\n'
                  '            letters = [c for c in clean if c.isalpha() or 0x0900 <= ord(c) <= '
                  '0x097F]\n'
                  '            if not letters:\n'
                  '                log.append({"at": w["start"], "from": text, "action": '
                  '"dropped"})\n'
                  '                continue\n'
                  '            log.append({"at": w["start"], "from": text, "to": clean, "action": '
                  '"stripped"})\n'
                  '            w = {**w, "word": clean, "sanitized_from": text}\n'
                  '        out.append(dict(w))\n'
                  '    return out, log\n'
                  '\n'
                  '\n'
                  'def repeated_runs(words: List[Dict[str, Any]], n: int = 3) -> List[Dict[str, '
                  'Any]]:\n'
                  '    """Same token n+ times in a row (classic ASR loop) -> review flag, never '
                  'auto-deleted."""\n'
                  '    flags, i = [], 0\n'
                  '    while i < len(words):\n'
                  '        j = i\n'
                  '        while j + 1 < len(words) and _bare(words[j + 1]["word"]) == '
                  '_bare(words[i]["word"]):\n'
                  '            j += 1\n'
                  '        if j - i + 1 >= n:\n'
                  '            flags.append({"i": i, "word": words[i]["word"], "count": j - i + 1, '
                  '"issue": "repeated_run"})\n'
                  '        i = j + 1\n'
                  '    return flags\n',
 'kang.py': '"""Kang AI CLI. One run = one folder under runs/ (or $KANG_RUNS_DIR).\n'
            '\n'
            'Approval gates (cut list, preview) ask y/n in the terminal. Pass --yes only when you\n'
            'have already reviewed what will be produced.\n'
            '\n'
            'Examples\n'
            '  python kang.py check                                   # Phase 0: PASS/FAIL report\n'
            '  python kang.py transcribe voice.mp3 --device cpu       # Phase 1: words.json + '
            'audit\n'
            '  python kang.py workflow voice.mp3 --start 0 --dur 60   # Phase 3: 1 command, 3 '
            'formats\n'
            '  python kang.py workflow voice.mp3 --full --yes          # whole file -> 16:9 long '
            'video\n'
            '  python kang.py fix-words runs/<id>/words.json          # glossary: द्वारी का -> '
            'द्वारका  [--spelling-policy config/caption_spelling_policy.json]\n'
            '  python kang.py subs video.mp4 runs/<id>/words.json     # F01: preview -> approve -> '
            'final\n'
            '  python kang.py promo voice.mp3 words.json --start 0 --dur 20   # Phase 2 exit '
            'proof\n'
            '  python kang.py cut raw.mp4 words.json                  # F03: cut list -> approve '
            '-> cut\n'
            '  python kang.py audio raw.mp4                           # F04: -14 LUFS / -1.5 dBTP\n'
            '  python kang.py formats final.mp4 --words words.json --to 9:16 16:9 1:1   # F05\n'
            '  python kang.py palette screenshot.png                  # F19-lite: '
            'brand_draft.json\n'
            '  python kang.py presets                                 # F02: list presets\n'
            '  python kang.py review voice.mp3 words.json --phrase "बेड द्वारका"   # review.csv + '
            'clips (no edits)\n'
            '  python kang.py apply-review words.json review.csv      # your fixes -> new '
            'words.json + accept gate\n'
            '  python kang.py package voice.mp3 words.json --win-root E:/kang/packages/x  # M3 '
            'Premiere package\n'
            '"""\n'
            'from __future__ import annotations\n'
            '\n'
            'import argparse\n'
            'import json\n'
            'import sys\n'
            'from pathlib import Path\n'
            '\n'
            'ROOT = Path(__file__).resolve().parent\n'
            'sys.path.insert(0, str(ROOT))\n'
            '\n'
            '\n'
            'def _approver(args):\n'
            '    from core.pipeline import cli_approver\n'
            '    if not args.yes:\n'
            '        return cli_approver\n'
            '\n'
            '    def auto(g, i):  # dev/automation only: recorded as auto(--yes) in '
            'approvals.json, never as a human review\n'
            '        return True\n'
            '    auto.approver_name = "auto(--yes)"\n'
            '    return auto\n'
            '\n'
            '\n'
            'def _print(obj):\n'
            '    print(json.dumps(obj, ensure_ascii=False, indent=2, default=str))\n'
            '\n'
            '\n'
            'def main(argv=None) -> int:\n'
            '    ap = argparse.ArgumentParser(prog="kang", description="Kang AI Creative '
            'Automation (Phase 0-3)")\n'
            '    _vf = Path(__file__).resolve().parent / "VERSION"\n'
            '    ap.add_argument("--version", action="version",\n'
            '                    version="kang " + (_vf.read_text(encoding="utf-8").strip() if '
            '_vf.exists() else "unknown"))\n'
            '    sub = ap.add_subparsers(dest="cmd", required=True)\n'
            '\n'
            '    sub.add_parser("check", help="Phase 0 build & environment check")\n'
            '    sub.add_parser("presets", help="List subtitle presets (F02)")\n'
            '\n'
            '    p = sub.add_parser("transcribe", help="Phase 1: audio -> words.json '
            '(cut=False)")\n'
            '    p.add_argument("audio"); p.add_argument("--lang", default="hi"); '
            'p.add_argument("--model", default="large-v3-turbo")\n'
            '    p.add_argument("--device", choices=["cuda", "cpu"]); p.add_argument("--prompt")\n'
            '    p.add_argument("--voters", help="spelling voter: name=ct2_dir,... (default: '
            'config/voters.json models)"); p.add_argument("--wordlist")\n'
            '    p.add_argument("--no-vote", action="store_true", help="turn the spelling voter '
            'off")\n'
            '    p.add_argument("--spelling-policy", help="per-video caption spelling json '
            '(opt-in, e.g. config/caption_spelling_policy.json)")\n'
            '\n'
            '    for name, hlp in (("subs", "F01 animated subtitles"), ("promo", "Phase 2 promo '
            'clip from voiceover")):\n'
            '        p = sub.add_parser(name, help=hlp)\n'
            '        p.add_argument("media"); p.add_argument("words")\n'
            '        p.add_argument("--fmt", default="9:16", choices=["9:16", "16:9", "1:1", '
            '"4:5"])\n'
            '        p.add_argument("--preset", default="vox_bold_yellow")\n'
            '        p.add_argument("--yes", action="store_true", help="approve the preview gate '
            'non-interactively")\n'
            '        if name == "subs":\n'
            '            p.add_argument("--preview-start", type=float, default=0.0)\n'
            '            p.add_argument("--option", choices=["A", "B"])\n'
            '        else:\n'
            '            p.add_argument("--start", type=float, default=0.0); '
            'p.add_argument("--dur", type=float, default=20.0)\n'
            '\n'
            '    p = sub.add_parser("workflow", help="Phase 3: audio/video -> cut -> clean -> '
            'subtitles in all formats")\n'
            '    p.add_argument("source"); p.add_argument("--start", type=float, default=0.0); '
            'p.add_argument("--dur", type=float, default=60.0)\n'
            '    p.add_argument("--to", nargs="+", default=None, help="formats (default: 9:16 16:9 '
            '1:1; --full: 16:9)")\n'
            '    p.add_argument("--full", action="store_true", help="whole file from --start (max '
            '1 h), --dur ignored")\n'
            '    p.add_argument("--preset", default="vox_bold_yellow")\n'
            '    p.add_argument("--lang", default="hi"); p.add_argument("--device", '
            'choices=["cuda", "cpu"]); p.add_argument("--prompt")\n'
            '    p.add_argument("--model", default="large-v3-turbo"); p.add_argument("--yes", '
            'action="store_true")\n'
            '    p.add_argument("--voters", help="spelling voter: name=ct2_dir,... (default: '
            'config/voters.json models)"); p.add_argument("--wordlist")\n'
            '    p.add_argument("--no-vote", action="store_true", help="turn the spelling voter '
            'off")\n'
            '    p.add_argument("--spelling-policy", help="per-video caption spelling json '
            '(opt-in, e.g. config/caption_spelling_policy.json)")\n'
            '\n'
            '    p = sub.add_parser("fix-words", help="Apply config/glossary.json to words.json '
            '(new run)")\n'
            '    p.add_argument("words"); p.add_argument("--glossary")\n'
            '    p.add_argument("--spelling-policy", help="per-video caption spelling json '
            '(opt-in)")\n'
            '\n'
            '    p = sub.add_parser("cut", help="F03 silence + filler cut")\n'
            '    p.add_argument("media"); p.add_argument("words"); p.add_argument("--noise-db", '
            'type=float, default=-34.0)\n'
            '    p.add_argument("--min-silence", type=float, default=0.45); '
            'p.add_argument("--yes", action="store_true")\n'
            '\n'
            '    p = sub.add_parser("audio", help="F04 cleanup + loudness")\n'
            '    p.add_argument("media"); p.add_argument("--lufs", type=float, default=-14.0); '
            'p.add_argument("--tp", type=float, default=-1.5)\n'
            '    p.add_argument("--rnnoise-model")\n'
            '\n'
            '    p = sub.add_parser("formats", help="F05 multi-format render")\n'
            '    p.add_argument("video"); p.add_argument("--words"); p.add_argument("--to", '
            'nargs="+", default=["9:16", "16:9", "1:1"])\n'
            '    p.add_argument("--preset", default="vox_bold_yellow"); '
            'p.add_argument("--reframe-x", type=float, default=0.5)\n'
            '\n'
            '    p = sub.add_parser("visuals-template", help="scenes.json -> visuals.csv '
            '(brief/asset/licence per scene)")\n'
            '    p.add_argument("scenes"); p.add_argument("--out")\n'
            '    p = sub.add_parser("visuals", help="approved local visuals -> preview mp4 (chosen '
            'scenes) + Premiere package with visual track")\n'
            '    p.add_argument("media"); p.add_argument("words"); p.add_argument("scenes"); '
            'p.add_argument("visuals_csv")\n'
            '    p.add_argument("--assets", required=True, help="folder holding your images/clips '
            '(asset paths are relative to it)")\n'
            '    p.add_argument("--preview", help="contiguous scenes for the integration preview, '
            'e.g. S01,S02,S03")\n'
            '    p.add_argument("--fps", type=int, default=25); p.add_argument("--win-root"); '
            'p.add_argument("--yes", action="store_true")\n'
            '\n'
            '    p = sub.add_parser("ingest", help="Asset Brain v0: scan assets dir -> index.json '
            '(tags, licence, sha256) [+ contact sheet]")\n'
            '    p.add_argument("assets"); p.add_argument("--briefs", nargs="*", default=[], '
            'help="visuals CSV(s) whose brief column adds tags")\n'
            '    p.add_argument("--out"); p.add_argument("--sheet", help="write a labelled contact '
            'sheet (jpg) here")\n'
            '    p = sub.add_parser("asset-search", help="Tag search in index.json (Hindi/English '
            'aliases); hints only, selects nothing")\n'
            '    p.add_argument("index"); p.add_argument("query"); p.add_argument("--top", '
            'type=int, default=3)\n'
            '    p.add_argument("--no-sharealike", action="store_true", help="skip CC BY-SA '
            'assets")\n'
            '    p = sub.add_parser("match", help="Asset Brain Phase 2: scenes + index.json -> '
            'broll_suggestions.json/.csv (hints, top-3 per scene)")\n'
            '    p.add_argument("scenes"); p.add_argument("index"); p.add_argument("--words"); '
            'p.add_argument("--top", type=int, default=3)\n'
            '    p.add_argument("--out", required=True, help="output prefix, writes <out>.json and '
            '<out>.csv")\n'
            '    p.add_argument("--no-sharealike", action="store_true"); '
            'p.add_argument("--eval-csv", help="visuals CSV actually used -> hit@k")\n'
            '    p = sub.add_parser("match-approve", help="your choices in broll_suggestions.csv '
            '(choose=1) -> visuals.csv for `visuals`")\n'
            '    p.add_argument("suggestions_csv"); p.add_argument("index"); '
            'p.add_argument("scenes"); p.add_argument("--out", required=True)\n'
            '    p.add_argument("--allow-missing", action="store_true")\n'
            '    p = sub.add_parser("match-check", help="structural QA of broll_suggestions.json '
            '(coverage, 3/scene, duplicates, reasons)")\n'
            '    p.add_argument("suggestions"); p.add_argument("--scenes", help="scenes.json for '
            'expected scene ids")\n'
            '    p.add_argument("--sheet", help="also write a spot-check image (narration + A/B/C '
            'thumbs); needs --scenes and --assets")\n'
            '    p.add_argument("--assets"); p.add_argument("--only", help="scene ids for the '
            'sheet, e.g. S04,S09,S13")\n'
            '    p = sub.add_parser("approve-broll", help="choose A/B/C per scene (prompt, or '
            '--choices S01=A,S02=AB,S22=skip) -> approved_broll.json + visuals.csv")\n'
            '    p.add_argument("suggestions"); p.add_argument("index"); p.add_argument("scenes")\n'
            '    p.add_argument("--out", required=True, help="visuals.csv to write"); '
            'p.add_argument("--approved", help="approved_broll.json (default next to --out)")\n'
            '    p.add_argument("--choices"); p.add_argument("--allow-missing", '
            'action="store_true")\n'
            '    p = sub.add_parser("package", help="M3 Workflow G: words -> scenes.json -> Gate 1 '
            '-> Premiere package (FCP7 XML)")\n'
            '    p.add_argument("media")\n'
            '    p.add_argument("words")\n'
            '    p.add_argument("--scenes", help="your edited scenes.json (default: generate)")\n'
            '    p.add_argument("--cuts", help="cuts.json from a run whose cut_list gate is '
            'approved (default: no cuts)")\n'
            '    p.add_argument("--fps", type=int, default=25)\n'
            '    p.add_argument("--win-root", help=r"Windows folder the package will live in, e.g. '
            'E:\\kang\\packages\\dwarika")\n'
            '    p.add_argument("--english-script", choices=["asr", "latin", "devanagari"], '
            'default="asr",\n'
            '                   help="captions only: English words as transcribed (asr), in Latin, '
            'or in Devanagari")\n'
            '    p.add_argument("--yes", action="store_true")\n'
            '    p = sub.add_parser("review", help="Transcript review sheet: review.csv + audio '
            'clips (changes no words)")\n'
            '    p.add_argument("media"); p.add_argument("words")\n'
            '    p.add_argument("--phrase", action="append", default=[], help="extra suspicious '
            'phrase to add (repeatable)")\n'
            '    p.add_argument("--no-clips", action="store_true")\n'
            '    p = sub.add_parser("apply-review", help="Apply your filled review.csv -> new '
            'words.json (+ accept gate)")\n'
            '    p.add_argument("words"); p.add_argument("review_csv"); p.add_argument("--yes", '
            'action="store_true")\n'
            '    p.add_argument("--source", choices=["human", "auto-extraction"], '
            'default="human",\n'
            '                   help="who decided the fixes: human listening (can be accepted) or '
            'automatic audio extraction (draft only)")\n'
            '    p = sub.add_parser("second-opinion", help="2nd ASR (e.g. Nemotron chunks.json) -> '
            'hint columns in review.csv (changes no words)")\n'
            '    p.add_argument("words"); p.add_argument("chunks"); '
            'p.add_argument("--review-csv")\n'
            '    p = sub.add_parser("consensus", help="3-model consensus hints (ours + timed ASR '
            'json + untimed text) -> consensus.csv (changes no words)")\n'
            '    p.add_argument("words"); p.add_argument("--timed", help="2nd ASR with word '
            'timestamps (NeMo-Speech.cpp json)")\n'
            '    p.add_argument("--text", help="untimed transcript text (e.g. MAI)"); '
            'p.add_argument("--review-csv")\n'
            '    p.add_argument("--names", help="name_suggestions.json (default config/)")\n'
            '    p.add_argument("--audio", help="source audio -> one mp3 clip per row to listen '
            'to")\n'
            '    p = sub.add_parser("hinglish", help="Devanagari words.json -> Hinglish (Roman) '
            'words/SRT/review sheet, same timings (master untouched)")\n'
            '    p.add_argument("words"); p.add_argument("--out", required=True, help="output '
            'prefix")\n'
            '    p.add_argument("--yapper-dir", help="clone of '
            'github.com/ABHISHEKgauti25/yapper_transliterator (romanizer)")\n'
            '    p.add_argument("--table", default=str(ROOT / "config" / "hinglish_names.json"))\n'
            '    p.add_argument("--review-csv", help="master review.csv: undecided rows flagged '
            'for listening")\n'
            '    p.add_argument("--apply-review-fixes", action="store_true", help="apply decided '
            'review.csv fixes in memory first (draft)")\n'
            '    p.add_argument("--other-asr", help="untimed Hinglish ASR chunks.json to compare '
            'against (hints only)")\n'
            '    p = sub.add_parser("hinglish-layer", help="Hinglish Transcript Layer v0: Hindi '
            'words.json -> new run with words_hinglish_v1.json, both SRTs, review CSV/HTML '
            '(+clips)")\n'
            '    p.add_argument("words"); p.add_argument("--media", help="audio for per-block '
            'review clips")\n'
            '    p.add_argument("--yapper-dir", required=True); p.add_argument("--table", '
            'default=str(ROOT / "config" / "hinglish_names.json"))\n'
            '    p.add_argument("--master-review-csv"); p.add_argument("--block-s", type=float, '
            'default=12.0)\n'
            '    p.add_argument("--no-clips", action="store_true"); p.add_argument("--version", '
            'default="v1")\n'
            '    p.add_argument("--other-asr", help="untimed Hinglish ASR chunks.json -> per-block '
            'hints (never applied)")\n'
            '    p = sub.add_parser("review-hinglish", help="Hinglish Review Assistant v0: layer '
            '-> review blocks CSV/HTML with audio clips (changes no words)")\n'
            '    p.add_argument("media"); p.add_argument("layer"); p.add_argument("--table", '
            'default=str(ROOT / "config" / "hinglish_names.json"))\n'
            '    p.add_argument("--block-s", type=float, default=12.0); '
            'p.add_argument("--other-asr")\n'
            '    p = sub.add_parser("apply-hinglish-review", help="your filled review_hinglish CSV '
            '-> words_hinglish_*_accepted.json + SRT (only if every block decided)")\n'
            '    p.add_argument("layer"); p.add_argument("review_csv"); '
            'p.add_argument("--partial", action="store_true", help="undecided blocks allowed -> '
            'draft_reviewed, NOT accepted")\n'
            '    p.add_argument("--out", help="output layer path (default '
            '<layer>_accepted.json)")\n'
            '    p.add_argument("--bulk-ok-by-user", metavar="STATEMENT", help="user explicitly '
            'approved ALL undecided blocks: writes a filled copy *_user_bulk_ok.csv '
            '(decided_by=user_bulk) and applies it")\n'
            '    p = sub.add_parser("fetch-plan", help="scenes.json -> editable scene,query CSV '
            'for kang.py fetch (from asset_aliases tags)")\n'
            '    p.add_argument("scenes"); p.add_argument("--out", required=True); '
            'p.add_argument("--place", default="", help="e.g. Dwarka (prefixed to every query)")\n'
            '    p = sub.add_parser("fetch", help="Wikimedia Commons stills + licence -> assets '
            'folder + CREDITS.json/.txt (free licences only; nothing approved)")\n'
            '    p.add_argument("--out", required=True, help="assets folder to fill"); '
            'p.add_argument("--queries", nargs="*", default=[])\n'
            '    p.add_argument("--queries-csv", help="CSV scene,query[,count] (from fetch-plan, '
            'edited)")\n'
            '    p.add_argument("--per-query", type=int, default=3); '
            'p.add_argument("--no-sharealike", action="store_true")\n'
            '    p.add_argument("--min-width", type=int, default=1280); '
            'p.add_argument("--dry-run", action="store_true")\n'
            '    p = sub.add_parser("fetch-pool", help="Commons category pool (e.g. '
            'deepcat:\\"Dwarka\\") ranked per scene -> assets + CREDITS + pool_ranking.csv")\n'
            '    p.add_argument("scenes"); p.add_argument("--pool", required=True, help=\'Commons '
            'search expr, e.g. deepcat:"Dwarka"\')\n'
            '    p.add_argument("--out", required=True); p.add_argument("--per-scene", type=int, '
            'default=2)\n'
            '    p.add_argument("--max-pool", type=int, default=800); '
            'p.add_argument("--min-score", type=float, default=1.0)\n'
            '    p.add_argument("--no-sharealike", action="store_true"); '
            'p.add_argument("--min-width", type=int, default=1280)\n'
            '    p.add_argument("--dry-run", action="store_true"); p.add_argument("--pool-cache", '
            'help="reuse a saved pool.json")\n'
            '    p = sub.add_parser("package-captions", help="copy a Premiere package into a new '
            'run with new captions.srt (old kept as captions_<tag>.srt)")\n'
            '    p.add_argument("package"); p.add_argument("srt"); p.add_argument("--old-tag", '
            'default="hindi")\n'
            '    p.add_argument("--extra", nargs="*", default=[], help="files to add to the '
            'package (layer json, acceptance json)")\n'
            '    p.add_argument("--note", default="")\n'
            '    p = sub.add_parser("reels", help="KANG Caption Adapter v0: Hinglish layer -> '
            'karaoke ASS (karaoke-subs, MIT) + 9:16 preview + qa_report")\n'
            '    p.add_argument("audio"); p.add_argument("layer"); p.add_argument("--style", '
            'default="karaoke_yellow")\n'
            '    p.add_argument("--format", default="9:16", choices=["9:16", "1:1", "16:9"]); '
            'p.add_argument("--start", type=float, default=0.0)\n'
            '    p.add_argument("--dur", type=float, default=30.0); p.add_argument("--visuals", '
            'help="approved visuals.json for the background")\n'
            '    p.add_argument("--allow-draft", action="store_true", help="render a non-accepted '
            'layer, outputs labelled DRAFT_")\n'
            '    p.add_argument("--no-render", action="store_true", help="ASS + QA only")\n'
            '    p = sub.add_parser("palette", help="F19-lite screenshot -> brand_draft.json")\n'
            '    p.add_argument("image"); p.add_argument("-k", type=int, default=5)\n'
            '\n'
            '    args = ap.parse_args(argv)\n'
            '    import os\n'
            '    if getattr(args, "voters", None):  # read by pipeline.transcribe_to_words (also '
            'inside workflow)\n'
            '        os.environ["ENGINE_VOTERS"] = args.voters\n'
            '    if getattr(args, "no_vote", False):\n'
            '        os.environ["ENGINE_VOTE"] = "0"\n'
            '    if getattr(args, "wordlist", None):\n'
            '        os.environ["ENGINE_WORDLIST"] = args.wordlist\n'
            '    from core import pipeline\n'
            '\n'
            '    if args.cmd == "check":\n'
            '        import build_and_check\n'
            '        return build_and_check.main([])\n'
            '    if args.cmd == "presets":\n'
            '        from core.subtitles import list_presets, load_preset\n'
            '        for n in list_presets(ROOT / "presets"):\n'
            '            print(f"{n:18s} {load_preset(n, ROOT / \'presets\').description}")\n'
            '        return 0\n'
            '    if args.cmd == "transcribe":\n'
            '        r = pipeline.transcribe_to_words(args.audio, language=args.lang, '
            'model_size=args.model,\n'
            '                                         initial_prompt=args.prompt, '
            'device=args.device,\n'
            '                                         spelling_policy=args.spelling_policy)\n'
            '        _print({k: r[k] for k in ("run_id", "dir", "words")} | {"audit_issues": '
            'r["audit"]["issue_count"]})\n'
            '        return 0\n'
            '    if args.cmd == "fix-words":\n'
            '        _print(pipeline.fix_words(args.words, glossary_path=args.glossary, '
            'spelling_policy=args.spelling_policy))\n'
            '        return 0\n'
            '    if args.cmd == "workflow":\n'
            '        r = pipeline.workflow_clip(args.source, args.start, args.dur, '
            'formats=args.to, preset_name=args.preset,\n'
            '                                   approver=_approver(args), language=args.lang, '
            'device=args.device,\n'
            '                                   model_size=args.model, initial_prompt=args.prompt, '
            'full=args.full,\n'
            '                                   spelling_policy=args.spelling_policy)\n'
            '        _print(r)\n'
            '        return 0 if r["pass"] else 1\n'
            '    if args.cmd == "subs":\n'
            '        r = pipeline.subtitle_video(args.media, args.words, fmt=args.fmt, '
            'preset_name=args.preset,\n'
            '                                    preview_start=args.preview_start, '
            'approver=_approver(args), option=args.option)\n'
            '    elif args.cmd == "promo":\n'
            '        r = pipeline.promo_clip(args.media, args.words, args.start, args.dur, '
            'fmt=args.fmt,\n'
            '                                preset_name=args.preset, approver=_approver(args))\n'
            '    elif args.cmd == "cut":\n'
            '        r = pipeline.plan_and_cut(args.media, args.words, approver=_approver(args),\n'
            '                                  noise_db=args.noise_db, '
            'min_silence=args.min_silence)\n'
            '    elif args.cmd == "audio":\n'
            '        r = pipeline.clean_audio(args.media, target_lufs=args.lufs, '
            'true_peak=args.tp, rnnoise_model=args.rnnoise_model)\n'
            '    elif args.cmd == "formats":\n'
            '        r = pipeline.multi_format(args.video, args.words, args.to, '
            'preset_name=args.preset, reframe_x=args.reframe_x)\n'
            '    elif args.cmd == "package":\n'
            '        r = pipeline.premiere_package(args.media, args.words, '
            'approver=_approver(args), scenes_path=args.scenes,\n'
            '                                      cuts_path=args.cuts, fps=args.fps, '
            'win_root=args.win_root,\n'
            '                                      english_script=args.english_script)\n'
            '    elif args.cmd == "visuals-template":\n'
            '        r = pipeline.visuals_template(args.scenes, args.out)\n'
            '        _print(r)\n'
            '        return 0 if r["ok"] else 1\n'
            '    elif args.cmd == "ingest":\n'
            '        from core import asset_index\n'
            '        r = asset_index.ingest(args.assets, args.briefs, args.out)\n'
            '        if r["ok"] and args.sheet:\n'
            '            r["sheet"] = asset_index.contact_sheet(r["index"], args.sheet)\n'
            '        _print(r)\n'
            '        return 0 if r["ok"] else 1\n'
            '    elif args.cmd == "asset-search":\n'
            '        from core import asset_index\n'
            '        idx = json.loads(Path(args.index).read_text(encoding="utf-8"))\n'
            '        _print({"query": args.query, "results": asset_index.search(idx, args.query, '
            'args.top, args.no_sharealike)})\n'
            '        return 0\n'
            '    elif args.cmd == "match-check":\n'
            '        from core import broll_match as bm\n'
            '        sg = json.loads(Path(args.suggestions).read_text(encoding="utf-8"))\n'
            '        exp = [x["id"] for x in '
            'json.loads(Path(args.scenes).read_text(encoding="utf-8"))["scenes"]] if args.scenes '
            'else None\n'
            '        r = bm.qa(sg, exp, sg.get("top", 3))\n'
            '        if args.sheet and args.scenes and args.assets:\n'
            '            r["sheet"] = bm.review_sheet(sg, args.assets, '
            'json.loads(Path(args.scenes).read_text(encoding="utf-8")),\n'
            '                                         args.sheet, args.only.split(",") if '
            'args.only else None)\n'
            '        _print(r)\n'
            '        return 0 if r["structure"] == "PASS" else 1\n'
            '    elif args.cmd == "approve-broll":\n'
            '        from core import broll_match as bm\n'
            '        import datetime as _dt, hashlib as _hl\n'
            '        sg_bytes = Path(args.suggestions).read_bytes()\n'
            '        sg = json.loads(sg_bytes.decode("utf-8"))\n'
            '        idx = json.loads(Path(args.index).read_text(encoding="utf-8"))\n'
            '        sc = json.loads(Path(args.scenes).read_text(encoding="utf-8"))\n'
            '        try:\n'
            '            picks = bm.choices_from_spec(sg, args.choices) if args.choices is not '
            'None else bm.interactive_choices(sg)\n'
            '        except ValueError as e:\n'
            '            _print({"ok": False, "problems": [str(e)]})\n'
            '            return 1\n'
            '        r = bm.approve(None, idx, args.out, sc, args.allow_missing, picks=picks)\n'
            '        if r["ok"]:\n'
            '            ap = Path(args.approved or '
            'Path(args.out).with_name("approved_broll.json"))\n'
            '            letters = {s_["scene_id"]: {c["path"]: bm.LETTERS[k] for k, c in '
            'enumerate(s_["candidates"])} for s_ in sg["scenes"]}\n'
            '            ap.write_text(json.dumps({"schema": "kang.approved_broll/1.0", '
            '"approved_by": "user",\n'
            '                "approved_at": '
            '_dt.datetime.now().astimezone().isoformat(timespec="seconds"),\n'
            '                "suggestions_sha256": _hl.sha256(sg_bytes).hexdigest(), '
            '"visuals_csv": str(args.out),\n'
            '                "scenes": [{"scene": k, "options": [letters.get(k, {}).get(a, "?") '
            'for a in v], "assets": v}\n'
            '                           for k, v in picks.items()],\n'
            '                "skipped": [s_["scene_id"] for s_ in sg["scenes"] if s_["scene_id"] '
            'not in picks]},\n'
            '                ensure_ascii=False, indent=1) + "\\n", encoding="utf-8")\n'
            '            r["approved_broll"] = str(ap)\n'
            '        _print(r)\n'
            '        return 0 if r["ok"] else 1\n'
            '    elif args.cmd in ("match", "match-approve"):\n'
            '        from core import broll_match as bm\n'
            '        idx = json.loads(Path(args.index).read_text(encoding="utf-8"))\n'
            '        sc = json.loads(Path(args.scenes).read_text(encoding="utf-8"))\n'
            '        if args.cmd == "match-approve":\n'
            '            r = bm.approve(args.suggestions_csv, idx, args.out, sc, '
            'args.allow_missing)\n'
            '            _print(r)\n'
            '            return 0 if r["ok"] else 1\n'
            '        words = json.loads(Path(args.words).read_text(encoding="utf-8")) if '
            'args.words else None\n'
            '        sg = bm.match(sc, idx, words, args.top, '
            'exclude_sharealike=args.no_sharealike)\n'
            '        Path(args.out + ".json").write_text(json.dumps(sg, ensure_ascii=False, '
            'indent=1) + "\\n", encoding="utf-8")\n'
            '        bm.write_csv(sg, args.out + ".csv")\n'
            '        r = {"ok": True, "json": args.out + ".json", "csv": args.out + ".csv", '
            '"scenes": len(sg["scenes"]),\n'
            '             "needs_manual": [s["scene_id"] for s in sg["scenes"] if '
            's["needs_manual"]],\n'
            '             "qa": bm.qa(sg, [x["id"] for x in sc["scenes"]], args.top)}\n'
            '        if args.eval_csv:\n'
            '            import csv as _csv\n'
            '            used = {}\n'
            '            with open(args.eval_csv, encoding="utf-8-sig", newline="") as f:\n'
            '                for row in _csv.DictReader(f):\n'
            '                    if row.get("asset"):\n'
            '                        used.setdefault(row["scene"], '
            'set()).add(Path(row["asset"]).name)\n'
            '            ev = bm.hit_at_k(sg, used)\n'
            '            r["eval"] = {k: v for k, v in ev.items() if k != "rows"} | {"misses": '
            '[x["scene"] for x in ev["rows"] if not x["hit"]]}\n'
            '        _print(r)\n'
            '        return 0\n'
            '    elif args.cmd == "visuals":\n'
            '        r = pipeline.visual_preview(args.media, args.words, args.scenes, '
            'args.visuals_csv, args.assets,\n'
            '                                    scene_ids=args.preview.split(",") if args.preview '
            'else None,\n'
            '                                    approver=_approver(args), fps=args.fps, '
            'win_root=args.win_root)\n'
            '        _print(r)\n'
            '        if r.get("error"):\n'
            '            return 1\n'
            '        return 2 if r.get("visual_selection") else 0\n'
            '    elif args.cmd == "review":\n'
            '        r = pipeline.review_transcript(args.media, args.words, manual=args.phrase, '
            'clips=not args.no_clips)\n'
            '        _print(r)\n'
            '        return 0\n'
            '    elif args.cmd == "apply-review":\n'
            '        r = pipeline.apply_transcript_review(args.words, args.review_csv, '
            'approver=_approver(args), source=args.source)\n'
            '        _print(r)\n'
            '        if not r["ok"]:\n'
            '            return 1\n'
            '        return 0 if r["transcript_accepted"] else 2\n'
            '    elif args.cmd == "second-opinion":\n'
            '        r = pipeline.second_opinion(args.words, args.chunks, '
            'review_csv=args.review_csv)\n'
            '    elif args.cmd == "consensus":\n'
            '        r = pipeline.consensus_review(args.words, args.timed, args.text, '
            'review_csv=args.review_csv, names_path=args.names, media=args.audio)\n'
            '    elif args.cmd == "hinglish":\n'
            '        from core.hinglish import run_hinglish\n'
            '        r = run_hinglish(args.words, args.out, yapper_dir=args.yapper_dir, '
            'table_path=args.table,\n'
            '                         review_csv=args.review_csv, '
            'apply_review_fixes=args.apply_review_fixes, other_asr=args.other_asr)\n'
            '    elif args.cmd == "hinglish-layer":\n'
            '        from core.hinglish_review import hinglish_layer_run\n'
            '        r = hinglish_layer_run(args.words, media=args.media, '
            'yapper_dir=args.yapper_dir, table_path=args.table,\n'
            '                               master_review_csv=args.master_review_csv, '
            'block_s=args.block_s,\n'
            '                               clips=not args.no_clips, version=args.version, '
            'other_asr=args.other_asr)\n'
            '    elif args.cmd == "review-hinglish":\n'
            '        from core.hinglish_review import review_pack\n'
            '        r = review_pack(args.layer, media=args.media, table_path=args.table, '
            'block_s=args.block_s, other_asr=args.other_asr)\n'
            '        r.pop("rows", None)\n'
            '    elif args.cmd == "apply-hinglish-review":\n'
            '        from core.hinglish_review import apply_hinglish_review, bulk_ok_csv\n'
            '        csv_in = args.review_csv\n'
            '        if args.bulk_ok_by_user:\n'
            '            csv_in = str(bulk_ok_csv(args.review_csv, args.bulk_ok_by_user))\n'
            '        r = apply_hinglish_review(args.layer, csv_in, allow_partial=args.partial, '
            'out=args.out)\n'
            '        if not r.get("ok"):\n'
            '            r["error"] = "not applied"\n'
            '    elif args.cmd == "fetch-plan":\n'
            '        from core.commons_fetch import plan_queries\n'
            '        r = plan_queries(args.scenes, args.out, place=args.place)\n'
            '    elif args.cmd == "fetch":\n'
            '        from core.commons_fetch import fetch, read_queries\n'
            '        qs = read_queries(args.queries_csv) if args.queries_csv else []\n'
            '        qs += [{"scene": "", "query": q} for q in args.queries]\n'
            '        r = fetch(qs, args.out, per_query=args.per_query, '
            'no_sharealike=args.no_sharealike,\n'
            '                  min_width=args.min_width, dry_run=args.dry_run) if qs else '
            '{"error": "give --queries or --queries-csv"}\n'
            '    elif args.cmd == "fetch-pool":\n'
            '        from core.commons_fetch import fetch_pool\n'
            '        r = fetch_pool(args.scenes, args.out, args.pool, per_scene=args.per_scene, '
            'max_pool=args.max_pool,\n'
            '                       min_score=args.min_score, no_sharealike=args.no_sharealike, '
            'min_width=args.min_width,\n'
            '                       dry_run=args.dry_run, pool_cache=args.pool_cache)\n'
            '    elif args.cmd == "package-captions":\n'
            '        from core.package_captions import repackage_captions\n'
            '        r = repackage_captions(args.package, args.srt, old_tag=args.old_tag, '
            'extras=args.extra, caption_note=args.note)\n'
            '    elif args.cmd == "reels":\n'
            '        from core.captions_adapter import make_reel\n'
            '        try:\n'
            '            r = make_reel(args.audio, args.layer, style=args.style, fmt=args.format, '
            'start=args.start, dur=args.dur,\n'
            '                          visuals=args.visuals, allow_draft=args.allow_draft, '
            'render=not args.no_render)\n'
            '        except (PermissionError, ValueError, RuntimeError) as exc:\n'
            '            r = {"error": str(exc)}\n'
            '    elif args.cmd == "palette":\n'
            '        r = pipeline.palette_from_screenshot(args.image, k=args.k)\n'
            '    _print(r)\n'
            '    return exit_code(args.cmd, r)\n'
            '\n'
            '\n'
            'def exit_code(cmd: str, r: dict) -> int:\n'
            '    """0 = done and QA passed, 1 = failed/QA failed, 2 = stopped at an approval gate '
            '(not a success)."""\n'
            '    if r.get("error"):\n'
            '        return 1\n'
            '    for key in ("qa", "report"):  # subs/promo/formats use "qa"; audio uses "report"\n'
            '        if isinstance(r.get(key), dict) and not r[key].get("pass", False):\n'
            '            return 1\n'
            '    if cmd == "cut":\n'
            '        if r.get("truncated_words"):\n'
            '            return 1\n'
            '        if not r.get("output"):\n'
            '            return 2\n'
            '        return 0 if r.get("duration_check", {}).get("pass", False) else 1\n'
            '    if cmd in ("subs", "promo") and not r.get("final"):\n'
            '        return 2\n'
            '    if cmd == "package" and not r.get("package"):\n'
            '        return 2\n'
            '    return 0\n'
            '\n'
            '\n'
            'if __name__ == "__main__":\n'
            '    raise SystemExit(main())\n',
 'presets/educational.json': '{\n'
                             '  "font_role": "hindi",\n'
                             '  "outline_px": 0,\n'
                             '  "shadow_px": 0,\n'
                             '  "uppercase": false,\n'
                             '  "max_chars": 32,\n'
                             '  "max_words": 5,\n'
                             '  "min_words": 2,\n'
                             '  "max_dur": 2.0,\n'
                             '  "pause_break": 0.35,\n'
                             '  "highlight": "kf",\n'
                             '  "position": "bottom",\n'
                             '  "bold": true,\n'
                             '  "box": true,\n'
                             '  "description": "Educational: boxed background for readability",\n'
                             '  "size_pct": 4.0,\n'
                             '  "name": "educational"\n'
                             '}',
 'presets/promo.json': '{\n'
                       '  "font_role": "hindi",\n'
                       '  "outline_px": 6,\n'
                       '  "shadow_px": 2,\n'
                       '  "uppercase": false,\n'
                       '  "max_chars": 24,\n'
                       '  "max_words": 3,\n'
                       '  "min_words": 2,\n'
                       '  "max_dur": 2.0,\n'
                       '  "pause_break": 0.35,\n'
                       '  "highlight": "pop",\n'
                       '  "position": "bottom",\n'
                       '  "bold": true,\n'
                       '  "box": false,\n'
                       '  "description": "Promo: big punchy text, word pop (Option B)",\n'
                       '  "size_pct": 5.4,\n'
                       '  "name": "promo"\n'
                       '}',
 'presets/travel.json': '{\n'
                        '  "font_role": "hindi",\n'
                        '  "outline_px": 4,\n'
                        '  "shadow_px": 4,\n'
                        '  "uppercase": false,\n'
                        '  "max_chars": 32,\n'
                        '  "max_words": 4,\n'
                        '  "min_words": 2,\n'
                        '  "max_dur": 2.0,\n'
                        '  "pause_break": 0.35,\n'
                        '  "highlight": "kf",\n'
                        '  "position": "bottom",\n'
                        '  "bold": true,\n'
                        '  "box": false,\n'
                        '  "description": "Travel vlogs: lighter outline, soft shadow",\n'
                        '  "size_pct": 4.2,\n'
                        '  "name": "travel"\n'
                        '}',
 'presets/vox_bold_yellow.json': '{\n'
                                 '  "font_role": "hindi",\n'
                                 '  "outline_px": 6,\n'
                                 '  "shadow_px": 2,\n'
                                 '  "uppercase": false,\n'
                                 '  "max_chars": 32,\n'
                                 '  "max_words": 4,\n'
                                 '  "min_words": 2,\n'
                                 '  "max_dur": 2.0,\n'
                                 '  "pause_break": 0.35,\n'
                                 '  "highlight": "kf",\n'
                                 '  "position": "bottom",\n'
                                 '  "bold": true,\n'
                                 '  "box": false,\n'
                                 '  "description": "Talking-head reels: bold white, yellow karaoke '
                                 'highlight",\n'
                                 '  "size_pct": 4.6,\n'
                                 '  "name": "vox_bold_yellow"\n'
                                 '}',
 'requirements.txt': '# Core (Phase 0-3, offline, CPU)\n'
                     'numpy>=1.24\n'
                     'pydantic>=2.5,<3\n'
                     'pillow>=10.0\n'
                     'fonttools>=4.40\n'
                     '# Transcription (Phase 1) - GPU on Kaggle, CPU int8 works for short tests\n'
                     'faster-whisper>=1.0\n'
                     '# API service\n'
                     'fastapi>=0.110,<1.0\n'
                     'uvicorn[standard]>=0.27,<1.0\n'
                     'python-multipart>=0.0.9\n'
                     '# Legacy silence cutter (engine.cut_silence); F03 uses FFmpeg silencedetect '
                     'instead\n'
                     'auto-editor>=26.0\n'
                     '# Tests\n'
                     'pytest>=7.4\n'
                     'karaoke-subs>=0.1  # MIT, Reels caption adapter (optional: only for kang.py '
                     'reels)\n',
 'scripts/__init__.py': '',
 'scripts/fetch_fonts.py': '"""Download the open-licensed fonts Kang needs into ./fonts '
                           '(idempotent).\n'
                           '\n'
                           'Mukta ExtraBold (OFL, Devanagari + Latin) for Hinglish subtitles, Noto '
                           'Sans Devanagari (OFL) fallback, Montserrat Black (OFL) for headings.\n'
                           '"""\n'
                           'from __future__ import annotations\n'
                           '\n'
                           'import sys\n'
                           'import urllib.request\n'
                           'from pathlib import Path\n'
                           '\n'
                           'ROOT = Path(__file__).resolve().parent.parent\n'
                           'FONTS = {\n'
                           '    "Mukta-ExtraBold.ttf":\n'
                           '        '
                           '"https://github.com/google/fonts/raw/main/ofl/mukta/Mukta-ExtraBold.ttf",\n'
                           '    "NotoSansDevanagari-Bold.ttf":\n'
                           '        '
                           '"https://github.com/notofonts/notofonts.github.io/raw/main/fonts/NotoSansDevanagari/hinted/ttf/NotoSansDevanagari-Bold.ttf",\n'
                           '    "Montserrat-Black.ttf":\n'
                           '        '
                           '"https://github.com/JulietaUla/Montserrat/raw/master/fonts/ttf/Montserrat-Black.ttf",\n'
                           '}\n'
                           '# Roadmap v2 §11.2: licence file per font family '
                           '(tests/test_licences.py checks they exist)\n'
                           'LICENCES = {\n'
                           '    "OFL-Mukta.txt": '
                           '"https://raw.githubusercontent.com/google/fonts/main/ofl/mukta/OFL.txt",\n'
                           '    "OFL-Noto.txt": '
                           '"https://raw.githubusercontent.com/notofonts/devanagari/main/OFL.txt",\n'
                           '    "OFL-Montserrat.txt": '
                           '"https://raw.githubusercontent.com/JulietaUla/Montserrat/master/OFL.txt",\n'
                           '}\n'
                           '\n'
                           '\n'
                           'def fetch(dest: Path = ROOT / "fonts") -> int:\n'
                           '    dest.mkdir(parents=True, exist_ok=True)\n'
                           '    rc = 0\n'
                           '    for name, url in {**FONTS, **LICENCES}.items():\n'
                           '        p = dest / name\n'
                           '        if p.is_file() and p.stat().st_size > (1000 if '
                           "name.endswith('.txt') else 10000):\n"
                           '            print(f"[fonts] ok  {name}")\n'
                           '            continue\n'
                           '        try:\n'
                           '            urllib.request.urlretrieve(url, p)\n'
                           '            print(f"[fonts] got {name} ({p.stat().st_size} bytes)")\n'
                           '        except Exception as exc:  # pragma: no cover\n'
                           '            print(f"[fonts] FAILED {name}: {exc}", file=sys.stderr)\n'
                           '            rc = 1\n'
                           '    return rc\n'
                           '\n'
                           '\n'
                           'if __name__ == "__main__":\n'
                           '    raise SystemExit(fetch())\n',
 'setup_kaggle.py': '"""\n'
                    'Kaggle Setup & Bootstrapper for AI Video Engine\n'
                    'Stage 1: Reproducible Environment & Recovery\n'
                    '\n'
                    'This script performs an idempotent setup in Kaggle:\n'
                    '1. Installs Hindi Devanagari fonts (Noto Sans Devanagari) so FFmpeg renders '
                    'Hindi subtitles cleanly.\n'
                    '2. Installs faster-whisper, auto-editor, and other required Python packages.\n'
                    '3. Creates the standard directory tree under '
                    '/kaggle/working/ai-video-engine.\n'
                    '4. Recovers the voiceover file from /kaggle/input if missing post-reset.\n'
                    '5. Verifies GPU, dependencies, fonts, and audio file integrity.\n'
                    '"""\n'
                    'import os\n'
                    'import sys\n'
                    'import shutil\n'
                    'import subprocess\n'
                    'import platform\n'
                    'import urllib.request\n'
                    'from pathlib import Path\n'
                    '\n'
                    'BASE_DIR = Path(os.environ.get("ENGINE_BASE_DIR", '
                    '"/kaggle/working/ai-video-engine"))\n'
                    'INPUT_ROOT = Path(os.environ.get("KAGGLE_INPUT_DIR", "/kaggle/input"))\n'
                    '\n'
                    'def run_cmd(cmd, desc="", check=True):\n'
                    '    """cmd as a list runs WITHOUT a shell (safe for \'pkg>=1.0\' specifiers); '
                    'a str keeps shell=True."""\n'
                    '    print(f"--> {desc}...")\n'
                    '    res = subprocess.run(cmd, shell=isinstance(cmd, str), '
                    'capture_output=True, text=True, encoding="utf-8", errors="replace")\n'
                    '    if check and res.returncode != 0:\n'
                    '        print(f"ERROR: {desc} failed with code {res.returncode}")\n'
                    '        print(res.stderr[:2000])\n'
                    '        raise RuntimeError(f"Step failed: {desc}")\n'
                    '    return res\n'
                    '\n'
                    '\n'
                    'def install_cloudflared():\n'
                    '    """Install the official Cloudflare quick-tunnel binary if it is '
                    'missing."""\n'
                    '    existing = shutil.which("cloudflared")\n'
                    '    if existing:\n'
                    '        print(f"--> cloudflared already available: {existing}")\n'
                    '        return existing\n'
                    '\n'
                    '    machine = platform.machine().lower()\n'
                    '    asset_by_machine = {\n'
                    '        "x86_64": "cloudflared-linux-amd64",\n'
                    '        "amd64": "cloudflared-linux-amd64",\n'
                    '        "aarch64": "cloudflared-linux-arm64",\n'
                    '        "arm64": "cloudflared-linux-arm64",\n'
                    '    }\n'
                    '    asset = asset_by_machine.get(machine)\n'
                    '    if not asset:\n'
                    '        raise RuntimeError(f"Unsupported architecture for cloudflared: '
                    '{machine}")\n'
                    '\n'
                    '    install_path = Path("/usr/local/bin/cloudflared")\n'
                    '    if not os.access(install_path.parent, os.W_OK):\n'
                    '        install_path = Path.home() / ".local" / "bin" / "cloudflared"\n'
                    '        install_path.parent.mkdir(parents=True, exist_ok=True)\n'
                    '\n'
                    '    url = '
                    'f"https://github.com/cloudflare/cloudflared/releases/latest/download/{asset}"\n'
                    '    print(f"--> Downloading cloudflared from official release: {url}")\n'
                    '    try:\n'
                    '        urllib.request.urlretrieve(url, install_path)\n'
                    '        install_path.chmod(0o755)\n'
                    '    except Exception as exc:\n'
                    '        install_path.unlink(missing_ok=True)\n'
                    '        raise RuntimeError(f"Could not install cloudflared from {url}: '
                    '{exc}") from exc\n'
                    '\n'
                    '    # Ensure a user-local install is discoverable by start.py subprocesses.\n'
                    '    os.environ["PATH"] = '
                    'f"{install_path.parent}{os.pathsep}{os.environ.get(\'PATH\', \'\')}"\n'
                    '    print(f"--> cloudflared installed at: {install_path}")\n'
                    '    return str(install_path)\n'
                    '\n'
                    'def setup_environment():\n'
                    '    print("=" * 60)\n'
                    '    print("AI VIDEO ENGINE - KAGGLE ENVIRONMENT SETUP (STAGE 1)")\n'
                    '    print("=" * 60)\n'
                    '\n'
                    '    # 1. Install System Fonts for Devanagari / Hindi subtitles\n'
                    '    # (Prevents tofu / blank square boxes in FFmpeg subtitle rendering)\n'
                    '    if os.path.exists("/usr/bin/apt-get"):\n'
                    '        run_cmd(\n'
                    '            "apt-get update -qq && apt-get install -y -qq fonts-noto-core '
                    'fonts-noto-extra ffmpeg",\n'
                    '            desc="Installing Hindi Devanagari fonts and FFmpeg"\n'
                    '        )\n'
                    '        run_cmd("fc-cache -f", desc="Refreshing font cache", check=False)\n'
                    '    else:\n'
                    '        print("--> apt-get not present (non-Debian / local environment). '
                    'Skipping apt-get.")\n'
                    '\n'
                    '    # 2. Install Python Dependencies\n'
                    '    packages = ["faster-whisper", "auto-editor", "fastapi", "uvicorn", '
                    '"python-multipart",\n'
                    '                "numpy", "pydantic>=2.5", "pillow", "fonttools", "pytest"]\n'
                    '    # list form: with shell=True, "pydantic>=2.5" was parsed as a redirect to '
                    'a file named "=2.5"\n'
                    '    run_cmd(\n'
                    '        [sys.executable, "-m", "pip", "install", "-q", *packages],\n'
                    '        desc="Installing Python dependencies (faster-whisper, auto-editor)"\n'
                    '    )\n'
                    '\n'
                    '    # 3. Install Cloudflare quick-tunnel binary for start.py\n'
                    '    # cloudflared is only needed for start.py (public tunnel). Offline Kaggle '
                    'sessions must still finish setup.\n'
                    '    try:\n'
                    '        install_cloudflared()\n'
                    '    except RuntimeError as exc:\n'
                    '        print(f"--> WARNING: cloudflared not installed ({exc}). start.py '
                    'tunnel unavailable; engine/tests unaffected.")\n'
                    '\n'
                    '    # 4. Create Project Directory Tree\n'
                    '    folders = [\n'
                    '        BASE_DIR / "core",\n'
                    '        BASE_DIR / "configs" / "presets",\n'
                    '        BASE_DIR / "storage" / "inputs",\n'
                    '        BASE_DIR / "storage" / "temp",\n'
                    '        BASE_DIR / "storage" / "outputs",\n'
                    '    ]\n'
                    '    for f in folders:\n'
                    '        f.mkdir(parents=True, exist_ok=True)\n'
                    '    print(f"--> Project directories verified under: {BASE_DIR}")\n'
                    '\n'
                    '    # 5. Deterministic Audio Discovery & Recovery from /kaggle/input\n'
                    '    target_audio = BASE_DIR / "storage" / "inputs" / "voiceover.mp3"\n'
                    '    selected_source = None\n'
                    '\n'
                    '    explicit = os.environ.get("KANG_INPUT_AUDIO", "").strip()\n'
                    '    if explicit:  # any client/audio: explicit path wins, no name guessing\n'
                    '        selected_source = Path(explicit)\n'
                    '        if not selected_source.is_file() or selected_source.stat().st_size == '
                    '0:\n'
                    '            raise FileNotFoundError(f"KANG_INPUT_AUDIO not found or empty: '
                    '{explicit}")\n'
                    '        print(f"--> Using explicit KANG_INPUT_AUDIO: {selected_source}")\n'
                    '        target_audio.parent.mkdir(parents=True, exist_ok=True)\n'
                    '        if selected_source.resolve() != target_audio.resolve():  # same file: '
                    'nothing to copy\n'
                    '            shutil.copy2(selected_source, target_audio)\n'
                    '    elif target_audio.is_file() and target_audio.stat().st_size > 0:\n'
                    '        print(f"--> Found existing voiceover in working directory: '
                    '{target_audio}")\n'
                    '        selected_source = target_audio\n'
                    '    else:\n'
                    '        print("--> storage/inputs/voiceover.mp3 not found. Performing '
                    'deterministic search in /kaggle/input...")\n'
                    '        input_root = INPUT_ROOT\n'
                    '        if not input_root.exists():\n'
                    '            raise FileNotFoundError("/kaggle/input directory does not exist. '
                    'Are datasets mounted?")\n'
                    '\n'
                    '        # Step 4a: Exact match for known Dwarika audio\n'
                    '        exact_matches = list(input_root.rglob("6 Dwarika enhanced.mp3"))\n'
                    '        if exact_matches:\n'
                    '            selected_source = exact_matches[0]\n'
                    '            print(f"--> Unambiguous exact match found: {selected_source}")\n'
                    '        else:\n'
                    '            # Step 4b: Search for Dwarika / voiceover candidates\n'
                    '            all_audios = [\n'
                    '                p for p in input_root.rglob("*")\n'
                    '                if p.is_file() and p.suffix.lower() in [".mp3", ".wav", '
                    '".m4a", ".aac"]\n'
                    '            ]\n'
                    '            dwarika_candidates = [\n'
                    '                p for p in all_audios\n'
                    '                if "dwarika" in p.name.lower() or "voiceover" in '
                    'p.name.lower()\n'
                    '            ]\n'
                    '\n'
                    '            if len(dwarika_candidates) == 1:\n'
                    '                selected_source = dwarika_candidates[0]\n'
                    '                print(f"--> Unambiguous Dwarika candidate found: '
                    '{selected_source}")\n'
                    '            elif len(dwarika_candidates) > 1:\n'
                    '                cand_list = "\\n  - ".join(f"{p} '
                    '({round(p.stat().st_size/(1024*1024), 2)} MB)" for p in dwarika_candidates)\n'
                    '                raise RuntimeError(\n'
                    '                    f"AMBIGUOUS AUDIO: Multiple Dwarika/voiceover candidates '
                    'found in /kaggle/input:\\n  - {cand_list}\\n"\n'
                    '                    f"Stopping automatically. Please specify which file to '
                    'use rather than guessing."\n'
                    '                )\n'
                    '            else:\n'
                    '                other_list = "\\n  - ".join(f"{p} '
                    '({round(p.stat().st_size/(1024*1024), 2)} MB)" for p in all_audios[:5])\n'
                    '                raise FileNotFoundError(\n'
                    '                    f"Expected \'6 Dwarika enhanced.mp3\' not found in '
                    '/kaggle/input.\\n"\n'
                    '                    f"Other files found:\\n  - {other_list if other_list else '
                    '\'None\'}\\n"\n'
                    '                    f"Please attach the Dwarika voiceover dataset to your '
                    'Kaggle session."\n'
                    '                )\n'
                    '\n'
                    '        print(f"--> Copying {selected_source} -> {target_audio}...")\n'
                    '        shutil.copy2(selected_source, target_audio)\n'
                    '\n'
                    '    # 6. Verify Environment\n'
                    '    print("\\n" + "=" * 60)\n'
                    '    print("STAGE 1 VERIFICATION REPORT")\n'
                    '    print("=" * 60)\n'
                    '    # Check PyTorch & GPU\n'
                    '    try:\n'
                    '        import torch\n'
                    '        cuda_ok = torch.cuda.is_available()\n'
                    '        gpu_name = torch.cuda.get_device_name(0) if cuda_ok else "None (CPU '
                    'only)"\n'
                    '        print(f"CUDA Available: {cuda_ok} | Device: {gpu_name}")\n'
                    '    except Exception as e:\n'
                    '        print(f"PyTorch Check: {e}")\n'
                    '\n'
                    '    # Check faster-whisper\n'
                    '    try:\n'
                    '        import faster_whisper\n'
                    '        print(f"faster-whisper: Installed (v{faster_whisper.__version__})")\n'
                    '    except Exception as e:\n'
                    '        print(f"faster-whisper: FAILED ({e})")\n'
                    '\n'
                    '    # Check auto-editor\n'
                    '    ae = shutil.which("auto-editor")\n'
                    '    print(f"auto-editor: {\'Available at \' + ae if ae else \'NOT FOUND\'}")\n'
                    '\n'
                    '    # Check ffmpeg\n'
                    '    ff = shutil.which("ffmpeg")\n'
                    '    print(f"ffmpeg: {\'Available at \' + ff if ff else \'NOT FOUND\'}")\n'
                    '\n'
                    '    # Check cloudflared\n'
                    '    cf = shutil.which("cloudflared")\n'
                    '    print(f"cloudflared: {\'Available at \' + cf if cf else \'NOT FOUND\'}")\n'
                    '\n'
                    '    # Check Hindi font\n'
                    '    font_res = subprocess.run("fc-match \'Noto Sans Devanagari\'", '
                    'shell=True, capture_output=True, text=True, encoding="utf-8", '
                    'errors="replace")\n'
                    '    out_font = font_res.stdout.strip()\n'
                    '    print(f"Devanagari Font: {out_font if out_font else \'NOT FOUND\'}")\n'
                    '\n'
                    '    # Check Voiceover Audio\n'
                    '    if target_audio.is_file():\n'
                    '        size_bytes = target_audio.stat().st_size\n'
                    '        size_mb = round(size_bytes / (1024 * 1024), 2)\n'
                    '        dur_sec = "Unknown"\n'
                    '        dur_min = "Unknown"\n'
                    '        if shutil.which("ffprobe"):\n'
                    '            dur_res = subprocess.run(\n'
                    '                ["ffprobe", "-v", "error", "-show_entries", '
                    '"format=duration", "-of", "default=nw=1:nk=1", str(target_audio)],\n'
                    '                capture_output=True, text=True, encoding="utf-8", '
                    'errors="replace"\n'
                    '            )\n'
                    '            dur_str = dur_res.stdout.strip()\n'
                    '            if dur_str:\n'
                    '                total_s = float(dur_str)\n'
                    '                dur_sec = round(total_s, 2)\n'
                    '                dur_min = round(total_s / 60, 2)\n'
                    '\n'
                    '        print("\\n--- DETERMINISTIC AUDIO INSPECTION ---")\n'
                    '        print(f"Selected Audio Path: {target_audio}")\n'
                    '        print(f"File Size          : {size_mb} MB ({size_bytes} bytes)")\n'
                    '        print(f"Audio Duration     : {dur_min} min ({dur_sec} seconds)")\n'
                    '        print(f"Expected Reference : ~11.54 min (~692s) | ~10.57 MB")\n'
                    '        if dur_min != "Unknown":\n'
                    '            match_ok = 11.0 <= float(dur_min) <= 12.0\n'
                    '            print(f"Integrity Check    : {\'PASS (Duration matches Dwarika '
                    "voiceover)' if match_ok else 'WARNING (Duration differs from ~11.54 "
                    'min)\'}")\n'
                    '        print("---------------------------------------")\n'
                    '    else:\n'
                    '        print("Audio File: Pending (not yet present in '
                    'storage/inputs/voiceover.mp3)")\n'
                    '    print("=" * 60)\n'
                    '\n'
                    'if __name__ == "__main__":\n'
                    '    setup_environment()\n'
                    '\n',
 'start.py': '"""Launch the API and Cloudflare quick tunnel as detached background processes."""\n'
             'from __future__ import annotations\n'
             '\n'
             'import os\n'
             'import re\n'
             'import secrets\n'
             'import shutil\n'
             'import subprocess\n'
             'import sys\n'
             'import time\n'
             'from pathlib import Path\n'
             '\n'
             'ROOT = Path(__file__).resolve().parent\n'
             'HOST = os.environ.get("ENGINE_HOST", "0.0.0.0")\n'
             'PORT = int(os.environ.get("ENGINE_PORT", "8000"))\n'
             'URL_RE = re.compile(r"https://[A-Za-z0-9.-]+\\.trycloudflare\\.com")\n'
             'STARTUP_TIMEOUT_SECONDS = int(os.environ.get("ENGINE_TUNNEL_STARTUP_TIMEOUT", '
             '"90"))\n'
             '\n'
             '\n'
             'def _stop_process(proc: subprocess.Popen | None) -> None:\n'
             '    if proc is not None and proc.poll() is None:\n'
             '        proc.terminate()\n'
             '        try:\n'
             '            proc.wait(timeout=5)\n'
             '        except subprocess.TimeoutExpired:\n'
             '            proc.kill()\n'
             '\n'
             '\n'
             'def main() -> int:\n'
             '    token = secrets.token_urlsafe(32)\n'
             '    env = os.environ.copy()\n'
             '    env["ENGINE_TOKEN"] = token\n'
             '    env.setdefault("ENGINE_BASE_DIR", str(ROOT))\n'
             '\n'
             '    cloudflared = shutil.which("cloudflared")\n'
             '    if not cloudflared:\n'
             '        print("ERROR: cloudflared not found. Run setup_kaggle.py first.", '
             'file=sys.stderr)\n'
             '        return 2\n'
             '\n'
             '    server_log_path = ROOT / "server.log"\n'
             '    tunnel_log_path = ROOT / "cloudflared.log"\n'
             '    server_log = server_log_path.open("a", encoding="utf-8")\n'
             '    tunnel_log = tunnel_log_path.open("a", encoding="utf-8")\n'
             '    server = None\n'
             '    tunnel = None\n'
             '    success = False\n'
             '    try:\n'
             '        server = subprocess.Popen(\n'
             '            [sys.executable, "-m", "uvicorn", "app:app", "--host", HOST, "--port", '
             'str(PORT), "--app-dir", str(ROOT)],\n'
             '            cwd=ROOT,\n'
             '            env=env,\n'
             '            stdin=subprocess.DEVNULL,\n'
             '            stdout=server_log,\n'
             '            stderr=subprocess.STDOUT,\n'
             '            start_new_session=True,\n'
             '        )\n'
             '        # Give Uvicorn a moment to bind before exposing it through the tunnel.\n'
             '        time.sleep(1.5)\n'
             '        if server.poll() is not None:\n'
             '            print(f"ERROR: API server exited early. Inspect {server_log_path}", '
             'file=sys.stderr)\n'
             '            return 3\n'
             '\n'
             '        tunnel = subprocess.Popen(\n'
             '            [cloudflared, "tunnel", "--url", f"http://127.0.0.1:{PORT}", '
             '"--no-autoupdate"],\n'
             '            cwd=ROOT,\n'
             '            env=env,\n'
             '            stdin=subprocess.DEVNULL,\n'
             '            stdout=tunnel_log,\n'
             '            stderr=subprocess.STDOUT,\n'
             '            start_new_session=True,\n'
             '        )\n'
             '\n'
             '        deadline = time.monotonic() + STARTUP_TIMEOUT_SECONDS\n'
             '        url = None\n'
             '        while time.monotonic() < deadline:\n'
             '            if server.poll() is not None:\n'
             '                print(f"ERROR: API server stopped. Inspect {server_log_path}", '
             'file=sys.stderr)\n'
             '                return 3\n'
             '            if tunnel.poll() is not None:\n'
             '                print(f"ERROR: cloudflared stopped. Inspect {tunnel_log_path}", '
             'file=sys.stderr)\n'
             '                return 4\n'
             '            try:\n'
             '                logs = tunnel_log_path.read_text(encoding="utf-8", '
             'errors="replace")\n'
             '            except OSError:\n'
             '                logs = ""\n'
             '            matches = URL_RE.findall(logs)\n'
             '            if matches:\n'
             '                url = matches[-1]\n'
             '                break\n'
             '            time.sleep(0.5)\n'
             '\n'
             '        if not url:\n'
             '            print(f"ERROR: No tunnel URL appeared within {STARTUP_TIMEOUT_SECONDS}s. '
             'Inspect {tunnel_log_path}", file=sys.stderr)\n'
             '            return 5\n'
             '\n'
             '        # Keep both children detached; this launcher can exit so notebook cells stay '
             'usable.\n'
             '        print(f"URL={url}", flush=True)\n'
             '        print(f"TOKEN={token}", flush=True)\n'
             '        print("WARNING: TOKEN is a secret. Do not share screenshots or logs '
             'containing it.", flush=True)\n'
             '        print(f"Server PID: {server.pid} | Tunnel PID: {tunnel.pid}", flush=True)\n'
             '        print(f"Logs: {server_log_path.name}, {tunnel_log_path.name}", flush=True)\n'
             '        success = True\n'
             '        return 0\n'
             '    finally:\n'
             '        server_log.close()\n'
             '        tunnel_log.close()\n'
             '        # Stop children if startup failed. On success they remain detached in '
             'background.\n'
             '        if not success:\n'
             '            _stop_process(tunnel)\n'
             '            _stop_process(server)\n'
             '\n'
             '\n'
             'if __name__ == "__main__":\n'
             '    raise SystemExit(main())\n',
 'test_30s_sample.py': '"""\n'
                       'AI Video Engine - 30-Second Audio Slice Test (Stage 4)\n'
                       'This script verifies Hindi transcription accuracy and SRT generation\n'
                       'on only the first 30 seconds of the voiceover audio.\n'
                       '"""\n'
                       '\n'
                       'import os\n'
                       'import sys\n'
                       'import subprocess\n'
                       'import shutil\n'
                       'from pathlib import Path\n'
                       '\n'
                       '# Add project root to path\n'
                       'sys.path.insert(0, str(Path(__file__).parent))\n'
                       '\n'
                       'from core.engine import (\n'
                       '    load_audio,\n'
                       '    transcribe,\n'
                       '    refine_transcript,\n'
                       '    chunk_words,\n'
                       '    generate_srt,\n'
                       '    save_outputs,\n'
                       '    BASE,\n'
                       '    DEFAULT_HINDI_PROMPT\n'
                       ')\n'
                       '\n'
                       'SAMPLE_DURATION_SEC = 30\n'
                       'TEST_OUTPUT_DIR = os.path.join(BASE, "storage", "outputs", "test_30s")\n'
                       'HINDI_PROMPT = DEFAULT_HINDI_PROMPT\n'
                       '\n'
                       '\n'
                       'def find_voiceover_audio() -> str:\n'
                       '    """\n'
                       '    Deterministically locate the Dwarika voiceover audio.\n'
                       '    Stops with clear error if ambiguous or missing.\n'
                       '    """\n'
                       '    # Priority 1: Check storage/inputs/voiceover.mp3\n'
                       '    target = os.path.join(BASE, "storage", "inputs", "voiceover.mp3")\n'
                       '    if os.path.isfile(target) and os.path.getsize(target) > 0:\n'
                       '        return target\n'
                       '\n'
                       '    # Priority 2: Check storage/inputs/voice.mp3\n'
                       '    legacy_target = os.path.join(BASE, "storage", "inputs", "voice.mp3")\n'
                       '    if os.path.isfile(legacy_target) and os.path.getsize(legacy_target) > '
                       '0:\n'
                       '        return legacy_target\n'
                       '\n'
                       '    # Priority 3: Search /kaggle/input deterministically\n'
                       '    input_path = Path(os.environ.get("KAGGLE_INPUT_DIR", '
                       '"/kaggle/input"))\n'
                       '    if not input_path.exists():\n'
                       '        raise FileNotFoundError(\n'
                       '            f"Voiceover not found in storage/inputs/voiceover.mp3 and '
                       '/kaggle/input does not exist."\n'
                       '        )\n'
                       '\n'
                       "    # 3a. Exact filename match for '6 Dwarika enhanced.mp3'\n"
                       '    exact = list(input_path.rglob("6 Dwarika enhanced.mp3"))\n'
                       '    if exact:\n'
                       '        return str(exact[0])\n'
                       '\n'
                       '    # 3b. Search for dwarika / voiceover candidates\n'
                       '    all_audios = [\n'
                       '        p for p in input_path.rglob("*")\n'
                       '        if p.is_file() and p.suffix.lower() in [".mp3", ".wav", ".m4a", '
                       '".aac"]\n'
                       '    ]\n'
                       '    dwarika_cand = [\n'
                       '        p for p in all_audios\n'
                       '        if "dwarika" in p.name.lower() or "voiceover" in p.name.lower()\n'
                       '    ]\n'
                       '\n'
                       '    if len(dwarika_cand) == 1:\n'
                       '        return str(dwarika_cand[0])\n'
                       '    elif len(dwarika_cand) > 1:\n'
                       '        cand_list = "\\n  - ".join(f"{p} '
                       '({round(p.stat().st_size/(1024*1024), 2)} MB)" for p in dwarika_cand)\n'
                       '        raise RuntimeError(\n'
                       '            f"AMBIGUOUS AUDIO: Multiple Dwarika/voiceover candidates found '
                       'in /kaggle/input:\\n  - {cand_list}\\n"\n'
                       '            f"Please specify the exact file path to proceed."\n'
                       '        )\n'
                       '    else:\n'
                       '        other_list = "\\n  - ".join(f"{p} '
                       '({round(p.stat().st_size/(1024*1024), 2)} MB)" for p in all_audios[:5])\n'
                       '        raise FileNotFoundError(\n'
                       '            f"Voiceover audio not found in storage/inputs/voiceover.mp3 or '
                       '/kaggle/input.\\n"\n'
                       '            f"Other audio files found:\\n  - {other_list if other_list '
                       'else \'None\'}\\n"\n'
                       '            f"Please mount the Dwarika voiceover dataset."\n'
                       '        )\n'
                       '\n'
                       '\n'
                       'def extract_30s_slice(source_audio: str, slice_path: str, duration: int = '
                       '30) -> str:\n'
                       '    """Extract first N seconds of audio using ffmpeg."""\n'
                       '    os.makedirs(os.path.dirname(slice_path), exist_ok=True)\n'
                       '    cmd = [\n'
                       '        "ffmpeg", "-y", "-v", "error",\n'
                       '        "-i", str(source_audio),\n'
                       '        "-t", str(duration),\n'
                       '        "-c:a", "libmp3lame", "-q:a", "2",\n'
                       '        str(slice_path)\n'
                       '    ]\n'
                       '    subprocess.run(cmd, check=True)\n'
                       '    return slice_path\n'
                       '\n'
                       '\n'
                       'def run_30s_test(model_size: str = "large-v3-turbo"):\n'
                       '    print("=" * 60)\n'
                       '    print(f"RUNNING 30-SECOND HINDI TRANSCRIPTION TEST (Stage 4)")\n'
                       '    print("=" * 60)\n'
                       '\n'
                       '    # 1. Deterministically Locate Audio\n'
                       '    source_audio = find_voiceover_audio()\n'
                       '    sz_bytes = os.path.getsize(source_audio)\n'
                       '    sz_mb = round(sz_bytes / (1024 * 1024), 2)\n'
                       '    dur_str, dur_min, dur_sec = "Unknown", "Unknown", "Unknown"\n'
                       '    if shutil.which("ffprobe"):\n'
                       '        d_res = subprocess.run(\n'
                       '            ["ffprobe", "-v", "error", "-show_entries", "format=duration", '
                       '"-of", "default=nw=1:nk=1", str(source_audio)],\n'
                       '            capture_output=True, text=True, encoding="utf-8", '
                       'errors="replace"\n'
                       '        )\n'
                       '        if d_res.stdout.strip():\n'
                       '            t_s = float(d_res.stdout.strip())\n'
                       '            dur_sec = round(t_s, 2)\n'
                       '            dur_min = round(t_s / 60, 2)\n'
                       '\n'
                       '    print("\\n--- DETERMINISTIC AUDIO SELECTION ---")\n'
                       '    print(f"Selected Audio Path: {source_audio}")\n'
                       '    print(f"File Size          : {sz_mb} MB ({sz_bytes} bytes)")\n'
                       '    print(f"Total Duration     : {dur_min} min ({dur_sec} seconds)")\n'
                       '    print(f"Expected Reference : ~11.54 min (~692s) | ~10.57 MB")\n'
                       '    print("-------------------------------------")\n'
                       '\n'
                       '    # 2. Extract 30s slice\n'
                       '    slice_path = os.path.join(BASE, "storage", "temp", '
                       '"voiceover_30s_test.mp3")\n'
                       '    print(f"Extracting {SAMPLE_DURATION_SEC}s test sample to: '
                       '{slice_path}...")\n'
                       '    extract_30s_slice(source_audio, slice_path, '
                       'duration=SAMPLE_DURATION_SEC)\n'
                       '    print(f"Sample created successfully '
                       '({round(os.path.getsize(slice_path) / 1024, 1)} KB)")\n'
                       '\n'
                       '    # 3. Transcribe only the sample\n'
                       '    print(f"\\nTranscribing 30s sample using {model_size}...")\n'
                       '    audio_data = load_audio(slice_path)\n'
                       '    rows, info = transcribe(\n'
                       '        audio=audio_data,\n'
                       '        language="hi",\n'
                       '        initial_prompt=HINDI_PROMPT,\n'
                       '        model_size=model_size\n'
                       '    )\n'
                       '\n'
                       '    # 3b. Detect and refine suspicious timestamps (>2.5s)\n'
                       '    rows, refine_stats = refine_transcript(\n'
                       '        rows=rows,\n'
                       '        audio_samples=audio_data,\n'
                       '        language="hi",\n'
                       '        initial_prompt=HINDI_PROMPT,\n'
                       '        model_size=model_size\n'
                       '    )\n'
                       '    if refine_stats.get("suspicious_count", 0) > 0:\n'
                       '        print(f"Refinement: {refine_stats[\'suspicious_count\']} '
                       'suspicious words detected, "\n'
                       '              f"{refine_stats[\'corrections_applied\']} corrections '
                       'applied.")\n'
                       '        if refine_stats.get("warnings"):\n'
                       '            for w in refine_stats["warnings"]:\n'
                       '                print(f"  [Warning] {w}")\n'
                       '\n'
                       '    # 4. Generate Subtitles\n'
                       '    chunks = chunk_words(rows)\n'
                       '    srt_p, json_p = save_outputs(rows, TEST_OUTPUT_DIR)\n'
                       '\n'
                       '    print("\\n" + "=" * 60)\n'
                       '    print("30-SECOND TEST RESULTS FOR ACCURACY VERIFICATION")\n'
                       '    print("=" * 60)\n'
                       '    print(f"Detected Language: {info.language} (probability: '
                       '{info.language_probability:.2f})")\n'
                       '    print(f"Total Segments: {len(rows)} | Total Subtitle Lines: '
                       '{len(chunks)}")\n'
                       '    print(f"SRT saved at: {srt_p}")\n'
                       '    print(f"JSON saved at: {json_p}\\n")\n'
                       '\n'
                       '    print("--- RAW SPOKEN TEXT EXTRACTED ---")\n'
                       '    full_text = " ".join(r["text"] for r in rows)\n'
                       '    print(full_text)\n'
                       '\n'
                       '    print("\\n--- FIRST 8 SUBTITLE LINES (TIMED) ---")\n'
                       '    for i, c in enumerate(chunks[:8], 1):\n'
                       '        print(f"[{c[\'start\']:.2f}s -> {c[\'end\']:.2f}s]: '
                       '{c[\'text\']}")\n'
                       '\n'
                       '    print("=" * 60)\n'
                       '    print("Test completed. Please review transcribed text against your '
                       'audio.")\n'
                       '    print("=" * 60)\n'
                       '    return rows, chunks\n'
                       '\n'
                       '\n'
                       'if __name__ == "__main__":\n'
                       '    run_30s_test()\n'
                       '\n'}

for rel, content in FILES.items():
    dest = BASE_DIR / rel
    dest.parent.mkdir(parents=True, exist_ok=True)
    dest.write_text(content, encoding='utf-8')
    print(f'[SYNC] Wrote {dest}')

print('\n[SYNC] Source files written. Fetching open fonts (Noto Sans Devanagari, Montserrat)...')
import subprocess, sys
subprocess.run([sys.executable, str(BASE_DIR / 'scripts' / 'fetch_fonts.py')], check=False)
print('\n[SYNC] Starting Kaggle environment setup...')
setup_path = BASE_DIR / 'setup_kaggle.py'
exec(compile(setup_path.read_text(encoding='utf-8'), str(setup_path), 'exec'), {'__name__': '__main__'})
````

````python
import sys, gc
sys.path.insert(0, "/kaggle/working/ai-video-engine")
from faster_whisper import WhisperModel
import core.engine as engine

audio = engine.load_audio("/kaggle/working/ai-video-engine/storage/temp/voiceover_30s_test.mp3")
PROMPT = "राधे राधे। प्यारे भक्तों, द्वारका, द्वारकाधीश, बेट द्वारका, गोपी तालाब, रुक्मिणी माता मंदिर, नागेश्वर ज्योतिर्लिंग, सुदर्शन सेतु।"

runs = {"turbo": ("large-v3-turbo", None),
        "turbo + prompt": ("large-v3-turbo", PROMPT),
        "large-v3": ("large-v3", None),
        "large-v3 + prompt": ("large-v3", PROMPT)}

for name, (m, p) in runs.items():
    model = WhisperModel(m, device="cuda", compute_type="float16")
    segs, _ = model.transcribe(audio, language="hi", vad_filter=True,
                               condition_on_previous_text=False, initial_prompt=p)
    print(f"\n=== {name} ===")
    print(" ".join(s.text.strip() for s in segs)[:260])
    del model
    gc.collect()
````

````python
import time
model = WhisperModel("large-v3", device="cuda", compute_type="float16")
for wt in (False, True):
    t0 = time.time()
    segs, _ = model.transcribe(audio, language="hi", vad_filter=True,
                               condition_on_previous_text=False,
                               initial_prompt=PROMPT, word_timestamps=wt)
    segs = list(segs)
    print(f"\n=== large-v3 + prompt | word_timestamps={wt} | {time.time()-t0:.1f}s ===")
    print(" ".join(s.text.strip() for s in segs))
    if wt:
        words = [w for s in segs for w in s.words]
        long = sorted(words, key=lambda w: w.end - w.start, reverse=True)[:5]
        print("Sabse lambe words:", [(w.word.strip(), round(w.end - w.start, 1)) for w in long])
````

````python
import time, subprocess, gc
SRC = "/kaggle/working/ai-video-engine/storage/inputs/voiceover.mp3"
S120 = "/kaggle/working/ai-video-engine/storage/temp/voiceover_120s.mp3"
subprocess.run(["ffmpeg", "-y", "-v", "error", "-ss", "0", "-t", "120", "-i", SRC, S120], check=True)
audio120 = engine.load_audio(S120)

configs = {
    "turbo + prompt":     ("large-v3-turbo", PROMPT),
    "large-v3 + prompt":  ("large-v3", PROMPT),
    "large-v3 no prompt": ("large-v3", None),
}
for name, (m, p) in configs.items():
    model = WhisperModel(m, device="cuda", compute_type="float16")
    t0 = time.time()
    segs, _ = model.transcribe(audio120, language="hi", vad_filter=True,
                               condition_on_previous_text=False,
                               initial_prompt=p, word_timestamps=True)
    segs = list(segs)
    sec = time.time() - t0
    words = [w for s in segs for w in s.words]
    long = sorted(words, key=lambda w: w.end - w.start, reverse=True)[:3]
    print(f"\n=== {name} | {sec:.0f}s | words={len(words)} | {len(words)/2:.0f} words/min ===")
    print("Lambe words:", [(w.word.strip(), round(w.end - w.start, 1)) for w in long])
    print("Aakhri words:", " ".join(w.word.strip() for w in words[-8:]))
    del model
    gc.collect()
````

````python
# Existing `large-v3 + prompt` result ko dobara check karne ke liye
# Agar `segs` loop ke baad available nahi hain, ye cell run karo.

model = WhisperModel("large-v3", device="cuda", compute_type="float16")

segs, _ = model.transcribe(
    audio120,
    language="hi",
    vad_filter=True,
    condition_on_previous_text=False,
    initial_prompt=PROMPT,
    word_timestamps=True
)
segs = list(segs)

words = [w for s in segs for w in s.words]

print("=== Suspicious word timings (>2.5 sec) ===")
for w in words:
    duration = w.end - w.start
    if duration > 2.5:
        print(
            f"{w.start:.2f}s - {w.end:.2f}s "
            f"({duration:.2f}s): {w.word.strip()!r}"
        )

print("\n=== Transcript around 10–30 seconds ===")
for s in segs:
    if s.end >= 10 and s.start <= 30:
        print(f"{s.start:.2f}–{s.end:.2f}: {s.text}")
````
